```python
from collections import defaultdict
from typing import Callable, Dict, Any, Set, Tuple, Optional
from langgraph.graph.state import StateNodeSpec
class StateGraphManager:
    def __init__(self, graph):
        self.graph = graph

        # Maintain references to key attributes
        self.nodes = self._nodes()
        self.edges = self._edges()
        self.all_edges = self._all_edges()
        self.is_compiled = self._is_compiled()
        self.input = self._input()
        self.output = self._output()
        self.schema = self._schema()
        self.branches = self._branches()

        # Track whether changes require recompilation
        self.needs_recompile = False

    ## ======== Internal Methods for Accessing Graph Properties ======== ##
    
    def _nodes(self):
        return self.graph.nodes

    def _edges(self):
        return self.graph.edges

    def _all_edges(self):
        return self.graph.edges  # TODO: If different from edges, adjust

    def _is_compiled(self):
        return self.graph.compiled

    def _input(self):
        return self.graph.input

    def _output(self):
        return self.graph.output

    def _schema(self):
        return self.graph.schema

    def _branches(self):
        return self.graph.branches

    ## ======== Graph Navigation Methods ======== ##

    def get_start_node(self) -> Optional[str]:
        """Finds the start node dynamically from edges."""
        for source, target in self.edges:
            if source == "__start__":  
                return target  # Explicit start node
        # If no __start__, find a node without incoming edges
        for node in self.nodes:
            if all(node != target for _, target in self.edges):
                return node
        return None  # No start node found

    def get_end_nodes(self) -> Set[str]:
        """Finds end nodes dynamically from edges and branches."""
        end_nodes = set()
        for source, target in self.edges:
            if target == "__end__":
                end_nodes.add(source)
        # If no explicit __end__, find nodes without outgoing edges
        if not end_nodes:
            end_nodes = {node for node in self.nodes if all(node != source for source, _ in self.edges)}
        return end_nodes

    ## ======== Core Graph Operations ======== ##

    def add_node(self, name: str, node: Callable):
        """ Adds a node but does not immediately recompile. """
        if name in self.nodes:
            raise ValueError(f"Node {name} already exists.")
        self.nodes[name] = node
        self.needs_recompile = True

    def remove_node(self, name: str):
        """ Removes a node, its associated edges, and marks for recompilation. """
        if name not in self.nodes:
            raise ValueError(f"Node {name} does not exist.")

        # Remove node
        del self.nodes[name]

        # Remove edges connected to the node
        self.edges = {edge for edge in self.edges if name not in edge}

        # Remove branches if they reference the node
        if name in self.branches:
            del self.branches[name]

        self.needs_recompile = True

    ## ======== Node Insertion Methods ======== ##

    from langgraph.graph.state import StateNodeSpec

    from langgraph.graph.state import StateNodeSpec

    def insert_node_before(self, existing_node: str, new_node: str, node_callable: Callable):
        """Inserts a node before an existing node, updating edges accordingly."""
        if new_node in self.nodes:
            raise ValueError(f"Node {new_node} already exists.")
    
        # Ensure existing_node has a valid input schema
        existing_node_spec = self.nodes.get(existing_node)
        if not isinstance(existing_node_spec, StateNodeSpec):
            raise ValueError(f"Cannot insert before {existing_node}: Not a valid StateNodeSpec.")
    
        # Use the same input schema as existing_node
        new_node_spec = StateNodeSpec(
            runnable=node_callable,
            metadata=None,
            input=existing_node_spec.input,  # Copy input schema from existing node
            retry_policy=None,
            ends=()
        )
    
        # Find all edges pointing to `existing_node`
        incoming_edges = {src for src, tgt in self.edges if tgt == existing_node}
    
        if not incoming_edges:
            print(f"Warning: {existing_node} is a start node, redirecting to {new_node} instead.")
            self.edges.add((new_node, existing_node))  # Connect new_node to existing node
        else:
            # Remove old incoming edges and redirect through new_node
            for src in incoming_edges:
                self.edges.discard((src, existing_node))
                self.edges.add((src, new_node))
    
            self.edges.add((new_node, existing_node))  # Connect new_node to existing node
    
        # Add new node as a proper StateNodeSpec
        self.nodes[new_node] = new_node_spec
        self.needs_recompile = True
    
    def insert_node_after(self, existing_node: str, new_node: str, node_callable: Callable):
        """Inserts a node after an existing node, updating edges accordingly."""
        if new_node in self.nodes:
            raise ValueError(f"Node {new_node} already exists.")
    
        # Ensure existing_node has a valid input schema
        existing_node_spec = self.nodes.get(existing_node)
        if not isinstance(existing_node_spec, StateNodeSpec):
            raise ValueError(f"Cannot insert after {existing_node}: Not a valid StateNodeSpec.")
    
        # Use the same input schema as existing_node
        new_node_spec = StateNodeSpec(
            runnable=node_callable,
            metadata=None,
            input=existing_node_spec.input,  # Copy input schema
            retry_policy=None,
            ends=()
        )
    
        # Find all edges coming from `existing_node`
        outgoing_edges = {tgt for src, tgt in self.edges if src == existing_node}
    
        if not outgoing_edges:
            print(f"Warning: {existing_node} is an end node, redirecting to {new_node} instead.")
            self.edges.add((existing_node, new_node))  # Add edge from existing node to new node
        else:
            # Remove old outgoing edges and redirect through new_node
            for tgt in outgoing_edges:
                self.edges.discard((existing_node, tgt))
                self.edges.add((new_node, tgt))
    
            self.edges.add((existing_node, new_node))  # Connect existing node to new node
    
        # Add new node as a proper StateNodeSpec
        self.nodes[new_node] = new_node_spec
        self.needs_recompile = True
    def remove_node(self, node_name: str):
        """Removes a node, updates edges, and marks for recompilation."""
        if node_name not in self.nodes:
            raise ValueError(f"Node {node_name} does not exist.")
    
        # Find edges involving the node
        incoming_edges = {src for src, tgt in self.edges if tgt == node_name}
        outgoing_edges = {tgt for src, tgt in self.edges if src == node_name}
    
        # If the node is completely isolated, remove it directly
        del self.nodes[node_name]
    
        # Remove all edges associated with the node
        self.edges = {(src, tgt) for src, tgt in self.edges if src != node_name and tgt != node_name}
    
        # If there were connections, link the previous and next nodes together
        if incoming_edges and outgoing_edges:
            for src in incoming_edges:
                for tgt in outgoing_edges:
                    self.edges.add((src, tgt))
    
        self.needs_recompile = True

    def replace_node(self, old_node: str, new_node: str, node_callable: Callable):
        """Replaces an existing node with a new one while keeping its connections."""
        if old_node not in self.nodes:
            raise ValueError(f"Node {old_node} does not exist.")
        if new_node in self.nodes:
            raise ValueError(f"Node {new_node} already exists.")
    
        # Preserve edges
        incoming_edges = {src for src, tgt in self.edges if tgt == old_node}
        outgoing_edges = {tgt for src, tgt in self.edges if src == old_node}
    
        # Create new node with same input type
        old_node_spec = self.nodes[old_node]
        new_node_spec = StateNodeSpec(
            runnable=node_callable,
            metadata=None,
            input=old_node_spec.input,  # Copy input schema
            retry_policy=None,
            ends=()
        )
    
        # Remove old node
        del self.nodes[old_node]
        self.nodes[new_node] = new_node_spec
    
        # Update edges
        self.edges = {(src, tgt) for src, tgt in self.edges if src != old_node and tgt != old_node}
        for src in incoming_edges:
            self.edges.add((src, new_node))
        for tgt in outgoing_edges:
            self.edges.add((new_node, tgt))
    
        self.needs_recompile = True
    
    def finalize_and_recompile(self):
        """Final step to ensure all modifications are applied before recompiling."""
        if self.needs_recompile:
            print("Recompiling modified graph...")
            self.recompile()
            self.graph.compile()
            print("Recompilation complete.")
        else:
            print("No changes detected. Skipping recompilation.")

        


    ## ======== Compilation Management ======== ##

    def recompile(self):
        """ Recompiles the graph if marked as needing recompilation. """
        if not self.needs_recompile:
            print("No recompilation needed.")
            return None

        print("Recompiling graph...")
        # TODO: Implement actual recompilation logic here
        self.is_compiled = True
        self.needs_recompile = False

    ## ======== Utility & Debugging ======== ##

    def pretty_print(self):
        """ Prints the current state of the graph manager. """
        items = {
            "nodes": self.nodes,
            "edges": self.edges,
            "branches": self.branches,
            "is_compiled": self.is_compiled,
            "needs_recompile": self.needs_recompile,
            "input": self.input,
            "schema": self.schema,
            "output": self.output,
        }
        for key, value in items.items():
            print(f"{key}: {value}")
```
