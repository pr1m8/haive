```python
from enum import Enum
from typing import List
from pydantic import BaseModel, Field

class MessageTag(str, Enum):
    TASK = "task"
    GLOBAL_MEMORY = "global_memory"
    ERROR_REVISION = "error_revision"
    TASK_SPECIFIC_MEMORY = "task_specific_memory"
    AGENT_CONTEXT = "agent_context"
    USER_CONTEXT = "user_context"
    SYSTEM_PROMPT_ADDITION = "system_prompt_addition"

class TagModel(BaseModel):
    """Model for tagging messages with relevant categories."""
    message: str = Field(..., description="Original message content.")
    tags: List[MessageTag] = Field(..., description="List of tags associated with the message.")

    @classmethod
    def add_tag(cls, tag: MessageTag):
        """Allows adding new tags dynamically if necessary."""
        if tag not in MessageTag.__members__.values():
            raise ValueError(f"Invalid tag: {tag}")
        return tag

class SummaryOutput(BaseModel):
    """Preserved model for summarization tasks."""
    summary: str = Field(..., description="Generated summary of the content.")

class TaskMemoryOutput(BaseModel):
    """Extracted task-specific information for in-task memory."""
    task_name: str = Field(..., description="The name of the task this memory belongs to.")
    details: str = Field(..., description="Specific task-related information.")

class GlobalMemoryOutput(BaseModel):
    """Extracts global memory relevant for future tasks."""
    fact: str = Field(..., description="Key fact that should persist across tasks.")
    relevance_score: float = Field(..., gt=0, lt=1, description="Relevance score for global memory retention.")

class ErrorTrackingOutput(BaseModel):
    """Tracks errors, contradictions, or necessary revisions."""
    error_type: MessageTag = Field(..., description="Categorized error type.")
    original_message: str = Field(..., description="Incorrect message content.")
    corrected_message: str = Field(..., description="Revised message content.")

class MessageSelectionOutput(BaseModel):
    """Determines message selection for deep in-task memory with global summaries."""
    message: str = Field(..., description="Message content.")
    action: str = Field(..., description="KEEP, DISCARD, or SUMMARIZE.")
    reason: str = Field(..., description="Why this action was chosen.")

class SystemPromptAdditionOutput(BaseModel):
    """Selects information to be added to the agent's system prompt."""
    content: str = Field(..., description="Important information to be injected into the system prompt.")
    source: MessageTag = Field(..., description="Where this information originated from.")
```

```python
from typing_extensions import TypedDict,Annotated,Any
import operator
class SummaryState(TypedDict):
    """State for the summarizer agent - combining structured outputs with operator.add."""

    # Input documents
    contents: List[Any] = Field(
        ..., description="The contents of the documents."
    )

    # Summaries at different stages
    summaries: Annotated[list, operator.add] = Field(
        default_factory=list, description="Summaries of individual documents."
    )
    collapsed_summaries: List[str] = Field(
        default_factory=list, description="Summaries collapsed to fit within token limits."
    )
    final_summary: str = Field(
        default="", description="Final summary for agent use."
    )

    # Tagged Messages
    tagged_messages: List[TagModel] = Field(
        default_factory=list, description="Categorized messages for better processing."
    )

    # Memory Storage
    task_memory: List[TaskMemoryOutput] = Field(
        default_factory=list, description="In-task memory storage."
    )
    global_memory: List[GlobalMemoryOutput] = Field(
        default_factory=list, description="Long-term knowledge storage."
    )

    # Error Tracking
    errors: List[ErrorTrackingOutput] = Field(
        default_factory=list, description="Tracked errors, contradictions, or necessary revisions."
    )

    # Message Selection
    message_selections: List[MessageSelectionOutput] = Field(
        default_factory=list, description="Which messages to keep, discard, or summarize."
    )

    # System Prompt Enhancements
    system_prompt_additions: List[SystemPromptAdditionOutput] = Field(
        default_factory=list, description="Important information injected into the system prompt."
    )

    # Token Control
    token_count: int = Field(
        default=0, description="Tracks total token count of processed documents."
    )
```

```python
from src.haive.agents.base import AgentArchitectureConfig,AgentArchitecture
from typing import Dict,Literal
from langgraph.graph import START,END
class SummarizerAgentConfig(AgentArchitectureConfig):
    """Configuration for the Summarizer Agent"""
    aug_llm_configs: Dict[str, AugLLMConfig] = Field(default_factory=lambda: {
        "map_chain": map_aug_llm_config,
        "reduce_chain": reduce_augllm_config,
        "task_categorization": task_categorization_config,
        "global_memory": global_memory_config,
        "error_tracking": error_tracking_config,
        "message_selection": message_selection_config,
        "system_prompt_addition": system_prompt_addition_config
    }, description="Configurations for all AugLLMs used in the workflow.")

    token_max: int = Field(default=1000, description="Max token limit before collapsing summaries.")
    state_schema: SummaryState = Field(default=SummaryState)
    should_visualize_graph: bool = True
    visualize_graph_output_name: str = "summarizer_agent_graph.png"

    def build_agent(self):
        return SummarizerAgent(self)
```

```python
from langgraph.types import Send
class SummarizerAgent(AgentArchitecture):
    """SummarizerAgent is responsible for extracting summaries and structured memory from input documents."""

    def __init__(self, config: SummarizerAgentConfig = SummarizerAgentConfig()):
        

        # Token constraints
        self.token_max = config.token_max  

        # Summarization chains (Preserved)
        self.map_chain = compose_runnable(map_aug_llm_config)
        self.reduce_chain = compose_runnable(reduce_augllm_config)

        # New structured processing chains
        
        self.task_categorization_chain = compose_runnable(task_categorization_config)
        self.global_memory_chain = compose_runnable(global_memory_config)
        self.error_tracking_chain = compose_runnable(error_tracking_config)
        self.message_selection_chain = compose_runnable(message_selection_config)
        self.system_prompt_addition_chain = compose_runnable(system_prompt_addition_config)

        super().__init__(config)

    def setup_workflow(self):
        """Construct the StateGraph for structured summarization and memory extraction."""
        
        #self.graph = StateGraph(self.state_schema)
    
        ### ✅ PHASE 1: MESSAGE SELECTION (Must Happen First)
        self.graph.add_node("select_relevant_messages", self.select_relevant_messages)
        
        ### ✅ PHASE 2: CATEGORIZATION & MEMORY PROCESSING (After Selection)
        self.graph.add_node("categorize_messages", self.categorize_messages)
        self.graph.add_node("extract_memory", self.extract_memory)
        self.graph.add_node("track_errors", self.track_errors)
    
        ### ✅ PHASE 3: SUMMARIZATION (Existing Map-Reduce Logic)
        self.graph.add_node("generate_summary", self.generate_summary)
        self.graph.add_node("collect_summaries", self.collect_summaries)
        self.graph.add_node("collapse_summaries", self.collapse_summaries)
        self.graph.add_node("generate_final_summary", self.generate_final_summary)
    
        ### ✅ PHASE 4: FINAL SYSTEM PROMPT UPDATE (Last Step)
        self.graph.add_node("add_system_prompt_content", self.add_system_prompt_content)
    
        ### ✅ REWORKED GRAPH FLOW ###
        self.graph.add_conditional_edges(START, self.map_messages, ["select_relevant_messages"])
        self.graph.add_edge("select_relevant_messages", "categorize_messages")
        self.graph.add_edge("categorize_messages", "extract_memory")
        self.graph.add_edge("categorize_messages", "track_errors")
        self.graph.add_edge("extract_memory", "generate_summary")
        self.graph.add_edge("track_errors", "generate_summary")
        self.graph.add_edge("generate_summary", "collect_summaries")
        self.graph.add_conditional_edges("collect_summaries", self.should_collapse)
        self.graph.add_edge("collapse_summaries", "generate_final_summary")
        self.graph.add_edge("generate_final_summary", "add_system_prompt_content")
        self.graph.add_edge("add_system_prompt_content", END)



    async def categorize_messages(self, state: SummaryState):
        print(state)
        """Categorize messages into tasks, memory types, and errors."""
        
        # Debugging: Check if `contents` exists in state
        if "contents" not in state or not state["contents"]:
            print(state.keys())
            raise KeyError("Missing 'contents' key in state. Ensure input is properly passed.")

        response = await self.task_categorization_chain.ainvoke({"context": state["contents"]})
        return Command(update={"tagged_messages": response})


    async def extract_memory(self, state: SummaryState):
        """Extract relevant task memory and global memory."""
        response = await self.global_memory_chain.ainvoke({"context": state["tagged_messages"]})
        return Command(update={"global_memory": response})

    async def track_errors(self, state: SummaryState):
        """Identify contradictions, mistakes, and required revisions."""
        response = await self.error_tracking_chain.ainvoke({"context": state["tagged_messages"]})
        return Command(update={"errors": response})

    async def select_relevant_messages(self, state: SummaryState):
        """Filter and decide which messages should be kept, discarded, or summarized."""
        response = await self.message_selection_chain.ainvoke({"context": state["tagged_messages"]})
        return Command(update={"message_selections": response})

    async def add_system_prompt_content(self, state: SummaryState):
        """Decide which extracted information should be inserted into the agent's system prompt."""
        response = await self.system_prompt_addition_chain.ainvoke({"context": state["tagged_messages"]})
        return Command(update={"system_prompt_additions": response})

    async def generate_summary(self, state: SummaryState):
        """Generate a summary for a single document."""
        response = await self.map_chain.ainvoke(state["contents"])
        return Command(update={"summaries": [response]})

    def collect_summaries(self, state: SummaryState):
        """Collect summaries into a list of documents."""
        return Command(update={"collapsed_summaries": [
            Document(summary) for summary in state["summaries"]
        ]})
    def map_messages(self, state: SummaryState):
        return [
            Send("categorize_messages", {"content": content})  # ✅ Sending single message at a time
            for content in state["contents"]
        ]



    async def collapse_summaries(self, state: SummaryState):
        """Collapse summaries if their total token count exceeds the limit."""
        doc_lists = split_list_of_docs(
            state["collapsed_summaries"], self.length_function, self.token_max
        )
        results = []
        for doc_list in doc_lists:
            results.append(await acollapse_docs(doc_list, self.reduce_chain.ainvoke))

        return Command(update={"collapsed_summaries": results})

    def should_collapse(self, state: SummaryState) -> Literal["collapse_summaries", "generate_final_summary"]:
        """Determine whether to collapse summaries further."""
        num_tokens = self.length_function(state["collapsed_summaries"])
        return "collapse_summaries" if num_tokens > self.token_max else "generate_final_summary"

    async def generate_final_summary(self, state: SummaryState):
        """Generate the final summary from collapsed summaries."""
        response = await self.reduce_chain.ainvoke(state["collapsed_summaries"])
        return Command(update={"final_summary": response})

    async def arun(self, contents: List[str]) -> str:
        """Run the summarization workflow and return the final summary."""
        async for output in self.app.astream({"contents": contents}, debug=True,config=self.runnable_config):
            print(output)
        return output.get("final_summary", "")
```

```python
import nest_asyncio
import asyncio
#from src.haive.agents.summarizer.agent import SummarizerAgent, SummarizerAgentConfig

# Ensure we can run async inside Jupyter
nest_asyncio.apply()

# Initialize summarizer agent
summarizer = SummarizerAgent(SummarizerAgentConfig())

# Test documents to summarize
test_documents = [
    "Artificial intelligence is transforming multiple industries, including healthcare, finance, and transportation. "
    "Recent advancements in deep learning and reinforcement learning have enabled AI systems to outperform humans in "
    "specific tasks, such as image recognition and game playing.",

    "In recent news, a groundbreaking study found that regular exercise significantly improves cognitive function "
    "in older adults. The researchers conducted a multi-year study with over 10,000 participants and discovered a "
    "strong correlation between physical activity and memory retention.",

    "User: Hey, can you tell me how to fix a bug in my Python code?\n"
    "Agent: Sure! Can you describe the error message you're seeing?\n"
    "User: I'm getting a 'KeyError' when accessing a dictionary key that I think should exist.\n"
    "Agent: You might want to check if the key exists before accessing it using `.get()` or handling it with `try-except`.",

    "NASA has announced a new mission to explore one of Jupiter’s moons, Europa. Scientists believe that Europa "
    "has a subsurface ocean that could harbor extraterrestrial life. The mission, set for 2030, aims to send a probe "
    "to collect and analyze ice samples from the moon’s surface."
]

# Run the summarization pipeline
async def test_summarizer():
    response = await summarizer.arun(test_documents)
    return response

# Execute in notebook
final_summary = asyncio.run(test_summarizer())

# Print the final summary
print("\n===== FINAL SUMMARY =====")
print(final_summary)
```
