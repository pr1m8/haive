```python
import inspect

def inspect_object(obj):
    """
    Prints all properties, methods, and docstrings of an object.
    - Properties: Attributes the object has.
    - Methods: Callable functions of the object.
    - Docstrings: If available, prints docstrings for methods.
    """
    obj_type = type(obj).__name__
    print(f"\n🔍 Inspecting Object: {obj_type}\n" + "=" * 50)

    # Get all attributes safely (handling class-only attributes)
    attributes = {}
    for attr in dir(obj):
        try:
            value = getattr(obj, attr)
            if not callable(value):  
                attributes[attr] = value
        except AttributeError:
            continue  # Skip attributes that cannot be accessed

    # Get all methods
    methods = {attr: getattr(obj, attr) for attr in dir(obj) if callable(getattr(obj, attr, None))}

    # Handle Pydantic models
    if hasattr(obj, "model_fields"):
        print("\n🛠 **Pydantic Model Fields:**\n" + "-" * 50)
        for field, field_info in obj.model_fields.items():
            print(f"  - {field}: {repr(field_info.annotation)}")

    # Print properties (non-callable attributes)
    if attributes:
        print("\n🛠 **Properties (Attributes):**\n" + "-" * 50)
        for attr, value in attributes.items():
            print(f"  - {attr}: {repr(value)}")
    else:
        print("\n(No properties found)")

    # Print methods
    if methods:
        print("\n⚡ **Methods:**\n" + "-" * 50)
        for method_name, method in methods.items():
            doc = inspect.getdoc(method)
            doc_preview = f"\n    📝 {doc}" if doc else " (No docstring available)"
            print(f"  - {method_name}(){doc_preview}")
    else:
        print("\n(No methods found)")

    print("\n" + "=" * 50 + "\n")


# Example Usage
from pydantic import BaseModel

class ExampleModel(BaseModel):
    """Pydantic Example Model"""
    name: str
    age: int

    def greet(self):
        """Returns a greeting."""
        return f"Hello, {self.name}!"


```

```python
from collections import defaultdict
from typing import Callable, Dict, Any, Set, Tuple, Optional
from langgraph.graph.state import StateNodeSpec

class StateGraphManager:
    def __init__(self, graph):
        self.graph = graph  # Directly store the graph

        # Maintain references to key attributes
        self.nodes = self.graph.nodes
        self.edges = self.graph.edges
        self.all_edges = self.graph.edges
        self.is_compiled = self.graph.compiled
        self.input = self.graph.input
        self.output = self.graph.output
        self.schema = self.graph.schema
        self.branches = self.graph.branches

        # Track whether changes require recompilation
        self.needs_recompile = False

    ## ======== Graph Navigation Methods ======== ##

    def get_start_node(self) -> Optional[str]:
        """Finds the start node dynamically from edges."""
        for source, target in self.edges:
            if source == "__start__":  
                return target  # Explicit start node
        for node in self.nodes:
            if all(node != target for _, target in self.edges):
                return node
        return None  # No start node found

    def get_end_nodes(self) -> Set[str]:
        """Finds end nodes dynamically from edges and branches."""
        end_nodes = set()
        for source, target in self.edges:
            if target == "__end__":
                end_nodes.add(source)
        if not end_nodes:
            end_nodes = {node for node in self.nodes if all(node != source for source, _ in self.edges)}
        return end_nodes

    ## ======== Helper Function: Always Updates `self.graph` ======== ##
    def _modify_graph(self, modify_fn: Callable):
        """Wrapper to modify the graph and trigger recompilation if needed."""
        modify_fn()
        if self.needs_recompile:
            self.finalize_and_recompile()

    ## ======== Core Graph Operations ======== ##

    def add_node(self, name: str, node: Callable):
        """Adds a node and updates `self.graph`."""
        def _add():
            if name in self.nodes:
                raise ValueError(f"Node {name} already exists.")
            self.graph.nodes[name] = node
            self.needs_recompile = True

        self._modify_graph(_add)

    def remove_node(self, name: str):
        """Removes a node, updates `self.graph`, and triggers recompilation."""
        def _remove():
            if name not in self.nodes:
                raise ValueError(f"Node {name} does not exist.")

            # Remove node
            del self.graph.nodes[name]

            # Remove edges
            self.graph.edges = {(src, tgt) for src, tgt in self.graph.edges if src != name and tgt != name}

            # Remove branches
            if name in self.graph.branches:
                del self.graph.branches[name]

            self.needs_recompile = True

        self._modify_graph(_remove)

    ## ======== Node Insertion Methods ======== ##

    def insert_node_before(self, existing_node: str, new_node: str, node_callable: Callable):
        """Inserts a node before an existing node, updating edges accordingly."""
        def _insert():
            if new_node in self.nodes:
                raise ValueError(f"Node {new_node} already exists.")

            existing_node_spec = self.nodes.get(existing_node)
            if not isinstance(existing_node_spec, StateNodeSpec):
                raise ValueError(f"Cannot insert before {existing_node}: Not a valid StateNodeSpec.")

            new_node_spec = StateNodeSpec(
                runnable=node_callable,
                metadata=None,
                input=existing_node_spec.input,
                retry_policy=None,
                ends=()
            )

            incoming_edges = {src for src, tgt in self.edges if tgt == existing_node}

            for src in incoming_edges:
                self.graph.edges.discard((src, existing_node))
                self.graph.edges.add((src, new_node))

            self.graph.edges.add((new_node, existing_node))
            self.graph.nodes[new_node] = new_node_spec
            self.needs_recompile = True

        self._modify_graph(_insert)

    def insert_node_after(self, existing_node: str, new_node: str, node_callable: Callable):
        """Inserts a node after an existing node, updating edges accordingly."""
        def _insert():
            if new_node in self.nodes:
                raise ValueError(f"Node {new_node} already exists.")

            existing_node_spec = self.nodes.get(existing_node)
            if not isinstance(existing_node_spec, StateNodeSpec):
                raise ValueError(f"Cannot insert after {existing_node}: Not a valid StateNodeSpec.")

            new_node_spec = StateNodeSpec(
                runnable=node_callable,
                metadata=None,
                input=existing_node_spec.input,
                retry_policy=None,
                ends=()
            )

            outgoing_edges = {tgt for src, tgt in self.edges if src == existing_node}

            for tgt in outgoing_edges:
                self.graph.edges.discard((existing_node, tgt))
                self.graph.edges.add((new_node, tgt))

            self.graph.edges.add((existing_node, new_node))
            self.graph.nodes[new_node] = new_node_spec
            self.needs_recompile = True

        self._modify_graph(_insert)

    ## ======== Node Replacement (Fixed to Update `self.graph`) ======== ##
    
    def replace_node(self, old_node: str, new_node: str, node_callable: Callable):
        """Replaces an existing node while keeping connections and branches intact."""
        def _replace():
            if old_node not in self.nodes:
                raise ValueError(f"Node {old_node} does not exist.")
            if new_node in self.nodes:
                raise ValueError(f"Node {new_node} already exists.")

            incoming_edges = {src for src, tgt in self.edges if tgt == old_node}
            outgoing_edges = {tgt for src, tgt in self.edges if src == old_node}

            old_node_spec = self.nodes[old_node]
            new_node_spec = StateNodeSpec(
                runnable=node_callable,
                metadata=None,
                input=old_node_spec.input,
                retry_policy=None,
                ends=()
            )

            # Remove old node
            del self.graph.nodes[old_node]

            # **Remove edges from `self.graph`**
            self.graph.edges = {(src, tgt) for src, tgt in self.graph.edges if src != old_node and tgt != old_node}

            # Add new node
            self.graph.nodes[new_node] = new_node_spec

            # Reconnect edges, replacing `old_node`
            for src in incoming_edges:
                self.graph.edges.add((src, new_node))
            for tgt in outgoing_edges:
                self.graph.edges.add((new_node, tgt))

            if old_node in self.graph.branches:
                self.graph.branches[new_node] = self.graph.branches.pop(old_node)

            self.needs_recompile = True

        self._modify_graph(_replace)

    ## ======== Compilation Management (Fixed to Update `self.graph`) ======== ##

    def finalize_and_recompile(self):
        """Final step to ensure all modifications are applied before recompiling."""
        if self.needs_recompile:
            print("Validating edges before recompilation...")

            start_node = self.get_start_node()
            if start_node:
                self.graph.edges.add(("__start__", start_node))
            else:
                raise ValueError("Graph is missing a valid start node.")

            end_nodes = self.get_end_nodes()
            if end_nodes:
                for node in end_nodes:
                    self.graph.edges.add((node, "__end__"))
            else:
                raise ValueError("Graph is missing a valid end node.")

            self.recompile()
            self.graph.compile()
            print("Recompilation complete.")
            self.needs_recompile = False
        else:
            print("No changes detected. Skipping recompilation.")

    def recompile(self):
        """ Recompiles the graph if marked as needing recompilation. """
        if not self.needs_recompile:
            print("No recompilation needed.")
            return
        print("Recompiling graph...")
        self.graph.is_compiled = True
        self.needs_recompile = False
```

```python
from typing import List, Optional, Dict, Any, Union
from pydantic import BaseModel, Field
from langchain_core.tools import BaseTool, StructuredTool


class ToolCall(BaseModel):
    name: str = Field(description="Name of the tool to use (or 'LLM')")
    input: Optional[Union[str, Dict[str, Any]]] = Field(
        description="Input data for the tool or reference like '$1'."
    )


class PlanStep(BaseModel):
    id: int = Field(description="Unique, strictly increasing step identifier.")
    description: str = Field(description="Detailed description of the step.")
    tool_calls: List[ToolCall] = Field(description="Tool invocations for the step.")
    evidence_ref: str = Field(description="Evidence reference like '#E1'.")
    dependencies: List[int] = Field(default_factory=list, description="Step IDs this step depends on.")
    result: Optional[str] = Field(default=None, description="Output/result of the step.")
    status: str = Field(default="not_started", description="Current execution status.")


class SequentialPlan(BaseModel):
    steps: List[PlanStep] = Field(description="Sequential steps in this individual plan.")

    def add_step(self, step: PlanStep):
        self.steps.append(step)


class GlobalJoinStep(BaseModel):
    id: int = Field(description="Unique ID for the global join step.")
    description: str = Field(default="Join results from all sub-plans.")
    input_refs: List[str] = Field(
        description="List of evidence references (#E) from sub-plan joins."
    )
    evidence_ref: str = Field(description="Evidence reference for the global join step.")
    dependencies: List[int] = Field(description="IDs of dependent join steps.")


class ParallelPlans(BaseModel):
    max_splits: int = Field(default=2, description="Maximum parallel splits allowed.")
    plans: Dict[str, SequentialPlan] = Field(
        default_factory=dict, description="Dictionary of parallel sub-plans."
    )
    global_join: GlobalJoinStep = Field(description="Final step joining all sub-plans.")

    def add_plan(self, name: str, plan: SequentialPlan):
        if len(self.plans) >= self.max_splits:
            raise ValueError("Exceeded maximum allowed splits")
        self.plans[name] = plan

    def set_global_join(self, join_step: GlobalJoinStep):
        self.global_join = join_step


# Example of extending the original RewooPlan for compatibility:
class ExtendedRewooPlan(BaseModel):
    parallel_plans: ParallelPlans = Field(description="Structured parallelizable Rewoo plan.")

    def from_rewoo_plan(rewoo_plan: 'RewooPlan', max_splits: int = 2) -> 'ExtendedRewooPlan':
        # Example transformation (you'll customize this logic as needed):
        parallel_plans = ParallelPlans(max_splits=max_splits, global_join=None)

        # Example single plan addition for illustration:
        sequential_plan = SequentialPlan(steps=rewoo_plan.steps)
        parallel_plans.add_plan("default", sequential_plan)

        # Example global join creation:
        join_step = GlobalJoinStep(
            id=len(rewoo_plan.steps) + 1,
            input_refs=[step.evidence_ref for step in rewoo_plan.steps],
            evidence_ref=f"#E{len(rewoo_plan.steps) + 1}",
            dependencies=[step.id for step in rewoo_plan.steps]
        )
        parallel_plans.set_global_join(join_step)

        return ExtendedRewooPlan(parallel_plans=parallel_plans)
```

```python
from typing import List, Optional, Dict, Any, Union
from pydantic import BaseModel, Field
from langchain_core.tools import BaseTool, StructuredTool


class ToolCall(BaseModel):
    name: str = Field(description="Name of the tool to use (or 'LLM')")
    input: Optional[Union[str, Dict[str, Any]]] = Field(
        description="Input for the tool or reference like '$1'."
    )


class ParallelPlanNode(BaseModel):
    """
    A recursive node that represents either:
    1. A single atomic step using a tool.
    2. A sub-plan containing multiple parallel steps.
    """
    id: int = Field(description="Unique step identifier.")
    description: str = Field(description="Detailed description of the step.")
    evidence_ref: str = Field(description="Evidence reference like '#E1'.")
    dependencies: List[int] = Field(default_factory=list, description="Step IDs this step depends on.")
    status: str = Field(default="not_started", description="Execution status.")

    # Either a tool call (leaf) or nested sub-steps (branching node)
    tool_calls: Optional[List[ToolCall]] = None
    sub_steps: Optional[List["ParallelPlanNode"]] = None  # Recursive reference

    def is_leaf(self) -> bool:
        """Determines if this node is a leaf (single tool call) or a branching node."""
        return self.tool_calls is not None and self.sub_steps is None


class GlobalJoinStep(BaseModel):
    id: int = Field(description="Unique ID for the global join step.")
    description: str = Field(default="Join results from all sub-plans.")
    input_refs: List[str] = Field(description="List of evidence references (#E) from sub-plan joins.")
    evidence_ref: str = Field(description="Evidence reference for the global join step.")
    dependencies: List[int] = Field(description="IDs of dependent join steps.")


class ParallelPlanTree(BaseModel):
    """
    A recursive parallel execution plan with tree-like branching and depth constraints.
    """
    root: ParallelPlanNode = Field(description="Root node of the execution plan.")
    max_depth: int = Field(default=3, description="Maximum depth allowed for branching execution.")
    global_join: Optional[GlobalJoinStep] = None

    def enforce_max_depth(self, current_depth: int = 0):
        """
        Recursively enforces max depth: If exceeded, child steps are collapsed into a join node.
        """
        def process_node(node: ParallelPlanNode, depth: int):
            if depth >= self.max_depth and node.sub_steps:
                # Collapse all sub-steps into a join() call
                join_id = max(step.id for step in node.sub_steps) + 1
                join_step = ParallelPlanNode(
                    id=join_id,
                    description="Intermediate join due to max depth reached.",
                    evidence_ref=f"#E{join_id}",
                    dependencies=[s.id for s in node.sub_steps],
                    tool_calls=[ToolCall(name="join", input=[s.evidence_ref for s in node.sub_steps])]
                )
                node.sub_steps = [join_step]

            elif node.sub_steps:
                for child in node.sub_steps:
                    process_node(child, depth + 1)

        process_node(self.root, 0)

    def set_global_join(self):
        """
        Creates a final join step combining all evidence from sub-branches.
        """
        all_refs = self.collect_all_evidence_refs()
        all_ids = [int(ref[2:]) for ref in all_refs]  # Extract IDs from #E<number>
        global_join_id = max(all_ids) + 1

        self.global_join = GlobalJoinStep(
            id=global_join_id,
            input_refs=all_refs,
            evidence_ref=f"#E{global_join_id}",
            dependencies=all_ids
        )

    def collect_all_evidence_refs(self) -> List[str]:
        """
        Collects all evidence references in the tree.
        """
        refs = []

        def collect(node: ParallelPlanNode):
            refs.append(node.evidence_ref)
            if node.sub_steps:
                for sub in node.sub_steps:
                    collect(sub)

        collect(self.root)
        return refs
```
