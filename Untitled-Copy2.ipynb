```python
from pydantic import BaseModel, Field
from typing import List, Dict, Optional, Any, Union, Literal

class ToolCall(BaseModel):
    """
    Represents an invocation of a specific tool or function.

    Attributes:
        name (str): Name of the tool being invoked.
        input (Optional[Union[str, dict]]): The input data provided to the tool.
    """
    name: str
    input: Optional[Union[str, dict]] = None


class PlanNode(BaseModel):
    """
    Represents a single node or step within a planning tree. It can either be a leaf node (with tool calls)
    or an intermediate node (with sub-steps).

    Attributes:
        id (int): Unique identifier for the plan node.
        description (str): Human-readable description of the task.
        evidence_ref (str): Reference to evidence or results associated with this node.
        status (str): Current execution status ('not_started', 'in_progress', 'completed').
        dependencies (List[int]): List of node IDs this step depends on.
        tool_calls (Optional[List[ToolCall]]): Tools to invoke if this node is a leaf node.
        sub_steps (Optional[List['PlanNode']]): Nested sub-steps if this is an intermediate node.
    """
    id: int
    description: str
    evidence_ref: str
    status: str = "not_started"
    dependencies: List[int] = Field(default_factory=list)
    tool_calls: Optional[List[ToolCall]] = None
    sub_steps: Optional[List['PlanNode']] = Field(default_factory=list)

    def is_leaf(self) -> bool:
        """Determines if the current node is a leaf node (no sub-steps, only tool calls)."""
        return self.tool_calls is not None and not self.sub_steps


class JoinStep(BaseModel):
    """
    Represents a join operation that aggregates results from multiple plan nodes.

    Attributes:
        id (int): Unique identifier for the join step.
        description (str): Description of the join operation.
        input_refs (List[str]): References to evidence from sub-plans being joined.
        evidence_ref (str): Evidence reference produced by this join step.
        dependencies (List[int]): Node IDs that must complete before this join step.
        status (str): Current execution status.
    """
    id: int
    description: str
    input_refs: List[str]
    evidence_ref: str
    dependencies: List[int]
    status: str = "not_started"


class PlanTree(BaseModel):
    """
    Represents the entire hierarchical execution plan, including all nodes, intermediate joins, and a final global join.

    Attributes:
        root (PlanNode): The root node of the execution plan.
        joins (List[JoinStep]): Intermediate join steps aggregating partial results.
        global_join (Optional[JoinStep]): Final join step aggregating results from all branches.
    """
    root: PlanNode
    joins: List[JoinStep] = Field(default_factory=list)
    global_join: Optional[JoinStep] = None

    def collect_all_steps(self, node: Optional[PlanNode] = None) -> List[PlanNode]:
        """Recursively collects all nodes within the planning tree."""
        node = node or self.root
        steps = [node]
        for sub in node.sub_steps:
            steps.extend(self.collect_all_steps(sub))
        return steps

    def find_step(self, step_id: int, node: Optional[PlanNode] = None) -> Optional[PlanNode]:
        """Finds and returns a node by its ID."""
        node = node or self.root
        if node.id == step_id:
            return node
        for sub in node.sub_steps:
            found = self.find_step(step_id, sub)
            if found:
                return found
        return None

    def add_join(self, join_step: JoinStep):
        """Adds a new intermediate join step to the plan."""
        self.joins.append(join_step)

    def set_global_join(self, join_step: JoinStep):
        """Sets the final global join step for the entire plan."""
        self.global_join = join_step


# Required for recursive models
PlanNode.model_rebuild()
PlanTree.model_rebuild()


def pretty_print_plan(node: PlanNode, depth: int = 0):
    """Recursively pretty-prints the plan tree with clear indentation representing hierarchy."""
    prefix = "│   " * depth
    print(f"{prefix}├── [{node.status}] {node.description} (ID: {node.id}, Ref: {node.evidence_ref})")
    if node.tool_calls:
        calls = ", ".join(f"{tc.name}({tc.input})" for tc in node.tool_calls)
        print(f"{prefix}│       🛠 {calls}")
    for sub_step in node.sub_steps:
        pretty_print_plan(sub_step, depth + 1)

    if depth == 0:
        print("\n")
```

```python
from typing import List, Dict, Any, Optional, Union, Tuple, Type,Callable
from pydantic import BaseModel, Field


class ToolCall(BaseModel):
    """
    Represents a tool invocation.
    """
    name: str = Field(description="Name of the tool to use (or 'LLM')")
    input: Optional[Union[str, Dict[str, Any]]] = Field(
        default=None,
        description="Input for the tool or reference like '$1'."
    )


class ParallelPlanNode(BaseModel):
    """
    A recursive node that represents:
    1. A single atomic step using a tool.
    2. A sub-plan containing multiple parallel steps.
    """
    id: int = Field(description="Unique step identifier.")
    description: str = Field(description="Detailed description of the step.")
    evidence_ref: str = Field(description="Evidence reference like '#E1'.")
    dependencies: List[int] = Field(default_factory=list, description="Step IDs this step depends on.")
    status: str = Field(default="not_started", description="Execution status.")

    # Either a tool call (leaf) or nested sub-steps (branching node)
    tool_calls: Optional[List[ToolCall]] = None
    sub_steps: Optional[List["ParallelPlanNode"]] = None  # Recursive reference

    def is_leaf(self) -> bool:
        """Determines if this node is a leaf (single tool call) or a branching node."""
        return self.tool_calls is not None and self.sub_steps is None

    def find_node_by_id(self, id: int) -> Optional["ParallelPlanNode"]:
        """Find a node by its ID."""
        if self.id == id:
            return self
        
        if self.sub_steps:
            for step in self.sub_steps:
                found = step.find_node_by_id(id)
                if found:
                    return found
        
        return None

    def get_executable_steps(self) -> List["ParallelPlanNode"]:
        """
        Get steps that are ready to be executed (dependencies satisfied).
        A step is executable if:
        1. It's not already completed or in progress
        2. All dependencies are completed
        """
        executable = []
        
        # Check if this node is executable
        if self.status == "not_started":
            # If no dependencies, it's executable
            if not self.dependencies:
                executable.append(self)
            # Otherwise, check if all dependencies are completed
            else:
                all_deps_completed = True
                for dep_id in self.dependencies:
                    dep_node = self.find_dep_node(dep_id)
                    if not dep_node or dep_node.status != "completed":
                        all_deps_completed = False
                        break
                
                if all_deps_completed:
                    executable.append(self)
        
        # Check sub-steps recursively
        if self.sub_steps:
            for step in self.sub_steps:
                executable.extend(step.get_executable_steps())
        
        return executable

    def find_dep_node(self, dep_id: int) -> Optional["ParallelPlanNode"]:
        """Find a dependency node by its ID."""
        # Start from the root (assuming self is the root or has access to it)
        root = self
        while hasattr(root, "parent") and root.parent:
            root = root.parent
        
        # Search from the root
        return root.find_node_by_id(dep_id)


class GlobalJoinStep(BaseModel):
    """
    Represents the final join step aggregating all sub-plans.
    """
    id: int = Field(description="Unique ID for the global join step.")
    description: str = Field(default="Join results from all sub-plans.")
    input_refs: List[str] = Field(description="List of evidence references (#E) from sub-plan joins.")
    evidence_ref: str = Field(description="Evidence reference for the global join step.")
    dependencies: List[int] = Field(description="IDs of dependent join steps.")
    status: str = Field(default="not_started", description="Execution status.")


class ParallelPlanTree(BaseModel):
    """
    A recursive parallel execution plan with tree-like branching and depth constraints.
    """
    root: ParallelPlanNode = Field(description="Root node of the plan tree.")
    max_depth: int = Field(default=10, description="Maximum depth allowed for branching execution.")
    global_join: Optional[GlobalJoinStep] = None
    evidence_store: Dict[str, Any] = Field(default_factory=dict, description="Store for evidence created during execution.")

    def enforce_max_depth(self):
        """
        Recursively enforces max depth: If exceeded, child steps are collapsed into a join node.
        """
        def process_node(node: ParallelPlanNode, depth: int):
            if depth >= self.max_depth and node.sub_steps:
                # Collapse all sub-steps into a join() call
                join_id = max(step.id for step in node.sub_steps) + 1
                join_step = ParallelPlanNode(
                    id=join_id,
                    description="Intermediate join due to max depth reached.",
                    evidence_ref=f"#E{join_id}",
                    dependencies=[s.id for s in node.sub_steps],
                    tool_calls=[ToolCall(name="join", input=[s.evidence_ref for s in node.sub_steps])]
                )
                node.sub_steps = [join_step]

            elif node.sub_steps:
                for child in node.sub_steps:
                    process_node(child, depth + 1)

        process_node(self.root, 0)

    def set_global_join(self):
        """
        Creates a final join step combining all evidence from sub-branches.
        """
        all_refs = self.collect_all_evidence_refs()
        all_ids = [int(ref[2:]) for ref in all_refs]  # Extract IDs from #E<number>
        global_join_id = max(all_ids) + 1

        self.global_join = GlobalJoinStep(
            id=global_join_id,
            input_refs=all_refs,
            evidence_ref=f"#E{global_join_id}",
            dependencies=all_ids
        )

    def collect_all_evidence_refs(self) -> List[str]:
        """
        Collects all evidence references in the tree.
        """
        refs = []

        def collect(node: ParallelPlanNode):
            refs.append(node.evidence_ref)
            if node.sub_steps:
                for sub in node.sub_steps:
                    collect(sub)

        collect(self.root)
        return refs

    def get_next_executable_step(self) -> Optional[ParallelPlanNode]:
        """Get the next step that can be executed."""
        executable_steps = self.root.get_executable_steps()
        return executable_steps[0] if executable_steps else None
    
    def update_step_status(self, step_id: int, status: str, result: Any = None):
        """Update status of a step and store its result if completed."""
        node = self.root.find_node_by_id(step_id)
        if node:
            node.status = status
            if status == "completed" and result is not None:
                self.evidence_store[node.evidence_ref] = result
        
        # Also check global join
        if self.global_join and self.global_join.id == step_id:
            self.global_join.status = status
            if status == "completed" and result is not None:
                self.evidence_store[self.global_join.evidence_ref] = result

    def is_complete(self) -> bool:
        """Check if the entire plan is complete."""
        return (self.global_join and self.global_join.status == "completed") or self._check_all_completed(self.root)
    
    def _check_all_completed(self, node: ParallelPlanNode) -> bool:
        """Check if a node and all its sub-nodes are completed."""
        if node.status != "completed":
            return False
        
        if node.sub_steps:
            return all(self._check_all_completed(step) for step in node.sub_steps)
        
        return True


# Rebuild models to handle recursive relationships
ParallelPlanNode.model_rebuild()
ParallelPlanTree.model_rebuild()


# State schema for plan-execute agent
class PlanExecuteState(BaseModel):
    """State schema for the plan-execute agent."""
    messages: List[Any] = Field(default_factory=list)
    plan: Optional[ParallelPlanTree] = None
    execution_status: str = Field(default="not_started")
    current_step_id: Optional[int] = None
    reflection: Optional[str] = None
    final_result: Optional[str] = None
    tools: Dict[str, Callable] = Field(default_factory=dict)
    replanning_count: int = Field(default=0)
    max_replanning: int = Field(default=10)
```

```python
from rich.tree import Tree
from rich.console import Console

def build_tree(node: ParallelPlanNode, rich_tree: Tree) -> None:
    """Recursively builds a rich tree representation of the ParallelPlanTree."""
    tool_calls = ", ".join(f"{tc.name}({tc.input})" for tc in node.tool_calls) if node.tool_calls else "None"
    step_text = f"[{node.status}] {node.description} (ID: {node.id}, Ref: {node.evidence_ref})"
    step_text += f"\n    Tool Calls: {tool_calls}"
    
    node_tree = rich_tree.add(step_text)
    
    if node.sub_steps:
        for sub in node.sub_steps:
            build_tree(sub, node_tree)

def display_plan_tree(plan_tree: ParallelPlanTree):
    """Displays the ParallelPlanTree in a tree structure using Rich."""
    console = Console()
    root_tree = Tree(f"🌳 **Parallel Execution Plan**\n[{plan_tree.root.status}] {plan_tree.root.description} (ID: {plan_tree.root.id}, Ref: {plan_tree.root.evidence_ref})")
    
    build_tree(plan_tree.root, root_tree)
    
    if plan_tree.global_join:
        root_tree.add(f"\n🔗 **Global Join Step** [{plan_tree.global_join.status}]\n    {plan_tree.global_join.description}\n    Evidence Ref: {plan_tree.global_join.evidence_ref}\n    Dependencies: {plan_tree.global_join.dependencies}")

    console.print(root_tree)

# Usage
display_plan_tree(result)
```

```python
import inspect

def inspect_object(obj):
    """
    Prints all properties, methods, and docstrings of an object.
    - Properties: Attributes the object has.
    - Methods: Callable functions of the object.
    - Docstrings: If available, prints docstrings for methods.
    """
    obj_type = type(obj).__name__
    print(f"\n🔍 Inspecting Object: {obj_type}\n" + "=" * 50)

    # Get all attributes safely (handling class-only attributes)
    attributes = {}
    for attr in dir(obj):
        try:
            value = getattr(obj, attr)
            if not callable(value):  
                attributes[attr] = value
        except AttributeError:
            continue  # Skip attributes that cannot be accessed

    # Get all methods
    methods = {attr: getattr(obj, attr) for attr in dir(obj) if callable(getattr(obj, attr, None))}

    # Handle Pydantic models
    if hasattr(obj, "model_fields"):
        print("\n🛠 **Pydantic Model Fields:**\n" + "-" * 50)
        for field, field_info in obj.model_fields.items():
            print(f"  - {field}: {repr(field_info.annotation)}")

    # Print properties (non-callable attributes)
    if attributes:
        print("\n🛠 **Properties (Attributes):**\n" + "-" * 50)
        for attr, value in attributes.items():
            print(f"  - {attr}: {repr(value)}")
    else:
        print("\n(No properties found)")

    # Print methods
    if methods:
        print("\n⚡ **Methods:**\n" + "-" * 50)
        for method_name, method in methods.items():
            doc = inspect.getdoc(method)
            doc_preview = f"\n    📝 {doc}" if doc else " (No docstring available)"
            print(f"  - {method_name}(){doc_preview}")
    else:
        print("\n(No methods found)")

    print("\n" + "=" * 50 + "\n")


# Example Usage
from pydantic import BaseModel

class ExampleModel(BaseModel):
    """Pydantic Example Model"""
    name: str
    age: int

    def greet(self):
        """Returns a greeting."""
        return f"Hello, {self.name}!"


```

```python
from collections import defaultdict
from typing import Callable, Dict, Any, Set, Tuple, Optional
from langgraph.graph.state import StateNodeSpec

class StateGraphManager:
    def __init__(self, graph):
        self.graph = graph  # Directly store the graph

        # Maintain references to key attributes
        self.nodes = self.graph.nodes
        self.edges = self.graph.edges
        self.all_edges = self.graph.edges
        self.is_compiled = self.graph.compiled
        self.input = self.graph.input
        self.output = self.graph.output
        self.schema = self.graph.schema
        self.branches = self.graph.branches

        # Track whether changes require recompilation
        self.needs_recompile = False

    ## ======== Graph Navigation Methods ======== ##

    def get_start_node(self) -> Optional[str]:
        """Finds the start node dynamically from edges."""
        for source, target in self.edges:
            if source == "__start__":  
                return target  # Explicit start node
        for node in self.nodes:
            if all(node != target for _, target in self.edges):
                return node
        return None  # No start node found

    def get_end_nodes(self) -> Set[str]:
        """Finds end nodes dynamically from edges and branches."""
        end_nodes = set()
        for source, target in self.edges:
            if target == "__end__":
                end_nodes.add(source)
        if not end_nodes:
            end_nodes = {node for node in self.nodes if all(node != source for source, _ in self.edges)}
        return end_nodes

    ## ======== Helper Function: Always Updates `self.graph` ======== ##
    def _modify_graph(self, modify_fn: Callable):
        """Wrapper to modify the graph and trigger recompilation if needed."""
        modify_fn()
        if self.needs_recompile:
            self.finalize_and_recompile()

    ## ======== Core Graph Operations ======== ##

    def add_node(self, name: str, node: Callable):
        """Adds a node and updates `self.graph`."""
        def _add():
            if name in self.nodes:
                raise ValueError(f"Node {name} already exists.")
            self.graph.nodes[name] = node
            self.needs_recompile = True

        self._modify_graph(_add)

    def remove_node(self, name: str):
        """Removes a node, updates `self.graph`, and triggers recompilation."""
        def _remove():
            if name not in self.nodes:
                raise ValueError(f"Node {name} does not exist.")

            # Remove node
            del self.graph.nodes[name]

            # Remove edges
            self.graph.edges = {(src, tgt) for src, tgt in self.graph.edges if src != name and tgt != name}

            # Remove branches
            if name in self.graph.branches:
                del self.graph.branches[name]

            self.needs_recompile = True

        self._modify_graph(_remove)

    ## ======== Node Insertion Methods ======== ##

    def insert_node_before(self, existing_node: str, new_node: str, node_callable: Callable):
        """Inserts a node before an existing node, updating edges accordingly."""
        def _insert():
            if new_node in self.nodes:
                raise ValueError(f"Node {new_node} already exists.")

            existing_node_spec = self.nodes.get(existing_node)
            if not isinstance(existing_node_spec, StateNodeSpec):
                raise ValueError(f"Cannot insert before {existing_node}: Not a valid StateNodeSpec.")

            new_node_spec = StateNodeSpec(
                runnable=node_callable,
                metadata=None,
                input=existing_node_spec.input,
                retry_policy=None,
                ends=()
            )

            incoming_edges = {src for src, tgt in self.edges if tgt == existing_node}

            for src in incoming_edges:
                self.graph.edges.discard((src, existing_node))
                self.graph.edges.add((src, new_node))

            self.graph.edges.add((new_node, existing_node))
            self.graph.nodes[new_node] = new_node_spec
            self.needs_recompile = True

        self._modify_graph(_insert)

    def insert_node_after(self, existing_node: str, new_node: str, node_callable: Callable):
        """Inserts a node after an existing node, updating edges accordingly."""
        def _insert():
            if new_node in self.nodes:
                raise ValueError(f"Node {new_node} already exists.")

            existing_node_spec = self.nodes.get(existing_node)
            if not isinstance(existing_node_spec, StateNodeSpec):
                raise ValueError(f"Cannot insert after {existing_node}: Not a valid StateNodeSpec.")

            new_node_spec = StateNodeSpec(
                runnable=node_callable,
                metadata=None,
                input=existing_node_spec.input,
                retry_policy=None,
                ends=()
            )

            outgoing_edges = {tgt for src, tgt in self.edges if src == existing_node}

            for tgt in outgoing_edges:
                self.graph.edges.discard((existing_node, tgt))
                self.graph.edges.add((new_node, tgt))

            self.graph.edges.add((existing_node, new_node))
            self.graph.nodes[new_node] = new_node_spec
            self.needs_recompile = True

        self._modify_graph(_insert)

    ## ======== Node Replacement (Fixed to Update `self.graph`) ======== ##
    
    def replace_node(self, old_node: str, new_node: str, node_callable: Callable):
        """Replaces an existing node while keeping connections and branches intact."""
        def _replace():
            if old_node not in self.nodes:
                raise ValueError(f"Node {old_node} does not exist.")
            if new_node in self.nodes:
                raise ValueError(f"Node {new_node} already exists.")

            incoming_edges = {src for src, tgt in self.edges if tgt == old_node}
            outgoing_edges = {tgt for src, tgt in self.edges if src == old_node}

            old_node_spec = self.nodes[old_node]
            new_node_spec = StateNodeSpec(
                runnable=node_callable,
                metadata=None,
                input=old_node_spec.input,
                retry_policy=None,
                ends=()
            )

            # Remove old node
            del self.graph.nodes[old_node]

            # **Remove edges from `self.graph`**
            self.graph.edges = {(src, tgt) for src, tgt in self.graph.edges if src != old_node and tgt != old_node}

            # Add new node
            self.graph.nodes[new_node] = new_node_spec

            # Reconnect edges, replacing `old_node`
            for src in incoming_edges:
                self.graph.edges.add((src, new_node))
            for tgt in outgoing_edges:
                self.graph.edges.add((new_node, tgt))

            if old_node in self.graph.branches:
                self.graph.branches[new_node] = self.graph.branches.pop(old_node)

            self.needs_recompile = True

        self._modify_graph(_replace)

    ## ======== Compilation Management (Fixed to Update `self.graph`) ======== ##

    def finalize_and_recompile(self):
        """Final step to ensure all modifications are applied before recompiling."""
        if self.needs_recompile:
            print("Validating edges before recompilation...")

            start_node = self.get_start_node()
            if start_node:
                self.graph.edges.add(("__start__", start_node))
            else:
                raise ValueError("Graph is missing a valid start node.")

            end_nodes = self.get_end_nodes()
            if end_nodes:
                for node in end_nodes:
                    self.graph.edges.add((node, "__end__"))
            else:
                raise ValueError("Graph is missing a valid end node.")

            self.recompile()
            self.graph.compile()
            print("Recompilation complete.")
            self.needs_recompile = False
        else:
            print("No changes detected. Skipping recompilation.")

    def recompile(self):
        """ Recompiles the graph if marked as needing recompilation. """
        if not self.needs_recompile:
            print("No recompilation needed.")
            return
        print("Recompiling graph...")
        self.graph.is_compiled = True
        self.needs_recompile = False
```
