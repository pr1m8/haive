```python
from typing import Type, List, Any, Union, Dict
import inspect
import logging

logger = logging.getLogger(__name__)

class SyntheticLiteral:
    """
    Utility class for creating synthetic Literal types at runtime.
    Enables dynamic typing for langgraph routes that aren't known at compile time.
    """
    
    @staticmethod
    def create(values: List[str]) -> Type:
        """
        Create a synthetic Literal type with the given values.
        
        Args:
            values: The list of string values to include in the Literal
            
        Returns:
            A Literal type with the specified values
        """
        from typing import Literal
        
        # Ensure we have at least one value and END is included
        if not values:
            values = ["END"]
        elif "END" not in values:
            values.append("END")
            
        # Create a string representation of the Literal type
        values_str = ", ".join([f'"{val}"' for val in values])
        type_str = f"Literal[{values_str}]"
        
        # Create the Literal type dynamically
        namespace = {"Literal": Literal}
        for val in values:
            namespace[val] = val
            
        # Execute the string to create the type
        try:
            literal_type = eval(type_str, namespace)
            return literal_type
        except Exception as e:
            logger.error(f"Error creating Literal type: {e}")
            # Return fallback literal with just END
            return Literal["END"]
    
    @staticmethod
    def validate_route(route: str, valid_routes: List[str], default_route: str = "END") -> str:
        """
        Validate a route against a list of valid routes.
        
        Args:
            route: The route to validate
            valid_routes: List of valid routes
            default_route: Default route to return if invalid
            
        Returns:
            Valid route or default route
        """
        if route == "END" or route in valid_routes:
            return "END" if route == "END" else route
        else:
            return default_route
    
    @staticmethod
    def get_route_type(state: Dict[str, Any]) -> Type:
        """
        Extract valid routes from state and create a Literal type.
        
        Args:
            state: State dictionary containing available_routes
            
        Returns:
            Literal type for routes
        """
        routes = list(state.get("available_routes", {}).keys())
        return SyntheticLiteral.create(routes)
```

```python
from typing import Dict, List, Any, Optional, Union, Type, Callable, Literal
from enum import Enum
from pydantic import BaseModel, Field
import logging
import json
import uuid
import re
from datetime import datetime

from langchain_core.messages import AIMessage, HumanMessage, SystemMessage, BaseMessage
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.runnables import Runnable

from langgraph.graph import END, StateGraph
from langgraph.types import Command

from src.haive.agents.base import Agent, AgentConfig, register_agent
from src.haive.agents.simple.agent import SimpleAgent, SimpleAgentConfig
from src.haive.core.aug_llm.base import AugLLMConfig
from src.haive.core.models.llm.base import AzureLLMConfig
from src.haive.core.graph.GraphBuilder import DynamicGraph
from langgraph.graph import add_messages
from typing import Annotated, Sequence

# Import the SyntheticLiteral utility
try:
    from synthetic_literal import SyntheticLiteral
except ImportError:
    # Define minimal SyntheticLiteral if import fails
    class SyntheticLiteral:
        @staticmethod
        def validate_route(route, valid_routes, default_route="END"):
            return route if route in valid_routes or route == "END" else default_route

logger = logging.getLogger(__name__)

# ==========================
# 🔹 Agent Command Types
# ==========================
class AgentCommand(str, Enum):
    """Command types for agent management via messages."""
    ADD_AGENT = "add_agent"
    REMOVE_AGENT = "remove_agent"
    LIST_AGENTS = "list_agents"
    HELP = "help"

# ==========================
# 🔹 Agent Configuration Model
# ==========================
class AgentConfigModel(BaseModel):
    """Model for agent configuration via messages."""
    name: str = Field(..., description="Name/route for the agent")
    description: str = Field(..., description="Description of what the agent does")
    system_prompt: str = Field(..., description="System prompt for the agent")
    model: str = Field(default="gpt-4o", description="Model to use for the agent")
    temperature: float = Field(default=0.7, description="Temperature for generation")
    
    def to_dict(self) -> Dict[str, Any]:
        """Convert to a configuration dictionary."""
        return {
            "type": "simple",
            "system_prompt": self.system_prompt,
            "model": self.model,
            "temperature": self.temperature,
            "description": self.description,
            "name": self.name
        }

# ==========================
# 🔹 Enhanced State Schema with Agent Management
# ==========================
class DynamicAgentState(BaseModel):
    """State schema with agent configurations and management capabilities."""
    
    # Standard message field using langgraph's add_messages reducer
    messages: Annotated[Sequence[BaseMessage], add_messages] = Field(
        default_factory=list,
        description="Messages in the conversation"
    )
    
    # Current route and execution state
    current_route: str = Field(
        default="supervisor", 
        description="Current active route/agent"
    )
    
    # Available routes and their descriptions
    available_routes: Dict[str, str] = Field(
        default_factory=dict,
        description="Map of route names to descriptions"
    )
    
    # Agent configurations by route name
    agent_configs: Dict[str, Dict[str, Any]] = Field(
        default_factory=dict,
        description="Map of agent configurations by route name"
    )
    
    # Runtime agent instances (not serialized)
    agent_instances: Dict[str, Agent] = Field(
        default_factory=dict,
        exclude=True,
        description="Runtime agent instances (not serialized)"
    )
    
    # Execution results
    results: Dict[str, Any] = Field(
        default_factory=dict,
        description="Results from agent execution"
    )
    
    # Error tracking
    error: Optional[str] = Field(
        default=None,
        description="Error message if any occurred"
    )
    
    # Execution metadata and tracing
    execution_trace: List[Dict[str, Any]] = Field(
        default_factory=list,
        description="Trace of execution steps"
    )
    
    # Command processing flag
    is_command: bool = Field(
        default=False,
        description="Flag indicating if current message is a command"
    )
    
    # Command result data
    command_result: Optional[Dict[str, Any]] = Field(
        default=None,
        description="Result data from command execution"
    )

# ==========================
# 🔹 Dynamic Supervisor Config
# ==========================
class DynamicSupervisorConfig(AgentConfig):
    """Configuration for a supervisor agent with truly dynamic routing capabilities."""
    
    # System prompt for the supervisor
    supervisor_system_prompt: str = Field(
        default=(
            "You are a supervisor agent responsible for routing requests to specialized agents "
            "and managing the agent pool. You can route requests to any available agent, or "
            "respond with 'END' if the task is complete.\n\n"
            "You can also respond to special commands:\n"
            "- To add an agent: `!add_agent name: Agent Name, description: Description, "
            "system_prompt: Agent system prompt, [model: model_name], [temperature: 0.7]`\n"
            "- To remove an agent: `!remove_agent name: Agent Name`\n"
            "- To list available agents: `!list_agents`\n"
            "- For help: `!help`\n\n"
            "When routing, respond with ONLY the name of an available agent or 'END'."
        ),
        description="System prompt template for the supervisor LLM."
    )
    
    # Default schema override
    state_schema: Type[BaseModel] = Field(
        default=DynamicAgentState,
        description="State schema for the agent"
    )
    
    # Default agents to register
    default_agents: Dict[str, Dict[str, Any]] = Field(
        default_factory=dict,
        description="Default agents to register at initialization"
    )
    
    # Command pattern regex
    command_pattern: str = Field(
        default=r"^\!(\w+)(?:\s+(.+))?$",
        description="Regex pattern for detecting commands in messages"
    )
    
    @classmethod
    def from_scratch(cls,
                    system_prompt: Optional[str] = None,
                    default_agents: Optional[Dict[str, Dict[str, Any]]] = None,
                    model: str = "gpt-4o",
                    temperature: float = 0.2,
                    name: Optional[str] = None,
                    **kwargs) -> 'DynamicSupervisorConfig':
        """
        Create a DynamicSupervisorConfig from scratch.
        
        Args:
            system_prompt: Optional system prompt override
            default_agents: Optional default agents to register
            model: Model name to use
            temperature: Temperature for generation
            name: Optional agent name
            **kwargs: Additional kwargs for the config
            
        Returns:
            DynamicSupervisorConfig instance
        """
        # Create LLM config
        llm_config = AzureLLMConfig(
            model=model,
            parameters={"temperature": temperature}
        )
        
        # Use provided system prompt or default
        system_prompt = system_prompt or cls.__fields__["supervisor_system_prompt"].default
        
        # Create prompt template
        messages = [
            SystemMessage(content=system_prompt),
            MessagesPlaceholder(variable_name="messages")
        ]
        prompt_template = ChatPromptTemplate.from_messages(messages)
        
        # Create AugLLM config
        aug_llm = AugLLMConfig(
            name=f"{name or 'supervisor'}_llm",
            llm_config=llm_config,
            prompt_template=prompt_template
        )
        
        # Create the config
        config = cls(
            name=name or f"dynamic_supervisor",
            engine=aug_llm,
            system_prompt=system_prompt,
            default_agents=default_agents or {},
            **kwargs
        )
        
        return config

# ==========================
# 🔹 Dynamic Supervisor Agent
# ==========================
@register_agent(DynamicSupervisorConfig)
class DynamicSupervisorAgent(Agent[DynamicSupervisorConfig]):
    """
    A truly dynamic supervisor agent with agent management through messages.
    
    This implementation:
    1. Uses a single generic agent node for all agent executions
    2. Stores agent configurations in state for runtime flexibility
    3. Can add/remove/modify agents without recompiling the graph
    4. Allows agent management through special message commands
    """
    
    def __init__(self, config: DynamicSupervisorConfig):
        """Initialize the supervisor agent with its configuration."""
        super().__init__(config)
        
        # Initialize with default agents if provided
        self._default_agents = config.default_agents or {}
        
        # Compile command pattern regex
        self._command_pattern = re.compile(config.command_pattern)
    
    def setup_workflow(self):
        """Set up a workflow with a supervisor, command handler, and generic agent node."""
        # Create DynamicGraph with proper component registration
        gb = DynamicGraph(
            components=[self.config.engine],
            state_schema=self.state_schema
        )
        
        # Add the command handler node that processes commands
        gb.add_node(
            name="command_handler",
            config=self._create_command_handler_function(),
            command_goto="supervisor"  # Return to supervisor after command processing
        )
        
        # Add the supervisor node that decides routing
        gb.add_node(
            name="supervisor",
            config=self._create_supervisor_node_function(),
            command_goto=None  # Will be determined by routing logic
        )
        
        # Add a generic agent execution node
        gb.add_node(
            name="agent_executor",
            config=self._create_agent_executor_function(),
            command_goto="supervisor"  # Always return to supervisor after execution
        )
        
        # Add END node for completion
        gb.add_node(
            name="finish",
            config=self._create_finish_node_function(),
            command_goto=END
        )
        
        # Set up conditional routing from supervisor to check for commands first
        gb.add_conditional_edges(
            "supervisor",
            self._check_for_command,
            {
                "command_handler": "command_handler",
                "continue": "supervisor_route"
            }
        )
        
        # Add supervisor routing node
        gb.add_node(
            name="supervisor_route",
            config=self._create_supervisor_routing_function(),
            command_goto=None  # Will be determined by routing logic
        )
        
        # Set up conditional routing from supervisor_route
        gb.add_conditional_edges(
            "supervisor_route",
            self._route_request,
            {
                "agent_executor": "agent_executor",
                "END": "finish"
            }
        )
        
        # Set the entry point to the supervisor
        gb.set_entry_point("supervisor")
        
        # Get the built graph
        self.graph = gb.build()
    
    def _create_command_handler_function(self) -> Callable:
        """Create the command handler node function."""
        def command_handler(state: Dict[str, Any]) -> Dict[str, Any]:
            """Handle special commands for agent management."""
            try:
                # Convert state to dictionary if necessary
                state_dict = state.model_dump() if hasattr(state, "model_dump") else state
                
                # Get the last message
                last_message = None
                if "messages" in state_dict and state_dict["messages"]:
                    last_message = state_dict["messages"][-1]
                
                if not last_message or not hasattr(last_message, "content"):
                    state_dict["is_command"] = False
                    return state_dict
                
                # Extract command and arguments
                content = last_message.content
                match = self._command_pattern.match(content)
                
                if not match:
                    state_dict["is_command"] = False
                    return state_dict
                
                command = match.group(1).lower()
                args_str = match.group(2) if match.group(2) else ""
                
                # Track execution
                state_dict["execution_trace"].append({
                    "node": "command_handler",
                    "timestamp": str(datetime.now()),
                    "command": command,
                    "args": args_str
                })
                
                # Process command
                result = self._process_command(command, args_str, state_dict)
                
                # Add result message
                result_message = AIMessage(content=result.get("message", "Command processed."))
                state_dict["messages"] = state_dict["messages"] + [result_message]
                
                # Store command result
                state_dict["command_result"] = result
                state_dict["is_command"] = True
                
                return state_dict
                
            except Exception as e:
                # Handle errors
                state_dict = state.model_dump() if hasattr(state, "model_dump") else state
                error_msg = f"Error processing command: {str(e)}"
                state_dict["error"] = error_msg
                
                # Add error message
                result_message = AIMessage(content=f"⚠️ {error_msg}")
                if "messages" in state_dict:
                    state_dict["messages"] = state_dict["messages"] + [result_message]
                
                state_dict["is_command"] = True
                return state_dict
        
        return command_handler
    
    def _process_command(self, command: str, args_str: str, state: Dict[str, Any]) -> Dict[str, Any]:
        """
        Process a command and return the result.
        
        Args:
            command: Command name
            args_str: Command arguments string
            state: Current state
            
        Returns:
            Result dictionary with message and any additional data
        """
        # Initialize result
        result = {"success": False, "message": "Unknown command."}
        
        if command == "add_agent":
            # Parse agent configuration arguments
            try:
                # Extract key-value pairs from args string
                arg_pairs = re.findall(r'(\w+)\s*:\s*([^,]+)(?:,|$)', args_str)
                config = {}
                
                for key, value in arg_pairs:
                    config[key.strip()] = value.strip()
                
                # Required fields
                required = ["name", "description", "system_prompt"]
                for field in required:
                    if field not in config:
                        return {
                            "success": False, 
                            "message": f"Missing required field '{field}' for add_agent command."
                        }
                
                # Add optional fields with defaults
                if "model" not in config:
                    config["model"] = "gpt-4o"
                if "temperature" not in config:
                    config["temperature"] = "0.7"
                
                # Convert temperature to float
                try:
                    config["temperature"] = float(config["temperature"])
                except ValueError:
                    config["temperature"] = 0.7
                
                # Create agent config
                agent_config = {
                    "type": "simple",
                    "system_prompt": config["system_prompt"],
                    "model": config["model"],
                    "temperature": config["temperature"],
                    "description": config["description"],
                    "name": config["name"]
                }
                
                # Add to state
                name = config["name"]
                description = config["description"]
                
                # Initialize state fields if needed
                if "available_routes" not in state:
                    state["available_routes"] = {}
                if "agent_configs" not in state:
                    state["agent_configs"] = {}
                
                # Add agent to state
                state["available_routes"][name] = description
                state["agent_configs"][name] = agent_config
                
                result = {
                    "success": True,
                    "message": f"✅ Agent '{name}' added successfully.\n\nDescription: {description}\nSystem prompt: {config['system_prompt'][:100]}...",
                    "agent": agent_config
                }
                
            except Exception as e:
                result = {
                    "success": False,
                    "message": f"⚠️ Error adding agent: {str(e)}\n\nUse format: !add_agent name: Name, description: Description, system_prompt: Prompt, [model: model_name], [temperature: 0.7]"
                }
        
        elif command == "remove_agent":
            # Parse agent name
            try:
                # Extract name from args
                match = re.search(r'name\s*:\s*([^,]+)(?:,|$)', args_str)
                if not match:
                    return {
                        "success": False,
                        "message": "Missing agent name. Use format: !remove_agent name: Agent Name"
                    }
                
                name = match.group(1).strip()
                
                # Check if agent exists
                if name not in state.get("available_routes", {}):
                    return {
                        "success": False,
                        "message": f"⚠️ Agent '{name}' not found."
                    }
                
                # Remove agent from state
                del state["available_routes"][name]
                if name in state.get("agent_configs", {}):
                    del state["agent_configs"][name]
                
                result = {
                    "success": True,
                    "message": f"✅ Agent '{name}' removed successfully."
                }
                
            except Exception as e:
                result = {
                    "success": False,
                    "message": f"⚠️ Error removing agent: {str(e)}\n\nUse format: !remove_agent name: Agent Name"
                }
        
        elif command == "list_agents":
            # List available agents
            try:
                available_routes = state.get("available_routes", {})
                
                if not available_routes:
                    result = {
                        "success": True,
                        "message": "No agents available. Add agents with !add_agent command."
                    }
                else:
                    # Format agent list
                    agent_list = "\n\n".join([
                        f"**{name}**: {description}"
                        for name, description in available_routes.items()
                    ])
                    
                    result = {
                        "success": True,
                        "message": f"📋 Available Agents:\n\n{agent_list}",
                        "agents": list(available_routes.keys())
                    }
                
            except Exception as e:
                result = {
                    "success": False,
                    "message": f"⚠️ Error listing agents: {str(e)}"
                }
        
        elif command == "help":
            # Show help message
            help_message = (
                "🔹 **Available Commands**:\n\n"
                "- `!add_agent name: Agent Name, description: Description, system_prompt: Agent system prompt, [model: model_name], [temperature: 0.7]`\n"
                "- `!remove_agent name: Agent Name`\n"
                "- `!list_agents`\n"
                "- `!help`\n\n"
                "To route to an agent, simply send a message and the supervisor will route it appropriately."
            )
            
            result = {
                "success": True,
                "message": help_message
            }
        
        return result
    
    def _create_supervisor_node_function(self) -> Callable:
        """Create the initial supervisor node function."""
        def supervisor_node(state: Dict[str, Any]) -> Dict[str, Any]:
            """Initial supervisor node that checks for commands."""
            try:
                # Convert state to dictionary if necessary
                state_dict = state.model_dump() if hasattr(state, "model_dump") else state
                
                # Track execution
                state_dict["execution_trace"].append({
                    "node": "supervisor",
                    "timestamp": str(datetime.now()),
                    "route": state_dict.get("current_route", "supervisor")
                })
                
                # Reset command flag
                state_dict["is_command"] = False
                state_dict["command_result"] = None
                
                # Update current route to supervisor
                state_dict["current_route"] = "supervisor"
                
                # Initialize agent data if needed
                if "available_routes" not in state_dict:
                    state_dict["available_routes"] = {}
                if "agent_configs" not in state_dict:
                    state_dict["agent_configs"] = {}
                
                # Add default agents if none exist
                if not state_dict["available_routes"] and self._default_agents:
                    for name, config in self._default_agents.items():
                        state_dict["available_routes"][name] = config.get("description", f"Agent {name}")
                        state_dict["agent_configs"][name] = config
                
                return state_dict
            
            except Exception as e:
                # Handle errors
                state_dict = state.model_dump() if hasattr(state, "model_dump") else state
                state_dict["error"] = f"Error in supervisor node: {str(e)}"
                return state_dict
        
        return supervisor_node
    
    def _check_for_command(self, state: Dict[str, Any]) -> str:
        """Check if the last message is a command."""
        try:
            # Convert state to dictionary if necessary
            state_dict = state.model_dump() if hasattr(state, "model_dump") else state
            
            # Get the last message
            last_message = None
            if "messages" in state_dict and state_dict["messages"]:
                last_message = state_dict["messages"][-1]
            
            if not last_message or not hasattr(last_message, "content"):
                return "continue"
            
            # Check if it's a command
            content = last_message.content
            if self._command_pattern.match(content):
                return "command_handler"
            
            return "continue"
            
        except Exception as e:
            # Error checking for command, continue with routing
            state_dict = state.model_dump() if hasattr(state, "model_dump") else state
            state_dict["error"] = f"Error checking for command: {str(e)}"
            return "continue"
    
    def _create_supervisor_routing_function(self) -> Callable:
        """Create the supervisor routing function."""
        def supervisor_routing(state: Dict[str, Any]) -> Dict[str, Any]:
            """Supervisor node that decides routing."""
            try:
                # Convert state to dictionary if necessary
                state_dict = state.model_dump() if hasattr(state, "model_dump") else state
                
                # Skip if we just processed a command
                if state_dict.get("is_command", False):
                    return state_dict
                
                # Get available routes with descriptions
                available_routes = state_dict.get("available_routes", {})
                
                # If no routes available, add default "help" response
                if not available_routes and not self._default_agents:
                    help_message = (
                        "👋 I'm your agent supervisor. I can route your requests to specialized agents, "
                        "but you don't have any agents configured yet.\n\n"
                        "To add an agent, use the command:\n"
                        "`!add_agent name: Agent Name, description: Description, system_prompt: Agent system prompt`\n\n"
                        "For more help, type `!help`"
                    )
                    
                    help_response = AIMessage(content=help_message)
                    state_dict["messages"] = state_dict["messages"] + [help_response]
                    state_dict["current_route"] = "END"
                    return state_dict
                
                # Track execution
                state_dict["execution_trace"].append({
                    "node": "supervisor_route",
                    "timestamp": str(datetime.now()),
                    "available_routes": list(available_routes.keys())
                })
                
                # Get the last user message for routing
                last_message = None
                if "messages" in state_dict and state_dict["messages"]:
                    for msg in reversed(state_dict["messages"]):
                        if hasattr(msg, "type") and msg.type == "human":
                            last_message = msg.content
                            break
                        elif isinstance(msg, HumanMessage):
                            last_message = msg.content
                            break
                        elif isinstance(msg, tuple) and msg[0] == "human":
                            last_message = msg[1]
                            break
                
                # If no message, return END
                if not last_message:
                    state_dict["current_route"] = "END"
                    return state_dict
                
                # Format options for the prompt
                options_str = ", ".join([f'"{name}": {desc}' for name, desc in available_routes.items()])
                
                # Get engine
                engine = self.config.engine.create_runnable()
                
                # Prepare input for the LLM
                engine_input = {
                    "messages": state_dict["messages"],
                    "options": options_str
                }
                
                # Get routing decision from LLM
                response = engine.invoke(engine_input)
                
                # Extract route from response
                route = self._extract_route_from_response(response)
                
                # Update the current route in state
                state_dict["current_route"] = route
                
                return state_dict
                
            except Exception as e:
                # Handle errors
                state_dict = state.model_dump() if hasattr(state, "model_dump") else state
                state_dict["error"] = f"Error in supervisor routing: {str(e)}"
                state_dict["current_route"] = "END"
                return state_dict
        
        return supervisor_routing
    
    def _route_request(self, state: Dict[str, Any]) -> str:
        """Route the request to either agent_executor or END."""
        try:
            # Convert state to dictionary if necessary
            state_dict = state.model_dump() if hasattr(state, "model_dump") else state
            
            # Get the current route
            route = state_dict.get("current_route", "END")
            
            # Validate the route
            valid_routes = list(state_dict.get("available_routes", {}).keys())
            
            if route == "END":
                return "END"
            elif route in valid_routes:
                return "agent_executor"
            else:
                # Invalid route, return END
                state_dict["error"] = f"Invalid route: {route}"
                return "END"
            
        except Exception as e:
            # Error in routing, return END
            state_dict = state.model_dump() if hasattr(state, "model_dump") else state
            state_dict["error"] = f"Error in routing: {str(e)}"
            return "END"
    
    def _extract_route_from_response(self, response: Any) -> str:
        """Extract the route name from the LLM response."""
        if isinstance(response, str):
            # Direct string response
            return response.strip()
        elif hasattr(response, "content") and isinstance(response.content, str):
            # AIMessage or similar with content
            return response.content.strip()
        elif isinstance(response, dict) and "route" in response:
            # Structured output with route field
            return response["route"]
        elif isinstance(response, dict) and "content" in response:
            # Dict with content field
            return response["content"].strip()
        else:
            # Try to convert to string and extract first word
            try:
                return str(response).strip().split()[0]
            except:
                return "END"
    
    def _create_agent_executor_function(self) -> Callable:
        """Create the generic agent executor node function."""
        def agent_executor(state: Dict[str, Any]) -> Dict[str, Any]:
            """Generic agent executor that runs the agent specified in current_route."""
            try:
                # Convert state to dictionary if necessary
                state_dict = state.model_dump() if hasattr(state, "model_dump") else state
                
                # Get the current route
                current_route = state_dict.get("current_route")
                if not current_route or current_route == "supervisor" or current_route == "END":
                    state_dict["error"] = f"Invalid route for agent_executor: {current_route}"
                    return state_dict
                
                # Track execution
                state_dict["execution_trace"].append({
                    "node": "agent_executor",
                    "timestamp": str(datetime.now()),
                    "route": current_route
                })
                
                # Get the agent configuration
                agent_config_dict = state_dict.get("agent_configs", {}).get(current_route)
                if not agent_config_dict:
                    state_dict["error"] = f"No agent configuration found for route: {current_route}"
                    # Add error message
                    error_msg = AIMessage(content=f"Error: No agent configuration found for '{current_route}'")
                    state_dict["messages"] = state_dict["messages"] + [error_msg]
                    return state_dict
                
                # Check if we already have an instantiated agent
                agent_instance = None
                agent_instances = state_dict.get("agent_instances", {})
                
                if current_route in agent_instances:
                    agent_instance = agent_instances[current_route]
                else:
                    # Try to instantiate the agent from configuration
                    try:
                        # Determine agent type from configuration
                        agent_type = agent_config_dict.get("type", "simple")
                        
                        if agent_type == "simple":
                            # Create a simple agent
                            from src.haive.agents.simple.agent import create_simple_agent
                            
                            agent_instance = create_simple_agent(
                                system_prompt=agent_config_dict.get("system_prompt", "You are a helpful assistant."),
                                model=agent_config_dict.get("model", "gpt-4o"),
                                temperature=agent_config_dict.get("temperature", 0.7),
                                name=agent_config_dict.get("name", f"agent_{current_route}")
                            )
                        else:
                            # For other agent types, would need appropriate creation logic
                            raise ValueError(f"Unsupported agent type: {agent_type}")
                        
                        # Store in agent_instances
                        if "agent_instances" not in state_dict:
                            state_dict["agent_instances"] = {}
                        state_dict["agent_instances"][current_route] = agent_instance
                        
                    except Exception as e:
                        state_dict["error"] = f"Error instantiating agent for route {current_route}: {str(e)}"
                        # Add error message
                        error_msg = AIMessage(content=f"Error: Failed to initialize agent '{current_route}': {str(e)}")
                        state_dict["messages"] = state_dict["messages"] + [error_msg]
                        return state_dict
                
                # Run the agent with the current state
                try:
                    # Clone the state for the agent to avoid modifying the original
                    agent_state = {
                        "messages": state_dict.get("messages", [])
                    }
                    
                    # Add a notification that we're switching to the specialized agent
                    transition_msg = AIMessage(
                        content=f"🔄 Routing to **{current_route}** agent: *{agent_config_dict.get('description', '')}*"
                    )
                    agent_state["messages"] = agent_state["messages"] + [transition_msg]
                    
                    # Run the agent
                    result = agent_instance.run(agent_state)
                    
                    # Update the state with the agent's result
                    if "messages" in result:
                        # Extract only new messages from the result
                        original_count = len(agent_state["messages"])
                        new_messages = result["messages"][original_count:]
                        
                        # Add a prefix to each AI message to indicate which agent responded
                        prefixed_messages = []
                        for msg in new_messages:
                            if isinstance(msg, AIMessage):
                                # Create a new message with the prefix
                                agent_msg = AIMessage(
                                    content=f"[{current_route}] {msg.content}",
                                    tool_calls=msg.tool_calls if hasattr(msg, "tool_calls") else None
                                )
                                prefixed_messages.append(agent_msg)
                            else:
                                prefixed_messages.append(msg)
                        
                        # Update state messages
                        state_dict["messages"] = state_dict["messages"] + [transition_msg] + prefixed_messages
                    
                    # Store any other results
                    state_dict["results"][current_route] = {
                        "timestamp": str(datetime.now()),
                        "result": {k: v for k, v in result.items() if k != "messages"}
                    }
                    
                    return state_dict
                    
                except Exception as e:
                    state_dict["error"] = f"Error executing agent for route {current_route}: {str(e)}"
                    # Add error message
                    error_msg = AIMessage(content=f"Error: Agent '{current_route}' execution failed: {str(e)}")
                    state_dict["messages"] = state_dict["messages"] + [error_msg]
                    return state_dict
            
            except Exception as e:
                # Handle errors
                state_dict = state.model_dump() if hasattr(state, "model_dump") else state
                state_dict["error"] = f"Error in agent_executor: {str(e)}"
                # Add error message if messages exist
                if "messages" in state_dict:
                    error_msg = AIMessage(content=f"Error in agent executor: {str(e)}")
                    state_dict["messages"] = state_dict["messages"] + [error_msg]
                return state_dict
```
