```python
from typing import Type, List, Any, Union, Dict
import inspect
import logging

logger = logging.getLogger(__name__)

class SyntheticLiteral:
    """
    Utility class for creating synthetic Literal types at runtime.
    Enables dynamic typing for langgraph routes that aren't known at compile time.
    """
    
    @staticmethod
    def create(values: List[str]) -> Type:
        """
        Create a synthetic Literal type with the given values.
        
        Args:
            values: The list of string values to include in the Literal
            
        Returns:
            A Literal type with the specified values
        """
        from typing import Literal
        
        # Ensure we have at least one value and END is included
        if not values:
            values = ["END"]
        elif "END" not in values:
            values.append("END")
            
        # Create a string representation of the Literal type
        values_str = ", ".join([f'"{val}"' for val in values])
        type_str = f"Literal[{values_str}]"
        
        # Create the Literal type dynamically
        namespace = {"Literal": Literal}
        for val in values:
            namespace[val] = val
            
        # Execute the string to create the type
        try:
            literal_type = eval(type_str, namespace)
            return literal_type
        except Exception as e:
            logger.error(f"Error creating Literal type: {e}")
            # Return fallback literal with just END
            return Literal["END"]
    
    @staticmethod
    def validate_route(route: str, valid_routes: List[str], default_route: str = "END") -> str:
        """
        Validate a route against a list of valid routes.
        
        Args:
            route: The route to validate
            valid_routes: List of valid routes
            default_route: Default route to return if invalid
            
        Returns:
            Valid route or default route
        """
        if route == "END" or route in valid_routes:
            return "END" if route == "END" else route
        else:
            return default_route
    
    @staticmethod
    def get_route_type(state: Dict[str, Any]) -> Type:
        """
        Extract valid routes from state and create a Literal type.
        
        Args:
            state: State dictionary containing available_routes
            
        Returns:
            Literal type for routes
        """
        routes = list(state.get("available_routes", {}).keys())
        return SyntheticLiteral.create(routes)
```

```python
"""
MetaStateSchema - A higher-level schema manager that provides state schema for dynamic agents.
This system allows for runtime schema modification and composition.
"""
import json
import logging
import inspect
from typing import Dict, Any, List, Optional, Union, Type, Callable, Set, Tuple
from enum import Enum
from copy import deepcopy
from datetime import datetime

from pydantic import BaseModel, Field, create_model, validator
from langchain_core.messages import BaseMessage
from langgraph.graph import add_messages

logger = logging.getLogger(__name__)

class SchemaField:
    """Represents a field in a dynamic schema."""
    def __init__(self, 
                field_type: Type, 
                default=None, 
                required: bool = False,
                description: Optional[str] = None,
                metadata: Optional[Dict[str, Any]] = None):
        """
        Initialize a schema field.
        
        Args:
            field_type: The type of the field
            default: Default value
            required: Whether the field is required
            description: Field description
            metadata: Additional metadata
        """
        self.field_type = field_type
        self.default = default
        self.required = required
        self.description = description or ""
        self.metadata = metadata or {}
    
    def to_dict(self) -> Dict[str, Any]:
        """Convert to dictionary representation."""
        return {
            "field_type": self.field_type.__name__,
            "default": self.default,
            "required": self.required,
            "description": self.description,
            "metadata": self.metadata
        }
    
    def to_field_info(self) -> Field:
        """Convert to Pydantic Field."""
        if self.required:
            return Field(..., description=self.description)
        else:
            if callable(self.default) and self.default.__name__ == "default_factory":
                return Field(default_factory=self.default, description=self.description)
            else:
                return Field(default=self.default, description=self.description)

class SchemaCategory(str, Enum):
    """Categories of schemas for organization."""
    AGENT = "agent"
    TOOL = "tool"
    ROUTER = "router"
    STATE = "state"
    NODE = "node"
    CUSTOM = "custom"

class MetaStateSchema:
    """
    A flexible schema manager that can:
    1. Create, modify, and compose schemas at runtime
    2. Generate Pydantic models dynamically
    3. Serialize/deserialize schemas to/from JSON
    4. Manage agent state schemas dynamically
    """
    
    def __init__(self, name: str, category: SchemaCategory = SchemaCategory.CUSTOM):
        """
        Initialize a new meta schema.
        
        Args:
            name: The name of the schema
            category: Schema category for organization
        """
        self.name = name
        self.category = category
        self.fields: Dict[str, SchemaField] = {}
        self.description = ""
        self.metadata = {}
        self.created_at = datetime.now()
        self.updated_at = datetime.now()
        self._model_cache = None
    
    def add_field(self, 
                name: str, 
                field_type: Type, 
                default=None, 
                required: bool = False,
                description: Optional[str] = None,
                metadata: Optional[Dict[str, Any]] = None) -> 'MetaStateSchema':
        """
        Add a field to the schema.
        
        Args:
            name: Field name
            field_type: Field type
            default: Default value or factory
            required: Whether the field is required
            description: Field description
            metadata: Additional metadata
            
        Returns:
            Self for chaining
        """
        field = SchemaField(
            field_type=field_type,
            default=default,
            required=required,
            description=description,
            metadata=metadata
        )
        
        self.fields[name] = field
        self.updated_at = datetime.now()
        # Invalidate model cache
        self._model_cache = None
        return self
    
    def remove_field(self, name: str) -> 'MetaStateSchema':
        """
        Remove a field from the schema.
        
        Args:
            name: Field name to remove
            
        Returns:
            Self for chaining
        """
        if name in self.fields:
            del self.fields[name]
            self.updated_at = datetime.now()
            # Invalidate model cache
            self._model_cache = None
        return self
    
    def has_field(self, name: str) -> bool:
        """
        Check if the schema has a field with the given name.
        
        Args:
            name: Field name to check
            
        Returns:
            Whether the field exists
        """
        return name in self.fields
    
    def get_field(self, name: str) -> Optional[SchemaField]:
        """
        Get a field by name.
        
        Args:
            name: Field name
            
        Returns:
            SchemaField instance or None if not found
        """
        return self.fields.get(name)
    
    def merge(self, other: 'MetaStateSchema') -> 'MetaStateSchema':
        """
        Merge this schema with another.
        
        Args:
            other: The schema to merge with
            
        Returns:
            New merged schema
        """
        # Create a new schema
        merged = MetaStateSchema(f"{self.name}_merged", self.category)
        
        # Copy fields from this schema
        for name, field in self.fields.items():
            merged.add_field(
                name=name,
                field_type=field.field_type,
                default=field.default,
                required=field.required,
                description=field.description,
                metadata=field.metadata
            )
        
        # Add fields from other schema
        for name, field in other.fields.items():
            # Skip "messages" field if it already exists
            if name == "messages" and name in merged.fields:
                continue
                
            # Add other fields
            if name not in merged.fields:
                merged.add_field(
                    name=name,
                    field_type=field.field_type,
                    default=field.default,
                    required=field.required,
                    description=field.description,
                    metadata=field.metadata
                )
        
        return merged
    
    def to_dict(self) -> Dict[str, Any]:
        """
        Convert to dictionary representation.
        
        Returns:
            Dictionary representation of the schema
        """
        fields_dict = {}
        for name, field in self.fields.items():
            field_dict = field.to_dict()
            field_dict["field_type"] = field.field_type.__name__
            fields_dict[name] = field_dict
            
        return {
            "name": self.name,
            "category": self.category,
            "fields": fields_dict,
            "description": self.description,
            "metadata": self.metadata,
            "created_at": self.created_at.isoformat(),
            "updated_at": self.updated_at.isoformat()
        }
    
    def to_json(self, indent: int = 2) -> str:
        """
        Convert to JSON string.
        
        Args:
            indent: Indentation level
            
        Returns:
            JSON string representation
        """
        return json.dumps(self.to_dict(), indent=indent, default=str)
    
    @classmethod
    def from_dict(cls, data: Dict[str, Any]) -> 'MetaStateSchema':
        """
        Create a schema from a dictionary.
        
        Args:
            data: Dictionary representation
            
        Returns:
            New MetaStateSchema instance
        """
        schema = cls(
            name=data["name"],
            category=data.get("category", SchemaCategory.CUSTOM)
        )
        
        schema.description = data.get("description", "")
        schema.metadata = data.get("metadata", {})
        
        # Parse created_at and updated_at if present
        if "created_at" in data:
            try:
                schema.created_at = datetime.fromisoformat(data["created_at"])
            except ValueError:
                pass
                
        if "updated_at" in data:
            try:
                schema.updated_at = datetime.fromisoformat(data["updated_at"])
            except ValueError:
                pass
        
        # Add fields
        for name, field_data in data.get("fields", {}).items():
            field_type_name = field_data.get("field_type", "str")
            
            # Convert field type name to actual type
            try:
                field_type = eval(field_type_name)
            except:
                # Default to str if type can't be resolved
                field_type = str
            
            schema.add_field(
                name=name,
                field_type=field_type,
                default=field_data.get("default"),
                required=field_data.get("required", False),
                description=field_data.get("description", ""),
                metadata=field_data.get("metadata", {})
            )
        
        return schema
    
    @classmethod
    def from_json(cls, json_str: str) -> 'MetaStateSchema':
        """
        Create a schema from a JSON string.
        
        Args:
            json_str: JSON string representation
            
        Returns:
            New MetaStateSchema instance
        """
        data = json.loads(json_str)
        return cls.from_dict(data)
    
    @classmethod
    def from_model(cls, model: Type[BaseModel], name: Optional[str] = None) -> 'MetaStateSchema':
        """
        Create a schema from a Pydantic model.
        
        Args:
            model: Pydantic model class
            name: Optional name (defaults to model class name)
            
        Returns:
            New MetaStateSchema instance
        """
        schema = cls(name=name or model.__name__)
        
        # Extract fields from model
        for field_name, field_info in model.model_fields.items():
            # Extract type information
            field_type = field_info.annotation
            
            # Extract default value
            default = field_info.default
            
            # Determine if the field is required
            required = field_info.is_required()
            
            # Extract description
            description = field_info.description or ""
            
            # Add field to schema
            schema.add_field(
                name=field_name,
                field_type=field_type,
                default=default,
                required=required,
                description=description
            )
        
        return schema
    
    def get_model(self) -> Type[BaseModel]:
        """
        Generate a Pydantic model from this schema.
        
        Returns:
            Pydantic model class
        """
        # Return cached model if available
        if self._model_cache is not None:
            return self._model_cache
        
        # Create field dictionary for create_model
        fields = {}
        for name, field in self.fields.items():
            fields[name] = (field.field_type, field.to_field_info())
        
        # Create model
        model = create_model(
            self.name,
            __doc__=self.description,
            **fields
        )
        
        # Set model config
        model.model_config = {
            "arbitrary_types_allowed": True,
            "extra": "ignore"
        }
        
        # Cache the model
        self._model_cache = model
        
        return model
    
    def create_instance(self, **kwargs) -> BaseModel:
        """
        Create an instance of the model with the given field values.
        
        Args:
            **kwargs: Field values
            
        Returns:
            Model instance
        """
        model_class = self.get_model()
        return model_class(**kwargs)
    
    def validate_data(self, data: Dict[str, Any]) -> Dict[str, Any]:
        """
        Validate data against this schema.
        
        Args:
            data: Data to validate
            
        Returns:
            Validated data (with defaults applied)
        """
        model_class = self.get_model()
        instance = model_class(**data)
        return instance.model_dump()
    
    def pretty_print(self) -> None:
        """Print a readable representation of the schema."""
        print(f"Schema: {self.name} ({self.category})")
        print(f"Description: {self.description}")
        print(f"Fields:")
        for name, field in self.fields.items():
            type_name = field.field_type.__name__
            default_str = str(field.default) if field.default is not None else "None"
            req_str = "Required" if field.required else "Optional"
            print(f"  - {name}: {type_name} ({req_str}, Default: {default_str})")
            if field.description:
                print(f"      {field.description}")
        print(f"Created: {self.created_at}")
        print(f"Updated: {self.updated_at}")

class MetaSchemaRegistry:
    """
    Registry for managing multiple schemas.
    """
    
    def __init__(self):
        """Initialize the registry."""
        self.schemas: Dict[str, MetaStateSchema] = {}
    
    def register(self, schema: MetaStateSchema) -> bool:
        """
        Register a schema with the registry.
        
        Args:
            schema: Schema to register
            
        Returns:
            Success status
        """
        self.schemas[schema.name] = schema
        return True
    
    def unregister(self, name: str) -> bool:
        """
        Unregister a schema.
        
        Args:
            name: Schema name
            
        Returns:
            Success status
        """
        if name in self.schemas:
            del self.schemas[name]
            return True
        return False
    
    def get(self, name: str) -> Optional[MetaStateSchema]:
        """
        Get a schema by name.
        
        Args:
            name: Schema name
            
        Returns:
            Schema instance or None if not found
        """
        return self.schemas.get(name)
    
    def list_schemas(self, category: Optional[SchemaCategory] = None) -> List[str]:
        """
        List registered schema names.
        
        Args:
            category: Optional category filter
            
        Returns:
            List of schema names
        """
        if category is None:
            return list(self.schemas.keys())
        else:
            return [
                name
                for name, schema in self.schemas.items()
                if schema.category == category
            ]
    
    def to_dict(self) -> Dict[str, Dict[str, Any]]:
        """
        Convert all schemas to a dictionary.
        
        Returns:
            Dictionary of schema dictionaries
        """
        return {
            name: schema.to_dict()
            for name, schema in self.schemas.items()
        }
    
    def to_json(self, indent: int = 2) -> str:
        """
        Convert all schemas to JSON.
        
        Args:
            indent: Indentation level
            
        Returns:
            JSON string representation
        """
        return json.dumps(self.to_dict(), indent=indent, default=str)
    
    @classmethod
    def from_dict(cls, data: Dict[str, Dict[str, Any]]) -> 'MetaSchemaRegistry':
        """
        Create a registry from a dictionary.
        
        Args:
            data: Dictionary of schema dictionaries
            
        Returns:
            New registry instance
        """
        registry = cls()
        
        for name, schema_data in data.items():
            schema = MetaStateSchema.from_dict(schema_data)
            registry.register(schema)
        
        return registry
    
    @classmethod
    def from_json(cls, json_str: str) -> 'MetaSchemaRegistry':
        """
        Create a registry from a JSON string.
        
        Args:
            json_str: JSON string representation
            
        Returns:
            New registry instance
        """
        data = json.loads(json_str)
        return cls.from_dict(data)

# Create default schemas for common use cases

def create_default_agent_state_schema() -> MetaStateSchema:
    """
    Create a default schema for agent state.
    
    Returns:
        Schema instance
    """
    from typing import Dict, Any, List, Optional, Sequence
    
    schema = MetaStateSchema("DefaultAgentState", SchemaCategory.AGENT)
    
    # Add standard fields
    schema.add_field(
        name="messages", 
        field_type=Annotated[Sequence[BaseMessage], add_messages],
        default=list,
        description="Messages in the conversation"
    )
    
    schema.add_field(
        name="current_route", 
        field_type=str,
        default="supervisor",
        description="Current active route/agent"
    )
    
    schema.add_field(
        name="available_routes", 
        field_type=Dict[str, str],
        default=dict,
        description="Map of route names to descriptions"
    )
    
    schema.add_field(
        name="agent_configs", 
        field_type=Dict[str, Dict[str, Any]],
        default=dict,
        description="Map of agent configurations by route name"
    )
    
    schema.add_field(
        name="results", 
        field_type=Dict[str, Any],
        default=dict,
        description="Results from agent execution"
    )
    
    schema.add_field(
        name="error", 
        field_type=Optional[str],
        default=None,
        description="Error message if any occurred"
    )
    
    schema.add_field(
        name="execution_trace", 
        field_type=List[Dict[str, Any]],
        default=list,
        description="Trace of execution steps"
    )
    
    schema.add_field(
        name="is_command", 
        field_type=bool,
        default=False,
        description="Flag indicating if current message is a command"
    )
    
    schema.add_field(
        name="command_result", 
        field_type=Optional[Dict[str, Any]],
        default=None,
        description="Result data from command execution"
    )
    
    return schema

# Create a global registry instance
schema_registry = MetaSchemaRegistry()

# Register default schemas
schema_registry.register(create_default_agent_state_schema())
```

```python
"""
AgentStateManager - A utility for managing agent instances outside the Pydantic model.
This solves the problem of serializing/deserializing Agent instances within state.
"""
import uuid
import logging
import threading
from typing import Dict, Any, Optional, Type, Union, List
from datetime import datetime

from src.haive.core.engine.agent.agent import Agent, AgentConfig
from src.haive.agents.simple.agent import SimpleAgent, create_simple_agent

logger = logging.getLogger(__name__)

class AgentStateManager:
    """
    Manages agent instances separately from the serializable state.
    
    This class:
    1. Stores agent instances in memory keyed by thread_id and route
    2. Creates agents on demand from configurations
    3. Provides thread-safe access to agent instances
    4. Acts as a cache to avoid recreating agents unnecessarily
    """
    
    _instance = None
    _lock = threading.RLock()
    
    def __new__(cls):
        """Implement as a singleton to ensure shared state across imports."""
        with cls._lock:
            if cls._instance is None:
                cls._instance = super(AgentStateManager, cls).__new__(cls)
                cls._instance._initialized = False
            return cls._instance
    
    def __init__(self):
        """Initialize the manager."""
        if self._initialized:
            return
            
        # Dict structure: thread_id -> route_name -> agent_instance
        self._agents = {}
        
        # Track last access time for potential cleanup
        self._last_access = {}
        
        self._initialized = True
        logger.info("AgentStateManager initialized as singleton")
    
    def get_agent(self, 
                thread_id: str, 
                route: str, 
                agent_config: Optional[Dict[str, Any]] = None) -> Optional[Agent]:
        """
        Get an agent instance for the given thread and route.
        Creates it if necessary using the provided config.
        
        Args:
            thread_id: Thread ID for this conversation
            route: Route/name of the agent
            agent_config: Optional configuration to create the agent if not found
            
        Returns:
            Agent instance or None if it can't be created/found
        """
        with self._lock:
            # Mark this thread/route as accessed
            now = datetime.now()
            if thread_id not in self._last_access:
                self._last_access[thread_id] = {}
            self._last_access[thread_id][route] = now
            
            # Check if we already have this agent
            if thread_id in self._agents and route in self._agents[thread_id]:
                return self._agents[thread_id][route]
            
            # No agent exists, check if we have config to create it
            if not agent_config:
                return None
                
            # Try to create agent from config
            try:
                agent_type = agent_config.get("type", "simple")
                
                if agent_type == "simple":
                    # Create a simple agent
                    agent = create_simple_agent(
                        system_prompt=agent_config.get("system_prompt", "You are a helpful assistant."),
                        model=agent_config.get("model", "gpt-4o"),
                        temperature=agent_config.get("temperature", 0.7),
                        name=agent_config.get("name", f"agent_{route}")
                    )
                else:
                    # For other agent types, would need appropriate creation logic
                    logger.error(f"Unsupported agent type: {agent_type}")
                    return None
                
                # Store the agent
                if thread_id not in self._agents:
                    self._agents[thread_id] = {}
                self._agents[thread_id][route] = agent
                
                return agent
                
            except Exception as e:
                logger.error(f"Error creating agent for route {route}: {str(e)}")
                return None
    
    def store_agent(self, thread_id: str, route: str, agent: Agent) -> bool:
        """
        Store an agent instance for the given thread and route.
        
        Args:
            thread_id: Thread ID for this conversation
            route: Route/name of the agent
            agent: Agent instance
            
        Returns:
            Success status
        """
        with self._lock:
            # Mark this thread/route as accessed
            now = datetime.now()
            if thread_id not in self._last_access:
                self._last_access[thread_id] = {}
            self._last_access[thread_id][route] = now
            
            # Store the agent
            if thread_id not in self._agents:
                self._agents[thread_id] = {}
            self._agents[thread_id][route] = agent
            
            return True
    
    def remove_agent(self, thread_id: str, route: str) -> bool:
        """
        Remove an agent instance for the given thread and route.
        
        Args:
            thread_id: Thread ID for this conversation
            route: Route/name of the agent
            
        Returns:
            Success status
        """
        with self._lock:
            if thread_id in self._agents and route in self._agents[thread_id]:
                del self._agents[thread_id][route]
                
                # Remove from last access too
                if thread_id in self._last_access and route in self._last_access[thread_id]:
                    del self._last_access[thread_id][route]
                
                # Clean up empty dictionaries
                if not self._agents[thread_id]:
                    del self._agents[thread_id]
                if thread_id in self._last_access and not self._last_access[thread_id]:
                    del self._last_access[thread_id]
                
                return True
            return False
    
    def cleanup_old_agents(self, max_age_minutes: int = 60) -> int:
        """
        Clean up agent instances that haven't been accessed for a while.
        
        Args:
            max_age_minutes: Maximum age in minutes before cleanup
            
        Returns:
            Number of agents cleaned up
        """
        with self._lock:
            now = datetime.now()
            count = 0
            
            # Identify old agents
            to_remove = []  # List of (thread_id, route) tuples
            
            for thread_id, routes in self._last_access.items():
                for route, last_access in routes.items():
                    age = (now - last_access).total_seconds() / 60.0
                    if age > max_age_minutes:
                        to_remove.append((thread_id, route))
            
            # Remove old agents
            for thread_id, route in to_remove:
                if self.remove_agent(thread_id, route):
                    count += 1
            
            return count
    
    def get_agent_counts(self) -> Dict[str, int]:
        """
        Get counts of agents by thread ID.
        
        Returns:
            Dictionary mapping thread IDs to agent counts
        """
        with self._lock:
            return {
                thread_id: len(routes)
                for thread_id, routes in self._agents.items()
            }

# Global instance of the manager
agent_manager = AgentStateManager()
```

```python
def agent_executor(state: Dict[str, Any]) -> Dict[str, Any]:
            """Generic agent executor that runs the agent specified in current_route."""
            try:
                # Import agent manager
                from agent_state_manager import agent_manager
                
                # Convert state to dictionary if necessary
                state_dict = state.model_dump() if hasattr(state, "model_dump") else state
                
                # Get the current route
                current_route = state_dict.get("current_route")
                if not current_route or current_route == "supervisor" or current_route == "END":
                    state_dict["error"] = f"Invalid route for agent_executor: {current_route}"
                    return state_dict
                
                # Track execution
                state_dict["execution_trace"].append({
                    "node": "agent_executor",
                    "timestamp": str(datetime.now()),
                    "route": current_route
                })
                
                # Get the agent configuration
                agent_config_dict = state_dict.get("agent_configs", {}).get(current_route)
                if not agent_config_dict:
                    state_dict["error"] = f"No agent configuration found for route: {current_route}"
                    # Add error message
                    error_msg = AIMessage(content=f"Error: No agent configuration found for '{current_route}'")
                    state_dict["messages"] = state_dict["messages"] + [error_msg]
                    return state_dict
                
                # Get thread_id from config
                thread_id = state_dict.get("thread_id", str(uuid.uuid4()))
                
                # Get or create agent instance using the manager
                agent_instance = agent_manager.get_agent(thread_id, current_route, agent_config_dict)
                
                if not agent_instance:
                    state_dict["error"] = f"Failed to create/get agent for route {current_route}"
                    # Add error message
                    error_msg = AIMessage(content=f"Error: Failed to initialize agent '{current_route}'")
                    state_dict["messages"] = state_dict["messages"] + [error_msg]
                    return state_dict
                
                # Run the agent with the current state
                try:
                    # Clone the state for the agent to avoid modifying the original
                    agent_state = {
                        "messages": state_dict.get("messages", [])
                    }
                    
                    # Add a notification that we're switching to the specialized agent
                    transition_msg = AIMessage(
                        content=f"🔄 Routing to **{current_route}** agent: *{agent_config_dict.get('description', '')}*"
                    )
                    agent_state["messages"] = agent_state["messages"] + [transition_msg]
                    
                    # Run the agent
                    result = agent_instance.run(agent_state)
                    
                    # Update the state with the agent's result
                    if "messages" in result:
                        # Extract only new messages from the result
                        original_count = len(agent_state["messages"])
                        new_messages = result["messages"][original_count:]
                        
                        # Add a prefix to each AI message to indicate which agent responded
                        prefixed_messages = []
                        for msg in new_messages:
                            if isinstance(msg, AIMessage):
                                # Create a new message with the prefix
                                agent_msg = AIMessage(
                                    content=f"[{current_route}] {msg.content}",
                                    tool_calls=msg.tool_calls if hasattr(msg, "tool_calls") else None
                                )
                                prefixed_messages.append(agent_msg)
                            else:
                                prefixed_messages.append(msg)
                        
                        # Update state messages
                        state_dict["messages"] = state_dict["messages"] + [transition_msg] + prefixed_messages
                    
                    # Store any other results
                    state_dict["results"][current_route] = {
                        "timestamp": str(datetime.now()),
                        "result": {k: v for k, v in result.items() if k != "messages"}
                    }
                    
                    # Store thread_id for future reference
                    state_dict["thread_id"] = thread_id
                    
                    return state_dict
                    
                except Exception as e:
                    state_dict["error"] = f"Error executing agent for route {current_route}: {str(e)}"
                    # Add error message
                    error_msg = AIMessage(content=f"Error: Agent '{current_route}' execution failed: {str(e)}")
                    state_dict["messages"] = state_dict["messages"] + [error_msg]
                    return state_dict
            
            except Exception as e:
                # Handle errors
                state_dict = state.model_dump() if hasattr(state, "model_dump") else state
                state_dict["error"] = f"Error in agent_executor: {str(e)}"
                # Add error message if messages exist
                if "messages" in state_dict:
                    error_msg = AIMessage(content=f"Error in agent executor: {str(e)}")
                    state_dict["messages"] = state_dict["messages"] + [error_msg]
                return state_dict
```

```python
"""
Fixed implementation of the Dynamic Supervisor with error corrections
"""
import logging
import uuid
import re
from datetime import datetime
from typing import Dict, List, Any, Optional, Union, Type, Callable, Literal
from enum import Enum

from pydantic import BaseModel, Field
from langchain_core.messages import AIMessage, HumanMessage, SystemMessage, BaseMessage
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.runnables import Runnable

from langgraph.graph import END, StateGraph
from langgraph.types import Command
from langgraph.graph import add_messages
from typing import Annotated, Sequence

from src.haive.core.engine.agent.agent import Agent, AgentConfig, register_agent
from src.haive.agents.simple.agent import SimpleAgent, SimpleAgentConfig, create_simple_agent
from src.haive.core.engine.aug_llm import AugLLMConfig
from src.haive.core.models.llm.base import AzureLLMConfig
from src.haive.core.graph.GraphBuilder import DynamicGraph

logger = logging.getLogger(__name__)

# ==========================
# 🔹 Agent Command Types
# ==========================
class AgentCommand(str, Enum):
    """Command types for agent management via messages."""
    ADD_AGENT = "add_agent"
    REMOVE_AGENT = "remove_agent"
    LIST_AGENTS = "list_agents"
    HELP = "help"

# ==========================
# 🔹 Agent Configuration Model
# ==========================
class AgentConfigModel(BaseModel):
    """Model for agent configuration via messages."""
    name: str = Field(..., description="Name/route for the agent")
    description: str = Field(..., description="Description of what the agent does")
    system_prompt: str = Field(..., description="System prompt for the agent")
    model: str = Field(default="gpt-4o", description="Model to use for the agent")
    temperature: float = Field(default=0.7, description="Temperature for generation")
    
    def to_dict(self) -> Dict[str, Any]:
        """Convert to a configuration dictionary."""
        return {
            "type": "simple",
            "system_prompt": self.system_prompt,
            "model": self.model,
            "temperature": self.temperature,
            "description": self.description,
            "name": self.name
        }

# ==========================
# 🔹 Enhanced State Schema with Agent Management
# ==========================
# Add this to your DynamicAgentState class definition:
class DynamicAgentState(BaseModel):
    """State schema with agent configurations and management capabilities."""
    
    # Standard message field using langgraph's add_messages reducer
    messages: Annotated[Sequence[BaseMessage], add_messages] = Field(
        default_factory=list,
        description="Messages in the conversation"
    )
    
    # Current route with reducer to handle concurrent updates - NEW ANNOTATION
    current_route: Annotated[str, select_highest] = Field(
        default="supervisor", 
        description="Current active route/agent"
    )
    
    # Available routes and their descriptions
    available_routes: Dict[str, str] = Field(
        default_factory=dict,
        description="Map of route names to descriptions"
    )
    
    # Agent configurations by route name
    agent_configs: Dict[str, Dict[str, Any]] = Field(
        default_factory=dict,
        description="Map of agent configurations by route name"
    )
    
    # Thread ID for this conversation
    thread_id: str = Field(
        default_factory=lambda: str(uuid.uuid4()),
        description="Thread ID for this conversation"
    )
    
    # Execution results
    results: Dict[str, Any] = Field(
        default_factory=dict,
        description="Results from agent execution"
    )
    
    # Error tracking - use add_to_each reducer to combine errors
    error: Annotated[Optional[str], add_to_each("\n")] = Field(
        default=None,
        description="Error message if any occurred"
    )
    
    # Execution metadata and tracing
    execution_trace: List[Dict[str, Any]] = Field(
        default_factory=list,
        description="Trace of execution steps"
    )
    
    # Command processing flag
    is_command: bool = Field(
        default=False,
        description="Flag indicating if current message is a command"
    )
    
    # Command result data
    command_result: Optional[Dict[str, Any]] = Field(
        default=None,
        description="Result data from command execution"
    )
    
    class Config:
        # Allow arbitrary types to be used in fields
        arbitrary_types_allowed = True
        # Exclude unset fields in serialization
        exclude_unset = True
# ==========================
# 🔹 Dynamic Supervisor Config
# ==========================
class DynamicSupervisorConfig(AgentConfig):
    """Configuration for a supervisor agent with truly dynamic routing capabilities."""
    
    # System prompt for the supervisor
    supervisor_system_prompt: str = Field(
        default=(
            "You are a supervisor agent responsible for routing requests to specialized agents "
            "and managing the agent pool. You can route requests to any available agent, or "
            "respond with 'END' if the task is complete.\n\n"
            "You can also respond to special commands:\n"
            "- To add an agent: `!add_agent name: Agent Name, description: Description, "
            "system_prompt: Agent system prompt, [model: model_name], [temperature: 0.7]`\n"
            "- To remove an agent: `!remove_agent name: Agent Name`\n"
            "- To list available agents: `!list_agents`\n"
            "- For help: `!help`\n\n"
            "When routing, respond with ONLY the name of an available agent or 'END'."
        ),
        description="System prompt template for the supervisor LLM."
    )
    
    # Default schema override
    state_schema: Type[BaseModel] = Field(
        default=DynamicAgentState,
        description="State schema for the agent"
    )
    
    # Default agents to register
    default_agents: Dict[str, Dict[str, Any]] = Field(
        default_factory=dict,
        description="Default agents to register at initialization"
    )
    
    # Command pattern regex
    command_pattern: str = Field(
        default=r"^\!(\w+)(?:\s+(.+))?$",
        description="Regex pattern for detecting commands in messages"
    )
    
    @classmethod
    def from_scratch(cls,
                    system_prompt: Optional[str] = None,
                    default_agents: Optional[Dict[str, Dict[str, Any]]] = None,
                    model: str = "gpt-4o",
                    temperature: float = 0.2,
                    name: Optional[str] = None,
                    **kwargs) -> 'DynamicSupervisorConfig':
        """
        Create a DynamicSupervisorConfig from scratch.
        
        Args:
            system_prompt: Optional system prompt override
            default_agents: Optional default agents to register
            model: Model name to use
            temperature: Temperature for generation
            name: Optional agent name
            **kwargs: Additional kwargs for the config
            
        Returns:
            DynamicSupervisorConfig instance
        """
        # Create LLM config
        llm_config = AzureLLMConfig(
            model=model,
            parameters={"temperature": temperature}
        )
        
        # Use provided system prompt or default
        # Fixed to use model_fields instead of __fields__
        system_prompt = system_prompt or cls.model_fields["supervisor_system_prompt"].default
        
        # Create prompt template
        messages = [
            SystemMessage(content=system_prompt),
            MessagesPlaceholder(variable_name="messages")
        ]
        prompt_template = ChatPromptTemplate.from_messages(messages)
        
        # Create AugLLM config
        aug_llm = AugLLMConfig(
            name=f"{name or 'supervisor'}_llm",
            llm_config=llm_config,
            prompt_template=prompt_template
        )
        
        # Create the config
        config = cls(
            name=name or f"dynamic_supervisor",
            engine=aug_llm,
            system_prompt=system_prompt,
            default_agents=default_agents or {},
            **kwargs
        )
        
        return config

# ==========================
# 🔹 Dynamic Supervisor Agent
# ==========================
@register_agent(DynamicSupervisorConfig)
class DynamicSupervisorAgent(Agent[DynamicSupervisorConfig]):
    """
    A truly dynamic supervisor agent with agent management through messages.
    
    This implementation:
    1. Uses a single generic agent node for all agent executions
    2. Stores agent configurations in state for runtime flexibility
    3. Can add/remove/modify agents without recompiling the graph
    4. Allows agent management through special message commands
    """
    
    def __init__(self, config: DynamicSupervisorConfig):
        """Initialize the supervisor agent with its configuration."""
        self._default_agents = config.default_agents or {}
        self._command_pattern = re.compile(config.command_pattern)
        
        # Initialize the agent manager for handling agent instances
        self._agent_instances = {}
        
        # Call parent constructor after setting our attributes
        super().__init__(config)
    
    def setup_workflow(self):
        """Set up a workflow with a supervisor, command handler, and generic agent node."""
        # Create DynamicGraph with proper component registration
        gb = DynamicGraph(
            components=[self.config.engine],
            state_schema=self.state_schema
        )
        
        # Add the command handler node that processes commands
        gb.add_node(
            name="command_handler",
            config=self._create_command_handler_function(),
            command_goto="supervisor"  # Return to supervisor after command processing
        )
        
        # Add the supervisor node that decides routing
        gb.add_node(
            name="supervisor",
            config=self._create_supervisor_node_function(),
            command_goto=None  # Will be determined by routing logic
        )
        
        # Explicitly create the agent executor function first
        agent_executor_fn = self._create_agent_executor_function()
        
        # Add a generic agent execution node
        gb.add_node(
            name="agent_executor",
            config=agent_executor_fn,  # Use the created function
            command_goto="supervisor"  # Always return to supervisor after execution
        )
        
        # Add END node for completion
        gb.add_node(
            name="finish",
            config=self._create_finish_node_function(),
            command_goto=END
        )
        
        # Set up conditional routing from supervisor to check for commands first
        gb.add_conditional_edges(
            "supervisor",
            self._check_for_command,
            {
                "command_handler": "command_handler",
                "continue": "supervisor_route"
            }
        )
        
        # Add supervisor routing node
        gb.add_node(
            name="supervisor_route",
            config=self._create_supervisor_routing_function(),
            command_goto=None  # Will be determined by routing logic
        )
        
        # Set up conditional routing from supervisor_route
        gb.add_conditional_edges(
            "supervisor_route",
            self._route_request,
            {
                "agent_executor": "agent_executor",
                "END": "finish"
            }
        )
        
        # Set the entry point to the supervisor
        gb.set_entry_point("supervisor")
        
        # Get the built graph
        self.graph = gb.build()
    
    def _create_command_handler_function(self) -> Callable:
        """Create the command handler node function."""
        def command_handler(state: Dict[str, Any]) -> Dict[str, Any]:
            """Handle special commands for agent management."""
            try:
                # Convert state to dictionary if necessary
                state_dict = state.model_dump() if hasattr(state, "model_dump") else state
                
                # Get the last message
                last_message = None
                if "messages" in state_dict and state_dict["messages"]:
                    last_message = state_dict["messages"][-1]
                
                if not last_message or not hasattr(last_message, "content"):
                    state_dict["is_command"] = False
                    return state_dict
                
                # Extract command and arguments
                content = last_message.content
                match = self._command_pattern.match(content)
                
                if not match:
                    state_dict["is_command"] = False
                    return state_dict
                
                command = match.group(1).lower()
                args_str = match.group(2) if match.group(2) else ""
                
                # Track execution
                state_dict["execution_trace"].append({
                    "node": "command_handler",
                    "timestamp": str(datetime.now()),
                    "command": command,
                    "args": args_str
                })
                
                # Process command
                result = self._process_command(command, args_str, state_dict)
                
                # Add result message
                result_message = AIMessage(content=result.get("message", "Command processed."))
                state_dict["messages"] = state_dict["messages"] + [result_message]
                
                # Store command result
                state_dict["command_result"] = result
                state_dict["is_command"] = True
                
                return state_dict
                
            except Exception as e:
                # Handle errors
                state_dict = state.model_dump() if hasattr(state, "model_dump") else state
                error_msg = f"Error processing command: {str(e)}"
                state_dict["error"] = error_msg
                
                # Add error message
                result_message = AIMessage(content=f"⚠️ {error_msg}")
                if "messages" in state_dict:
                    state_dict["messages"] = state_dict["messages"] + [result_message]
                
                state_dict["is_command"] = True
                return state_dict
        
        return command_handler
    
    def _process_command(self, command: str, args_str: str, state: Dict[str, Any]) -> Dict[str, Any]:
        """
        Process a command and return the result.
        
        Args:
            command: Command name
            args_str: Command arguments string
            state: Current state
            
        Returns:
            Result dictionary with message and any additional data
        """
        # Initialize result
        result = {"success": False, "message": "Unknown command."}
        
        if command == "add_agent":
            # Parse agent configuration arguments
            try:
                # Extract key-value pairs from args string
                arg_pairs = re.findall(r'(\w+)\s*:\s*([^,]+)(?:,|$)', args_str)
                config = {}
                
                for key, value in arg_pairs:
                    config[key.strip()] = value.strip()
                
                # Required fields
                required = ["name", "description", "system_prompt"]
                for field in required:
                    if field not in config:
                        return {
                            "success": False, 
                            "message": f"Missing required field '{field}' for add_agent command."
                        }
                
                # Add optional fields with defaults
                if "model" not in config:
                    config["model"] = "gpt-4o"
                if "temperature" not in config:
                    config["temperature"] = "0.7"
                
                # Convert temperature to float
                try:
                    config["temperature"] = float(config["temperature"])
                except ValueError:
                    config["temperature"] = 0.7
                
                # Create agent config
                agent_config = {
                    "type": "simple",
                    "system_prompt": config["system_prompt"],
                    "model": config["model"],
                    "temperature": config["temperature"],
                    "description": config["description"],
                    "name": config["name"]
                }
                
                # Add to state
                name = config["name"]
                description = config["description"]
                
                # Initialize state fields if needed
                if "available_routes" not in state:
                    state["available_routes"] = {}
                if "agent_configs" not in state:
                    state["agent_configs"] = {}
                
                # Add agent to state
                state["available_routes"][name] = description
                state["agent_configs"][name] = agent_config
                
                result = {
                    "success": True,
                    "message": f"✅ Agent '{name}' added successfully.\n\nDescription: {description}\nSystem prompt: {config['system_prompt'][:100]}...",
                    "agent": agent_config
                }
                
            except Exception as e:
                result = {
                    "success": False,
                    "message": f"⚠️ Error adding agent: {str(e)}\n\nUse format: !add_agent name: Name, description: Description, system_prompt: Prompt, [model: model_name], [temperature: 0.7]"
                }
        
        elif command == "remove_agent":
            # Parse agent name
            try:
                # Extract name from args
                match = re.search(r'name\s*:\s*([^,]+)(?:,|$)', args_str)
                if not match:
                    return {
                        "success": False,
                        "message": "Missing agent name. Use format: !remove_agent name: Agent Name"
                    }
                
                name = match.group(1).strip()
                
                # Check if agent exists
                if name not in state.get("available_routes", {}):
                    return {
                        "success": False,
                        "message": f"⚠️ Agent '{name}' not found."
                    }
                
                # Remove agent from state
                del state["available_routes"][name]
                if name in state.get("agent_configs", {}):
                    del state["agent_configs"][name]
                
                result = {
                    "success": True,
                    "message": f"✅ Agent '{name}' removed successfully."
                }
                
            except Exception as e:
                result = {
                    "success": False,
                    "message": f"⚠️ Error removing agent: {str(e)}\n\nUse format: !remove_agent name: Agent Name"
                }
        
        elif command == "list_agents":
            # List available agents
            try:
                available_routes = state.get("available_routes", {})
                
                if not available_routes:
                    result = {
                        "success": True,
                        "message": "No agents available. Add agents with !add_agent command."
                    }
                else:
                    # Format agent list
                    agent_list = "\n\n".join([
                        f"**{name}**: {description}"
                        for name, description in available_routes.items()
                    ])
                    
                    result = {
                        "success": True,
                        "message": f"📋 Available Agents:\n\n{agent_list}",
                        "agents": list(available_routes.keys())
                    }
                
            except Exception as e:
                result = {
                    "success": False,
                    "message": f"⚠️ Error listing agents: {str(e)}"
                }
        
        elif command == "help":
            # Show help message
            help_message = (
                "🔹 **Available Commands**:\n\n"
                "- `!add_agent name: Agent Name, description: Description, system_prompt: Agent system prompt, [model: model_name], [temperature: 0.7]`\n"
                "- `!remove_agent name: Agent Name`\n"
                "- `!list_agents`\n"
                "- `!help`\n\n"
                "To route to an agent, simply send a message and the supervisor will route it appropriately."
            )
            
            result = {
                "success": True,
                "message": help_message
            }
        
        return result
    
    def _create_supervisor_node_function(self) -> Callable:
        """Create the initial supervisor node function."""
        def supervisor_node(state: Dict[str, Any]) -> Dict[str, Any]:
            """Initial supervisor node that checks for commands."""
            try:
                # Convert state to dictionary if necessary
                state_dict = state.model_dump() if hasattr(state, "model_dump") else state
                
                # Track execution
                state_dict["execution_trace"].append({
                    "node": "supervisor",
                    "timestamp": str(datetime.now()),
                    "route": state_dict.get("current_route", "supervisor")
                })
                
                # Reset command flag
                state_dict["is_command"] = False
                state_dict["command_result"] = None
                
                # DON'T update current_route here to avoid conflict
                # state_dict["current_route"] = "supervisor"
                
                # Initialize agent data if needed
                if "available_routes" not in state_dict:
                    state_dict["available_routes"] = {}
                if "agent_configs" not in state_dict:
                    state_dict["agent_configs"] = {}
                
                # Add default agents if none exist
                if not state_dict["available_routes"] and self._default_agents:
                    for name, config in self._default_agents.items():
                        state_dict["available_routes"][name] = config.get("description", f"Agent {name}")
                        state_dict["agent_configs"][name] = config
                
                return state_dict
            
            except Exception as e:
                # Handle errors
                state_dict = state.model_dump() if hasattr(state, "model_dump") else state
                state_dict["error"] = f"Error in supervisor node: {str(e)}"
                return state_dict
        
        return supervisor_node
    
    def _check_for_command(self, state: Dict[str, Any]) -> str:
        """Check if the last message is a command."""
        try:
            # Convert state to dictionary if necessary
            state_dict = state.model_dump() if hasattr(state, "model_dump") else state
            
            # Get the last message
            last_message = None
            if "messages" in state_dict and state_dict["messages"]:
                last_message = state_dict["messages"][-1]
            
            if not last_message or not hasattr(last_message, "content"):
                return "continue"
            
            # Check if it's a command
            content = last_message.content
            if self._command_pattern.match(content):
                return "command_handler"
            
            return "continue"
            
        except Exception as e:
            # Error checking for command, continue with routing
            state_dict = state.model_dump() if hasattr(state, "model_dump") else state
            state_dict["error"] = f"Error checking for command: {str(e)}"
            return "continue"
    
    def _create_supervisor_routing_function(self) -> Callable:
        """Create the supervisor routing function."""
        def supervisor_routing(state: Dict[str, Any]) -> Dict[str, Any]:
            """Supervisor node that decides routing."""
            try:
                # Convert state to dictionary if necessary
                state_dict = state.model_dump() if hasattr(state, "model_dump") else state
                
                # Skip if we just processed a command
                if state_dict.get("is_command", False):
                    return state_dict
                
                # Get available routes with descriptions
                available_routes = state_dict.get("available_routes", {})
                
                # If no routes available, add default "help" response
                if not available_routes and not self._default_agents:
                    help_message = (
                        "👋 I'm your agent supervisor. I can route your requests to specialized agents, "
                        "but you don't have any agents configured yet.\n\n"
                        "To add an agent, use the command:\n"
                        "`!add_agent name: Agent Name, description: Description, system_prompt: Agent system prompt`\n\n"
                        "For more help, type `!help`"
                    )
                    
                    help_response = AIMessage(content=help_message)
                    state_dict["messages"] = state_dict["messages"] + [help_response]
                    state_dict["current_route"] = "END"
                    return state_dict
                
                # Track execution
                state_dict["execution_trace"].append({
                    "node": "supervisor_route",
                    "timestamp": str(datetime.now()),
                    "available_routes": list(available_routes.keys())
                })
                
                # Get the last user message for routing
                last_message = None
                if "messages" in state_dict and state_dict["messages"]:
                    for msg in reversed(state_dict["messages"]):
                        if hasattr(msg, "type") and msg.type == "human":
                            last_message = msg.content
                            break
                        elif isinstance(msg, HumanMessage):
                            last_message = msg.content
                            break
                        elif isinstance(msg, tuple) and msg[0] == "human":
                            last_message = msg[1]
                            break
                
                # If no message, return END
                if not last_message:
                    state_dict["current_route"] = "END"
                    return state_dict
                
                # Format options for the prompt
                options_str = ", ".join([f'"{name}": {desc}' for name, desc in available_routes.items()])
                
                # Get engine
                engine = self.config.engine.create_runnable()
                
                # Prepare input for the LLM
                engine_input = {
                    "messages": state_dict["messages"],
                    "options": options_str
                }
                
                # Get routing decision from LLM
                response = engine.invoke(engine_input)
                
                # Extract route from response
                route = self._extract_route_from_response(response)
                
                # Update the current route in state
                state_dict["current_route"] = route
                
                return state_dict
                
            except Exception as e:
                # Handle errors
                state_dict = state.model_dump() if hasattr(state, "model_dump") else state
                state_dict["error"] = f"Error in supervisor routing: {str(e)}"
                state_dict["current_route"] = "END"
                return state_dict
        
        return supervisor_routing
    
    def _route_request(self, state: Dict[str, Any]) -> str:
        """Route the request to either agent_executor or END."""
        try:
            # Convert state to dictionary if necessary
            state_dict = state.model_dump() if hasattr(state, "model_dump") else state
            
            # Get the current route
            route = state_dict.get("current_route", "END")
            
            # Validate the route
            valid_routes = list(state_dict.get("available_routes", {}).keys())
            
            if route == "END":
                return "END"
            elif route in valid_routes:
                return "agent_executor"
            else:
                # Invalid route, return END
                state_dict["error"] = f"Invalid route: {route}"
                return "END"
            
        except Exception as e:
            # Error in routing, return END
            state_dict = state.model_dump() if hasattr(state, "model_dump") else state
            state_dict["error"] = f"Error in routing: {str(e)}"
            return "END"
    
    def _extract_route_from_response(self, response: Any) -> str:
        """Extract the route name from the LLM response."""
        if isinstance(response, str):
            # Direct string response
            return response.strip()
        elif hasattr(response, "content") and isinstance(response.content, str):
            # AIMessage or similar with content
            return response.content.strip()
        elif isinstance(response, dict) and "route" in response:
            # Structured output with route field
            return response["route"]
        elif isinstance(response, dict) and "content" in response:
            # Dict with content field
            return response["content"].strip()
        else:
            # Try to convert to string and extract first word
            try:
                return str(response).strip().split()[0]
            except:
                return "END"
    
    def _create_agent_executor_function(self) -> Callable:
        """Create the generic agent executor node function."""
        def agent_executor(state: Dict[str, Any]) -> Dict[str, Any]:
            """Generic agent executor that runs the agent specified in current_route."""
            try:
                # Convert state to dictionary if necessary
                state_dict = state.model_dump() if hasattr(state, "model_dump") else state
                
                # Get the current route
                current_route = state_dict.get("current_route")
                if not current_route or current_route == "supervisor" or current_route == "END":
                    state_dict["error"] = f"Invalid route for agent_executor: {current_route}"
                    return state_dict
                
                # Track execution
                state_dict["execution_trace"].append({
                    "node": "agent_executor",
                    "timestamp": str(datetime.now()),
                    "route": current_route
                })
                
                # Get the agent configuration
                agent_config_dict = state_dict.get("agent_configs", {}).get(current_route)
                if not agent_config_dict:
                    state_dict["error"] = f"No agent configuration found for route: {current_route}"
                    # Add error message
                    error_msg = AIMessage(content=f"Error: No agent configuration found for '{current_route}'")
                    state_dict["messages"] = state_dict["messages"] + [error_msg]
                    return state_dict
                
                # Check if we already have an instantiated agent
                # Use thread_id + route as the key for the agent
                thread_id = state_dict.get("thread_id", str(uuid.uuid4()))
                agent_key = f"{thread_id}_{current_route}"
                
                if agent_key in self._agent_instances:
                    agent_instance = self._agent_instances[agent_key]
                else:
                    # Try to instantiate the agent from configuration
                    try:
                        # Determine agent type from configuration
                        agent_type = agent_config_dict.get("type", "simple")
                        
                        if agent_type == "simple":
                            # Create a simple agent
                            agent_instance = create_simple_agent(
                                system_prompt=agent_config_dict.get("system_prompt", "You are a helpful assistant."),
                                model=agent_config_dict.get("model", "gpt-4o"),
                                temperature=agent_config_dict.get("temperature", 0.7),
                                name=agent_config_dict.get("name", f"agent_{current_route}")
                            )
                        else:
                            # For other agent types, would need appropriate creation logic
                            raise ValueError(f"Unsupported agent type: {agent_type}")
                        
                        # Store in agent_instances
                        self._agent_instances[agent_key] = agent_instance
                        
                    except Exception as e:
                        state_dict["error"] = f"Error instantiating agent for route {current_route}: {str(e)}"
                        # Add error message
                        error_msg = AIMessage(content=f"Error: Failed to initialize agent '{current_route}': {str(e)}")
                        state_dict["messages"] = state_dict["messages"] + [error_msg]
                        return state_dict
                
                # Run the agent with the current state
                try:
                    # Clone the state for the agent to avoid modifying the original
                    agent_state = {
                        "messages": state_dict.get("messages", [])
                    }
                    
                    # Add a notification that we're switching to the specialized agent
                    transition_msg = AIMessage(
                        content=f"🔄 Routing to **{current_route}** agent: *{agent_config_dict.get('description', '')}*"
                    )
                    agent_state["messages"] = agent_state["messages"] + [transition_msg]
                    
                    # Run the agent
                    result = agent_instance.run(agent_state)
                    
                    # Update the state with the agent's result
                    if "messages" in result:
                        # Extract only new messages from the result
                        original_count = len(agent_state["messages"])
                        new_messages = result["messages"][original_count:]
                        
                        # Add a prefix to each AI message to indicate which agent responded
                        prefixed_messages = []
                        for msg in new_messages:
                            if isinstance(msg, AIMessage):
                                # Create a new message with the prefix
                                agent_msg = AIMessage(
                                    content=f"[{current_route}] {msg.content}",
                                    tool_calls=msg.tool_calls if hasattr(msg, "tool_calls") else None
                                )
                                prefixed_messages.append(agent_msg)
                            else:
                                prefixed_messages.append(msg)
                        
                        # Update state messages
                        state_dict["messages"] = state_dict["messages"] + [transition_msg] + prefixed_messages
                    
                    # Store any other results
                    state_dict["results"][current_route] = {
                        "timestamp": str(datetime.now()),
                        "result": {k: v for k, v in result.items() if k != "messages"}
                    }
                    
                    # Store thread_id for future reference
                    state_dict["thread_id"] = thread_id
                    
                    return state_dict
                    
                except Exception as e:
                    state_dict["error"] = f"Error executing agent for route {current_route}: {str(e)}"
                    # Add error message
                    error_msg = AIMessage(content=f"Error: Agent '{current_route}' execution failed: {str(e)}")
                    state_dict["messages"] = state_dict["messages"] + [error_msg]
                    return state_dict
            
            except Exception as e:
                # Handle errors
                state_dict = state.model_dump() if hasattr(state, "model_dump") else state
                state_dict["error"] = f"Error in agent_executor: {str(e)}"
                # Add error message if messages exist
                if "messages" in state_dict:
                    error_msg = AIMessage(content=f"Error in agent executor: {str(e)}")
                    state_dict["messages"] = state_dict["messages"] + [error_msg]
                return state_dict
        
        return agent_executor
        
    def _create_finish_node_function(self) -> Callable:
        """Create the finish node function."""
        def finish_node(state: Dict[str, Any]) -> Dict[str, Any]:
            """Finish node that completes execution."""
            try:
                # Convert state to dictionary if necessary
                state_dict = state.model_dump() if hasattr(state, "model_dump") else state
                
                # Track execution
                state_dict["execution_trace"].append({
                    "node": "finish",
                    "timestamp": str(datetime.now()),
                    "route": "END"
                })
                
                # Update current route to END
                state_dict["current_route"] = "END"
                
                # Add completion message if needed
                if not state_dict.get("is_command", False):
                    last_message = state_dict["messages"][-1] if state_dict.get("messages") else None
                    if not (isinstance(last_message, AIMessage) or 
                            (isinstance(last_message, tuple) and last_message[0] == "ai")):
                        # No AI response yet, add one
                        completion_msg = AIMessage(
                            content="I've completed processing your request. If you need further assistance, please let me know."
                        )
                        state_dict["messages"] = state_dict["messages"] + [completion_msg]
                
                return state_dict
            
            except Exception as e:
                # Handle errors
                state_dict = state.model_dump() if hasattr(state, "model_dump") else state
                state_dict["error"] = f"Error in finish node: {str(e)}"
                return state_dict
        
        return finish_node
# Helper function to create a dynamic supervisor
def create_dynamic_supervisor(
    system_prompt: Optional[str] = None,
    default_agents: Optional[Dict[str, Dict[str, Any]]] = None,
    model: str = "gpt-4o",
    temperature: float = 0.2,
    name: Optional[str] = None,
    **kwargs
) -> DynamicSupervisorAgent:
    """
    Create a dynamic supervisor agent.
    
    Args:
        system_prompt: Optional system prompt override
        default_agents: Optional default agents to register at initialization
        model: Model name to use
        temperature: Temperature for generation
        name: Optional name for the agent
        **kwargs: Additional kwargs for the config
        
    Returns:
        DynamicSupervisorAgent instance
    """
    # Create the configuration
    config = DynamicSupervisorConfig.from_scratch(
        system_prompt=system_prompt,
        default_agents=default_agents or {},
        model=model,
        temperature=temperature,
        name=name,
        **kwargs
    )
    
    # Build and return the agent
    return config.build_agent()
```

```python
"""
Dynamic Supervisor with proper reducers for concurrent state updates
"""
import logging
import uuid
import re
from datetime import datetime
from typing import Dict, List, Any, Optional, Union, Type, Callable
from enum import Enum

from pydantic import BaseModel, Field, create_model
from langchain_core.messages import AIMessage, HumanMessage, SystemMessage, BaseMessage
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.output_parsers import StrOutputParser

from langgraph.graph import END, StateGraph
from langgraph.graph import add_messages
from typing import Annotated, Sequence

from src.haive.core.engine.agent.agent import Agent, AgentConfig, register_agent
from src.haive.agents.simple.agent import create_simple_agent
from src.haive.core.engine.aug_llm import AugLLMConfig
from src.haive.core.models.llm.base import AzureLLMConfig
from src.haive.core.graph.GraphBuilder import DynamicGraph

logger = logging.getLogger(__name__)

# ==============================================
# Reducer functions with proper signatures
# ==============================================
"""
Final fix for the DynamicAgentState class with all fields properly annotated.
"""
from typing import Dict, List, Any, Optional, Sequence
from pydantic import BaseModel, Field
from langchain_core.messages import BaseMessage
from langgraph.graph import add_messages
from typing import Annotated
import uuid

# ==============================================
# Reducer functions with proper signatures
# ==============================================

def select_most_recent(values, new_value):
    """Reducer that selects the most recent value."""
    return new_value

def dict_merge(values, new_value):
    """Reducer that merges dictionaries."""
    if values is None:
        return new_value
    if new_value is None:
        return values
    
    # Create a copy to avoid modifying the original
    result = values.copy()
    # Update with new values
    result.update(new_value)
    return result

def list_append(values, new_value):
    """Reducer that appends to lists."""
    if values is None:
        return new_value if new_value is not None else []
    if new_value is None:
        return values
    
    # Create a copy and append new values
    result = list(values)
    if isinstance(new_value, list):
        result.extend(new_value)
    else:
        result.append(new_value)
    return result

def add_to_each(separator: str = "\n"):
    """Reducer that combines strings with a separator."""
    def _reducer(values, new_value):
        if values is None:
            return new_value
        if new_value is None:
            return values
            
        if isinstance(values, str) and isinstance(new_value, str):
            if not values:
                return new_value
            if not new_value:
                return values
            return f"{values}{separator}{new_value}"
        return new_value
    
    return _reducer

# ==============================================
# Fully corrected DynamicAgentState Schema
# ==============================================
class DynamicAgentState(BaseModel):
    """State schema with agent configurations and management capabilities."""
    
    # Standard message field using langgraph's add_messages reducer
    messages: Annotated[Sequence[BaseMessage], add_messages] = Field(
        default_factory=list,
        description="Messages in the conversation"
    )
    
    # Current route with reducer to handle concurrent updates
    current_route: Annotated[str, select_most_recent] = Field(
        default="supervisor", 
        description="Current active route/agent"
    )
    
    # Available routes with dict_merge reducer for concurrent updates
    available_routes: Annotated[Dict[str, str], dict_merge] = Field(
        default_factory=dict,
        description="Map of route names to descriptions"
    )
    
    # Agent configurations with dict_merge reducer
    agent_configs: Annotated[Dict[str, Dict[str, Any]], dict_merge] = Field(
        default_factory=dict,
        description="Map of agent configurations by route name"
    )
    
    # Thread ID with select_most_recent reducer - THIS FIXES THE THREAD_ID ISSUE
    thread_id: Annotated[str, select_most_recent] = Field(
        default_factory=lambda: str(uuid.uuid4()),
        description="Thread ID for this conversation"
    )
    
    # Execution results with dict_merge reducer
    results: Annotated[Dict[str, Any], dict_merge] = Field(
        default_factory=dict,
        description="Results from agent execution"
    )
    
    # Error tracking with add_to_each reducer
    error: Annotated[Optional[str], add_to_each("\n")] = Field(
        default=None,
        description="Error message if any occurred"
    )
    
    # Execution metadata and tracing with list_append reducer
    execution_trace: Annotated[List[Dict[str, Any]], list_append] = Field(
        default_factory=list,
        description="Trace of execution steps"
    )
    
    # Command processing flag with select_most_recent reducer
    is_command: Annotated[bool, select_most_recent] = Field(
        default=False,
        description="Flag indicating if current message is a command"
    )
    
    # Command result data with select_most_recent reducer
    command_result: Annotated[Optional[Dict[str, Any]], select_most_recent] = Field(
        default=None,
        description="Result data from command execution"
    )
    
    class Config:
        arbitrary_types_allowed = True
        exclude_unset = True
# ==============================================
# Agent State Manager
# ==============================================
class AgentStateManager:
    """Manages agent instances separately from serializable state."""
    
    def __init__(self):
        """Initialize the manager."""
        self._agents = {}
        self._last_access = {}
    
    def get_agent(self, 
                thread_id: str, 
                route: str, 
                agent_config: Optional[Dict[str, Any]] = None) -> Optional[Agent]:
        """Get or create agent instance for thread/route."""
        # Mark this thread/route as accessed
        now = datetime.now()
        if thread_id not in self._last_access:
            self._last_access[thread_id] = {}
        self._last_access[thread_id][route] = now
        
        # Check if we already have this agent
        if thread_id in self._agents and route in self._agents[thread_id]:
            return self._agents[thread_id][route]
        
        # No agent exists, check if we have config to create it
        if not agent_config:
            return None
            
        # Try to create agent from config
        try:
            agent_type = agent_config.get("type", "simple")
            
            if agent_type == "simple":
                # Create a simple agent
                agent = create_simple_agent(
                    system_prompt=agent_config.get("system_prompt", "You are a helpful assistant."),
                    model=agent_config.get("model", "gpt-4o"),
                    temperature=agent_config.get("temperature", 0.7),
                    name=agent_config.get("name", f"agent_{route}")
                )
            else:
                logger.error(f"Unsupported agent type: {agent_type}")
                return None
            
            # Store the agent
            if thread_id not in self._agents:
                self._agents[thread_id] = {}
            self._agents[thread_id][route] = agent
            
            return agent
                
        except Exception as e:
            logger.error(f"Error creating agent for route {route}: {str(e)}")
            return None
    
    def remove_agent(self, thread_id: str, route: str) -> bool:
        """Remove an agent instance for the given thread and route."""
        if thread_id in self._agents and route in self._agents[thread_id]:
            del self._agents[thread_id][route]
            
            # Remove from last access too
            if thread_id in self._last_access and route in self._last_access[thread_id]:
                del self._last_access[thread_id][route]
            
            # Clean up empty dictionaries
            if not self._agents[thread_id]:
                del self._agents[thread_id]
            if thread_id in self._last_access and not self._last_access[thread_id]:
                del self._last_access[thread_id]
            
            return True
        return False

# ==============================================
# Dynamic Supervisor Config
# ==============================================
class DynamicSupervisorConfig(AgentConfig):
    """Configuration for a supervisor agent with truly dynamic routing capabilities."""
    
    # System prompt for the supervisor
    supervisor_system_prompt: str = Field(
        default=(
            "You are a supervisor agent responsible for routing requests to specialized agents "
            "and managing the agent pool. You can route requests to any available agent, or "
            "respond with 'END' if the task is complete.\n\n"
            "You can also respond to special commands:\n"
            "- To add an agent: `!add_agent name: Agent Name, description: Description, "
            "system_prompt: Agent system prompt, [model: model_name], [temperature: 0.7]`\n"
            "- To remove an agent: `!remove_agent name: Agent Name`\n"
            "- To list available agents: `!list_agents`\n"
            "- For help: `!help`\n\n"
            "When routing, respond with ONLY the name of an available agent or 'END'."
        ),
        description="System prompt template for the supervisor LLM."
    )
    
    # Default schema override
    state_schema: Type[BaseModel] = Field(
        default=DynamicAgentState,
        description="State schema for the agent"
    )
    
    # Default agents to register
    default_agents: Dict[str, Dict[str, Any]] = Field(
        default_factory=dict,
        description="Default agents to register at initialization"
    )
    
    # Command pattern regex
    command_pattern: str = Field(
        default=r"^\!(\w+)(?:\s+(.+))?$",
        description="Regex pattern for detecting commands in messages"
    )
    
    @classmethod
    def from_scratch(cls,
                    system_prompt: Optional[str] = None,
                    default_agents: Optional[Dict[str, Dict[str, Any]]] = None,
                    model: str = "gpt-4o",
                    temperature: float = 0.2,
                    name: Optional[str] = None,
                    **kwargs) -> 'DynamicSupervisorConfig':
        """Create a DynamicSupervisorConfig from scratch."""
        # Create LLM config
        llm_config = AzureLLMConfig(
            model=model,
            parameters={"temperature": temperature}
        )
        
        # Use provided system prompt or default
        system_prompt = system_prompt or cls.model_fields["supervisor_system_prompt"].default
        
        # Create prompt template
        messages = [
            SystemMessage(content=system_prompt),
            MessagesPlaceholder(variable_name="messages")
        ]
        prompt_template = ChatPromptTemplate.from_messages(messages)
        
        # Create AugLLM config
        aug_llm = AugLLMConfig(
            name=f"{name or 'supervisor'}_llm",
            llm_config=llm_config,
            prompt_template=prompt_template
        )
        
        # Create the config
        config = cls(
            name=name or f"dynamic_supervisor",
            engine=aug_llm,
            system_prompt=system_prompt,
            default_agents=default_agents or {},
            **kwargs
        )
        
        return config

# ==============================================
# Dynamic Supervisor Agent Implementation
# ==============================================
@register_agent(DynamicSupervisorConfig)
class DynamicSupervisorAgent(Agent[DynamicSupervisorConfig]):
    """
    A truly dynamic supervisor agent with agent management through messages.
    """
    
    def __init__(self, config: DynamicSupervisorConfig):
        """Initialize the supervisor agent with its configuration."""
        self._default_agents = config.default_agents or {}
        self._command_pattern = re.compile(config.command_pattern)
        
        # Initialize the agent manager for handling agent instances
        self._agent_manager = AgentStateManager()
        
        # Call parent constructor after setting our attributes
        super().__init__(config)
    
    def setup_workflow(self):
        """Set up a workflow with a supervisor, command handler, and generic agent node."""
        # Create DynamicGraph with proper component registration
        gb = DynamicGraph(
            components=[self.config.engine],
            state_schema=self.state_schema
        )
        
        # Add the command handler node that processes commands
        gb.add_node(
            name="command_handler",
            config=self._create_command_handler_function(),
            command_goto="supervisor"  # Return to supervisor after command processing
        )
        
        # Add the supervisor node that decides routing
        gb.add_node(
            name="supervisor",
            config=self._create_supervisor_node_function(),
            command_goto=None  # Will be determined by routing logic
        )
        
        # Add a generic agent execution node
        gb.add_node(
            name="agent_executor",
            config=self._create_agent_executor_function(),
            command_goto="supervisor"  # Always return to supervisor after execution
        )
        
        # Add END node for completion
        gb.add_node(
            name="finish",
            config=self._create_finish_node_function(),
            command_goto=END
        )
        
        # Set up conditional routing from supervisor to check for commands first
        gb.add_conditional_edges(
            "supervisor",
            self._check_for_command,
            {
                "command_handler": "command_handler",
                "continue": "supervisor_route"
            }
        )
        
        # Add supervisor routing node
        gb.add_node(
            name="supervisor_route",
            config=self._create_supervisor_routing_function(),
            command_goto=None  # Will be determined by routing logic
        )
        
        # Set up conditional routing from supervisor_route
        gb.add_conditional_edges(
            "supervisor_route",
            self._route_request,
            {
                "agent_executor": "agent_executor",
                "END": "finish"
            }
        )
        
        # Set the entry point to the supervisor
        gb.set_entry_point("supervisor")
        
        # Get the built graph
        self.graph = gb.build()
    
    def _create_command_handler_function(self) -> Callable:
        """Create the command handler node function."""
        def command_handler(state: Dict[str, Any]) -> Dict[str, Any]:
            """Handle special commands for agent management."""
            try:
                # Convert state to dictionary if necessary
                state_dict = state.model_dump() if hasattr(state, "model_dump") else state
                
                # Get the last message
                last_message = None
                if "messages" in state_dict and state_dict["messages"]:
                    last_message = state_dict["messages"][-1]
                
                if not last_message or not hasattr(last_message, "content"):
                    state_dict["is_command"] = False
                    return state_dict
                
                # Extract command and arguments
                content = last_message.content
                match = self._command_pattern.match(content)
                
                if not match:
                    state_dict["is_command"] = False
                    return state_dict
                
                command = match.group(1).lower()
                args_str = match.group(2) if match.group(2) else ""
                
                # Track execution
                state_dict["execution_trace"] = [
                    {
                        "node": "command_handler",
                        "timestamp": str(datetime.now()),
                        "command": command,
                        "args": args_str
                    }
                ]
                
                # Process command
                result = self._process_command(command, args_str, state_dict)
                
                # Add result message
                result_message = AIMessage(content=result.get("message", "Command processed."))
                state_dict["messages"] = state_dict["messages"] + [result_message]
                
                # Store command result
                state_dict["command_result"] = result
                state_dict["is_command"] = True
                
                return state_dict
                
            except Exception as e:
                # Handle errors
                state_dict = state.model_dump() if hasattr(state, "model_dump") else state
                error_msg = f"Error processing command: {str(e)}"
                state_dict["error"] = error_msg
                
                # Add error message
                result_message = AIMessage(content=f"⚠️ {error_msg}")
                if "messages" in state_dict:
                    state_dict["messages"] = state_dict["messages"] + [result_message]
                
                state_dict["is_command"] = True
                return state_dict
        
        return command_handler
    
    def _process_command(self, command: str, args_str: str, state: Dict[str, Any]) -> Dict[str, Any]:
        """Process a command and return the result."""
        # Initialize result
        result = {"success": False, "message": "Unknown command."}
        
        if command == "add_agent":
            # Parse agent configuration arguments
            try:
                # Extract key-value pairs from args string
                arg_pairs = re.findall(r'(\w+)\s*:\s*([^,]+)(?:,|$)', args_str)
                config = {}
                
                for key, value in arg_pairs:
                    config[key.strip()] = value.strip()
                
                # Required fields
                required = ["name", "description", "system_prompt"]
                for field in required:
                    if field not in config:
                        return {
                            "success": False, 
                            "message": f"Missing required field '{field}' for add_agent command."
                        }
                
                # Add optional fields with defaults
                if "model" not in config:
                    config["model"] = "gpt-4o"
                if "temperature" not in config:
                    config["temperature"] = "0.7"
                
                # Convert temperature to float
                try:
                    config["temperature"] = float(config["temperature"])
                except ValueError:
                    config["temperature"] = 0.7
                
                # Create agent config
                agent_config = {
                    "type": "simple",
                    "system_prompt": config["system_prompt"],
                    "model": config["model"],
                    "temperature": config["temperature"],
                    "description": config["description"],
                    "name": config["name"]
                }
                
                # Add to state
                name = config["name"]
                description = config["description"]
                
                # Add agent to state - Use dictionary updates for concurrent updates
                state["available_routes"] = {name: description}
                state["agent_configs"] = {name: agent_config}
                
                result = {
                    "success": True,
                    "message": f"✅ Agent '{name}' added successfully.\n\nDescription: {description}\nSystem prompt: {config['system_prompt'][:100]}...",
                    "agent": agent_config
                }
                
            except Exception as e:
                result = {
                    "success": False,
                    "message": f"⚠️ Error adding agent: {str(e)}\n\nUse format: !add_agent name: Name, description: Description, system_prompt: Prompt, [model: model_name], [temperature: 0.7]"
                }
        
        elif command == "remove_agent":
            # Parse agent name
            try:
                # Extract name from args
                match = re.search(r'name\s*:\s*([^,]+)(?:,|$)', args_str)
                if not match:
                    return {
                        "success": False,
                        "message": "Missing agent name. Use format: !remove_agent name: Agent Name"
                    }
                
                name = match.group(1).strip()
                
                # Check if agent exists
                if name not in state.get("available_routes", {}):
                    return {
                        "success": False,
                        "message": f"⚠️ Agent '{name}' not found."
                    }
                
                # Create new dictionaries without the agent
                new_routes = {k: v for k, v in state.get("available_routes", {}).items() if k != name}
                new_configs = {k: v for k, v in state.get("agent_configs", {}).items() if k != name}
                
                # Update state with new dictionaries
                state["available_routes"] = new_routes
                state["agent_configs"] = new_configs
                
                # Remove from agent manager too
                thread_id = state.get("thread_id", "")
                if thread_id:
                    self._agent_manager.remove_agent(thread_id, name)
                
                result = {
                    "success": True,
                    "message": f"✅ Agent '{name}' removed successfully."
                }
                
            except Exception as e:
                result = {
                    "success": False,
                    "message": f"⚠️ Error removing agent: {str(e)}\n\nUse format: !remove_agent name: Agent Name"
                }
        
        elif command == "list_agents":
            # List available agents
            try:
                available_routes = state.get("available_routes", {})
                
                if not available_routes:
                    result = {
                        "success": True,
                        "message": "No agents available. Add agents with !add_agent command."
                    }
                else:
                    # Format agent list
                    agent_list = "\n\n".join([
                        f"**{name}**: {description}"
                        for name, description in available_routes.items()
                    ])
                    
                    result = {
                        "success": True,
                        "message": f"📋 Available Agents:\n\n{agent_list}",
                        "agents": list(available_routes.keys())
                    }
                
            except Exception as e:
                result = {
                    "success": False,
                    "message": f"⚠️ Error listing agents: {str(e)}"
                }
        
        elif command == "help":
            # Show help message
            help_message = (
                "🔹 **Available Commands**:\n\n"
                "- `!add_agent name: Agent Name, description: Description, system_prompt: Agent system prompt, [model: model_name], [temperature: 0.7]`\n"
                "- `!remove_agent name: Agent Name`\n"
                "- `!list_agents`\n"
                "- `!help`\n\n"
                "To route to an agent, simply send a message and the supervisor will route it appropriately."
            )
            
            result = {
                "success": True,
                "message": help_message
            }
        
        return result
    
    def _create_supervisor_node_function(self) -> Callable:
        """Create the initial supervisor node function."""
        def supervisor_node(state: Dict[str, Any]) -> Dict[str, Any]:
            """Initial supervisor node that checks for commands."""
            try:
                # Convert state to dictionary if necessary
                state_dict = state.model_dump() if hasattr(state, "model_dump") else state
                
                # Track execution
                state_dict["execution_trace"] = [
                    {
                        "node": "supervisor",
                        "timestamp": str(datetime.now()),
                        "route": state_dict.get("current_route", "supervisor")
                    }
                ]
                
                # Reset command flag
                state_dict["is_command"] = False
                state_dict["command_result"] = None
                
                # Add default agents if none exist (only run once)
                if not state_dict.get("available_routes") and self._default_agents:
                    # Create a new dictionary for all default agents
                    new_routes = {}
                    new_configs = {}
                    
                    for name, config in self._default_agents.items():
                        new_routes[name] = config.get("description", f"Agent {name}")
                        new_configs[name] = config
                    
                    # Update state with all default agents at once
                    state_dict["available_routes"] = new_routes
                    state_dict["agent_configs"] = new_configs
                
                return state_dict
            
            except Exception as e:
                # Handle errors
                state_dict = state.model_dump() if hasattr(state, "model_dump") else state
                state_dict["error"] = f"Error in supervisor node: {str(e)}"
                return state_dict
        
        return supervisor_node
    
    def _check_for_command(self, state: Dict[str, Any]) -> str:
        """Check if the last message is a command."""
        try:
            # Convert state to dictionary if necessary
            state_dict = state.model_dump() if hasattr(state, "model_dump") else state
            
            # Get the last message
            last_message = None
            if "messages" in state_dict and state_dict["messages"]:
                last_message = state_dict["messages"][-1]
            
            if not last_message or not hasattr(last_message, "content"):
                return "continue"
            
            # Check if it's a command
            content = last_message.content
            if self._command_pattern.match(content):
                return "command_handler"
            
            return "continue"
            
        except Exception as e:
            # Error checking for command, continue with routing
            logger.error(f"Error checking for command: {e}")
            return "continue"
    
    def _create_supervisor_routing_function(self) -> Callable:
        """Create the supervisor routing function."""
        def supervisor_routing(state: Dict[str, Any]) -> Dict[str, Any]:
            """Supervisor node that decides routing."""
            try:
                # Convert state to dictionary if necessary
                state_dict = state.model_dump() if hasattr(state, "model_dump") else state
                
                # Skip if we just processed a command
                if state_dict.get("is_command", False):
                    return state_dict
                
                # Get available routes with descriptions
                available_routes = state_dict.get("available_routes", {})
                
                # If no routes available, add default "help" response
                if not available_routes:
                    help_message = (
                        "👋 I'm your agent supervisor. I can route your requests to specialized agents, "
                        "but you don't have any agents configured yet.\n\n"
                        "To add an agent, use the command:\n"
                        "`!add_agent name: Agent Name, description: Description, system_prompt: Agent system prompt`\n\n"
                        "For more help, type `!help`"
                    )
                    
                    help_response = AIMessage(content=help_message)
                    state_dict["messages"] = state_dict["messages"] + [help_response]
                    state_dict["current_route"] = "END"
                    return state_dict
                
                # Add trace
                state_dict["execution_trace"] = [
                    {
                        "node": "supervisor_route",
                        "timestamp": str(datetime.now()),
                        "available_routes": list(available_routes.keys())
                    }
                ]
                
                # Get the last user message for routing
                last_message = None
                if "messages" in state_dict and state_dict["messages"]:
                    for msg in reversed(state_dict["messages"]):
                        if hasattr(msg, "type") and msg.type == "human":
                            last_message = msg.content
                            break
                        elif isinstance(msg, HumanMessage):
                            last_message = msg.content
                            break
                        elif isinstance(msg, tuple) and msg[0] == "human":
                            last_message = msg[1]
                            break
                
                # If no message, return END
                if not last_message:
                    state_dict["current_route"] = "END"
                    return state_dict
                
                # Format options for the prompt
                options_str = "\n".join([
                    f"- {name}: {desc}" 
                    for name, desc in available_routes.items()
                ])
                
                # Create a simple routing prompt
                prompt = f"""Based on the user message, determine which agent is best suited to handle it.

Available agents:
{options_str}

If the task is complete or no agent is appropriate, respond with 'END'.
Respond ONLY with the name of an agent or 'END'. No explanation or additional text.

User message: {last_message}

Your routing decision (respond with exactly one word):"""

                # Get engine
                llm = self.config.engine.llm_config.instantiate_llm()
                parser = StrOutputParser()
                
                # Get routing decision from LLM
                response = llm.invoke(prompt) | parser
                route = response.strip()
                
                # Validate the route against available agents
                valid_routes = list(available_routes.keys())
                if route not in valid_routes and route != "END":
                    logger.warning(f"Invalid route '{route}', defaulting to END")
                    route = "END"
                
                # Update the current route in state
                state_dict["current_route"] = route
                
                return state_dict
                
            except Exception as e:
                # Handle errors
                state_dict = state.model_dump() if hasattr(state, "model_dump") else state
                state_dict["error"] = f"Error in supervisor routing: {str(e)}"
                state_dict["current_route"] = "END"
                return state_dict
        
        return supervisor_routing
    
    def _route_request(self, state: Dict[str, Any]) -> str:
        """Route the request to either agent_executor or END."""
        try:
            # Convert state to dictionary if necessary
            state_dict = state.model_dump() if hasattr(state, "model_dump") else state
            
            # Get the current route
            route = state_dict.get("current_route", "END")
            
            # Validate the route
            valid_routes = list(state_dict.get("available_routes", {}).keys())
            
            if route == "END":
                return "END"
            elif route in valid_routes:
                return "agent_executor"
            else:
                # Invalid route, return END
                return "END"
            
        except Exception as e:
            # Error in routing, return END
            logger.error(f"Error in route_request: {e}")
            return "END"
    
    def _create_agent_executor_function(self) -> Callable:
        """Create the generic agent executor node function."""
        def agent_executor(state: Dict[str, Any]) -> Dict[str, Any]:
            """Generic agent executor that runs the agent specified in current_route."""
            try:
                # Convert state to dictionary if necessary
                state_dict = state.model_dump() if hasattr(state, "model_dump") else state
                
                # Get the current route
                current_route = state_dict.get("current_route")
                if not current_route or current_route == "supervisor" or current_route == "END":
                    state_dict["error"] = f"Invalid route for agent_executor: {current_route}"
                    return state_dict
                
                # Track execution
                state_dict["execution_trace"] = [
                    {
                        "node": "agent_executor",
                        "timestamp": str(datetime.now()),
                        "route": current_route
                    }
                ]
                
                # Get the agent configuration
                agent_config_dict = state_dict.get("agent_configs", {}).get(current_route)
                if not agent_config_dict:
                    state_dict["error"] = f"No agent configuration found for route: {current_route}"
                    # Add error message
                    error_msg = AIMessage(content=f"Error: No agent configuration found for '{current_route}'")
                    state_dict["messages"] = state_dict["messages"] + [error_msg]
                    return state_dict
                
                # Get thread_id for agent persistence
                thread_id = state_dict.get("thread_id", str(uuid.uuid4()))
                
                # Get agent from the manager
                agent_instance = self._agent_manager.get_agent(thread_id, current_route, agent_config_dict)
                
                if not agent_instance:
                    state_dict["error"] = f"Failed to create/get agent for route {current_route}"
                    # Add error message
                    error_msg = AIMessage(content=f"Error: Failed to initialize agent '{current_route}'")
                    state_dict["messages"] = state_dict["messages"] + [error_msg]
                    return state_dict
                
                # Run the agent with the current state
                try:
                    # Clone the state for the agent to avoid modifying the original
                    agent_state = {
                        "messages": state_dict.get("messages", [])
                    }
                    
                    # Add a notification that we're switching to the specialized agent
                    transition_msg = AIMessage(
                        content=f"🔄 Routing to **{current_route}** agent: *{agent_config_dict.get('description', '')}*"
                    )
                    agent_state["messages"] = agent_state["messages"] + [transition_msg]
                    
                    # Run the agent
                    result = agent_instance.run(agent_state)
                    
                    # Update the state with the agent's result
                    if "messages" in result:
                        # Extract only new messages from the result
                        original_count = len(agent_state["messages"])
                        new_messages = result["messages"][original_count:]
                        
                        # Add a prefix to each AI message to indicate which agent responded
                        prefixed_messages = []
                        for msg in new_messages:
                            if isinstance(msg, AIMessage):
                                # Create a new message with the prefix
                                agent_msg = AIMessage(
                                    content=f"[{current_route}] {msg.content}",
                                    tool_calls=msg.tool_calls if hasattr(msg, "tool_calls") else None
                                )
                                prefixed_messages.append(agent_msg)
                            else:
                                prefixed_messages.append(msg)
                        
                        # Update state messages
                        state_dict["messages"] = state_dict["messages"] + [transition_msg] + prefixed_messages
                    
                    # Store any other results
                    state_dict["results"] = {
                        current_route: {
                            "timestamp": str(datetime.now()),
                            "result": {k: v for k, v in result.items() if k != "messages"}
                        }
                    }
                    
                    # Store thread_id for future reference
                    state_dict["thread_id"] = thread_id
                    
                    return state_dict
                    
                except Exception as e:
                    state_dict["error"] = f"Error executing agent for route {current_route}: {str(e)}"
                    # Add error message
                    error_msg = AIMessage(content=f"Error: Agent '{current_route}' execution failed: {str(e)}")
                    state_dict["messages"] = state_dict["messages"] + [error_msg]
                    return state_dict
            
            except Exception as e:
                # Handle errors
                state_dict = state.model_dump() if hasattr(state, "model_dump") else state
                state_dict["error"] = f"Error in agent_executor: {str(e)}"
                # Add error message if messages exist
                if "messages" in state_dict:
                    error_msg = AIMessage(content=f"Error in agent executor: {str(e)}")
                    state_dict["messages"] = state_dict["messages"] + [error_msg]
                return state_dict
        
        return agent_executor
    
    def _create_finish_node_function(self) -> Callable:
        """Create the finish node function."""
        def finish_node(state: Dict[str, Any]) -> Dict[str, Any]:
            """Finish node that completes execution."""
            try:
                # Convert state to dictionary if necessary
                state_dict = state.model_dump() if hasattr(state, "model_dump") else state
                
                # Track execution
                state_dict["execution_trace"] = [
                    {
                        "node": "finish",
                        "timestamp": str(datetime.now()),
                        "route": "END"
                    }
                ]
                
                # Update current route to END
                state_dict["current_route"] = "END"
                
                # Add completion message if needed
                if not state_dict.get("is_command", False):
                    last_message = state_dict["messages"][-1] if state_dict.get("messages") else None
                    if not (isinstance(last_message, AIMessage) or 
                            (isinstance(last_message, tuple) and last_message[0] == "ai")):
                        # No AI response yet, add one
                        completion_msg = AIMessage(
                            content="I've completed processing your request. If you need further assistance, please let me know."
                        )
                        state_dict["messages"] = state_dict["messages"] + [completion_msg]
                
                return state_dict
            
            except Exception as e:
                # Handle errors
                state_dict = state.model_dump() if hasattr(state, "model_dump") else state
                state_dict["error"] = f"Error in finish node: {str(e)}"
                return state_dict
        
        return finish_node

# ==============================================
# Helper function to create a dynamic supervisor
# ==============================================
def create_dynamic_supervisor(
    system_prompt: Optional[str] = None,
    default_agents: Optional[Dict[str, Dict[str, Any]]] = None,
    model: str = "gpt-4o",
    temperature: float = 0.2,
    name: Optional[str] = None,
    **kwargs
) -> DynamicSupervisorAgent:
    """
    Create a dynamic supervisor agent with structured output routing.
    
    Args:
        system_prompt: Optional system prompt override
        default_agents: Optional default agents to register at initialization
        model: Model name to use
        temperature: Temperature for generation
        name: Optional name for the agent
        **kwargs: Additional kwargs for the config
        
    Returns:
        DynamicSupervisorAgent instance
    """
    # Create the configuration
    config = DynamicSupervisorConfig.from_scratch(
        system_prompt=system_prompt,
        default_agents=default_agents or {},
        model=model,
        temperature=temperature,
        name=name,
        **kwargs
    )
    
    # Build and return the agent
    return config.build_agent()

# ==============================================
# Example usage with default agents
# ==============================================
def example():
    """Example of how to use the dynamic supervisor."""
    
    # Define some specialized agents
    default_agents = {
        "coder": {
            "type": "simple",
            "name": "coder",
            "description": "Helps with coding and programming tasks",
            "system_prompt": "You are an expert programmer who specializes in writing clean, efficient code. You explain your code clearly and provide examples when helpful.",
            "model": "gpt-4o",
            "temperature": 0.3
        },
        "writer": {
            "type": "simple",
            "name": "writer",
            "description": "Assists with writing and content creation",
            "system_prompt": "You are a skilled writer who helps create clear, engaging content. You adapt your tone and style to match the specific needs of each request.",
            "model": "gpt-4o",
            "temperature": 0.7
        }
    }
    
    # Create a supervisor with default agents
    supervisor = create_dynamic_supervisor(
        name="example_supervisor",
        model="gpt-4o",
        temperature=0.2,
        default_agents=default_agents
    )
    
    # Create a thread ID for conversation persistence
    thread_id = str(uuid.uuid4())
    config = {"configurable": {"thread_id": thread_id}}
    
    # Run the supervisor with a message
    result = supervisor.run("Can you help me write a Python function to calculate Fibonacci numbers?", config=config)
    
    # Print messages
    for msg in result.get("messages", []):
        if hasattr(msg, "content"):
            role = "User" if hasattr(msg, "type") and msg.type == "human" else "AI"
            print(f"{role}: {msg.content}")
            
    # Continue the conversation
    result = supervisor.run("Now can you help me write a blog post about AI?", config=config)
    
    # Print messages
    for msg in result.get("messages", [])[-3:]:  # Just the last 3 messages
        if hasattr(msg, "content"):
            role = "User" if hasattr(msg, "type") and msg.type == "human" else "AI"
            print(f"{role}: {msg.content}")
    
    return result, supervisor
```

```python
"""
Simple example showing how to use the Dynamic Supervisor Agent
"""
import uuid
import logging
import sys

# Import the dynamic supervisor
#from working_dynamic_supervisor import create_dynamic_supervisor

# Set up logging
logging.basicConfig(
    level=logging.INFO,
    format='%(levelname)s: %(message)s',
    stream=sys.stdout
)

def run_simple_example():
    """Run a simple example demonstrating the dynamic supervisor."""
    print("=" * 80)
    print("Simple Dynamic Supervisor Example")
    print("=" * 80)
    
    # Create a basic supervisor (without default agents)
    supervisor = create_dynamic_supervisor(
        name="simple_supervisor",
        model="gpt-4o",
        temperature=0.2
    )
    
    # Create a thread ID for conversation persistence
    thread_id = str(uuid.uuid4())
    config = {"configurable": {"thread_id": thread_id}}
    
    # Helper to print messages
    def print_messages(result):
        if "messages" in result:
            for msg in result["messages"]:
                if hasattr(msg, "content"):
                    role = "User" if hasattr(msg, "type") and msg.type == "human" else "AI"
                    print(f"{role}: {msg.content}")
                elif isinstance(msg, tuple):
                    role = "User" if msg[0] == "human" else "AI"
                    print(f"{role}: {msg[1]}")
        print("-" * 80)
    
    # Step 1: Initial greeting
    print("\nStep 1: Initial greeting")
    result = supervisor.run("Hello! I need help with several different tasks.", config=config)
    print_messages(result)
    
    # Step 2: Ask for help
    print("\nStep 2: Getting help")
    result = supervisor.run("!help", config=config)
    print_messages(result)
    
    # Step 3: Add a Python coding agent
    print("\nStep 3: Adding a Python coding agent")
    result = supervisor.run(
        "!add_agent name: python_coder, description: Helps with Python coding tasks, "
        "system_prompt: You are an expert Python programmer. Write clean, efficient code "
        "with good documentation. Always provide explanations of how the code works.",
        config=config
    )
    print_messages(result)
    
    # Step 4: Add a writing agent
    print("\nStep 4: Adding a writing agent")
    result = supervisor.run(
        "!add_agent name: writer, description: Helps with writing and content creation, "
        "system_prompt: You are a skilled writer specializing in clear, engaging content. "
        "Adapt your tone and style to match the specific content needs of the user.",
        config=config
    )
    print_messages(result)
    
    # Step 5: List agents
    print("\nStep 5: Listing available agents")
    result = supervisor.run("!list_agents", config=config)
    print_messages(result)
    
    # Step 6: Ask for Python code
    print("\nStep 6: Asking for Python code (should route to python_coder)")
    result = supervisor.run(
        "I need a Python function that calculates the Fibonacci sequence up to n terms.",
        config=config
    )
    print_messages(result)
    
    # Step 7: Ask for writing help
    print("\nStep 7: Asking for writing help (should route to writer)")
    result = supervisor.run(
        "Can you help me write a professional email to reschedule a meeting?",
        config=config
    )
    print_messages(result)
    
    # Step 8: Remove an agent
    print("\nStep 8: Removing the writing agent")
    result = supervisor.run("!remove_agent name: writer", config=config)
    print_messages(result)
    
    print("\nExample completed successfully.")
    return thread_id, supervisor

if __name__ == "__main__":
    try:
        thread_id, supervisor = run_simple_example()
        print(f"Example thread ID: {thread_id}")
    except Exception as e:
        import traceback
        print(f"Error running example: {e}")
        traceback.print_exc()
```
