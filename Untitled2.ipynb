````python
from typing import List, Dict, Any, TypedDict, Optional, Union, Literal

from langchain_core.prompts import ChatPromptTemplate
from langchain_openai import ChatOpenAI
from langchain.schema import Document
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough

from pydantic import BaseModel, Field
from langgraph.graph import StateGraph, END, START
from langgraph.checkpoint.memory import MemorySaver
# ===== LLM Components =====
from src.haive.core.models.llm.base import AzureLLMConfig
llm = AzureLLMConfig().instantiate_llm()
````

````python

# ===== State Schema =====

class WorkingMemory(BaseModel):
    """Short-term memory for current conversation"""
    current_query: str
    decomposed_queries: List[str] = Field(default_factory=list)
    retrieval_context: Dict[str, Any] = Field(default_factory=dict)
    retrieved_documents: List[Document] = Field(default_factory=list)
    graded_documents: List[Document] = Field(default_factory=list)

class EpisodicMemory(BaseModel):
    """Memory of past interactions"""
    past_queries: List[str] = Field(default_factory=list)
    past_responses: List[str] = Field(default_factory=list)
    interaction_feedback: Dict[str, Any] = Field(default_factory=dict)

class SemanticMemory(BaseModel):
    """Long-term consolidated knowledge"""
    knowledge_graph: Dict[str, Any] = Field(default_factory=dict)
    entity_summaries: Dict[str, str] = Field(default_factory=dict)
    topic_contexts: Dict[str, str] = Field(default_factory=dict)

class PerformanceMetrics(BaseModel):
    """Metrics for self-improvement"""
    retrieval_metrics: Dict[str, float] = Field(default_factory=dict)
    query_success_rate: Dict[str, float] = Field(default_factory=dict)
    strategy_effectiveness: Dict[str, float] = Field(default_factory=dict)

class EnhancedRAGState(TypedDict):
    """Complete state for enhanced RAG system"""
    query: str
    working_memory: WorkingMemory
    episodic_memory: EpisodicMemory
    semantic_memory: SemanticMemory
    performance_metrics: PerformanceMetrics
    generation: str
    route_path: str



# ===== Query Analysis & Decomposition =====

class QueryAnalysisOutput(BaseModel):
    """Analysis of user query"""
    query_type: Literal["factual", "conceptual", "recent_event", "opinion", "code", "multi_hop"] = Field(
        description="The type of query being asked"
    )
    domain: str = Field(description="The knowledge domain the query belongs to")
    entities: List[str] = Field(description="Key entities mentioned in the query")
    temporal_relevance: Literal["historical", "current", "future"] = Field(
        description="Temporal relevance of the query"
    )
    complexity: int = Field(description="Complexity score from 1-5")

query_analyzer = llm.with_structured_output(QueryAnalysisOutput)
query_analysis_prompt = ChatPromptTemplate.from_messages([
    ("system", "Analyze the user query to determine its type, domain, entities, and complexity."),
    ("human", "{query}")
])
query_analysis_chain = query_analysis_prompt | query_analyzer

# ===== Query Routing =====

class RoutingDecision(BaseModel):
    """Routing decision for the query"""
    primary_route: Literal["vector_search", "knowledge_graph", "web_search", "code_search", "structured_data", "multi_hop"] = Field(
        description="Primary retrieval method to use"
    )
    secondary_routes: List[Literal["vector_search", "knowledge_graph", "web_search", "code_search", "structured_data"]] = Field(
        description="Backup retrieval methods if primary fails"
    )
    reasoning: str = Field(description="Reasoning behind the routing decision")

query_router = llm.with_structured_output(RoutingDecision)
routing_prompt = ChatPromptTemplate.from_messages([
    ("system", "Based on the query analysis, determine the optimal routing strategy."),
    ("human", "Query: {query}\nAnalysis: {analysis}")
])
routing_chain = routing_prompt | query_router

# ===== Query Decomposition =====

class QueryDecomposition(BaseModel):
    """Decomposition of complex queries into simpler sub-queries"""
    sub_queries: List[str] = Field(description="List of sub-queries to answer the main query")
    execution_order: List[int] = Field(description="Order in which to execute sub-queries (indices)")
    dependencies: Dict[int, List[int]] = Field(description="Dependencies between sub-queries (index -> [indices])")

query_decomposer = llm.with_structured_output(QueryDecomposition)
decomposition_prompt = ChatPromptTemplate.from_messages([
    ("system", "Decompose the complex query into simpler sub-queries that can be answered independently."),
    ("human", "Query: {query}\nAnalysis: {analysis}")
])
decomposition_chain = decomposition_prompt | query_decomposer

# ===== Document Relevance Grading =====

class DocumentGrade(BaseModel):
    """Grading of document relevance"""
    relevance_score: float = Field(description="Relevance score from 0-1")
    contains_answer: bool = Field(description="Whether document likely contains answer to query")
    key_information: List[str] = Field(description="Key pieces of information in the document")
    missing_information: List[str] = Field(description="Important information missing from the document")

document_grader = llm.with_structured_output(DocumentGrade)
grading_prompt = ChatPromptTemplate.from_messages([
    ("system", "Assess the relevance of this document to the query. Grade its usefulness."),
    ("human", "Query: {query}\nDocument: {document}")
])
grading_chain = grading_prompt | document_grader

# ===== Hallucination Detection =====

class HallucinationCheck(BaseModel):
    """Check for hallucinations in generated content"""
    hallucination_detected: bool = Field(description="Whether hallucination was detected")
    ungrounded_claims: List[str] = Field(description="Claims not supported by retrieved context")
    correction_needed: bool = Field(description="Whether correction is needed")
    confidence: float = Field(description="Confidence in assessment from 0-1")

hallucination_checker = llm.with_structured_output(HallucinationCheck)
hallucination_prompt = ChatPromptTemplate.from_messages([
    ("system", "Check whether the generated response contains hallucinations not supported by the context."),
    ("human", "Query: {query}\nContext: {context}\nGeneration: {generation}")
])
hallucination_chain = hallucination_prompt | hallucination_checker

# ===== Graph Implementation Functions =====

def initialize_state(query: str) -> EnhancedRAGState:
    """Initialize the state with a query"""
    return {
        "query": query,
        "working_memory": WorkingMemory(current_query=query),
        "episodic_memory": EpisodicMemory(),
        "semantic_memory": SemanticMemory(),
        "performance_metrics": PerformanceMetrics(),
        "generation": "",
        "route_path": ""
    }

def analyze_query(state: EnhancedRAGState) -> EnhancedRAGState:
    """Analyze the query and update state"""
    query = state["query"]
    analysis = query_analysis_chain.invoke({"query": query})
    
    # Update working memory
    working_memory = state["working_memory"]
    working_memory.retrieval_context["query_analysis"] = analysis.model_dump()
    
    return {**state, "working_memory": working_memory}

def route_query(state: EnhancedRAGState) -> str:
    """Route the query to the appropriate retrieval method"""
    query = state["query"]
    analysis = state["working_memory"].retrieval_context["query_analysis"]
    routing = routing_chain.invoke({"query": query, "analysis": analysis})
    
    # Update working memory
    working_memory = state["working_memory"]
    working_memory.retrieval_context["routing"] = routing.model_dump()
    state["working_memory"] = working_memory
    
    # Determine next node
    route = routing.primary_route
    state["route_path"] = route
    return route

def decompose_query(state: EnhancedRAGState) -> EnhancedRAGState:
    """Decompose complex queries into sub-queries"""
    query = state["query"]
    analysis = state["working_memory"].retrieval_context["query_analysis"]
    decomposition = decomposition_chain.invoke({"query": query, "analysis": analysis})
    
    # Update working memory
    working_memory = state["working_memory"]
    working_memory.retrieval_context["decomposition"] = decomposition.model_dump()
    working_memory.decomposed_queries = decomposition.sub_queries
    
    return {**state, "working_memory": working_memory}

def vector_retrieval(state: EnhancedRAGState) -> EnhancedRAGState:
    """Retrieve documents from vector store"""
    # This would connect to your vector database
    # For demonstration, we're creating placeholder documents
    documents = [
        Document(page_content="Example vector document 1", metadata={"source": "vector_db"}),
        Document(page_content="Example vector document 2", metadata={"source": "vector_db"}),
    ]
    
    # Update working memory
    working_memory = state["working_memory"]
    working_memory.retrieved_documents.extend(documents)
    
    return {**state, "working_memory": working_memory}

def knowledge_graph_retrieval(state: EnhancedRAGState) -> EnhancedRAGState:
    """Retrieve information from knowledge graph"""
    # This would connect to your knowledge graph
    # For demonstration, we're creating placeholder documents
    documents = [
        Document(page_content="Example knowledge graph result 1", metadata={"source": "knowledge_graph"}),
        Document(page_content="Example knowledge graph result 2", metadata={"source": "knowledge_graph"}),
    ]
    
    # Update working memory
    working_memory = state["working_memory"]
    working_memory.retrieved_documents.extend(documents)
    
    return {**state, "working_memory": working_memory}

def web_search(state: EnhancedRAGState) -> EnhancedRAGState:
    """Retrieve information from web search"""
    # This would connect to a web search API
    # For demonstration, we're creating placeholder documents
    documents = [
        Document(page_content="Example web search result 1", metadata={"source": "web_search"}),
        Document(page_content="Example web search result 2", metadata={"source": "web_search"}),
    ]
    
    # Update working memory
    working_memory = state["working_memory"]
    working_memory.retrieved_documents.extend(documents)
    
    return {**state, "working_memory": working_memory}

def code_search(state: EnhancedRAGState) -> EnhancedRAGState:
    """Retrieve code examples"""
    # This would connect to a code repository
    # For demonstration, we're creating placeholder documents
    documents = [
        Document(page_content="```python\ndef example_function():\n    return 'example'\n```", 
                 metadata={"source": "code_repo"}),
        Document(page_content="```python\nclass ExampleClass:\n    def __init__(self):\n        pass\n```", 
                 metadata={"source": "code_repo"}),
    ]
    
    # Update working memory
    working_memory = state["working_memory"]
    working_memory.retrieved_documents.extend(documents)
    
    return {**state, "working_memory": working_memory}

def multi_hop_retrieval(state: EnhancedRAGState) -> EnhancedRAGState:
    """Execute multi-hop retrieval for complex queries"""
    # First decompose the query if not already done
    if not state["working_memory"].decomposed_queries:
        state = decompose_query(state)
    
    # For each sub-query, perform appropriate retrieval
    all_documents = []
    for sub_query in state["working_memory"].decomposed_queries:
        # Determine retrieval method for sub-query
        # This is simplified - in practice would use a more sophisticated approach
        if "code" in sub_query.lower():
            docs = [Document(page_content=f"Code result for: {sub_query}", metadata={"source": "code_repo"})]
        elif "recent" in sub_query.lower():
            docs = [Document(page_content=f"Web result for: {sub_query}", metadata={"source": "web_search"})]
        else:
            docs = [Document(page_content=f"Vector result for: {sub_query}", metadata={"source": "vector_db"})]
        all_documents.extend(docs)
    
    # Update working memory
    working_memory = state["working_memory"]
    working_memory.retrieved_documents.extend(all_documents)
    
    return {**state, "working_memory": working_memory}

def grade_documents(state: EnhancedRAGState) -> EnhancedRAGState:
    """Grade retrieved documents for relevance"""
    query = state["query"]
    documents = state["working_memory"].retrieved_documents
    
    graded_docs = []
    for doc in documents:
        grade = grading_chain.invoke({"query": query, "document": doc.page_content})
        if grade.relevance_score > 0.5:  # Threshold for relevance
            doc.metadata["grade"] = grade.model_dump()
            graded_docs.append(doc)
    
    # Update working memory
    working_memory = state["working_memory"]
    working_memory.graded_documents = graded_docs
    
    return {**state, "working_memory": working_memory}

def decide_to_generate(state: EnhancedRAGState) -> str:
    """Decide whether to generate answer or reformulate query"""
    graded_docs = state["working_memory"].graded_documents
    
    if len(graded_docs) > 0:
        return "generate"
    else:
        return "reformulate"

def reformulate_query(state: EnhancedRAGState) -> EnhancedRAGState:
    """Reformulate the query for better retrieval"""
    query = state["query"]
    
    # Simple reformulation - in practice would use an LLM chain
    reformulated_query = f"Alternative formulation of: {query}"
    
    # Update state
    new_state = {**state, "query": reformulated_query}
    new_state["working_memory"].current_query = reformulated_query
    
    return new_state

def generate_response(state: EnhancedRAGState) -> EnhancedRAGState:
    """Generate response based on graded documents"""
    query = state["query"]
    documents = state["working_memory"].graded_documents
    
    # Format documents for context
    context = "\n\n".join([doc.page_content for doc in documents])
    
    # Generate response - this would be your RAG response generation
    generation_prompt = ChatPromptTemplate.from_messages([
        ("system", "Answer the question based on the provided context."),
        ("human", "Question: {question}\n\nContext: {context}")
    ])
    generation_chain = generation_prompt | llm | StrOutputParser()
    generation = generation_chain.invoke({"question": query, "context": context})
    
    return {**state, "generation": generation}

def check_hallucination(state: EnhancedRAGState) -> str:
    """Check for hallucinations in generated response"""
    query = state["query"]
    context = "\n\n".join([doc.page_content for doc in state["working_memory"].graded_documents])
    generation = state["generation"]
    
    check = hallucination_chain.invoke({
        "query": query, 
        "context": context, 
        "generation": generation
    })
    
    if check.hallucination_detected and check.correction_needed:
        return "correct"
    else:
        return "finalize"

def correct_generation(state: EnhancedRAGState) -> EnhancedRAGState:
    """Correct hallucinations in the generated response"""
    query = state["query"]
    context = "\n\n".join([doc.page_content for doc in state["working_memory"].graded_documents])
    generation = state["generation"]
    
    # Get hallucination details
    check = hallucination_chain.invoke({
        "query": query, 
        "context": context, 
        "generation": generation
    })
    
    # Correction prompt
    correction_prompt = ChatPromptTemplate.from_messages([
        ("system", "Correct the following response to remove any ungrounded claims."),
        ("human", "Original Response: {response}\n\nUngrounded Claims: {claims}\n\nContext: {context}")
    ])
    correction_chain = correction_prompt | llm | StrOutputParser()
    
    corrected_generation = correction_chain.invoke({
        "response": generation,
        "claims": "\n".join(check.ungrounded_claims),
        "context": context
    })
    
    return {**state, "generation": corrected_generation}

def finalize_response(state: EnhancedRAGState) -> EnhancedRAGState:
    """Finalize response and update episodic memory"""
    query = state["query"]
    generation = state["generation"]
    
    # Update episodic memory
    episodic_memory = state["episodic_memory"]
    episodic_memory.past_queries.append(query)
    episodic_memory.past_responses.append(generation)
    
    # Update performance metrics
    metrics = state["performance_metrics"]
    route_path = state["route_path"]
    if route_path in metrics.retrieval_metrics:
        metrics.retrieval_metrics[route_path] += 1
    else:
        metrics.retrieval_metrics[route_path] = 1
    
    return {**state, "episodic_memory": episodic_memory, "performance_metrics": metrics}

# ===== Graph Definition =====

workflow = StateGraph(EnhancedRAGState)

# Add nodes
workflow.add_node("analyze_query", analyze_query)
workflow.add_node("decompose_query", decompose_query)
workflow.add_node("vector_retrieval", vector_retrieval)
workflow.add_node("knowledge_graph_retrieval", knowledge_graph_retrieval)
workflow.add_node("web_search", web_search)
workflow.add_node("code_search", code_search)
workflow.add_node("multi_hop_retrieval", multi_hop_retrieval)
workflow.add_node("grade_documents", grade_documents)
workflow.add_node("reformulate_query", reformulate_query)
workflow.add_node("generate_response", generate_response)
workflow.add_node("correct_generation", correct_generation)
workflow.add_node("finalize_response", finalize_response)

# Define edges
workflow.add_edge(START, "analyze_query")
workflow.add_conditional_edges(
    "analyze_query",
    route_query,
    {
        "vector_search": "vector_retrieval",
        "knowledge_graph": "knowledge_graph_retrieval",
        "web_search": "web_search",
        "code_search": "code_search",
        "multi_hop": "multi_hop_retrieval",
    }
)

# Connect retrieval nodes to document grading
workflow.add_edge("vector_retrieval", "grade_documents")
workflow.add_edge("knowledge_graph_retrieval", "grade_documents")
workflow.add_edge("web_search", "grade_documents")
workflow.add_edge("code_search", "grade_documents")
workflow.add_edge("multi_hop_retrieval", "grade_documents")

# Decide based on document grading
workflow.add_conditional_edges(
    "grade_documents",
    decide_to_generate,
    {
        "generate": "generate_response",
        "reformulate": "reformulate_query",
    }
)

# Handle query reformulation
workflow.add_edge("reformulate_query", "analyze_query")

# Generate and check for hallucinations
workflow.add_conditional_edges(
    "generate_response",
    check_hallucination,
    {
        "correct": "correct_generation",
        "finalize": "finalize_response",
    }
)

# Correction might need additional hallucination checking
workflow.add_edge("correct_generation", "finalize_response")

# End after finalization
workflow.add_edge("finalize_response", END)

# Compile the graph
enhanced_rag_app = workflow.compile()

````

````python
# Example usage
def query_enhanced_rag(question: str) -> Dict[str, Any]:
    """Query the enhanced RAG system"""
    state = initialize_state(question)
    result = enhanced_rag_app.invoke(state)
    return result
````
