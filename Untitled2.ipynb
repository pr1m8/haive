````python
from typing import List, Dict, Any, Optional, Union, Tuple, Type
from pydantic import BaseModel, Field


class ToolCall(BaseModel):
    """
    Represents a tool invocation.
    """
    name: str = Field(description="Name of the tool to use (or 'LLM')")
    input: Optional[Union[str, Dict[str, Any],Any]] = Field(
        default=None,
        description="Input for the tool or reference like '$1'."
    )


class ParallelPlanNode(BaseModel):
    """
    A recursive node that represents:
    1. A single atomic step using a tool.
    2. A sub-plan containing multiple parallel steps.
    """
    id: int = Field(description="Unique step identifier.")
    description: str = Field(description="Detailed description of the step.")
    evidence_ref: str = Field(description="Evidence reference like '#E1'.")
    dependencies: List[int] = Field(default_factory=list, description="Step IDs this step depends on.")
    status: str = Field(default="not_started", description="Execution status.")

    # Either a tool call (leaf) or nested sub-steps (branching node)
    tool_calls: Optional[List[ToolCall]] = None
    sub_steps: Optional[List["ParallelPlanNode"]] = None  # Recursive reference

    def is_leaf(self) -> bool:
        """Determines if this node is a leaf (single tool call) or a branching node."""
        return self.tool_calls is not None and self.sub_steps is None

    def find_node_by_id(self, id: int) -> Optional["ParallelPlanNode"]:
        """Find a node by its ID."""
        if self.id == id:
            return self
        
        if self.sub_steps:
            for step in self.sub_steps:
                found = step.find_node_by_id(id)
                if found:
                    return found
        
        return None

    def get_executable_steps(self) -> List["ParallelPlanNode"]:
        """
        Get steps that are ready to be executed (dependencies satisfied).
        A step is executable if:
        1. It's not already completed or in progress
        2. All dependencies are completed
        """
        executable = []
        
        # Check if this node is executable
        if self.status == "not_started":
            # If no dependencies, it's executable
            if not self.dependencies:
                executable.append(self)
            # Otherwise, check if all dependencies are completed
            else:
                all_deps_completed = True
                for dep_id in self.dependencies:
                    dep_node = self.find_dep_node(dep_id)
                    if not dep_node or dep_node.status != "completed":
                        all_deps_completed = False
                        break
                
                if all_deps_completed:
                    executable.append(self)
        
        # Check sub-steps recursively
        if self.sub_steps:
            for step in self.sub_steps:
                executable.extend(step.get_executable_steps())
        
        return executable

    def find_dep_node(self, dep_id: int) -> Optional["ParallelPlanNode"]:
        """Find a dependency node by its ID."""
        # Start from the root (assuming self is the root or has access to it)
        root = self
        while hasattr(root, "parent") and root.parent:
            root = root.parent
        
        # Search from the root
        return root.find_node_by_id(dep_id)


class GlobalJoinStep(BaseModel):
    """
    Represents the final join step aggregating all sub-plans.
    """
    id: int = Field(description="Unique ID for the global join step.")
    description: str = Field(default="Join results from all sub-plans.")
    input_refs: List[str] = Field(description="List of evidence references (#E) from sub-plan joins.")
    evidence_ref: str = Field(description="Evidence reference for the global join step.")
    dependencies: List[int] = Field(description="IDs of dependent join steps.")
    status: str = Field(default="not_started", description="Execution status.")


class ParallelPlanTree(BaseModel):
    """
    A recursive parallel execution plan with tree-like branching and depth constraints.
    """
    root: ParallelPlanNode = Field(description="Root node of the plan tree.")
    max_depth: Optional[int] = Field(default=3, description="Maximum depth allowed for branching execution.")
    global_join: Optional[GlobalJoinStep] = None
    evidence_store: Dict[str, Any] = Field(default_factory=dict, description="Store for evidence created during execution.")

    def enforce_max_depth(self):
        """
        Recursively enforces max depth: If exceeded, child steps are collapsed into a join node.
        """
        def process_node(node: ParallelPlanNode, depth: int):
            if depth >= self.max_depth and node.sub_steps:
                # Collapse all sub-steps into a join() call
                join_id = max(step.id for step in node.sub_steps) + 1
                join_step = ParallelPlanNode(
                    id=join_id,
                    description="Intermediate join due to max depth reached.",
                    evidence_ref=f"#E{join_id}",
                    dependencies=[s.id for s in node.sub_steps],
                    tool_calls=[ToolCall(name="join", input=[s.evidence_ref for s in node.sub_steps])]
                )
                node.sub_steps = [join_step]

            elif node.sub_steps:
                for child in node.sub_steps:
                    process_node(child, depth + 1)

        process_node(self.root, 0)

    def set_global_join(self):
        """
        Creates a final join step combining all evidence from sub-branches.
        """
        all_refs = self.collect_all_evidence_refs()
        all_ids = [int(ref[2:]) for ref in all_refs]  # Extract IDs from #E<number>
        global_join_id = max(all_ids) + 1

        self.global_join = GlobalJoinStep(
            id=global_join_id,
            input_refs=all_refs,
            evidence_ref=f"#E{global_join_id}",
            dependencies=all_ids
        )

    def collect_all_evidence_refs(self) -> List[str]:
        """
        Collects all evidence references in the tree.
        """
        refs = []

        def collect(node: ParallelPlanNode):
            refs.append(node.evidence_ref)
            if node.sub_steps:
                for sub in node.sub_steps:
                    collect(sub)

        collect(self.root)
        return refs

    def get_next_executable_step(self) -> Optional[ParallelPlanNode]:
        """Get the next step that can be executed."""
        executable_steps = self.root.get_executable_steps()
        return executable_steps[0] if executable_steps else None
    
    def update_step_status(self, step_id: int, status: str, result: Any = None):
        """Update status of a step and store its result if completed."""
        node = self.root.find_node_by_id(step_id)
        if node:
            node.status = status
            if status == "completed" and result is not None:
                self.evidence_store[node.evidence_ref] = result
        
        # Also check global join
        if self.global_join and self.global_join.id == step_id:
            self.global_join.status = status
            if status == "completed" and result is not None:
                self.evidence_store[self.global_join.evidence_ref] = result

    def is_complete(self) -> bool:
        """Check if the entire plan is complete."""
        return (self.global_join and self.global_join.status == "completed") or self._check_all_completed(self.root)
    
    def _check_all_completed(self, node: ParallelPlanNode) -> bool:
        """Check if a node and all its sub-nodes are completed."""
        if node.status != "completed":
            return False
        
        if node.sub_steps:
            return all(self._check_all_completed(step) for step in node.sub_steps)
        
        return True

from typing import Callable
# Rebuild models to handle recursive relationships
ParallelPlanNode.model_rebuild()
ParallelPlanTree.model_rebuild()


# State schema for plan-execute agent
class PlanExecuteState(BaseModel):
    """State schema for the plan-execute agent."""
    messages: List[Any] = Field(default_factory=list)
    plan: Optional[ParallelPlanTree] = None
    execution_status: str = Field(default="not_started")
    current_step_id: Optional[int] = None
    reflection: Optional[str] = None
    final_result: Optional[str] = None
    tools: Dict[str, Callable] = Field(default_factory=dict)
    replanning_count: int = Field(default=0)
    max_replanning: int = Field(default=3)
````

````python
from typing import Any, Dict, List, Optional, Union, Type, cast
from pydantic import BaseModel
import json
import re

from langchain_core.output_parsers import BaseOutputParser
from langchain_core.outputs import Generation




class ParallelPlanTreeOutputParser(BaseOutputParser[ParallelPlanTree]):
    """
    Custom output parser for ParallelPlanTree.
    Handles various output formats from LLMs and ensures proper conversion to ParallelPlanTree.
    """
    
    @property
    def _type(self) -> str:
        return "parallel_plan_tree"

    def parse(self, result: List[Generation]) -> ParallelPlanTree:
        """Parse the LLM output into a ParallelPlanTree object."""
        if not result:
            raise ValueError("No generations to parse")
        
        text = result[0].text
        
        # Try to extract JSON from the text if it's not already pure JSON
        json_str = self._extract_json(text)
        
        try:
            # Try to parse as direct ParallelPlanTree
            return ParallelPlanTree.model_validate_json(json_str)
        except Exception as e1:
            try:
                # Try to parse as dict then convert
                data = json.loads(json_str)
                return ParallelPlanTree.model_validate(data)
            except Exception as e2:
                # If both approaches fail, try to extract a valid tree structure
                try:
                    tree_data = self._extract_tree_structure(text)
                    return ParallelPlanTree.model_validate(tree_data)
                except Exception as e3:
                    # Provide detailed error information
                    raise ValueError(
                        f"Failed to parse into ParallelPlanTree: \n"
                        f"Error 1: {str(e1)}\n"
                        f"Error 2: {str(e2)}\n" 
                        f"Error 3: {str(e3)}\n"
                        f"Original text: {text[:500]}..."
                    )

    def _extract_json(self, text: str) -> str:
        """Extract JSON from text that might contain markdown or other formatting."""
        # Check if the text is already valid JSON
        try:
            json.loads(text)
            return text
        except json.JSONDecodeError:
            pass
        
        # Try to extract JSON from markdown code blocks
        json_matches = re.findall(r"```(?:json)?\s*([\s\S]*?)```", text)
        if json_matches:
            for match in json_matches:
                try:
                    json.loads(match.strip())
                    return match.strip()
                except json.JSONDecodeError:
                    continue
        
        # Try to find JSON-like content with braces
        try:
            start_idx = text.find('{')
            end_idx = text.rfind('}')
            if start_idx != -1 and end_idx != -1 and end_idx > start_idx:
                json_candidate = text[start_idx:end_idx+1]
                json.loads(json_candidate)  # Validate it's proper JSON
                return json_candidate
        except json.JSONDecodeError:
            pass
        
        # Return the original text if extraction fails
        return text

    def _extract_tree_structure(self, text: str) -> Dict[str, Any]:
        """
        Extract a tree structure from text when JSON parsing fails.
        This is a fallback mechanism for poorly formatted outputs.
        """
        # Default minimal valid structure
        default_structure = {
            "root": {
                "id": 1,
                "description": "Root node extracted from text",
                "evidence_ref": "#E1",
                "dependencies": [],
                "status": "not_started"
            },
            "max_depth": 3,
            "evidence_store": {}
        }
        
        # Try to extract node information using regex patterns
        nodes = []
        node_patterns = [
            r"id:\s*(\d+)[,\s]+description:\s*\"([^\"]+)\"[,\s]+evidence_ref:\s*\"?(#E\d+)\"?",
            r"Step\s+(\d+):\s+([^\n#]+)\s+→\s+Evidence:\s+(#E\d+)",
            r"(\d+)\.\s+([^\n#]+)\s+\|\s+evidence_ref=(#E\d+)"
        ]
        
        for pattern in node_patterns:
            matches = re.findall(pattern, text)
            if matches:
                for match in matches:
                    node_id = int(match[0])
                    description = match[1].strip()
                    evidence_ref = match[2].strip()
                    nodes.append({
                        "id": node_id,
                        "description": description,
                        "evidence_ref": evidence_ref,
                        "dependencies": [],
                        "status": "not_started"
                    })
                break  # Use the first pattern that works
        
        # Extract dependencies
        for node in nodes:
            dep_pattern = rf"id:\s*{node['id']}[^d]*dependencies:\s*\[([\d,\s]+)\]"
            dep_matches = re.findall(dep_pattern, text)
            if dep_matches:
                try:
                    deps_str = dep_matches[0].strip()
                    if deps_str:
                        node["dependencies"] = [int(d.strip()) for d in deps_str.split(",")]
                except Exception:
                    pass
        
        # Build tree structure
        if nodes:
            # Sort by ID to find root (usually ID 1)
            nodes.sort(key=lambda n: n["id"])
            root_node = nodes[0]
            
            # Organize nodes into tree structure
            node_map = {node["id"]: node for node in nodes}
            for node in nodes:
                node_id = node["id"]
                for other_node in nodes:
                    if node_id in other_node["dependencies"]:
                        if "sub_steps" not in node:
                            node["sub_steps"] = []
                        node["sub_steps"].append(other_node)
            
            return {
                "root": root_node,
                "max_depth": 3,
                "evidence_store": {}
            }
        
        return default_structure


# Alternative approach: Define a structured tools parser for OpenAI
class ParallelPlanTreeToolsParser(BaseOutputParser[ParallelPlanTree]):
    """
    Parser specifically designed to work with OpenAI's structured tool outputs.
    """
    
    def __init__(self):
        super().__init__()
        self.tool_name = "ParallelPlanTree"
    
    @property
    def _type(self) -> str:
        return "parallel_plan_tree_tools"
    
    def parse(self, text: str) -> ParallelPlanTree:
        """Direct parse method for compatibility with OpenAI function calling."""
        try:
            # Try to parse as JSON first
            data = json.loads(text)
            return ParallelPlanTree.model_validate(data)
        except json.JSONDecodeError:
            # Fall back to regex extraction if not valid JSON
            parser = ParallelPlanTreeOutputParser()
            return parser._extract_tree_structure(text)
    
    def parse_result(self, result: List[Generation]) -> ParallelPlanTree:
        """Parse Generation objects into ParallelPlanTree."""
        if not result:
            raise ValueError("No generations to parse")
        
        text = result[0].text
        return self.parse(text)
````

````python
from langchain_core.messages import SystemMessage
from src.haive.core.models.llm.base import AzureLLMConfig
def create_planner_aug_llm(
    system_prompt: Optional[str] = None,
    model: str = "gpt-4o",
    tools: Optional[Dict[str, Callable]] = None,
    temperature: float = 0.2,
    max_depth: int = 3,
    **kwargs
) -> AugLLMConfig:
    """
    Create an AugLLM configuration for the planning component.
    
    Args:
        system_prompt: Optional system prompt override
        model: LLM model to use
        tools: Dictionary of available tools
        temperature: Temperature for planning (lower is more deterministic)
        max_depth: Maximum plan depth to allow
        **kwargs: Additional configuration parameters
        
    Returns:
        AugLLMConfig for the planner
    """
    # Generate tool descriptions
    tools_description = ""
    if tools:
        tools_description = "\n".join([
            f"- {name}: {func.__doc__ or f'Tool: {name}'}"
            for name, func in tools.items()
        ])
    
    # Default system prompt if none provided
    if system_prompt is None:
        system_prompt = f"""
You are an advanced planning agent responsible for transforming a user query into a structured, parallelizable execution plan. Your plan will follow a strict **recursive tree structure**, allowing for efficient parallel execution.

## **CORE OBJECTIVE**
- Break down the **user query** into **multiple parallelizable sub-plans**.
- Each plan must be structured **recursively**, with each **sub-plan** containing:
  - **Sequential actions** that depend on previous steps.
  - **Independent actions** that can be parallelized.
  - **A final join() step** aggregating all necessary results.

## **PLAN STRUCTURE**
You must create a plan with the following structure:
- A root node decomposing the main query
- Sub-plans with tool calls and dependencies
- A global join step combining all results

## **MAX DEPTH CONSTRAINT**
- The maximum depth of the plan tree is {max_depth}
- Once this depth is reached, steps must be combined into a join operation

## **AVAILABLE TOOLS**
{tools_description}

## **RESPONSE FORMAT**
You MUST create a valid JSON object that matches this example structure:
```json
{{
  "root": {{
    "id": 1,
    "description": "Analyze customer feedback data",
    "evidence_ref": "#E1",
    "dependencies": [],
    "status": "not_started",
    "sub_steps": [
      {{
        "id": 2,
        "description": "Extract sentiment from customer reviews",
        "evidence_ref": "#E2",
        "dependencies": [],
        "status": "not_started",
        "tool_calls": [
          {{
            "name": "sentiment_analysis",
            "input": "customer_reviews.csv"
          }}
        ]
      }},
      {{
        "id": 3,
        "description": "Categorize feedback by topic",
        "evidence_ref": "#E3",
        "dependencies": [],
        "status": "not_started",
        "tool_calls": [
          {{
            "name": "topic_modeling",
            "input": "customer_reviews.csv"
          }}
        ]
      }},
      {{
        "id": 4,
        "description": "Join sentiment and topic analysis",
        "evidence_ref": "#E4",
        "dependencies": [2, 3],
        "status": "not_started",
        "tool_calls": [
          {{
            "name": "join",
            "input": ["#E2", "#E3"]
          }}
        ]
      }}
    ]
  }},
  "max_depth": {max_depth},
  "evidence_store": {{}}
}}
```
"""

    # Create prompt template
    prompt = ChatPromptTemplate.from_messages([
        SystemMessage(content=system_prompt),
        ("human", "{messages}")
    ])
    
    # Create LLM config with lower temperature for planning
    llm_config = AzureLLMConfig(
        model=model,
        parameters={
            "temperature": temperature,
            #"response_format": {"type": "json_object"}  # Force JSON format
        }
    )
    
    # Create output parser
    output_parser = ParallelPlanTreeOutputParser()
    
    # Create and return AugLLM config
    return AugLLMConfig(
        name="planner",
        llm_config=llm_config,
        prompt_template=prompt,
        structured_output_model=ParallelPlanTree,
        #output_parser=output_parser,
        **kwargs
    )
````

````python
import asyncio
import inspect
import textwrap
from typing import Any, Dict, List, Callable, Optional, get_type_hints, Type
import libcst as cst
from pydantic import BaseModel

# --- Import your actual models and AugLLM components ---
from src.haive.agents.plan_and_execute.models import Plan, Response, Act
from src.haive.core.aug_llm.base import AugLLMConfig, compose_runnable
from langchain.prompts import ChatPromptTemplate, MessagesPlaceholder

# Import Command and Send from langgraph.types
from langgraph.types import Command, Send

# Global flag to control verbose model info
VERBOSE_MODEL_INFO = True

def verbose_model_info(model: Type[BaseModel]) -> str:
    """Return a verbose string with detailed info about the Pydantic model."""
    info_lines = []
    info_lines.append(f"Model Name: {model.__name__}")
    info_lines.append("Fields:")
    # Use Pydantic V2 API if available.
    fields = model.model_fields if hasattr(model, "model_fields") else model.__fields__
    for field_name, field in fields.items():
        field_type = getattr(field, "annotation", "Unknown")
        required = getattr(field, "is_required", getattr(field, "required", "Unknown"))
        default = field.default if hasattr(field, "default") and field.default is not None else "None"
        try:
            description = field.field_info.description
        except AttributeError:
            description = getattr(field, "description", "")
        info_lines.append(
            f"  {field_name}: type={field_type}, required={required}, default={default}, description={description}"
        )
    properties = [name for name, _ in inspect.getmembers(model, predicate=lambda m: isinstance(m, property))]
    if properties:
        info_lines.append("Properties: " + ", ".join(properties))
    methods = [name for name, _ in inspect.getmembers(model, predicate=inspect.isfunction) if not name.startswith("__")]
    if methods:
        info_lines.append("Methods: " + ", ".join(methods))
    return "\n".join(info_lines)

# --- Create a real planner configuration using your actual prompt templates ---
EXECUTOR_PROMPT = "You are a helpful assistant"
EXECUTOR_PROMPT_TEMPLATE = ChatPromptTemplate.from_messages(
    [
        ("system", EXECUTOR_PROMPT),
        MessagesPlaceholder(variable_name="messages"),
    ]
)
PLANNER_PROMPT = (
    "For the given objective, come up with a simple step by step plan. "
    "This plan should involve individual tasks, that if executed correctly will yield the correct answer. "
    "Do not add any superfluous steps. "
    "The result of the final step should be the final answer. "
    "Make sure that each step has all the information needed - do not skip steps."
)
PLANNER_PROMPT_TEMPLATE = ChatPromptTemplate.from_messages(
    [
        ("system", PLANNER_PROMPT),
        ("placeholder", "{messages}"),
    ]
)
planner_aug_llm_config = AugLLMConfig(
    name="planner",
    prompt_template=PLANNER_PROMPT_TEMPLATE,
    structured_output_model=Plan,
)

# --- Dynamic Node Function Factory using AugLLMConfig properties ---
def infer_input_field(config: AugLLMConfig) -> str:
    """
    Determine the input field based on the prompt template.
    Looks for 'messages' in optional_variables; otherwise, uses the first input variable or defaults to 'input'.
    """
    pt = config.prompt_template
    if hasattr(pt, "optional_variables") and pt.optional_variables:
        for var in pt.optional_variables:
            if var.lower() == "messages":
                return var
        return pt.optional_variables[0]
    elif hasattr(pt, "input_variables") and pt.input_variables:
        return pt.input_variables[0]
    return "input"

def infer_output_mapper(config: AugLLMConfig) -> Callable[[Any], Dict[str, Any]]:
    """
    Build an output mapper using the structured output model (or output parser) from the AugLLMConfig.
    
    If a structured output model is provided and VERBOSE_MODEL_INFO is enabled,
    prints detailed info about the model before returning a mapper that parses the result.
    """
    if config.structured_output_model:
        Model = config.structured_output_model
        if VERBOSE_MODEL_INFO:
            print("Using structured output model:")
            print(verbose_model_info(Model))
        def mapper(result: Any) -> Dict[str, Any]:
            if not isinstance(result, Model):
                try:
                    parsed = Model.parse_obj(result)
                except Exception as e:
                    raise ValueError(f"Error parsing result with {Model}: {e}")
            else:
                parsed = result
            # Use the 'plan' attribute if present; otherwise, default to the model instance.
            plan_val = getattr(parsed, "plan", parsed)
            steps_val = getattr(parsed, "steps", None)
            return {"plan": plan_val, "plan_steps": steps_val}
        return mapper
    elif config.output_parser:
        def mapper(result: Any) -> Dict[str, Any]:
            parsed = config.output_parser.parse(result)
            return {"plan": parsed, "plan_steps": parsed}
        return mapper
    else:
        def mapper(result: Any) -> Dict[str, Any]:
            return {"plan": result, "plan_steps": result}
        return mapper

def create_node_function(
    config: AugLLMConfig,
    command_goto: str = "execute_step",
    async_mode: bool = True,
) -> Callable[[Dict[str, Any]], Any]:
    """
    Creates a node function from an AugLLMConfig using its prompt template and structured output model.
    
    - It grabs and prints detailed info about the output model (or parser) if verbose is enabled.
    - It composes the runnable via compose_runnable.
    - It infers the input field and wraps the state input accordingly.
    - It maps the output (parsing the raw result into the structured model)
      and returns a Command (from langgraph.types) with update, goto, and resume fields.
    """
    output_model = config.structured_output_model or config.output_parser
    if output_model:
        print(f"Output model/parser detected: {output_model}")
    else:
        print("No structured output model or output parser provided.")
    
    node_runnable = compose_runnable(config)
    
    input_field = infer_input_field(config)
    if hasattr(config.prompt_template, "input_types") and input_field in config.prompt_template.input_types:
        expected_type = config.prompt_template.input_types[input_field]
        print(f"Inferred input field: '{input_field}' with expected type: {expected_type}")
    else:
        print(f"Inferred input field: '{input_field}' (no type info available)")
    
    def input_mapper(state: Dict[str, Any]) -> Dict[str, Any]:
        value = state.get(input_field)
        if value is None:
            raise ValueError(f"State does not contain required key '{input_field}'")
        if input_field.lower() == "messages":
            if isinstance(value, list):
                if not all(isinstance(x, tuple) and len(x) == 2 for x in value):
                    value = [("user", str(value))]
            else:
                value = [("user", str(value))]
            return {"messages": value}
        return {input_field: value}
    
    output_mapper = infer_output_mapper(config)
    
    if async_mode:
        async def node(state: Dict[str, Any]):
            payload = input_mapper(state)
            result = await node_runnable.ainvoke(payload)
            out = output_mapper(result)
            return Command(
                update={"plan": out["plan"]},
                goto=command_goto,
                resume={"plan": out["plan_steps"]}
            )
        return node
    else:
        def node(state: Dict[str, Any]):
            payload = input_mapper(state)
            result = node_runnable.invoke(payload)
            out = output_mapper(result)
            return Command(
                update={"plan": out["plan"]},
                goto=command_goto,
                resume={"plan": out["plan_steps"]}
            )
        return node

# --- Glue Nodes Functions ---
def glue_nodes_sync(
    nodes: List[Callable[[Dict[str, Any]], Command]],
    initial_state: Dict[str, Any]
) -> Command:
    """
    Synchronously chains a list of node functions.
    Each node receives the current state; its output update is merged into the state for the next node.
    Returns the final Command.
    """
    state = initial_state.copy()
    last_command = None
    for node in nodes:
        last_command = node(state)
        if last_command.update:
            state.update(last_command.update)
    return last_command

async def glue_nodes_async(
    nodes: List[Callable[[Dict[str, Any]], Command]],
    initial_state: Dict[str, Any]
) -> Command:
    """
    Asynchronously chains a list of node functions.
    Each node receives the current state; its output update is merged into the state for the next node.
    Returns the final Command.
    """
    state = initial_state.copy()
    last_command = None
    for node in nodes:
        last_command = await node(state)
        if last_command.update:
            state.update(last_command.update)
    return last_command

# --- Create Node Functions ---
# Create two node functions (async and sync) using the same planner configuration.
node1_async = create_node_function(planner_aug_llm_config, async_mode=True)
node2_async = create_node_function(planner_aug_llm_config, async_mode=True)
node1_sync = create_node_function(planner_aug_llm_config, async_mode=False)
node2_sync = create_node_function(planner_aug_llm_config, async_mode=False)

# --- Function Inspector (using heavy introspection via inspect and LibCST) ---
class FunctionInspector:
    """
    Inspects and analyzes a Python function.
    Prints signature, parameter details, type hints, docstring, source code, and a truncated CST.
    """
    def __init__(self, func: Any):
        if not inspect.isfunction(func):
            raise TypeError(f"Provided object {func} is not a function.")
        self.func = func
        self.signature = inspect.signature(func)
        self.type_hints = get_type_hints(func)
        self.docstring = inspect.getdoc(func)
        source = inspect.getsource(func)
        self.source_code = textwrap.dedent(source)
    
    def get_signature(self) -> inspect.Signature:
        return self.signature
    
    def get_parameters_info(self) -> Dict[str, Dict[str, Any]]:
        params_info = {}
        for name, param in self.signature.parameters.items():
            params_info[name] = {
                "annotation": param.annotation,
                "default": param.default,
                "kind": param.kind,
            }
        return params_info
    
    def get_return_annotation(self) -> Any:
        ret = self.signature.return_annotation
        return ret if ret is not inspect._empty else "No return annotation specified"
    
    def get_type_hints(self) -> Dict[str, Any]:
        return self.type_hints
    
    def get_docstring(self) -> Optional[str]:
        return self.docstring
    
    def get_source_code(self) -> str:
        return self.source_code
    
    def analyze_ast(self) -> cst.Module:
        try:
            tree = cst.parse_module(self.source_code)
            return tree
        except Exception as e:
            raise RuntimeError(f"Error parsing source code with LibCST: {e}")
    
    def print_info(self) -> None:
        print(f"Function: {self.func.__name__}")
        print("Signature:", self.get_signature())
        print("\nParameters:")
        for name, info in self.get_parameters_info().items():
            print(f"  {name}:")
            print(f"    Annotation: {info['annotation']}")
            print(f"    Default: {info['default']}")
            print(f"    Kind: {info['kind']}")
        print("\nReturn Annotation:", self.get_return_annotation())
        print("\nType Hints:", self.get_type_hints())
        print("\nDocstring:")
        print(self.get_docstring())
        print("\nSource Code:")
        print(self.get_source_code())
        print("\n--- Analyzing AST with LibCST ---")
        ast_tree = self.analyze_ast()
        ast_str = str(ast_tree)
        print(ast_str[:500] + "\n...")

# --- Inspect one of the asynchronous node functions ---
inspector = FunctionInspector(node1_async)
inspector.print_info()

# --- Test Execution of the Async Node Function ---
async def test_planner_node():
    state = {"messages": [("user", "Test input for planner node")]}
    result = await node1_async(state)
    print("\nPlanner node execution result (async):", result)

# --- Test Merging Two Nodes Synchronously ---
def test_glue_sync_nodes():
    initial_state = {"messages": [("user", "Initial sync input")]}
    final_command = glue_nodes_sync([node1_sync, node2_sync], initial_state)
    print("Final Command from sync pipeline:", final_command)

# --- Test Merging Two Nodes Asynchronously ---
async def test_glue_async_nodes():
    initial_state = {"messages": [("user", "Initial async input")]}
    final_command = await glue_nodes_async([node1_async, node2_async], initial_state)
    print("Final Command from async pipeline:", final_command)

if __name__ == "__main__":
    try:
        import nest_asyncio
        nest_asyncio.apply()
        print("=== Testing Async Planner Node ===")
        asyncio.run(test_planner_node())
        print("\n=== Testing Synchronous Glue Nodes ===")
        test_glue_sync_nodes()
        print("\n=== Testing Asynchronous Glue Nodes ===")
        asyncio.run(test_glue_async_nodes())
    except RuntimeError as e:
        print("RuntimeError encountered:", e)
        print("If you're in a notebook, try running:\n\n    await test_planner_node()\n")
````

````python
import asyncio
import inspect
import textwrap
from typing import Any, Dict, List, Type, Union, Optional, Callable, get_type_hints
import libcst as cst
from pydantic import BaseModel, Field, create_model, ValidationError

# --- Import your actual models and AugLLM components ---
from src.haive.agents.plan_and_execute.models import Plan, Response, Act
from src.haive.core.aug_llm.base import AugLLMConfig, compose_runnable
from langchain.prompts import ChatPromptTemplate, MessagesPlaceholder

# Import Command and Send from langgraph.types
from langgraph.types import Command, Send

# Global flag to control verbose model info
VERBOSE_MODEL_INFO = True

def verbose_model_info(model: Type[BaseModel]) -> str:
    """Return a verbose string with detailed info about the Pydantic model."""
    info_lines = []
    info_lines.append(f"Model Name: {model.__name__}")
    info_lines.append("Fields:")
    # Use Pydantic V2 API if available.
    fields = model.model_fields if hasattr(model, "model_fields") else model.__fields__
    for field_name, field in fields.items():
        field_type = getattr(field, "annotation", "Unknown")
        required = getattr(field, "is_required", getattr(field, "required", "Unknown"))
        default = field.default if hasattr(field, "default") and field.default is not None else "None"
        try:
            description = field.field_info.description
        except AttributeError:
            description = getattr(field, "description", "")
        info_lines.append(
            f"  {field_name}: type={field_type}, required={required}, default={default}, description={description}"
        )
    properties = [name for name, _ in inspect.getmembers(model, predicate=lambda m: isinstance(m, property))]
    if properties:
        info_lines.append("Properties: " + ", ".join(properties))
    methods = [name for name, _ in inspect.getmembers(model, predicate=inspect.isfunction) if not name.startswith("__")]
    if methods:
        info_lines.append("Methods: " + ", ".join(methods))
    return "\n".join(info_lines)

# --- Create a real planner configuration using your actual prompt templates ---
EXECUTOR_PROMPT = "You are a helpful assistant"
EXECUTOR_PROMPT_TEMPLATE = ChatPromptTemplate.from_messages(
    [
        ("system", EXECUTOR_PROMPT),
        MessagesPlaceholder(variable_name="messages"),
    ]
)
PLANNER_PROMPT = (
    "For the given objective, come up with a simple step by step plan. "
    "This plan should involve individual tasks, that if executed correctly will yield the correct answer. "
    "Do not add any superfluous steps. "
    "The result of the final step should be the final answer. "
    "Make sure that each step has all the information needed - do not skip steps."
)
PLANNER_PROMPT_TEMPLATE = ChatPromptTemplate.from_messages(
    [
        ("system", PLANNER_PROMPT),
        ("placeholder", "{messages}"),
    ]
)
planner_aug_llm_config = AugLLMConfig(
    name="planner",
    prompt_template=PLANNER_PROMPT_TEMPLATE,
    structured_output_model=Plan,
)

# --- Dynamic Node Function Factory using AugLLMConfig properties ---
def infer_input_field(config: AugLLMConfig) -> str:
    """
    Determine the input field based on the prompt template.
    Looks for 'messages' in optional_variables; otherwise, uses the first input variable or defaults to 'input'.
    """
    pt = config.prompt_template
    if hasattr(pt, "optional_variables") and pt.optional_variables:
        for var in pt.optional_variables:
            if var.lower() == "messages":
                return var
        return pt.optional_variables[0]
    elif hasattr(pt, "input_variables") and pt.input_variables:
        return pt.input_variables[0]
    return "input"

def infer_output_mapper(config: AugLLMConfig) -> Callable[[Any], Dict[str, Any]]:
    """
    Build an output mapper using the structured output model (or output parser) from the AugLLMConfig.
    
    If a structured output model is provided and VERBOSE_MODEL_INFO is enabled,
    prints detailed info about the model before returning a mapper that parses the result.
    """
    if config.structured_output_model:
        Model = config.structured_output_model
        if VERBOSE_MODEL_INFO:
            print("Using structured output model:")
            print(verbose_model_info(Model))
        def mapper(result: Any) -> Dict[str, Any]:
            if not isinstance(result, Model):
                try:
                    parsed = Model.parse_obj(result)
                except Exception as e:
                    raise ValueError(f"Error parsing result with {Model}: {e}")
            else:
                parsed = result
            # If the model has a 'plan' attribute, use it; otherwise, use the model instance.
            plan_val = getattr(parsed, "plan", parsed)
            steps_val = getattr(parsed, "steps", None)
            return {"plan": plan_val, "plan_steps": steps_val}
        return mapper
    elif config.output_parser:
        def mapper(result: Any) -> Dict[str, Any]:
            parsed = config.output_parser.parse(result)
            return {"plan": parsed, "plan_steps": parsed}
        return mapper
    else:
        def mapper(result: Any) -> Dict[str, Any]:
            return {"plan": result, "plan_steps": result}
        return mapper

def create_node_function(
    config: AugLLMConfig,
    command_goto: str = "execute_step",
    async_mode: bool = True,
) -> Callable[[Dict[str, Any]], Any]:
    """
    Creates a node function from an AugLLMConfig using its prompt template and structured output model.
    
    - It grabs and prints detailed info about the output model (or parser) if verbose is enabled.
    - It composes the runnable via compose_runnable.
    - It infers the input field and wraps the state input accordingly.
    - It maps the output (parsing the raw result into the structured model)
      and returns a Command (from langgraph.types) with update, goto, and resume fields.
    """
    output_model = config.structured_output_model or config.output_parser
    if output_model:
        print(f"Output model/parser detected: {output_model}")
    else:
        print("No structured output model or output parser provided.")
    
    node_runnable = compose_runnable(config)
    
    input_field = infer_input_field(config)
    if hasattr(config.prompt_template, "input_types") and input_field in config.prompt_template.input_types:
        expected_type = config.prompt_template.input_types[input_field]
        print(f"Inferred input field: '{input_field}' with expected type: {expected_type}")
    else:
        print(f"Inferred input field: '{input_field}' (no type info available)")
    
    def input_mapper(state: Dict[str, Any]) -> Dict[str, Any]:
        value = state.get(input_field)
        if value is None:
            raise ValueError(f"State does not contain required key '{input_field}'")
        if input_field.lower() == "messages":
            if isinstance(value, list):
                if not all(isinstance(x, tuple) and len(x) == 2 for x in value):
                    value = [("user", str(value))]
            else:
                value = [("user", str(value))]
            return {"messages": value}
        return {input_field: value}
    
    output_mapper = infer_output_mapper(config)
    
    if async_mode:
        async def node(state: Dict[str, Any]):
            payload = input_mapper(state)
            result = await node_runnable.ainvoke(payload)
            out = output_mapper(result)
            return Command(
                update={"plan": out["plan"]},
                goto=command_goto,
                resume={"plan": out["plan_steps"]}
            )
        return node
    else:
        def node(state: Dict[str, Any]):
            payload = input_mapper(state)
            result = node_runnable.invoke(payload)
            out = output_mapper(result)
            return Command(
                update={"plan": out["plan"]},
                goto=command_goto,
                resume={"plan": out["plan_steps"]}
            )
        return node

# ------------------------------
# StateSchemaManager (modified for base model)
# ------------------------------
class StateSchemaManager:
    """
    A dynamic schema manager that:
    - Builds schemas from dicts, lists, or Pydantic models.
    - Merges schemas while preserving uniqueness.
    - Tracks validators, properties, computed attributes, class/static methods.
    - Supports recursive field merging for nested schemas.
    - Uses a config dictionary for settings and overrides.
    - If initialized with a BaseModel, uses that as the base state model.
    - Can pretty-print the schema.
    """

    def __init__(self, 
                 data: Optional[Union[Dict[str, Any], Type[BaseModel]]] = None,
                 name: Optional[str] = None,
                 config: Optional[Dict[str, Any]] = None):
        self.fields = {}
        self.validators = {}
        self.properties = {}
        self.class_methods = {}
        self.static_methods = {}
        self.config = config or {}
        self.locked = False
        self.original_model = None

        if data is None:
            self.name = name or self.config.get("default_schema_name", "UnnamedSchema")
            return  

        if isinstance(data, type) and issubclass(data, BaseModel):
            self.original_model = data
            self.name = data.__name__
            self._load_from_model(data)
        else:
            self.name = name or "CustomState"
            if isinstance(data, dict):
                self._load_from_dict(data)
            else:
                raise TypeError(f"Unsupported data type: {type(data)}")

    def _load_from_dict(self, data: Dict[str, Any]) -> None:
        for key, value in data.items():
            inferred_type = self._infer_type(value)
            self.fields[key] = (inferred_type, Field(default=value))

    def _load_from_model(self, model_cls: Type[BaseModel]) -> None:
        fields = getattr(model_cls, "model_fields", model_cls.__fields__)
        for field_name, field_info in fields.items():
            self.fields[field_name] = (field_info.annotation, field_info)
        for name, method in inspect.getmembers(model_cls, predicate=inspect.isfunction):
            if hasattr(method, "__validator_config__"):
                self.validators[name] = method
            elif isinstance(getattr(model_cls, name, None), property):
                self.properties[name] = method
            elif isinstance(getattr(model_cls, name, None), classmethod):
                self.class_methods[name] = method
            elif isinstance(getattr(model_cls, name, None), staticmethod):
                self.static_methods[name] = method

    def _infer_type(self, value: Any) -> Type:
        if isinstance(value, str):
            return str
        elif isinstance(value, int):
            return int
        elif isinstance(value, float):
            return float
        elif isinstance(value, bool):
            return bool
        elif isinstance(value, list):
            return List[self._infer_type(value[0])] if value else List[Any]
        elif isinstance(value, dict):
            return Dict[str, Any]
        return Any

    def add_field(self, name: str, type_hint: Type, default: Any = None, required: bool = False) -> 'StateSchemaManager':
        if self.locked:
            raise ValueError("Schema is locked and cannot be modified.")
        field_info = Field(...) if required else Field(default=default)
        self.fields[name] = (type_hint, field_info)
        return self

    def remove_field(self, name: str) -> 'StateSchemaManager':
        if self.locked:
            raise ValueError("Schema is locked and cannot be modified.")
        if name in self.fields:
            del self.fields[name]
        return self

    def modify_field(self, name: str, new_type: Type, new_default: Any = None) -> 'StateSchemaManager':
        if self.locked:
            raise ValueError("Schema is locked and cannot be modified.")
        if name in self.fields:
            self.fields[name] = (new_type, Field(default=new_default))
        return self

    def merge(self, other: Union['StateSchemaManager', Type[BaseModel]]) -> 'StateSchemaManager':
        if self.locked:
            raise ValueError("Schema is locked and cannot be modified.")

        merged = StateSchemaManager(name=f"{self.name}_Merged", config=self.config)
        merged.fields = self.fields.copy()
        merged.validators = self.validators.copy()
        merged.properties = self.properties.copy()
        merged.class_methods = self.class_methods.copy()
        merged.static_methods = self.static_methods.copy()

        if isinstance(other, StateSchemaManager):
            for field, (field_type, field_info) in other.fields.items():
                if field == "messages":
                    continue  
                if field not in merged.fields:
                    merged.fields[field] = (field_type, field_info)
            merged.validators.update(other.validators)
            merged.properties.update(other.properties)
            merged.class_methods.update(other.class_methods)
            merged.static_methods.update(other.static_methods)
        elif isinstance(other, type) and issubclass(other, BaseModel):
            return merged.merge(StateSchemaManager(other, config=self.config))
        return merged

    def get_model(self, lock: bool = False) -> Type[BaseModel]:
        if lock:
            self.locked = True
        if self.original_model is not None:
            return self.original_model
        model = create_model(self.name, **self.fields)
        for name, validator in self.validators.items():
            setattr(model, name, validator)
        for name, prop in self.properties.items():
            setattr(model, name, prop)
        for name, method in self.class_methods.items():
            setattr(model, name, method)
        for name, method in self.static_methods.items():
            setattr(model, name, method)
        return model

    def pretty_print(self) -> None:
        print(f"class {self.name}(BaseModel):")
        if not any([self.fields, self.properties, self.class_methods, self.static_methods]):
            print("    pass  # No fields defined\n")
            return
        for field_name, (field_type, field_info) in self.fields.items():
            type_str = str(field_type).replace("typing.", "").replace("NoneType", "Optional")
            default_val = field_info.default if field_info.default != ... else "..."
            default_str = f" = {default_val}" if default_val != "..." else ""
            print(f"    {field_name}: {type_str}{default_str}")
        for prop_name in self.properties.keys():
            print(f"\n    @property")
            print(f"    def {prop_name}(self): ...  # Computed property")
        for method_name in self.class_methods.keys():
            print(f"\n    @classmethod")
            print(f"    def {method_name}(cls): ...  # Class method")
        for method_name in self.static_methods.keys():
            print(f"\n    @staticmethod")
            print(f"    def {method_name}(): ...  # Static method")
        print()

    # --- New method to integrate node functionality ---
    def create_node_from_config(
        self,
        config: AugLLMConfig,
        command_goto: str = "execute_step",
        async_mode: bool = True
    ) -> Callable[[Dict[str, Any]], Any]:
        """
        Create a node function from an AugLLMConfig, wrapping it with state validation
        using this schema's base model. The incoming state is validated against the schema's model.
        Returns a node function that accepts a dict, validates it, and then calls the underlying node.
        """
        base_node = create_node_function(config, command_goto, async_mode)
        Model = self.get_model()
        if async_mode:
            async def wrapped_node(state: Dict[str, Any]):
                try:
                    valid_state = Model.parse_obj(state).dict()
                except ValidationError as e:
                    raise ValueError(f"State validation failed: {e}")
                return await base_node(valid_state)
            return wrapped_node
        else:
            def wrapped_node(state: Dict[str, Any]):
                try:
                    valid_state = Model.parse_obj(state).dict()
                except ValidationError as e:
                    raise ValueError(f"State validation failed: {e}")
                return base_node(valid_state)
            return wrapped_node

    # --- New helper methods using langgraph.types Command and Send ---
    def create_default_command(
        self,
        update: Optional[Dict[str, Any]] = None,
        goto: Optional[Union[str, Send, List[Union[str, Send]]]] = None,
        resume: Optional[Union[Any, Dict[str, Any]]] = None
    ) -> Command:
        update = update or {}
        goto = goto or ""
        resume = resume or {}
        return Command(update=update, goto=goto, resume=resume)

    def create_send(self, node: str, arg: Any) -> Send:
        return Send(node, arg)

# ------------------------------
# Glue Nodes Functions (as before)
# ------------------------------
def glue_nodes_sync(
    nodes: List[Callable[[Dict[str, Any]], Command]],
    initial_state: Dict[str, Any]
) -> Command:
    state = initial_state.copy()
    last_command = None
    for node in nodes:
        last_command = node(state)
        if last_command.update:
            state.update(last_command.update)
    return last_command

async def glue_nodes_async(
    nodes: List[Callable[[Dict[str, Any]], Command]],
    initial_state: Dict[str, Any]
) -> Command:
    state = initial_state.copy()
    last_command = None
    for node in nodes:
        last_command = await node(state)
        if last_command.update:
            state.update(last_command.update)
    return last_command

# ------------------------------
# Example usage of node integration
# ------------------------------
# Create a StateSchemaManager from a simple dict schema.
schema_data = {
    "messages": [("user", "default input")],
    "extra": "initial extra"
}
schema_manager = StateSchemaManager(schema_data, name="ExampleState")
schema_manager.pretty_print()

# Now create a node function from an AugLLMConfig, wrapping it with state validation.
node_from_schema = schema_manager.create_node_from_config(planner_aug_llm_config, async_mode=True)

# ------------------------------
# Function Inspector (for node functions)
# ------------------------------
class FunctionInspector:
    """
    Inspects and analyzes a Python function.
    Prints signature, parameter details, type hints, docstring, source code, and a truncated CST.
    """
    def __init__(self, func: Any):
        if not inspect.isfunction(func):
            raise TypeError(f"Provided object {func} is not a function.")
        self.func = func
        self.signature = inspect.signature(func)
        self.type_hints = get_type_hints(func)
        self.docstring = inspect.getdoc(func)
        source = inspect.getsource(func)
        self.source_code = textwrap.dedent(source)
    
    def get_signature(self) -> inspect.Signature:
        return self.signature
    
    def get_parameters_info(self) -> Dict[str, Dict[str, Any]]:
        params_info = {}
        for name, param in self.signature.parameters.items():
            params_info[name] = {
                "annotation": param.annotation,
                "default": param.default,
                "kind": param.kind,
            }
        return params_info
    
    def get_return_annotation(self) -> Any:
        ret = self.signature.return_annotation
        return ret if ret is not inspect._empty else "No return annotation specified"
    
    def get_type_hints(self) -> Dict[str, Any]:
        return self.type_hints
    
    def get_docstring(self) -> Optional[str]:
        return self.docstring
    
    def get_source_code(self) -> str:
        return self.source_code
    
    def analyze_ast(self) -> cst.Module:
        try:
            tree = cst.parse_module(self.source_code)
            return tree
        except Exception as e:
            raise RuntimeError(f"Error parsing source code with LibCST: {e}")
    
    def print_info(self) -> None:
        print(f"Function: {self.func.__name__}")
        print("Signature:", self.get_signature())
        print("\nParameters:")
        for name, info in self.get_parameters_info().items():
            print(f"  {name}:")
            print(f"    Annotation: {info['annotation']}")
            print(f"    Default: {info['default']}")
            print(f"    Kind: {info['kind']}")
        print("\nReturn Annotation:", self.get_return_annotation())
        print("\nType Hints:", self.get_type_hints())
        print("\nDocstring:")
        print(self.get_docstring())
        print("\nSource Code:")
        print(self.get_source_code())
        print("\n--- Analyzing AST with LibCST ---")
        ast_tree = self.analyze_ast()
        ast_str = str(ast_tree)
        print(ast_str[:500] + "\n...")

# Inspect the created node function from the schema manager.
inspector = FunctionInspector(node_from_schema)
inspector.print_info()

# ------------------------------
# Test Execution
# ------------------------------
async def test_node():
    # Define initial state with an overridden 'extra' field.
    state = {"messages": [("user", "Plan a dinner party for 10")], "extra": "overridden extra"}
    result = await node_from_schema(state)
    print("\nNode execution result:", result)

# ------------------------------
# Test merging two nodes (glue)
# ------------------------------
async def test_glue_async_nodes():
    initial_state = {"messages": [("user", "Initial async input")], "extra": "initial extra"}
    # Create two nodes from the same configuration and schema.
    node_a = schema_manager.create_node_from_config(planner_aug_llm_config, async_mode=True)
    node_b = schema_manager.create_node_from_config(planner_aug_llm_config, async_mode=True)
    final_command = await glue_nodes_async([node_a, node_b], initial_state)
    print("Final Command from async glue pipeline:", final_command)

if __name__ == "__main__":
    import nest_asyncio
    nest_asyncio.apply()
    print("=== Testing Single Node Execution ===")
    asyncio.run(test_node())
    print("\n=== Testing Asynchronous Glue Nodes ===")
    asyncio.run(test_glue_async_nodes())
````

````python
import asyncio
import inspect
import textwrap
from typing import Any, Dict, List, Type, Union, Optional, Callable, get_type_hints
import libcst as cst
from pydantic import BaseModel, Field, create_model, ValidationError

# --- Import your actual models and AugLLM components ---
from src.haive.agents.plan_and_execute.models import Plan, Response, Act
from src.haive.core.aug_llm.base import AugLLMConfig, compose_runnable
from langchain.prompts import ChatPromptTemplate, MessagesPlaceholder

# Import Command and Send from langgraph.types
from langgraph.types import Command, Send

# Global flag for verbose model info
VERBOSE_MODEL_INFO = True

def verbose_model_info(model: Type[BaseModel]) -> str:
    """Return a verbose string with detailed info about the Pydantic model."""
    info_lines = []
    info_lines.append(f"Model Name: {model.__name__}")
    info_lines.append("Fields:")
    # Use Pydantic V2 API if available.
    fields = model.model_fields if hasattr(model, "model_fields") else model.__fields__
    for field_name, field in fields.items():
        field_type = getattr(field, "annotation", "Unknown")
        required = getattr(field, "is_required", getattr(field, "required", "Unknown"))
        default = field.default if hasattr(field, "default") and field.default is not None else "None"
        try:
            description = field.field_info.description
        except AttributeError:
            description = getattr(field, "description", "")
        info_lines.append(
            f"  {field_name}: type={field_type}, required={required}, default={default}, description={description}"
        )
    properties = [name for name, _ in inspect.getmembers(model, predicate=lambda m: isinstance(m, property))]
    if properties:
        info_lines.append("Properties: " + ", ".join(properties))
    methods = [name for name, _ in inspect.getmembers(model, predicate=inspect.isfunction) if not name.startswith("__")]
    if methods:
        info_lines.append("Methods: " + ", ".join(methods))
    return "\n".join(info_lines)

# --- Create a real planner configuration using your actual prompt templates ---
EXECUTOR_PROMPT = "You are a helpful assistant"
EXECUTOR_PROMPT_TEMPLATE = ChatPromptTemplate.from_messages(
    [
        ("system", EXECUTOR_PROMPT),
        MessagesPlaceholder(variable_name="messages"),
    ]
)
PLANNER_PROMPT = (
    "For the given objective, come up with a simple step by step plan. "
    "This plan should involve individual tasks, that if executed correctly will yield the correct answer. "
    "Do not add any superfluous steps. "
    "The result of the final step should be the final answer. "
    "Make sure that each step has all the information needed - do not skip steps."
)
PLANNER_PROMPT_TEMPLATE = ChatPromptTemplate.from_messages(
    [
        ("system", PLANNER_PROMPT),
        ("placeholder", "{messages}"),
    ]
)
planner_aug_llm_config = AugLLMConfig(
    name="planner",
    prompt_template=PLANNER_PROMPT_TEMPLATE,
    structured_output_model=Plan,
)

# --- Dynamic Node Function Factory using AugLLMConfig properties ---
def infer_input_field(config: AugLLMConfig) -> str:
    """Determine the input field based on the prompt template."""
    pt = config.prompt_template
    if hasattr(pt, "optional_variables") and pt.optional_variables:
        for var in pt.optional_variables:
            if var.lower() == "messages":
                return var
        return pt.optional_variables[0]
    elif hasattr(pt, "input_variables") and pt.input_variables:
        return pt.input_variables[0]
    return "input"

def infer_output_mapper(config: AugLLMConfig) -> Callable[[Any], Dict[str, Any]]:
    """
    Build an output mapper using the structured output model (or output parser) from the AugLLMConfig.
    If VERBOSE_MODEL_INFO is enabled, print detailed model info.
    """
    if config.structured_output_model:
        Model = config.structured_output_model
        if VERBOSE_MODEL_INFO:
            print("Using structured output model:")
            print(verbose_model_info(Model))
        def mapper(result: Any) -> Dict[str, Any]:
            if not isinstance(result, Model):
                try:
                    parsed = Model.parse_obj(result)
                except Exception as e:
                    raise ValueError(f"Error parsing result with {Model}: {e}")
            else:
                parsed = result
            plan_val = getattr(parsed, "plan", parsed)
            steps_val = getattr(parsed, "steps", None)
            return {"plan": plan_val, "plan_steps": steps_val}
        return mapper
    elif config.output_parser:
        def mapper(result: Any) -> Dict[str, Any]:
            parsed = config.output_parser.parse(result)
            return {"plan": parsed, "plan_steps": parsed}
        return mapper
    else:
        def mapper(result: Any) -> Dict[str, Any]:
            return {"plan": result, "plan_steps": result}
        return mapper

def create_node_function(
    config: AugLLMConfig,
    command_goto: str = "execute_step",
    async_mode: bool = True,
) -> Callable[[Dict[str, Any]], Any]:
    """
    Creates a node function from an AugLLMConfig.
    Returns a node function that calls the runnable and returns a Command.
    """
    output_model = config.structured_output_model or config.output_parser
    if output_model:
        print(f"Output model/parser detected: {output_model}")
    else:
        print("No structured output model or output parser provided.")
    
    node_runnable = compose_runnable(config)
    
    input_field = infer_input_field(config)
    if hasattr(config.prompt_template, "input_types") and input_field in config.prompt_template.input_types:
        expected_type = config.prompt_template.input_types[input_field]
        print(f"Inferred input field: '{input_field}' with expected type: {expected_type}")
    else:
        print(f"Inferred input field: '{input_field}' (no type info available)")
    
    def input_mapper(state: Dict[str, Any]) -> Dict[str, Any]:
        value = state.get(input_field)
        if value is None:
            raise ValueError(f"State does not contain required key '{input_field}'")
        if input_field.lower() == "messages":
            if isinstance(value, list):
                if not all(isinstance(x, tuple) and len(x) == 2 for x in value):
                    value = [("user", str(value))]
            else:
                value = [("user", str(value))]
            return {"messages": value}
        return {input_field: value}
    
    output_mapper = infer_output_mapper(config)
    
    if async_mode:
        async def node(state: Dict[str, Any]):
            payload = input_mapper(state)
            result = await node_runnable.ainvoke(payload)
            out = output_mapper(result)
            return Command(
                update={"plan": out["plan"]},
                goto=command_goto,
                resume={"plan": out["plan_steps"]}
            )
        return node
    else:
        def node(state: Dict[str, Any]):
            payload = input_mapper(state)
            result = node_runnable.invoke(payload)
            out = output_mapper(result)
            return Command(
                update={"plan": out["plan"]},
                goto=command_goto,
                resume={"plan": out["plan_steps"]}
            )
        return node

# --- Glue Nodes Functions with goto override ---
def glue_nodes_sync(
    nodes: List[Callable[[Dict[str, Any]], Command]],
    initial_state: Dict[str, Any]
) -> Command:
    state = initial_state.copy()
    last_command = None
    for i, node in enumerate(nodes):
        cmd = node(state)
        # If there's a next node, set the goto field to its __name__
        if i < len(nodes) - 1:
            next_node_name = nodes[i+1].__name__
            # Create a new Command with updated goto (assuming Command is immutable)
            cmd = Command(update=cmd.update, goto=next_node_name, resume=cmd.resume)
        last_command = cmd
        if last_command.update:
            state.update(last_command.update)
    return last_command

async def glue_nodes_async(
    nodes: List[Callable[[Dict[str, Any]], Command]],
    initial_state: Dict[str, Any]
) -> Command:
    state = initial_state.copy()
    last_command = None
    for i, node in enumerate(nodes):
        cmd = await node(state)
        if i < len(nodes) - 1:
            next_node_name = nodes[i+1].__name__
            cmd = Command(update=cmd.update, goto=next_node_name, resume=cmd.resume)
        last_command = cmd
        if last_command.update:
            state.update(last_command.update)
    return last_command

# ------------------------------
# StateSchemaManager (modified for base model)
# ------------------------------
class StateSchemaManager:
    """
    A dynamic schema manager that:
    - Builds schemas from dicts, lists, or Pydantic models.
    - Merges schemas while preserving uniqueness.
    - Tracks validators, properties, computed attributes, class/static methods.
    - Supports recursive field merging for nested schemas.
    - Uses a config dictionary for settings and overrides.
    - If initialized with a BaseModel, uses that as the base state model.
    - Can pretty-print the schema.
    """

    def __init__(self, 
                 data: Optional[Union[Dict[str, Any], Type[BaseModel]]] = None,
                 name: Optional[str] = None,
                 config: Optional[Dict[str, Any]] = None):
        self.fields = {}
        self.validators = {}
        self.properties = {}
        self.class_methods = {}
        self.static_methods = {}
        self.config = config or {}
        self.locked = False
        self.original_model = None

        if data is None:
            self.name = name or self.config.get("default_schema_name", "UnnamedSchema")
            return  

        if isinstance(data, type) and issubclass(data, BaseModel):
            self.original_model = data
            self.name = data.__name__
            self._load_from_model(data)
        else:
            self.name = name or "CustomState"
            if isinstance(data, dict):
                self._load_from_dict(data)
            else:
                raise TypeError(f"Unsupported data type: {type(data)}")

    def _load_from_dict(self, data: Dict[str, Any]) -> None:
        for key, value in data.items():
            inferred_type = self._infer_type(value)
            self.fields[key] = (inferred_type, Field(default=value))

    def _load_from_model(self, model_cls: Type[BaseModel]) -> None:
        fields = getattr(model_cls, "model_fields", model_cls.__fields__)
        for field_name, field_info in fields.items():
            self.fields[field_name] = (field_info.annotation, field_info)
        for name, method in inspect.getmembers(model_cls, predicate=inspect.isfunction):
            if hasattr(method, "__validator_config__"):
                self.validators[name] = method
            elif isinstance(getattr(model_cls, name, None), property):
                self.properties[name] = method
            elif isinstance(getattr(model_cls, name, None), classmethod):
                self.class_methods[name] = method
            elif isinstance(getattr(model_cls, name, None), staticmethod):
                self.static_methods[name] = method

    def _infer_type(self, value: Any) -> Type:
        if isinstance(value, str):
            return str
        elif isinstance(value, int):
            return int
        elif isinstance(value, float):
            return float
        elif isinstance(value, bool):
            return bool
        elif isinstance(value, list):
            return List[self._infer_type(value[0])] if value else List[Any]
        elif isinstance(value, dict):
            return Dict[str, Any]
        return Any

    def add_field(self, name: str, type_hint: Type, default: Any = None, required: bool = False) -> 'StateSchemaManager':
        if self.locked:
            raise ValueError("Schema is locked and cannot be modified.")
        field_info = Field(...) if required else Field(default=default)
        self.fields[name] = (type_hint, field_info)
        return self

    def remove_field(self, name: str) -> 'StateSchemaManager':
        if self.locked:
            raise ValueError("Schema is locked and cannot be modified.")
        if name in self.fields:
            del self.fields[name]
        return self

    def modify_field(self, name: str, new_type: Type, new_default: Any = None) -> 'StateSchemaManager':
        if self.locked:
            raise ValueError("Schema is locked and cannot be modified.")
        if name in self.fields:
            self.fields[name] = (new_type, Field(default=new_default))
        return self

    def merge(self, other: Union['StateSchemaManager', Type[BaseModel]]) -> 'StateSchemaManager':
        if self.locked:
            raise ValueError("Schema is locked and cannot be modified.")

        merged = StateSchemaManager(name=f"{self.name}_Merged", config=self.config)
        merged.fields = self.fields.copy()
        merged.validators = self.validators.copy()
        merged.properties = self.properties.copy()
        merged.class_methods = self.class_methods.copy()
        merged.static_methods = self.static_methods.copy()

        if isinstance(other, StateSchemaManager):
            for field, (field_type, field_info) in other.fields.items():
                if field == "messages":
                    continue  
                if field not in merged.fields:
                    merged.fields[field] = (field_type, field_info)
            merged.validators.update(other.validators)
            merged.properties.update(other.properties)
            merged.class_methods.update(other.class_methods)
            merged.static_methods.update(other.static_methods)
        elif isinstance(other, type) and issubclass(other, BaseModel):
            return merged.merge(StateSchemaManager(other, config=self.config))
        return merged

    def get_model(self, lock: bool = False) -> Type[BaseModel]:
        if lock:
            self.locked = True
        if self.original_model is not None:
            return self.original_model
        model = create_model(self.name, **self.fields)
        for name, validator in self.validators.items():
            setattr(model, name, validator)
        for name, prop in self.properties.items():
            setattr(model, name, prop)
        for name, method in self.class_methods.items():
            setattr(model, name, method)
        for name, method in self.static_methods.items():
            setattr(model, name, method)
        return model

    def pretty_print(self) -> None:
        print(f"class {self.name}(BaseModel):")
        if not any([self.fields, self.properties, self.class_methods, self.static_methods]):
            print("    pass  # No fields defined\n")
            return
        for field_name, (field_type, field_info) in self.fields.items():
            type_str = str(field_type).replace("typing.", "").replace("NoneType", "Optional")
            default_val = field_info.default if field_info.default != ... else "..."
            default_str = f" = {default_val}" if default_val != "..." else ""
            print(f"    {field_name}: {type_str}{default_str}")
        for prop_name in self.properties.keys():
            print(f"\n    @property")
            print(f"    def {prop_name}(self): ...  # Computed property")
        for method_name in self.class_methods.keys():
            print(f"\n    @classmethod")
            print(f"    def {method_name}(cls): ...  # Class method")
        for method_name in self.static_methods.keys():
            print(f"\n    @staticmethod")
            print(f"    def {method_name}(): ...  # Static method")
        print()

    # --- New method to integrate node functionality ---
    def create_node_from_config(
        self,
        config: AugLLMConfig,
        command_goto: str = "execute_step",
        async_mode: bool = True
    ) -> Callable[[Dict[str, Any]], Any]:
        """
        Create a node function from an AugLLMConfig, wrapping it with state validation
        using this schema's base model.
        Returns a node function that accepts a dict, validates it, and then calls the underlying node.
        """
        base_node = create_node_function(config, command_goto, async_mode)
        Model = self.get_model()
        if async_mode:
            async def wrapped_node(state: Dict[str, Any]):
                try:
                    valid_state = Model.parse_obj(state).dict()
                except ValidationError as e:
                    raise ValueError(f"State validation failed: {e}")
                return await base_node(valid_state)
            return wrapped_node
        else:
            def wrapped_node(state: Dict[str, Any]):
                try:
                    valid_state = Model.parse_obj(state).dict()
                except ValidationError as e:
                    raise ValueError(f"State validation failed: {e}")
                return base_node(valid_state)
            return wrapped_node

    # --- New helper methods using langgraph.types Command and Send ---
    def create_default_command(
        self,
        update: Optional[Dict[str, Any]] = None,
        goto: Optional[Union[str, Send, List[Union[str, Send]]]] = None,
        resume: Optional[Union[Any, Dict[str, Any]]] = None
    ) -> Command:
        update = update or {}
        goto = goto or ""
        resume = resume or {}
        return Command(update=update, goto=goto, resume=resume)

    def create_send(self, node: str, arg: Any) -> Send:
        return Send(node, arg)

# ------------------------------
# Glue Nodes Functions with goto override
# ------------------------------
def glue_nodes_sync(
    nodes: List[Callable[[Dict[str, Any]], Command]],
    initial_state: Dict[str, Any]
) -> Command:
    state = initial_state.copy()
    last_command = None
    for i, node in enumerate(nodes):
        cmd = node(state)
        if i < len(nodes) - 1:
            next_node_name = nodes[i+1].__name__
            # Create a new Command with updated goto field.
            cmd = Command(update=cmd.update, goto=next_node_name, resume=cmd.resume)
        last_command = cmd
        if last_command.update:
            state.update(last_command.update)
    return last_command

async def glue_nodes_async(
    nodes: List[Callable[[Dict[str, Any]], Command]],
    initial_state: Dict[str, Any]
) -> Command:
    state = initial_state.copy()
    last_command = None
    for i, node in enumerate(nodes):
        cmd = await node(state)
        if i < len(nodes) - 1:
            next_node_name = nodes[i+1].__name__
            cmd = Command(update=cmd.update, goto=next_node_name, resume=cmd.resume)
        last_command = cmd
        if last_command.update:
            state.update(last_command.update)
    return last_command

# ------------------------------
# Example usage of node integration
# ------------------------------
# Create a StateSchemaManager from a simple dict schema.
schema_data = {
    "messages": [("user", "default input")],
    "extra": "initial extra"
}
schema_manager = StateSchemaManager(schema_data, name="ExampleState")
schema_manager.pretty_print()

# Now create a node function from an AugLLMConfig, wrapping it with state validation.
node_from_schema = schema_manager.create_node_from_config(planner_aug_llm_config, async_mode=True)

# ------------------------------
# Function Inspector for Node Functions
# ------------------------------
class FunctionInspector:
    """
    Inspects and analyzes a Python function.
    Prints signature, parameter details, type hints, docstring, source code, and a truncated CST.
    """
    def __init__(self, func: Any):
        if not inspect.isfunction(func):
            raise TypeError(f"Provided object {func} is not a function.")
        self.func = func
        self.signature = inspect.signature(func)
        self.type_hints = get_type_hints(func)
        self.docstring = inspect.getdoc(func)
        source = inspect.getsource(func)
        self.source_code = textwrap.dedent(source)
    
    def get_signature(self) -> inspect.Signature:
        return self.signature
    
    def get_parameters_info(self) -> Dict[str, Dict[str, Any]]:
        params_info = {}
        for name, param in self.signature.parameters.items():
            params_info[name] = {
                "annotation": param.annotation,
                "default": param.default,
                "kind": param.kind,
            }
        return params_info
    
    def get_return_annotation(self) -> Any:
        ret = self.signature.return_annotation
        return ret if ret is not inspect._empty else "No return annotation specified"
    
    def get_type_hints(self) -> Dict[str, Any]:
        return self.type_hints
    
    def get_docstring(self) -> Optional[str]:
        return self.docstring
    
    def get_source_code(self) -> str:
        return self.source_code
    
    def analyze_ast(self) -> cst.Module:
        try:
            tree = cst.parse_module(self.source_code)
            return tree
        except Exception as e:
            raise RuntimeError(f"Error parsing source code with LibCST: {e}")
    
    def print_info(self) -> None:
        print(f"Function: {self.func.__name__}")
        print("Signature:", self.get_signature())
        print("\nParameters:")
        for name, info in self.get_parameters_info().items():
            print(f"  {name}:")
            print(f"    Annotation: {info['annotation']}")
            print(f"    Default: {info['default']}")
            print(f"    Kind: {info['kind']}")
        print("\nReturn Annotation:", self.get_return_annotation())
        print("\nType Hints:", self.get_type_hints())
        print("\nDocstring:")
        print(self.get_docstring())
        print("\nSource Code:")
        print(self.get_source_code())
        print("\n--- Analyzing AST with LibCST ---")
        ast_tree = self.analyze_ast()
        ast_str = str(ast_tree)
        print(ast_str[:500] + "\n...")

# Inspect the created node function from the schema manager.
inspector = FunctionInspector(node_from_schema)
inspector.print_info()

# ------------------------------
# Test Execution
# ------------------------------
async def test_node():
    state = {"messages": [("user", "Plan a dinner party for 10")], "extra": "overridden extra"}
    result = await node_from_schema(state)
    print("\nNode execution result:", result)

# Test merging two nodes (glue)
async def test_glue_async_nodes():
    initial_state = {"messages": [("user", "Initial async input")], "extra": "initial extra"}
    node_a = schema_manager.create_node_from_config(planner_aug_llm_config, async_mode=True)
    node_b = schema_manager.create_node_from_config(planner_aug_llm_config, async_mode=True)
    final_command = await glue_nodes_async([node_a, node_b], initial_state)
    print("Final Command from async glue pipeline:", final_command)

if __name__ == "__main__":
    import nest_asyncio
    nest_asyncio.apply()
    print("=== Testing Single Node Execution ===")
    asyncio.run(test_node())
    print("\n=== Testing Asynchronous Glue Nodes ===")
    asyncio.run(test_glue_async_nodes())
````

````python
import asyncio
import inspect
import textwrap
from typing import Any, Dict, List, Type, Union, Optional, Callable, get_type_hints
import libcst as cst
from pydantic import BaseModel, Field, create_model, ValidationError

# --- Import your actual models and AugLLM components ---
from src.haive.agents.plan_and_execute.models import Plan, Response, Act
from src.haive.core.aug_llm.base import AugLLMConfig, compose_runnable
from langchain.prompts import ChatPromptTemplate, MessagesPlaceholder

# Import Command and Send from langgraph.types
from langgraph.types import Command, Send

# Global flag for verbose model info
VERBOSE_MODEL_INFO = True

def verbose_model_info(model: Type[BaseModel]) -> str:
    """Return a verbose string with detailed info about the Pydantic model."""
    info_lines = []
    info_lines.append(f"Model Name: {model.__name__}")
    info_lines.append("Fields:")
    # Use Pydantic V2 API if available.
    fields = model.model_fields if hasattr(model, "model_fields") else model.__fields__
    for field_name, field in fields.items():
        field_type = getattr(field, "annotation", "Unknown")
        required = getattr(field, "is_required", getattr(field, "required", "Unknown"))
        default = field.default if hasattr(field, "default") and field.default is not None else "None"
        try:
            description = field.field_info.description
        except AttributeError:
            description = getattr(field, "description", "")
        info_lines.append(
            f"  {field_name}: type={field_type}, required={required}, default={default}, description={description}"
        )
    properties = [name for name, _ in inspect.getmembers(model, predicate=lambda m: isinstance(m, property))]
    if properties:
        info_lines.append("Properties: " + ", ".join(properties))
    methods = [name for name, _ in inspect.getmembers(model, predicate=inspect.isfunction) if not name.startswith("__")]
    if methods:
        info_lines.append("Methods: " + ", ".join(methods))
    return "\n".join(info_lines)

# --- Create a real planner configuration using your actual prompt templates ---
EXECUTOR_PROMPT = "You are a helpful assistant"
EXECUTOR_PROMPT_TEMPLATE = ChatPromptTemplate.from_messages(
    [
        ("system", EXECUTOR_PROMPT),
        MessagesPlaceholder(variable_name="messages"),
    ]
)
PLANNER_PROMPT = (
    "For the given objective, come up with a simple step by step plan. "
    "This plan should involve individual tasks, that if executed correctly will yield the correct answer. "
    "Do not add any superfluous steps. "
    "The result of the final step should be the final answer. "
    "Make sure that each step has all the information needed - do not skip steps."
)
PLANNER_PROMPT_TEMPLATE = ChatPromptTemplate.from_messages(
    [
        ("system", PLANNER_PROMPT),
        ("placeholder", "{messages}"),
    ]
)
planner_aug_llm_config = AugLLMConfig(
    name="planner",
    prompt_template=PLANNER_PROMPT_TEMPLATE,
    structured_output_model=Plan,
)

# --- Dynamic Node Function Factory using AugLLMConfig properties ---
def infer_input_field(config: AugLLMConfig) -> str:
    """Determine the input field based on the prompt template."""
    pt = config.prompt_template
    if hasattr(pt, "optional_variables") and pt.optional_variables:
        for var in pt.optional_variables:
            if var.lower() == "messages":
                return var
        return pt.optional_variables[0]
    elif hasattr(pt, "input_variables") and pt.input_variables:
        return pt.input_variables[0]
    return "input"

def infer_output_mapper(config: AugLLMConfig) -> Callable[[Any], Dict[str, Any]]:
    """
    Build an output mapper using the structured output model (or output parser) from the AugLLMConfig.
    If VERBOSE_MODEL_INFO is enabled, prints detailed info about the model.
    """
    if config.structured_output_model:
        Model = config.structured_output_model
        if VERBOSE_MODEL_INFO:
            print("Using structured output model:")
            print(verbose_model_info(Model))
        def mapper(result: Any) -> Dict[str, Any]:
            if not isinstance(result, Model):
                try:
                    # Use model_validate instead of parse_obj in Pydantic V2.
                    parsed = Model.model_validate(result)
                except Exception as e:
                    raise ValueError(f"Error parsing result with {Model}: {e}")
            else:
                parsed = result
            plan_val = getattr(parsed, "plan", parsed)
            steps_val = getattr(parsed, "steps", None)
            return {"plan": plan_val, "plan_steps": steps_val}
        return mapper
    elif config.output_parser:
        def mapper(result: Any) -> Dict[str, Any]:
            parsed = config.output_parser.parse(result)
            return {"plan": parsed, "plan_steps": parsed}
        return mapper
    else:
        def mapper(result: Any) -> Dict[str, Any]:
            return {"plan": result, "plan_steps": result}
        return mapper

def create_node_function(
    config: AugLLMConfig,
    command_goto: str = "execute_step",
    async_mode: bool = True,
) -> Callable[[Dict[str, Any]], Any]:
    """
    Creates a node function from an AugLLMConfig.
    Returns a node function that calls the runnable and returns a Command.
    """
    output_model = config.structured_output_model or config.output_parser
    if output_model:
        print(f"Output model/parser detected: {output_model}")
    else:
        print("No structured output model or output parser provided.")
    
    node_runnable = compose_runnable(config)
    input_field = infer_input_field(config)
    if hasattr(config.prompt_template, "input_types") and input_field in config.prompt_template.input_types:
        expected_type = config.prompt_template.input_types[input_field]
        print(f"Inferred input field: '{input_field}' with expected type: {expected_type}")
    else:
        print(f"Inferred input field: '{input_field}' (no type info available)")
    
    def input_mapper(state: Dict[str, Any]) -> Dict[str, Any]:
        value = state.get(input_field)
        if value is None:
            raise ValueError(f"State does not contain required key '{input_field}'")
        if input_field.lower() == "messages":
            if isinstance(value, list):
                if not all(isinstance(x, tuple) and len(x) == 2 for x in value):
                    value = [("user", str(value))]
            else:
                value = [("user", str(value))]
            return {"messages": value}
        return {input_field: value}
    
    output_mapper = infer_output_mapper(config)
    
    if async_mode:
        async def node(state: Dict[str, Any]):
            payload = input_mapper(state)
            result = await node_runnable.ainvoke(payload)
            out = output_mapper(result)
            return Command(
                update={"plan": out["plan"]},
                goto=command_goto,
                resume={"plan": out["plan_steps"]}
            )
        return node
    else:
        def node(state: Dict[str, Any]):
            payload = input_mapper(state)
            result = node_runnable.invoke(payload)
            out = output_mapper(result)
            return Command(
                update={"plan": out["plan"]},
                goto=command_goto,
                resume={"plan": out["plan_steps"]}
            )
        return node

# --- Glue Nodes Functions with goto override ---
def glue_nodes_sync(
    nodes: List[Callable[[Dict[str, Any]], Command]],
    initial_state: Dict[str, Any]
) -> Command:
    state = initial_state.copy()
    last_command = None
    for i, node in enumerate(nodes):
        cmd = node(state)
        if i < len(nodes) - 1:
            next_node_name = nodes[i+1].__name__
            cmd = Command(update=cmd.update, goto=next_node_name, resume=cmd.resume)
        last_command = cmd
        if last_command.update:
            state.update(last_command.update)
    return last_command

async def glue_nodes_async(
    nodes: List[Callable[[Dict[str, Any]], Command]],
    initial_state: Dict[str, Any]
) -> Command:
    state = initial_state.copy()
    last_command = None
    for i, node in enumerate(nodes):
        cmd = await node(state)
        if i < len(nodes) - 1:
            next_node_name = nodes[i+1].__name__
            cmd = Command(update=cmd.update, goto=next_node_name, resume=cmd.resume)
        last_command = cmd
        if last_command.update:
            state.update(last_command.update)
    return last_command

# ------------------------------
# StateSchemaManager (modified for base model)
# ------------------------------
class StateSchemaManager:
    """
    A dynamic schema manager that:
    - Builds schemas from dicts, lists, or Pydantic models.
    - Merges schemas while preserving uniqueness.
    - Tracks validators, properties, computed attributes, class/static methods.
    - Supports recursive field merging for nested schemas.
    - Uses a config dictionary for settings and overrides.
    - If initialized with a BaseModel, uses that as the base state model.
    - Can pretty-print the schema.
    """

    def __init__(self, 
                 data: Optional[Union[Dict[str, Any], Type[BaseModel]]] = None,
                 name: Optional[str] = None,
                 config: Optional[Dict[str, Any]] = None):
        self.fields = {}
        self.validators = {}
        self.properties = {}
        self.class_methods = {}
        self.static_methods = {}
        self.config = config or {}
        self.locked = False
        self.original_model = None

        if data is None:
            self.name = name or self.config.get("default_schema_name", "UnnamedSchema")
            return  

        if isinstance(data, type) and issubclass(data, BaseModel):
            self.original_model = data
            self.name = data.__name__
            self._load_from_model(data)
        else:
            self.name = name or "CustomState"
            if isinstance(data, dict):
                self._load_from_dict(data)
            else:
                raise TypeError(f"Unsupported data type: {type(data)}")

    def _load_from_dict(self, data: Dict[str, Any]) -> None:
        for key, value in data.items():
            inferred_type = self._infer_type(value)
            self.fields[key] = (inferred_type, Field(default=value))

    def _load_from_model(self, model_cls: Type[BaseModel]) -> None:
        fields = getattr(model_cls, "model_fields", model_cls.__fields__)
        for field_name, field_info in fields.items():
            self.fields[field_name] = (field_info.annotation, field_info)
        for name, method in inspect.getmembers(model_cls, predicate=inspect.isfunction):
            if hasattr(method, "__validator_config__"):
                self.validators[name] = method
            elif isinstance(getattr(model_cls, name, None), property):
                self.properties[name] = method
            elif isinstance(getattr(model_cls, name, None), classmethod):
                self.class_methods[name] = method
            elif isinstance(getattr(model_cls, name, None), staticmethod):
                self.static_methods[name] = method

    def _infer_type(self, value: Any) -> Type:
        if isinstance(value, str):
            return str
        elif isinstance(value, int):
            return int
        elif isinstance(value, float):
            return float
        elif isinstance(value, bool):
            return bool
        elif isinstance(value, list):
            return List[self._infer_type(value[0])] if value else List[Any]
        elif isinstance(value, dict):
            return Dict[str, Any]
        return Any

    def add_field(self, name: str, type_hint: Type, default: Any = None, required: bool = False) -> 'StateSchemaManager':
        if self.locked:
            raise ValueError("Schema is locked and cannot be modified.")
        field_info = Field(...) if required else Field(default=default)
        self.fields[name] = (type_hint, field_info)
        return self

    def remove_field(self, name: str) -> 'StateSchemaManager':
        if self.locked:
            raise ValueError("Schema is locked and cannot be modified.")
        if name in self.fields:
            del self.fields[name]
        return self

    def modify_field(self, name: str, new_type: Type, new_default: Any = None) -> 'StateSchemaManager':
        if self.locked:
            raise ValueError("Schema is locked and cannot be modified.")
        if name in self.fields:
            self.fields[name] = (new_type, Field(default=new_default))
        return self

    def merge(self, other: Union['StateSchemaManager', Type[BaseModel]]) -> 'StateSchemaManager':
        if self.locked:
            raise ValueError("Schema is locked and cannot be modified.")

        merged = StateSchemaManager(name=f"{self.name}_Merged", config=self.config)
        merged.fields = self.fields.copy()
        merged.validators = self.validators.copy()
        merged.properties = self.properties.copy()
        merged.class_methods = self.class_methods.copy()
        merged.static_methods = self.static_methods.copy()

        if isinstance(other, StateSchemaManager):
            for field, (field_type, field_info) in other.fields.items():
                if field == "messages":
                    continue  
                if field not in merged.fields:
                    merged.fields[field] = (field_type, field_info)
            merged.validators.update(other.validators)
            merged.properties.update(other.properties)
            merged.class_methods.update(other.class_methods)
            merged.static_methods.update(other.static_methods)
        elif isinstance(other, type) and issubclass(other, BaseModel):
            return merged.merge(StateSchemaManager(other, config=self.config))
        return merged

    def get_model(self, lock: bool = False) -> Type[BaseModel]:
        if lock:
            self.locked = True
        if self.original_model is not None:
            return self.original_model
        model = create_model(self.name, **self.fields)
        for name, validator in self.validators.items():
            setattr(model, name, validator)
        for name, prop in self.properties.items():
            setattr(model, name, prop)
        for name, method in self.class_methods.items():
            setattr(model, name, method)
        for name, method in self.static_methods.items():
            setattr(model, name, method)
        return model

    def pretty_print(self) -> None:
        print(f"class {self.name}(BaseModel):")
        if not any([self.fields, self.properties, self.class_methods, self.static_methods]):
            print("    pass  # No fields defined\n")
            return
        for field_name, (field_type, field_info) in self.fields.items():
            type_str = str(field_type).replace("typing.", "").replace("NoneType", "Optional")
            default_val = field_info.default if field_info.default != ... else "..."
            default_str = f" = {default_val}" if default_val != "..." else ""
            print(f"    {field_name}: {type_str}{default_str}")
        for prop_name in self.properties.keys():
            print(f"\n    @property")
            print(f"    def {prop_name}(self): ...  # Computed property")
        for method_name in self.class_methods.keys():
            print(f"\n    @classmethod")
            print(f"    def {method_name}(cls): ...  # Class method")
        for method_name in self.static_methods.keys():
            print(f"\n    @staticmethod")
            print(f"    def {method_name}(): ...  # Static method")
        print()

    # --- New method to integrate node functionality ---
    def create_node_from_config(
        self,
        config: AugLLMConfig,
        command_goto: str = "execute_step",
        async_mode: bool = True
    ) -> Callable[[Dict[str, Any]], Any]:
        """
        Create a node function from an AugLLMConfig, wrapping it with state validation
        using this schema's base model.
        Returns a node function that accepts a dict, validates it, and then calls the underlying node.
        """
        base_node = create_node_function(config, command_goto, async_mode)
        Model = self.get_model()
        if async_mode:
            async def wrapped_node(state: Dict[str, Any]):
                try:
                    # Use model_validate and model_dump in Pydantic V2 instead of parse_obj and dict.
                    valid_state = Model.model_validate(state).model_dump()
                except ValidationError as e:
                    raise ValueError(f"State validation failed: {e}")
                return await base_node(valid_state)
            return wrapped_node
        else:
            def wrapped_node(state: Dict[str, Any]):
                try:
                    valid_state = Model.model_validate(state).model_dump()
                except ValidationError as e:
                    raise ValueError(f"State validation failed: {e}")
                return base_node(valid_state)
            return wrapped_node

    # --- New helper methods using langgraph.types Command and Send ---
    def create_default_command(
        self,
        update: Optional[Dict[str, Any]] = None,
        goto: Optional[Union[str, Send, List[Union[str, Send]]]] = None,
        resume: Optional[Union[Any, Dict[str, Any]]] = None
    ) -> Command:
        update = update or {}
        goto = goto or ""
        resume = resume or {}
        return Command(update=update, goto=goto, resume=resume)

    def create_send(self, node: str, arg: Any) -> Send:
        return Send(node, arg)

# ------------------------------
# Glue Nodes Functions with goto override
# ------------------------------
def glue_nodes_sync(
    nodes: List[Callable[[Dict[str, Any]], Command]],
    initial_state: Dict[str, Any]
) -> Command:
    state = initial_state.copy()
    last_command = None
    for i, node in enumerate(nodes):
        cmd = node(state)
        if i < len(nodes) - 1:
            next_node_name = nodes[i+1].__name__
            cmd = Command(update=cmd.update, goto=next_node_name, resume=cmd.resume)
        last_command = cmd
        if last_command.update:
            state.update(last_command.update)
    return last_command

async def glue_nodes_async(
    nodes: List[Callable[[Dict[str, Any]], Command]],
    initial_state: Dict[str, Any]
) -> Command:
    state = initial_state.copy()
    last_command = None
    for i, node in enumerate(nodes):
        cmd = await node(state)
        if i < len(nodes) - 1:
            next_node_name = nodes[i+1].__name__
            cmd = Command(update=cmd.update, goto=next_node_name, resume=cmd.resume)
        last_command = cmd
        if last_command.update:
            state.update(last_command.update)
    return last_command

# ------------------------------
# Example usage: Build State Graph and Process I/O
# ------------------------------
# Create a StateSchemaManager from an initial dict schema.
schema_data = {
    "messages": [("user", "default input")],
    "extra": "initial extra"
}
schema_manager = StateSchemaManager(schema_data, name="ExampleState")
print("State Schema:")
schema_manager.pretty_print()

# Create a node function from an AugLLMConfig, wrapping it with state validation.
node_from_schema = schema_manager.create_node_from_config(planner_aug_llm_config, async_mode=True)

# ------------------------------
# Function Inspector for Node Functions
# ------------------------------
class FunctionInspector:
    """
    Inspects and analyzes a Python function.
    Prints signature, parameter details, type hints, docstring, source code, and a truncated CST.
    """
    def __init__(self, func: Any):
        if not inspect.isfunction(func):
            raise TypeError(f"Provided object {func} is not a function.")
        self.func = func
        self.signature = inspect.signature(func)
        self.type_hints = get_type_hints(func)
        self.docstring = inspect.getdoc(func)
        source = inspect.getsource(func)
        self.source_code = textwrap.dedent(source)
    
    def get_signature(self) -> inspect.Signature:
        return self.signature
    
    def get_parameters_info(self) -> Dict[str, Dict[str, Any]]:
        params_info = {}
        for name, param in self.signature.parameters.items():
            params_info[name] = {
                "annotation": param.annotation,
                "default": param.default,
                "kind": param.kind,
            }
        return params_info
    
    def get_return_annotation(self) -> Any:
        ret = self.signature.return_annotation
        return ret if ret is not inspect._empty else "No return annotation specified"
    
    def get_type_hints(self) -> Dict[str, Any]:
        return self.type_hints
    
    def get_docstring(self) -> Optional[str]:
        return self.docstring
    
    def get_source_code(self) -> str:
        return self.source_code
    
    def analyze_ast(self) -> cst.Module:
        try:
            tree = cst.parse_module(self.source_code)
            return tree
        except Exception as e:
            raise RuntimeError(f"Error parsing source code with LibCST: {e}")
    
    def print_info(self) -> None:
        print(f"Function: {self.func.__name__}")
        print("Signature:", self.get_signature())
        print("\nParameters:")
        for name, info in self.get_parameters_info().items():
            print(f"  {name}:")
            print(f"    Annotation: {info['annotation']}")
            print(f"    Default: {info['default']}")
            print(f"    Kind: {info['kind']}")
        print("\nReturn Annotation:", self.get_return_annotation())
        print("\nType Hints:", self.get_type_hints())
        print("\nDocstring:")
        print(self.get_docstring())
        print("\nSource Code:")
        print(self.get_source_code())
        print("\n--- Analyzing AST with LibCST ---")
        ast_tree = self.analyze_ast()
        ast_str = str(ast_tree)
        print(ast_str[:500] + "\n...")

# Inspect the created node function from the schema manager.
inspector = FunctionInspector(node_from_schema)
inspector.print_info()

# ------------------------------
# Test Execution: Build State Graph and Process I/O
# ------------------------------
async def test_state_graph():
    # Start with a minimal state.
    initial_state = {"messages": [("user", "Plan a dinner party for 10")]}
    # Simulate a chain by gluing two nodes together.
    node_a = schema_manager.create_node_from_config(planner_aug_llm_config, async_mode=True)
    node_b = schema_manager.create_node_from_config(planner_aug_llm_config, async_mode=True)
    final_command = await glue_nodes_async([node_a, node_b], initial_state)
    # Merge the final command's update into the state.
    final_state = initial_state.copy()
    if final_command.update:
        final_state.update(final_command.update)
    # Get our state model and create an instance from the final state.
    StateModel = schema_manager.get_model()
    # Use model_validate and model_dump_json in Pydantic V2.
    state_instance = StateModel.model_validate(final_state)
    print("Final State Model:")
    state_instance.pretty_print()
    print(state_instance.model_dump_json(indent=4))

if __name__ == "__main__":
    import nest_asyncio
    nest_asyncio.apply()
    print("=== Testing Single Node Execution ===")
    asyncio.run(test_state_graph())
````

````python
import asyncio
import inspect
import textwrap
from typing import Any, Dict, List, Type, Union, Optional, Callable, get_type_hints
import libcst as cst
from pydantic import BaseModel, Field, create_model, ValidationError

# --- Import your actual models and AugLLM components ---
from src.haive.agents.plan_and_execute.models import Plan, Response, Act
from src.haive.core.aug_llm.base import AugLLMConfig, compose_runnable
from langchain.prompts import ChatPromptTemplate, MessagesPlaceholder

# Import Command and Send from langgraph.types
from langgraph.types import Command, Send

# Global flag for verbose model info
VERBOSE_MODEL_INFO = True

def verbose_model_info(model: Type[BaseModel]) -> str:
    """Return a verbose string with detailed info about the Pydantic model."""
    info_lines = []
    info_lines.append(f"Model Name: {model.__name__}")
    info_lines.append("Fields:")
    fields = model.model_fields if hasattr(model, "model_fields") else model.__fields__
    for field_name, field in fields.items():
        field_type = getattr(field, "annotation", "Unknown")
        required = getattr(field, "is_required", getattr(field, "required", "Unknown"))
        default = field.default if hasattr(field, "default") and field.default is not None else "None"
        try:
            description = field.field_info.description
        except AttributeError:
            description = getattr(field, "description", "")
        info_lines.append(
            f"  {field_name}: type={field_type}, required={required}, default={default}, description={description}"
        )
    properties = [name for name, _ in inspect.getmembers(model, predicate=lambda m: isinstance(m, property))]
    if properties:
        info_lines.append("Properties: " + ", ".join(properties))
    methods = [name for name, _ in inspect.getmembers(model, predicate=inspect.isfunction) if not name.startswith("__")]
    if methods:
        info_lines.append("Methods: " + ", ".join(methods))
    return "\n".join(info_lines)

# --- Create AugLLM configurations ---
EXECUTOR_PROMPT = "You are a helpful assistant"
EXECUTOR_PROMPT_TEMPLATE = ChatPromptTemplate.from_messages(
    [
        ("system", EXECUTOR_PROMPT),
        MessagesPlaceholder(variable_name="messages"),
    ]
)
PLANNER_PROMPT = (
    "For the given objective, come up with a simple step by step plan. "
    "This plan should involve individual tasks, that if executed correctly will yield the correct answer. "
    "Do not add any superfluous steps. "
    "The result of the final step should be the final answer. "
    "Make sure that each step has all the information needed - do not skip steps."
)
PLANNER_PROMPT_TEMPLATE = ChatPromptTemplate.from_messages(
    [
        ("system", PLANNER_PROMPT),
        ("placeholder", "{messages}"),
    ]
)
planner_aug_llm_config = AugLLMConfig(
    name="planner",
    prompt_template=PLANNER_PROMPT_TEMPLATE,
    structured_output_model=Plan,
)

# Define a second AugLLM config with a modified output model.
class PlanWithExtra(Plan):
    notes: Optional[str] = Field(default=None, description="Extra notes for the plan")

planner_aug_llm_config_2 = AugLLMConfig(
    name="planner_extra",
    prompt_template=PLANNER_PROMPT_TEMPLATE,
    structured_output_model=PlanWithExtra,
)

# --- Node Factory Functions (same as before) ---
def infer_input_field(config: AugLLMConfig) -> str:
    pt = config.prompt_template
    if hasattr(pt, "optional_variables") and pt.optional_variables:
        for var in pt.optional_variables:
            if var.lower() == "messages":
                return var
        return pt.optional_variables[0]
    elif hasattr(pt, "input_variables") and pt.input_variables:
        return pt.input_variables[0]
    return "input"

def infer_output_mapper(config: AugLLMConfig) -> Callable[[Any], Dict[str, Any]]:
    if config.structured_output_model:
        Model = config.structured_output_model
        if VERBOSE_MODEL_INFO:
            print("Using structured output model:")
            print(verbose_model_info(Model))
        def mapper(result: Any) -> Dict[str, Any]:
            if not isinstance(result, Model):
                try:
                    parsed = Model.model_validate(result)
                except Exception as e:
                    raise ValueError(f"Error parsing result with {Model}: {e}")
            else:
                parsed = result
            plan_val = getattr(parsed, "plan", parsed)
            steps_val = getattr(parsed, "steps", None)
            return {"plan": plan_val, "plan_steps": steps_val}
        return mapper
    elif config.output_parser:
        def mapper(result: Any) -> Dict[str, Any]:
            parsed = config.output_parser.parse(result)
            return {"plan": parsed, "plan_steps": parsed}
        return mapper
    else:
        def mapper(result: Any) -> Dict[str, Any]:
            return {"plan": result, "plan_steps": result}
        return mapper

def create_node_function(
    config: AugLLMConfig,
    command_goto: str = "execute_step",
    async_mode: bool = True,
) -> Callable[[Dict[str, Any]], Any]:
    output_model = config.structured_output_model or config.output_parser
    if output_model:
        print(f"Output model/parser detected: {output_model}")
    else:
        print("No structured output model or output parser provided.")
    
    node_runnable = compose_runnable(config)
    input_field = infer_input_field(config)
    if hasattr(config.prompt_template, "input_types") and input_field in config.prompt_template.input_types:
        expected_type = config.prompt_template.input_types[input_field]
        print(f"Inferred input field: '{input_field}' with expected type: {expected_type}")
    else:
        print(f"Inferred input field: '{input_field}' (no type info available)")
    
    def input_mapper(state: Dict[str, Any]) -> Dict[str, Any]:
        value = state.get(input_field)
        if value is None:
            raise ValueError(f"State does not contain required key '{input_field}'")
        if input_field.lower() == "messages":
            if isinstance(value, list):
                if not all(isinstance(x, tuple) and len(x) == 2 for x in value):
                    value = [("user", str(value))]
            else:
                value = [("user", str(value))]
            return {"messages": value}
        return {input_field: value}
    
    output_mapper = infer_output_mapper(config)
    
    if async_mode:
        async def node(state: Dict[str, Any]):
            payload = input_mapper(state)
            result = await node_runnable.ainvoke(payload)
            out = output_mapper(result)
            return Command(
                update={"plan": out["plan"]},
                goto=command_goto,
                resume={"plan": out["plan_steps"]}
            )
        return node
    else:
        def node(state: Dict[str, Any]):
            payload = input_mapper(state)
            result = node_runnable.invoke(payload)
            out = output_mapper(result)
            return Command(
                update={"plan": out["plan"]},
                goto=command_goto,
                resume={"plan": out["plan_steps"]}
            )
        return node

# --- Glue Functions (unchanged) ---
async def glue_nodes_async(
    nodes: List[Callable[[Dict[str, Any]], Command]],
    initial_state: Dict[str, Any]
) -> Command:
    state = initial_state.copy()
    last_command = None
    for i, node in enumerate(nodes):
        cmd = await node(state)
        if i < len(nodes) - 1:
            next_node_name = nodes[i+1].__name__
            cmd = Command(update=cmd.update, goto=next_node_name, resume=cmd.resume)
        last_command = cmd
        if last_command.update:
            state.update(last_command.update)
    return last_command

# ------------------------------
# Use the existing StateSchemaManager (assume it's already defined)
# ------------------------------
# Here we use the manager as defined earlier.
schema_data = {
    "messages": [("user", "default input")],
    "extra": "initial extra"
}
schema_manager = StateSchemaManager(schema_data, name="ExampleState")
print("State Schema:")
schema_manager.pretty_print()

# Get the state model
StateModel = schema_manager.get_model()

# Create two node functions from different AugLLM configs:
node1 = schema_manager.create_node_from_config(planner_aug_llm_config, async_mode=True)
node2 = schema_manager.create_node_from_config(planner_aug_llm_config_2, async_mode=True)

# Assume that in our state graph the nodes are chained:
# For this test, we'll glue the nodes together using our glue function.
async def test_state_graph():
    # Start with an initial state that has only the messages field.
    initial_state = {"messages": [("user", "Plan a dinner party for 10")]}
    # Chain node1 and node2 together.
    final_command = await glue_nodes_async([node1, node2], initial_state)
    # Merge the final command's update into the state.
    final_state = initial_state.copy()
    if final_command.update:
        final_state.update(final_command.update)
    # Validate final state with our state model and pretty-print.
    state_instance = StateModel.model_validate(final_state)
    print("Final State Model:")
    print(state_instance.model_dump_json(indent=4))

if __name__ == "__main__":
    import nest_asyncio
    nest_asyncio.apply()
    print("=== Testing State Graph Execution ===")
    asyncio.run(test_state_graph())
````
