```python

from typing import List, Dict, Any, Optional, Union, Tuple, Type, Callable
from pydantic import BaseModel, Field
from enum import Enum
import math
import copy
import json
from collections import deque

from langchain_core.prompts import ChatPromptTemplate
from langchain_core.messages import AIMessage, SystemMessage, HumanMessage, BaseMessage, ToolMessage,AnyMessage
from langchain_core.output_parsers import PydanticOutputParser
from langchain_core.tools import BaseTool, StructuredTool, Tool

from langgraph.graph import StateGraph, END
from langgraph.types import Command
from langgraph.prebuilt import ToolNode

from src.haive.core.aug_llm.base import AugLLMConfig
from src.haive.core.models.llm.base import AzureLLMConfig
from src.haive.core.graph import NodeFactory
from src.haive.agents.base import AgentConfig, AgentArchitecture, register_agent
from src.haive.core.tools.search_tools import tavily_search_tool

# Define models for LATS
class Reflection(BaseModel):
    """Reflection on a candidate solution or action."""
    reflections: str = Field(
        description="The critique and reflections on the sufficiency, superfluency,"
        " and general quality of the response"
    )
    score: int = Field(
        description="Score from 0-10 on the quality of the candidate response.",
        gte=0,
        lte=10,
    )
    found_solution: bool = Field(
        description="Whether the response has fully solved the question or task."
    )

    def as_message(self):
        return HumanMessage(
            content=f"Reasoning: {self.reflections}\nScore: {self.score}"
        )

    @property
    def normalized_score(self) -> float:
        return self.score / 10.0


class MCTSNode(BaseModel):
    """
    Node in the Monte Carlo Tree Search.
    Made serializable with Pydantic for LangGraph state.
    """
    id: str = Field(default_factory=lambda: f"node_{hash(str(id))}")
    messages: List[Dict[str, Any]] = Field(default_factory=list)
    parent_id: Optional[str] = None
    children_ids: List[str] = Field(default_factory=list)
    value: float = 0.0
    visits: int = 0
    reflection_data: Optional[Dict[str, Any]] = None
    depth: int = 1
    is_solved: bool = False

    def to_dict(self) -> Dict[str, Any]:
        """Serialize to dictionary."""
        return {
            "id": self.id,
            "messages": self.messages,
            "parent_id": self.parent_id,
            "children_ids": self.children_ids,
            "value": self.value,
            "visits": self.visits,
            "reflection_data": self.reflection_data,
            "depth": self.depth,
            "is_solved": self.is_solved
        }


class LATSState(BaseModel):
    """State for LATS agent."""
    # User input
    input: str = Field(default="")
    # All messages in the conversation
    messages: List[AnyMessage] = Field(default_factory=list)
    # MCTS nodes
    nodes: Dict[str, MCTSNode] = Field(default_factory=dict)
    # Root node ID
    root_id: Optional[str] = None
    # Current node ID
    current_id: Optional[str] = None
    # Iteration count
    iteration: int = 0
    # Max iterations
    max_iterations: int = 10
    # Max depth
    max_depth: int = 5
    # Exploration constant for UCB
    exploration_weight: float = 1.0
    # Number of expansions per node
    num_expansions: int = 5
    # Final result
    final_result: Optional[str] = None
    # Available tools
    #tools: Union[List[Union[BaseTool,StructuredTool,BaseModel,Tool]] = [tavily_search_tool]
    # Tool results (similar to React agent)
    tool_results: List[Dict[str, Any]] = Field(default_factory=list)
    # Current tool call
    current_tool_call: Optional[Dict[str, Any]] = None
    # Execution status
    status: str = Field(default="thinking")  # thinking, tool_calling, or done


class LATSAgentConfig(AgentConfig):
    """
    Configuration for LATS agent.
    """
    # Schema with LATS state
    schema: Type[LATSState] = LATSState
    
    # System prompts
    initial_system_prompt: str = Field(
        default="You are an AI assistant. Answer the following question or complete the requested task. Use tools when necessary.",
        description="System prompt for initial response."
    )
    
    reflection_system_prompt: str = Field(
        default="""
You are evaluating a candidate response to a user query.
Consider the following aspects:
1. Relevance: Does the response address the query?
2. Accuracy: Is the information correct?
3. Completeness: Does it fully answer the question?
4. Clarity: Is it well-structured and easy to understand?
5. Tool Usage: Were tools used appropriately when needed?

Score from 0-10 and indicate if the response fully solves the task.
""",
        description="System prompt for reflection."
    )
    
    expansion_system_prompt: str = Field(
        default="""
You are generating potential next steps for solving a problem.
Based on the current conversation and the original query, generate a new step 
that might advance the solution in a different way than previous steps.

Consider using tools when you need to get information or perform calculations.
""",
        description="System prompt for expansion."
    )
    
    # Exploration settings
    exploration_weight: float = Field(
        default=1.0,
        description="Exploration weight for UCB calculation."
    )
    
    num_expansions: int = Field(
        default=5,
        description="Number of expansions to generate per node."
    )
    
    max_iterations: int = Field(
        default=10,
        description="Maximum number of MCTS iterations."
    )
    
    max_depth: int = Field(
        default=5,
        description="Maximum depth of the MCTS tree."
    )
    
    # Tools configuration
    tools: Union[List[Union[BaseTool,Tool,StructuredTool]],Dict[str, Callable]] = Field(
        default_factory=dict,
        description="Dictionary of tools available to the agent."
    )
    
    @classmethod
    def from_scratch(cls,
                    system_prompt: Optional[str] = None,
                    reflection_prompt: Optional[str] = None,
                    expansion_prompt: Optional[str] = None,
                    model: str = "gpt-4o",
                    max_iterations: int = 10,
                    max_depth: int = 5,
                    exploration_weight: float = 1.0,
                    num_expansions: int = 5,
                    tools: Optional[Union[List[Union[BaseTool,Tool,StructuredTool]],Dict[str, Callable]]] = None,
                    name: Optional[str] = None,
                    **kwargs) -> 'LATSAgentConfig':
        """
        Create a LATSAgentConfig from scratch.
        
        Args:
            system_prompt: Optional system prompt override
            reflection_prompt: Optional reflection prompt override
            expansion_prompt: Optional expansion prompt override
            model: Model name
            max_iterations: Maximum number of iterations
            max_depth: Maximum depth of the tree
            exploration_weight: Exploration weight for UCB
            num_expansions: Number of expansions per node
            tools: Dictionary of tools to make available
            name: Optional agent name
            **kwargs: Additional configuration
            
        Returns:
            LATSAgentConfig instance
        """
        # Create the config with basic settings
        config = cls(
            name=name or "lats_agent",
            model=model,
            max_iterations=max_iterations,
            max_depth=max_depth,
            exploration_weight=exploration_weight,
            num_expansions=num_expansions,
            tools=tools or {},
            **kwargs
        )
        
        # Override prompts if provided
        if system_prompt:
            config.initial_system_prompt = system_prompt
        if reflection_prompt:
            config.reflection_system_prompt = reflection_prompt
        if expansion_prompt:
            config.expansion_system_prompt = expansion_prompt
        
        # Prepare tools description for system prompt
        tools_description = ""
        if tools:
            tools_description = "Available tools:\n"
            for tool in tools:
                doc = tool.__doc__ or f"Tool: {tool.name if isinstance(tool,StructuredTool) or isinstance(tool,BaseTool) else tool.__name__}"
                tools_description += f"- {tool.name if isinstance(tool,StructuredTool) or isinstance(tool,BaseTool) else tool.__name__}: {doc}\n"
            
            # Update system prompt with tools
            config.initial_system_prompt += f"\n\n{tool.description}"
        
        # Create the initial response LLM
        initial_prompt = ChatPromptTemplate.from_messages([
            ("system", config.initial_system_prompt),
            ("human", "{input}")
        ])
        
        llm_config = AzureLLMConfig(
            model=model,
            parameters={"temperature": 0.7}
        )
        
        # Convert function tools to LangChain tools
        langchain_tools = []
        for name, func in (tools or {}).items():
            # Create a tool from the function
            tool = Tool(
                name=name,
                func=func,
                description=func.__doc__ or f"Tool: {name}"
            )
            langchain_tools.append(tool)
        
        # Create AugLLM config with tools
        initial_aug_llm = AugLLMConfig(
            name="initial_llm",
            llm_config=llm_config,
            prompt_template=initial_prompt,
            tools=langchain_tools
        )
        
        # Set engine
        config.engine = initial_aug_llm
        
        return config


@register_agent(LATSAgentConfig)
class LATSAgent(AgentArchitecture[LATSAgentConfig]):
    """
    Language Agent Tree Search (LATS) implementation.
    
    This agent uses Monte Carlo Tree Search to explore multiple solution paths
    and chooses the best based on reflections and evaluations.
    """
    def setup_workflow(self) -> None:
        """Set up the LATS workflow with conditional edges."""
        # Create the initial node
        initial_node = self._create_initial_node()
        self.graph.add_node("initial", initial_node)
        
        # Create selection node
        select_node = self._create_select_node()
        self.graph.add_node("select", select_node)
        
        # Create expansion node
        expand_node = self._create_expand_node()
        self.graph.add_node("expand", expand_node)
        
        # Create backpropagation node
        backprop_node = self._create_backprop_node()
        self.graph.add_node("backpropagate", backprop_node)
        
        # Create termination check node
        check_termination_node = self._create_check_termination_node()
        self.graph.add_node("check_termination", check_termination_node)
        
        # Create final response node
        final_response_node = self._create_final_response_node()
        self.graph.add_node("final_response", final_response_node)
        
        # Create tool execution node
        tool_node = self._create_tool_node()
        self.graph.add_node("tool", tool_node)
        
        # Set entry point
        self.graph.set_entry_point("initial")
        
        # Add edges for the workflow
        self.graph.add_edge("initial", "check_termination")
        self.graph.add_edge("select", "expand")
        self.graph.add_edge("expand", "backpropagate")
        self.graph.add_edge("backpropagate", "check_termination")
        
        # Add conditional edge from check_termination
        def termination_router(state):
            """Route based on termination conditions."""
            # Check if we need to execute a tool
            if state.status == "tool_calling":
                return "tool"
            
            # Check if we have a final result
            if state.final_result is not None:
                return "final_response"
            else:
                return "select"
        
        self.graph.add_conditional_edges(
            "check_termination",
            termination_router,
            {
                "select": "select",
                "tool": "tool",
                "final_response": "final_response"
            }
        )
        
        # Add edge from tool back to check_termination
        self.graph.add_edge("tool", "check_termination")
        
        # Add edge from final_response to END
        self.graph.add_edge("final_response", END)
    
    def _create_initial_node(self) -> Callable:
        """Create the initial node that starts the MCTS process."""
        def initial_node(state):
            """Initialize the MCTS process with the first node."""
            # Convert state to dict if it's a model
            state_dict = state.model_dump() if hasattr(state, "model_dump") else state
            
            # Get the user input
            input_text = state_dict.get("input", "")
            if not input_text:
                messages = state_dict.get("messages", [])
                for msg in messages:
                    if getattr(msg, "type", None) == "human":
                        input_text = msg.content
                        break
            
            if not input_text:
                return Command(
                    update={"final_result": "No input provided."},
                    goto="final_response"
                )
            
            # Initialize state variables
            nodes = {}
            
            # Create initial AugLLM
            initial_prompt = ChatPromptTemplate.from_messages([
                ("system", self.config.initial_system_prompt),
                ("human", "{input}")
            ])
            
            llm_config = AzureLLMConfig(
                #model=self.config.model,
                parameters={"temperature": 0.7}
            )
            
            initial_aug_llm = AugLLMConfig(
                name="initial_llm",
                llm_config=llm_config,
                prompt_template=initial_prompt
            )
            
            # Generate initial response
            runnable = initial_aug_llm.create_runnable()
            response = runnable.invoke({"input": input_text})
            
            # Convert to serializable format
            if isinstance(response, BaseMessage):
                response_dict = {
                    "type": response.type,
                    "content": response.content
                }
            else:
                response_dict = {
                    "type": "ai",
                    "content": str(response)
                }
            
            # Create reflection AugLLM
            reflection_prompt = ChatPromptTemplate.from_messages([
                ("system", self.config.reflection_system_prompt),
                ("human", "{input}"),
                ("ai", "{response}")
            ])
            
            reflection_parser = PydanticOutputParser(pydantic_object=Reflection)
            
            reflection_llm_config = AzureLLMConfig(
                #model=self.config.model,
                parameters={"temperature": 0.3}
            )
            
            reflection_aug_llm = AugLLMConfig(
                name="reflection_llm",
                llm_config=reflection_llm_config,
                prompt_template=reflection_prompt,
                structured_output_model=Reflection
            )
            
            # Generate reflection
            reflection_runnable = reflection_aug_llm.create_runnable()
            reflection = reflection_runnable.invoke({
                "input": input_text,
                "response": response_dict["content"]
            })
            
            # Create root node
            root_node = MCTSNode(
                id="root",
                messages=[{"type": "human", "content": input_text}, response_dict],
                reflection_data=reflection.model_dump(),
                is_solved=reflection.found_solution
            )
            
            # Initialize with backpropagation
            root_node.visits = 1
            root_node.value = reflection.normalized_score
            
            # Add to nodes dictionary
            nodes[root_node.id] = root_node
            
            # Update state
            return Command(
                update={
                    "nodes": nodes,
                    "root_id": root_node.id,
                    "current_id": root_node.id,
                    "iteration": 1,
                    "input": input_text,
                    "messages": [
                        HumanMessage(content=input_text),
                        AIMessage(content=response_dict["content"])
                    ]
                },
                goto="check_termination"
            )
        
        return initial_node
    
    def _create_select_node(self) -> Callable:
        """Create the selection node for MCTS."""
        def upper_confidence_bound(node, parent, exploration_weight):
            """Calculate UCB score."""
            if node.visits == 0:
                return float('inf')
            exploitation = node.value / node.visits
            exploration = math.sqrt(math.log(parent.visits) / node.visits)
            return exploitation + exploration_weight * exploration
        
        def select_node(state):
            """Select the best node for expansion based on UCB."""
            # Convert state to dict if it's a model
            state_dict = state.model_dump() if hasattr(state, "model_dump") else state
            
            nodes = state_dict.get("nodes", {})
            root_id = state_dict.get("root_id")
            
            if not root_id or root_id not in nodes:
                return Command(
                    update={"final_result": "Invalid tree state."},
                    goto="final_response"
                )
            
            # Start from root
            current_node = nodes[root_id]
            
            # Traverse down the tree
            while current_node.children_ids:
                # Select best child by UCB
                best_child_id = max(
                    current_node.children_ids,
                    key=lambda cid: upper_confidence_bound(
                        nodes[cid], 
                        current_node, 
                        state_dict.get("exploration_weight", 1.0)
                    )
                )
                current_node = nodes[best_child_id]
                
                # Check if we've reached a solved node
                if current_node.is_solved:
                    return Command(
                        update={
                            "current_id": current_node.id,
                            "final_result": current_node.messages[-1]["content"]
                        },
                        goto="check_termination"
                    )
                
                # Check if we've reached the max depth
                if current_node.depth >= state_dict.get("max_depth", 5):
                    return Command(
                        update={
                            "current_id": current_node.id,
                            "final_result": current_node.messages[-1]["content"]
                        },
                        goto="check_termination"
                    )
            
            # Return the selected leaf node
            return Command(
                update={"current_id": current_node.id},
                goto="expand"
            )
        
        return select_node
    
    def _create_expand_node(self) -> Callable:
        """Create the expansion node for MCTS."""
        def expand_node(state):
            """Expand the current node with new candidates."""
            # Convert state to dict if it's a model
            state_dict = state.model_dump() if hasattr(state, "model_dump") else state
            
            nodes = state_dict.get("nodes", {})
            current_id = state_dict.get("current_id")
            
            if not current_id or current_id not in nodes:
                return Command(
                    update={"final_result": "Invalid node selection."},
                    goto="final_response"
                )
            
            current_node = nodes[current_id]
            input_text = state_dict.get("input", "")
            
            # Extract messages for prompt
            messages = []
            node = current_node
            while node:
                for msg in node.messages:
                    messages.insert(0, msg)
                if node.parent_id:
                    node = nodes.get(node.parent_id)
                else:
                    node = None
            
            # Create expansion AugLLM
            expansion_prompt = ChatPromptTemplate.from_messages([
                ("system", self.config.expansion_system_prompt),
                *[(msg["type"], msg["content"]) for msg in messages],
                ("human", "Generate a new response that advances the solution.")
            ])
            
            llm_config = AzureLLMConfig(
                model=self.config.model,
                parameters={"temperature": 0.8}  # Higher temperature for diversity
            )
            
            expansion_aug_llm = AugLLMConfig(
                name="expansion_llm",
                llm_config=llm_config,
                prompt_template=expansion_prompt
            )
            
            # Generate expansion candidates
            runnable = expansion_aug_llm.create_runnable()
            num_expansions = min(state_dict.get("num_expansions", 5), 5)  # Limit to 5
            
            # We'll generate one at a time for simplicity, but could be parallelized
            expansion_results = []
            for _ in range(num_expansions):
                response = runnable.invoke({})
                
                # Convert to serializable format
                if isinstance(response, BaseMessage):
                    response_dict = {
                        "type": response.type,
                        "content": response.content
                    }
                else:
                    response_dict = {
                        "type": "ai",
                        "content": str(response)
                    }
                    
                expansion_results.append(response_dict)
            
            # Create reflection AugLLM
            reflection_prompt = ChatPromptTemplate.from_messages([
                ("system", self.config.reflection_system_prompt),
                ("human", "{input}"),
                ("ai", "{response}")
            ])
            
            reflection_parser = PydanticOutputParser(pydantic_object=Reflection)
            
            reflection_llm_config = AzureLLMConfig(
                model=self.config.model,
                parameters={"temperature": 0.3}
            )
            
            reflection_aug_llm = AugLLMConfig(
                name="reflection_llm",
                llm_config=reflection_llm_config,
                prompt_template=reflection_prompt,
                structured_output_model=Reflection
            )
            
            # Generate reflections for each expansion
            reflection_runnable = reflection_aug_llm.create_runnable()
            
            child_nodes = []
            for i, response_dict in enumerate(expansion_results):
                # Generate reflection
                reflection = reflection_runnable.invoke({
                    "input": input_text,
                    "response": response_dict["content"]
                })
                
                # Create child node
                child_node = MCTSNode(
                    id=f"{current_id}_child_{i}",
                    messages=[response_dict],
                    parent_id=current_id,
                    reflection_data=reflection.model_dump(),
                    is_solved=reflection.found_solution,
                    depth=current_node.depth + 1
                )
                
                # Initialize with reflection score
                child_node.visits = 1
                child_node.value = reflection.normalized_score
                
                # Add to nodes
                nodes[child_node.id] = child_node
                child_nodes.append(child_node.id)
            
            # Update parent's children
            current_node.children_ids.extend(child_nodes)
            nodes[current_id] = current_node
            
            # Return all child IDs for backpropagation
            return Command(
                update={
                    "nodes": nodes,
                    "child_ids": child_nodes
                },
                goto="backpropagate"
            )
        
        return expand_node
   
    def _create_tool_node(self) -> Callable:
        """Create the tool execution node."""
        def tool_node(state):
            """Execute a tool and update the state."""
            # Convert state to dict if it's a model
            state_dict = state.model_dump() if hasattr(state, "model_dump") else state
            
            # Get the tool call
            tool_call = state_dict.get("current_tool_call")
            if not tool_call:
                return Command(
                    update={"status": "thinking"},
                    goto="check_termination"
                )
            
            # Get the tool function
            tools = state_dict.get("tools", {})
            tool_name = tool_call.get("name")
            tool_input = tool_call.get("input")
            
            if tool_name not in tools:
                tool_result = f"Tool '{tool_name}' not found."
            else:
                # Execute the tool
                try:
                    tool_fn = tools[tool_name]
                    tool_result = tool_fn(tool_input)
                except Exception as e:
                    tool_result = f"Error executing tool '{tool_name}': {str(e)}"
            
            # Create tool message
            tool_message = {
                "type": "tool",
                "name": tool_name,
                "input": tool_input,
                "content": tool_result
            }
            
            # Update tool results
            tool_results = state_dict.get("tool_results", [])
            tool_results.append({
                "name": tool_name,
                "input": tool_input,
                "result": tool_result
            })
            
            # Get current node and update its messages
            nodes = state_dict.get("nodes", {})
            current_id = state_dict.get("current_id")
            
            if current_id and current_id in nodes:
                current_node = nodes[current_id]
                current_node.messages.append(tool_message)
                nodes[current_id] = current_node
            
            # Update state
            return Command(
                update={
                    "tool_results": tool_results,
                    "nodes": nodes,
                    "status": "thinking",
                    "current_tool_call": None  # Clear the current tool call
                },
                goto="check_termination"
            )
        
        return tool_node
    def _create_backprop_node(self) -> Callable:
        """Create the backpropagation node for MCTS."""
        def backprop_node(state):
            """Backpropagate values through the tree."""
            # Convert state to dict if it's a model
            state_dict = state.model_dump() if hasattr(state, "model_dump") else state
            
            nodes = state_dict.get("nodes", {})
            child_ids = state_dict.get("child_ids", [])
            
            if not child_ids:
                return Command(
                    update={"iteration": state_dict.get("iteration", 0) + 1},
                    goto="check_termination"
                )
            
            # Backpropagate from each child
            for child_id in child_ids:
                if child_id not in nodes:
                    continue
                
                child = nodes[child_id]
                
                # Check if solution found
                if child.is_solved:
                    # Mark parent chain as solved
                    current_id = child.parent_id
                    while current_id:
                        parent = nodes[current_id]
                        parent.is_solved = True
                        nodes[current_id] = parent
                        current_id = parent.parent_id
                
                # Backpropagate values
                current_id = child.parent_id
                while current_id:
                    parent = nodes[current_id]
                    parent.visits += 1
                    # Update value (average of children)
                    child_values = [nodes[cid].value for cid in parent.children_ids if cid in nodes]
                    if child_values:
                        parent.value = sum(child_values) / len(child_values)
                    nodes[current_id] = parent
                    current_id = parent.parent_id
            
            # Increment iteration counter
            iteration = state_dict.get("iteration", 0) + 1
            
            return Command(
                update={
                    "nodes": nodes,
                    "iteration": iteration
                },
                goto="check_termination"
            )
        
        return backprop_node
    
    def _create_check_termination_node(self) -> Callable:
        """Create the termination check node."""
        def check_termination_node(state):
            """Check if we should terminate the search."""
            # Convert state to dict if it's a model
            state_dict = state.model_dump() if hasattr(state, "model_dump") else state
            
            nodes = state_dict.get("nodes", {})
            root_id = state_dict.get("root_id")
            
            if not root_id or root_id not in nodes:
                return Command(
                    update={"final_result": "Invalid tree state."},
                    goto="final_response"
                )
            
            root_node = nodes[root_id]
            
            # Check if solution found
            if root_node.is_solved:
                # Find the best solution
                best_solution = self._find_best_solution(nodes, root_id)
                return Command(
                    update={"final_result": best_solution},
                    goto="final_response"
                )
            
            # Check max iterations
            iteration = state_dict.get("iteration", 0)
            max_iterations = state_dict.get("max_iterations", self.config.max_iterations)
            
            if iteration >= max_iterations:
                # Find the best solution so far
                best_solution = self._find_best_solution(nodes, root_id)
                return Command(
                    update={"final_result": best_solution},
                    goto="final_response"
                )
            
            # Continue search
            return Command(goto="select")
        
        return check_termination_node
    
    def _create_final_response_node(self) -> Callable:
        """Create the final response node."""
        def final_response_node(state):
            """Format the final response."""
            # Convert state to dict if it's a model
            state_dict = state.model_dump() if hasattr(state, "model_dump") else state
            
            final_result = state_dict.get("final_result", "No solution found.")
            messages = state_dict.get("messages", [])
            
            # Add final result to messages
            updated_messages = messages + [AIMessage(content=final_result)]
            
            return Command(
                update={"messages": updated_messages},
                goto=END
            )
        
        return final_response_node
    
    def _find_best_solution(self, nodes, root_id):
        """Find the best solution in the tree."""
        # Calculate scores for all nodes
        node_scores = {}
        for node_id, node in nodes.items():
            # Score is a combination of value, visits, and solved status
            score = node.value * (1 + 0.5 * node.visits) * (2 if node.is_solved else 1)
            node_scores[node_id] = score
        
        # Find node with highest score
        best_node_id = max(node_scores.items(), key=lambda x: x[1])[0]
        best_node = nodes[best_node_id]
        
        # Extract the content from the best node
        if best_node.messages:
            return best_node.messages[-1]["content"]
        
        # Fallback to the original answer
        return nodes[root_id].messages[-1]["content"]


# Helper function to create a LATS agent
def create_lats_agent(
    system_prompt: Optional[str] = None,
    reflection_prompt: Optional[str] = None,
    expansion_prompt: Optional[str] = None,
    model: str = "gpt-4o",
    max_iterations: int = 10,
    max_depth: int = 5,
    exploration_weight: float = 1.0,
    num_expansions: int = 5,
    name: Optional[str] = None,
    visualize: bool = False,
    tools:List[Union[BaseTool,StructuredTool,Tool,BaseModel]]=[tavily_search_tool]
) -> LATSAgent:
    """
    Create a LATS agent with the specified configuration.
    
    Args:
        system_prompt: Optional system prompt override
        reflection_prompt: Optional reflection prompt override
        expansion_prompt: Optional expansion prompt override
        model: LLM model to use
        max_iterations: Maximum number of MCTS iterations
        max_depth: Maximum depth of the MCTS tree
        exploration_weight: Exploration weight for UCB
        num_expansions: Number of expansions per node
        name: Optional name for the agent
        visualize: Whether to generate a graph visualization
        
    Returns:
        A LATSAgent instance
    """
    # Create config
    config = LATSAgentConfig.from_scratch(
        system_prompt=system_prompt,
        reflection_prompt=reflection_prompt,
        expansion_prompt=expansion_prompt,
        model=model,
        max_iterations=max_iterations,
        max_depth=max_depth,
        exploration_weight=exploration_weight,
        num_expansions=num_expansions,
        name=name or "lats_agent",
        visualize=visualize,
        tools=[]
    )
    
    # Build and return agent
    return config.build_agent()
```

```python
from typing import Dict, Any, List, Optional
import json
import math

#from src.haive.agents.lats_agent import create_lats_agent
from langchain_core.messages import HumanMessage, AIMessage


# Define some tools that can be used by the LATS agent
def search_tool(query: str) -> str:
    """
    Search for information on a topic.
    
    Args:
        query: The search query
        
    Returns:
        Search results as text
    """
    # Mock implementation
    mock_results = {
        "ai": "Artificial intelligence (AI) is intelligence demonstrated by machines, as opposed to intelligence of humans and other animals.",
        "weather": "Current weather information varies by location. Generally, Earth's climate is experiencing warming trends.",
        "python": "Python is a high-level, general-purpose programming language. Its design philosophy emphasizes code readability.",
        "chess": "Chess is a board game played between two players with 16 pieces each on a checkered board. Magnus Carlsen is the current world champion.",
        "mathematics": "Mathematics includes the study of quantity, structure, space, and change. It's essential for many fields including science and engineering."
    }
    
    # Find the most relevant mock result
    for key, value in mock_results.items():
        if key.lower() in query.lower():
            return f"Search results for '{query}':\n{value}"
    
    return f"No specific information found for: {query}"


def calculator_tool(expression: str) -> str:
    """
    Calculate the result of a mathematical expression.
    
    Args:
        expression: Mathematical expression as string
        
    Returns:
        Calculation result
    """
    try:
        # Simple safe evaluation - in a real tool, use a proper parser
        allowed_names = {
            "sqrt": math.sqrt, 
            "pow": pow, 
            "pi": math.pi,
            "sin": math.sin,
            "cos": math.cos,
            "tan": math.tan,
            "abs": abs,
            "round": round
        }
        # Remove any dangerous functions/modules
        expression = expression.replace("__", "").replace("import", "")
        result = eval(expression, {"__builtins__": {}}, allowed_names)
        return f"Result of '{expression}': {result}"
    except Exception as e:
        return f"Error calculating '{expression}': {str(e)}"


def main():
    """Run the LATS agent example."""
    # Create tool dictionary
    tools = {
        "search": search_tool,
        "calculator": calculator_tool
    }
    
    # Create a LATS agent with default settings
    agent = create_lats_agent(
        system_prompt="You are a helpful AI assistant that can solve complex reasoning problems step by step. Use tools when needed.",
        reflection_prompt="""
You are evaluating a candidate solution to a user's query.
Consider:
1. Accuracy: Is the information correct?
2. Relevance: Does it address the query?
3. Coherence: Is it logically sound?
4. Completeness: Does it fully solve the problem?
5. Tool Usage: Were tools used appropriately?

Score from 0-10 where:
0-3: Poor, major issues or mostly irrelevant
4-6: Mediocre, partially addresses the problem with some issues
7-9: Good, addresses most aspects with minor issues
10: Excellent, completely solves the problem correctly

Also indicate if this completely solves the problem.
""",
        expansion_prompt="""
You are exploring alternative solutions to a problem.
Based on the conversation history and the original query, generate a different approach 
or refine the existing one in a way that might lead to a better solution.

Consider using tools like search or calculator if appropriate.
Be creative but precise, and focus on advancing towards the correct solution.
""",
        model="gpt-4o",
        max_iterations=5,  # Limited for the example
        num_expansions=3,  # Generate 3 alternatives per expansion
        tools=tools,  # Add tools
        name="example_lats_agent",
        visualize=True  # Generate graph visualization
    )
    
    # Example queries that benefit from exploration
    queries = [
        "What is the square root of 144 and how does it relate to the number of chess pieces on a board?",
        "Explain the concept of neural networks in AI and how they calculate outputs from inputs",
        "If I have a 3x3x3 cube, what is its volume and surface area?"
    ]
    
    # Run each query and display the results
    for i, query in enumerate(queries):
        print(f"\n\n===== QUERY {i+1}: {query} =====\n")
        
        # Run the agent
        print(agent.app)
        print(agent.app.input_schema)
        #result = agent.run({'messages':[query]})
        result = agent.app.invoke({'input':query},config={'configurable':{'thread_id':'s'}},debug=True)
        # Print the conversation
        print("--- CONVERSATION ---")
        for msg in result.get("messages", []):
            if hasattr(msg, "type") and msg.type == "human":
                print(f"User: {msg.content}")
            elif hasattr(msg, "type") and msg.type == "ai":
                print(f"AI: {msg.content}")
            elif isinstance(msg, tuple):
                print(f"{msg[0].capitalize()}: {msg[1]}")
        
        # Print exploration statistics
        print("\n--- EXPLORATION STATISTICS ---")
        nodes = result.get("nodes", {})
        if nodes:
            print(f"Total nodes explored: {len(nodes)}")
            
            solved_nodes = sum(1 for node in nodes.values() if node.get("is_solved", False))
            print(f"Solution nodes found: {solved_nodes}")
            
            max_depth = max(node.get("depth", 0) for node in nodes.values())
            print(f"Maximum tree depth: {max_depth}")
            
            avg_score = sum(node.get("value", 0) for node in nodes.values()) / len(nodes)
            print(f"Average node score: {avg_score:.2f}")


if __name__ == "__main__":
    main()
```
