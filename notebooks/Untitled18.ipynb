```python
from pydantic import BaseModel,Field,HttpUrl
from typing_extensions import Union,Optional
```

```python
class KYCInputState(BaseModel):
    company_name: str = Field(...,description='name of the company')
    company_url: Optional[Union[HttpUrl,str]] = Field(default='',description='base url of the company')
    country_of_company:Optional
    description: Optional[str] = Field(default='',description='Description of the company')
```

```python
import pycountry
from pydantic import BaseModel, field_validator

class CountryModel(BaseModel):
    country: str

    @field_validator("country")
    def check_valid_country(cls, v):
        if not pycountry.countries.get(name=v):
            raise ValueError("Invalid country name")
        return v
```
