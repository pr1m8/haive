```python
from langchain_community.document_loaders import (
    PDFMinerLoader,
    PDFMinerPDFasHTMLLoader,
    PDFPlumberLoader,
    UnstructuredPDFLoader,
    PagedPDFSplitter,
)
from pprint import pprint
import os

# Setup
dl_dir = '/mnt/n/downloads'
pdf_test_file = 'sat-practice-test-4-digital.pdf'
pdf_path = os.path.join(dl_dir, pdf_test_file)
print(f"Loading PDF from: {pdf_path}")

# Initialize loader classes
loaders = [
    PDFMinerLoader,
    UnstructuredPDFLoader,
    PagedPDFSplitter,
    PDFPlumberLoader,
    PDFMinerPDFasHTMLLoader
]

# Load and store results
load_dict = {}
for loader_cls in loaders:
    try:
        loader = loader_cls(pdf_path)
        docs = loader.load()
        load_dict[loader_cls.__name__] = docs
    except Exception as e:
        print(f"❌ Error with {loader_cls.__name__}: {e}")
        load_dict[loader_cls.__name__] = []

# Pretty print first 10 pages per loader (truncated)
for loader_name, docs in load_dict.items():
    print(f"\n=== {loader_name} ===")
    for i, doc in enumerate(docs[:10]):
        content = doc.page_content.strip().replace("\n", " ")
        print(f"Page {i+1}: {content[:30000]}{'...' if len(content) > 30000 else ''}")
```
