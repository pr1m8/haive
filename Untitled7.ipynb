````python
# src/haive/agents/react_agent/agent.py

from typing import Any, Dict, List, Optional, Union, Callable, Type, Tuple
from pydantic import BaseModel, Field, field_validator
import json
import re

from langchain_core.messages import AIMessage, HumanMessage, SystemMessage
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.output_parsers import BaseOutputParser
from langchain_core.output_parsers.openai_tools import PydanticToolsParser

from langgraph.graph import StateGraph, END
from langgraph.prebuilt import ToolNode

from src.haive.agents.base_agent import BaseAgentConfig, BaseAgent, register_agent
from src.haive.core.aug_llm.base import AugLLMConfig
from src.haive.core.models.llm.base import AzureLLMConfig
from src.haive.core.node.node_factory import NodeFactory


class ReactAgentState(BaseModel):
    """State schema for React Agent."""
    messages: List[Any] = Field(default_factory=list)
    tool_results: List[Dict[str, Any]] = Field(default_factory=list)
    current_tool_calls: List[Dict[str, Any]] = Field(default_factory=list)
    structured_output: Optional[Dict[str, Any]] = None
    error: Optional[str] = None
    done: bool = False


class ReactAgentConfig(BaseAgentConfig):
    """
    Configuration for a React Agent that implements the ReAct pattern.
    
    The agent can:
    1. Reason about the input and decide whether to use tools
    2. Conditionally execute tools when needed
    3. Process tool results and continue reasoning
    4. Optionally provide structured output
    """
    # Override state schema
    state_schema: Type[BaseModel] = ReactAgentState
    
    # Node names
    reasoning_node_name: str = Field(
        default="reasoning",
        description="Name for the reasoning node in the graph."
    )
    
    tool_node_name: str = Field(
        default="tool_execution",
        description="Name for the tool execution node in the graph."
    )
    
    output_node_name: str = Field(
        default="structured_output",
        description="Name for the structured output node in the graph (if used)."
    )
    
    # Tools configuration
    tools: Optional[List[Any]] = Field(
        default=None,
        description="List of tools available to the agent."
    )
    
    tool_choice: str = Field(
        default="auto",
        description="Tool choice strategy: 'auto', 'any', or 'required'."
    )
    
    # Structured output configuration
    structured_output_model: Optional[Type[BaseModel]] = Field(
        default=None,
        description="Optional model for structured output."
    )
    
    max_iterations: int = Field(
        default=10,
        description="Maximum number of iterations before forcing completion."
    )
    
    # System prompts
    reasoning_system_prompt: str = Field(
        default="""You are an intelligent agent that can use tools to help solve problems.
When you need information or want to perform an action, use the appropriate tool.
Think step-by-step and decide if you need to use any tools to fulfill the user's request.
If you need to use a tool, specify the tool name and parameters.
If you don't need a tool, simply respond directly to the user.""",
        description="System prompt for the reasoning node."
    )
    
    # Helper class method for creating from scratch
    @classmethod
    def from_scratch(cls, 
                   system_prompt: Optional[str] = None,
                   model: str = "gpt-4o", 
                   temperature: float = 0.7,
                   tools: Optional[List[Any]] = None,
                   structured_output_model: Optional[Type[BaseModel]] = None,
                   name: Optional[str] = None,
                   **kwargs) -> 'ReactAgentConfig':
        """
        Create a ReactAgentConfig from scratch with basic parameters.
        
        Args:
            system_prompt: Optional system prompt (overrides default reasoning prompt)
            model: Model name
            temperature: Temperature
            tools: Optional list of tools
            structured_output_model: Optional structured output model
            name: Optional agent name
            **kwargs: Additional configuration options
            
        Returns:
            ReactAgentConfig instance
        """
        # Use provided system prompt or default reasoning prompt
        reasoning_system_prompt = system_prompt or cls.__fields__["reasoning_system_prompt"].default
        
        # Create prompt template
        messages = [
            SystemMessage(content=reasoning_system_prompt),
            MessagesPlaceholder(variable_name="messages")
        ]
        prompt = ChatPromptTemplate.from_messages(messages)
        
        # Create LLM config
        llm_config = AzureLLMConfig(
            model=model,
            parameters={"temperature": temperature}
        )
        
        # Create AugLLM config for reasoning
        aug_llm = AugLLMConfig(
            name=name or "react_llm",
            llm_config=llm_config,
            prompt_template=prompt,
            tools=tools
        )
        
        # Create and return config
        return cls(
            name=name or "react_agent",
            engine=aug_llm,
            tools=tools,
            structured_output_model=structured_output_model,
            reasoning_system_prompt=reasoning_system_prompt,
            model=model,
            temperature=temperature,
            **kwargs
        )


@register_agent(ReactAgentConfig)
class ReactAgent(BaseAgent[ReactAgentConfig]):
    """
    React agent that implements the ReAct pattern for reasoning and tool use.
    
    This agent has a graph structure that:
    1. Uses a reasoning node to analyze input and decide on next steps
    2. Conditionally routes to tool execution when needed
    3. Optionally routes to structured output processing
    4. Terminates when the reasoning is complete
    """
    
    def setup_workflow(self) -> None:
        """
        Set up the ReAct workflow with conditional routing based on tool use.
        """
        # Extract configuration
        engine = self.config.engine
        tools = self.config.tools
        structured_output_model = self.config.structured_output_model
        max_iterations = self.config.max_iterations
        
        # Set up tracking for iterations
        self.graph.add_node("track_iterations", self._track_iterations)
        
        # Set up reasoning node
        self.graph.add_node(
            self.config.reasoning_node_name, 
            self._create_reasoning_node()
        )
        
        # Set up tool execution node if tools are provided
        if tools:
            self.graph.add_node(
                self.config.tool_node_name,
                ToolNode(tools)
            )
        
        # Set up structured output node if a model is provided
        if structured_output_model:
            self.graph.add_node(
                self.config.output_node_name,
                self._create_structured_output_node()
            )
        
        # Set the entry point to iteration tracking
        self.graph.set_entry_point("track_iterations")
        
        # Add edge from iteration tracking to reasoning
        self.graph.add_edge("track_iterations", self.config.reasoning_node_name)
        
        # Add conditional edges from reasoning node
        self.graph.add_conditional_edges(
            self.config.reasoning_node_name,
            self._route_from_reasoning,
            {
                "tool_execution": self.config.tool_node_name if tools else END,
                "structured_output": self.config.output_node_name if structured_output_model else END,
                "complete": END
            }
        )
        
        # Add edges from tool execution back to reasoning
        if tools:
            self.graph.add_edge(self.config.tool_node_name, self.config.reasoning_node_name)
        
        # Add edge from structured output to END
        if structured_output_model:
            self.graph.add_edge(self.config.output_node_name, END)
    
    def _track_iterations(self, state: ReactAgentState) -> ReactAgentState:
        """Track iterations to prevent infinite loops."""
        # Initialize iteration counter if not present
        if "iteration_count" not in state.__dict__:
            state.__dict__["iteration_count"] = 0
        else:
            state.__dict__["iteration_count"] += 1
            
        # Check if we've exceeded max iterations
        if state.__dict__["iteration_count"] >= self.config.max_iterations:
            # Add a message indicating we're forcing completion
            state.messages.append(
                AIMessage(content=f"I've reached the maximum number of iterations ({self.config.max_iterations}). "
                          "I'll provide my best response based on what I've learned so far.")
            )
            state.done = True
        
        return state
    
    def _create_reasoning_node(self) -> Callable:
        """Create the reasoning node function."""
        # Extract configuration
        engine = self.config.engine
        
        def reasoning_node(state: ReactAgentState) -> ReactAgentState:
            """Reason about the current state and decide next steps."""
            try:
                # Check if there are new tool results to incorporate
                if state.tool_results:
                    # Get the latest tool results
                    latest_results = state.tool_results[-1]
                    
                    # Add tool result as a message
                    tool_message = f"Tool result: {json.dumps(latest_results, indent=2)}"
                    state.messages.append(AIMessage(content=tool_message))
                
                # Invoke the reasoning engine
                response = engine.invoke({"messages": state.messages})
                
                # Add AI response to messages
                if isinstance(response, AIMessage):
                    state.messages.append(response)
                else:
                    state.messages.append(AIMessage(content=str(response)))
                
                # Extract tool calls from response if present
                latest_message = state.messages[-1]
                if hasattr(latest_message, "tool_calls") and latest_message.tool_calls:
                    state.current_tool_calls = latest_message.tool_calls
                elif hasattr(latest_message, "additional_kwargs") and "tool_calls" in latest_message.additional_kwargs:
                    state.current_tool_calls = latest_message.additional_kwargs["tool_calls"]
                else:
                    # Try to parse tool calls from text content
                    state.current_tool_calls = self._extract_tool_calls_from_text(latest_message.content)
                
                return state
                
            except Exception as e:
                state.error = f"Error in reasoning: {str(e)}"
                state.messages.append(AIMessage(content=f"I encountered an error: {str(e)}"))
                state.done = True
                return state
        
        return reasoning_node
    
    def _extract_tool_calls_from_text(self, text: str) -> List[Dict[str, Any]]:
        """Extract tool calls from text using regex patterns."""
        tool_calls = []
        
        # Look for patterns like: Tool: tool_name({"param": "value"})
        pattern = r'Tool:\s+(\w+)\(({.*?})\)'
        matches = re.findall(pattern, text, re.DOTALL)
        
        for tool_name, args_str in matches:
            try:
                args = json.loads(args_str)
                tool_calls.append({
                    "name": tool_name,
                    "args": args
                })
            except json.JSONDecodeError:
                # If JSON parsing fails, try to extract key-value pairs
                args = {}
                kv_pattern = r'(\w+):\s*"([^"]*)"'
                kv_matches = re.findall(kv_pattern, args_str)
                for key, value in kv_matches:
                    args[key] = value
                
                tool_calls.append({
                    "name": tool_name,
                    "args": args
                })
        
        return tool_calls
    
    def _create_structured_output_node(self) -> Callable:
        """Create the structured output node function."""
        # Extract configuration
        structured_output_model = self.config.structured_output_model
        
        def structured_output_node(state: ReactAgentState) -> ReactAgentState:
            """Process the final AI message into structured output."""
            try:
                # Get the final AI message
                final_messages = [msg for msg in state.messages if isinstance(msg, AIMessage)]
                if not final_messages:
                    state.error = "No AI messages found for structured output."
                    return state
                
                final_message = final_messages[-1]
                
                # Create a parser for the structured output model
                parser = PydanticToolsParser(tools=[structured_output_model])
                
                # Parse the message content
                try:
                    parsed_output = parser.parse(final_message.content)
                    # Take the first result if it's a list
                    if isinstance(parsed_output, list) and parsed_output:
                        state.structured_output = parsed_output[0].dict()
                    else:
                        state.structured_output = parsed_output.dict()
                except Exception as parse_error:
                    # If parsing fails, try to use the model directly
                    try:
                        # Try to extract JSON content
                        json_pattern = r'```json\n(.*?)\n```'
                        json_matches = re.findall(json_pattern, final_message.content, re.DOTALL)
                        
                        if json_matches:
                            json_content = json_matches[0]
                            parsed_data = json.loads(json_content)
                            state.structured_output = structured_output_model(**parsed_data).dict()
                        else:
                            # Try direct instantiation of the model
                            state.structured_output = structured_output_model(
                                **{"content": final_message.content}
                            ).dict()
                    except Exception as e:
                        state.error = f"Failed to create structured output: {str(e)}"
                
                return state
                
            except Exception as e:
                state.error = f"Error in structured output: {str(e)}"
                return state
        
        return structured_output_node
    
    def _route_from_reasoning(self, state: ReactAgentState) -> str:
        """
        Determine where to route next based on reasoning output.
        
        Returns:
            "tool_execution": If there are tool calls to execute
            "structured_output": If we're done and have a structured output model
            "complete": If we're done with no need for structured output
        """
        # Check if we're done (forced completion or no tool calls)
        if state.done:
            if self.config.structured_output_model:
                return "structured_output"
            return "complete"
        
        # Check if there are tool calls to execute
        if state.current_tool_calls:
            return "tool_execution"
        
        # No tool calls, we're done with this request
        if self.config.structured_output_model:
            return "structured_output"
        return "complete"


# Helper functions for easy React agent creation
def create_react_agent(
    system_prompt: Optional[str] = None,
    model: str = "gpt-4o",
    temperature: float = 0.7,
    tools: Optional[List[Any]] = None,
    structured_output_model: Optional[Type[BaseModel]] = None,
    name: Optional[str] = None,
    max_iterations: int = 10,
    visualize: bool = False
) -> ReactAgent:
    """
    Create a React agent from scratch.
    
    Args:
        system_prompt: Optional system prompt
        model: Model name
        temperature: Temperature
        tools: Optional list of tools
        structured_output_model: Optional structured output model
        name: Optional agent name
        max_iterations: Maximum iterations
        visualize: Whether to generate visualization
        
    Returns:
        ReactAgent instance
    """
    config = ReactAgentConfig.from_scratch(
        system_prompt=system_prompt,
        model=model,
        temperature=temperature,
        tools=tools,
        structured_output_model=structured_output_model,
        name=name or "react_agent",
        max_iterations=max_iterations,
        visualize=visualize
    )
    
    return config.build_agent()
````

````python
# src/haive/agents/react_agent/agent.py

from typing import Any, Dict, List, Optional, Union, Callable, Type, Tuple
from pydantic import BaseModel, Field, field_validator
import json
import re

from langchain_core.messages import AIMessage, HumanMessage, SystemMessage
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.output_parsers import BaseOutputParser
from langchain_core.output_parsers.openai_tools import PydanticToolsParser

from langgraph.graph import StateGraph, END
#from langgraph.prebuilt import ToolNode
from langgraph.prebuilt import ToolNode,tools_condition
from src.haive.agents.base_agent import BaseAgentConfig, BaseAgent, register_agent
from src.haive.core.aug_llm.base import AugLLMConfig
from src.haive.core.models.llm.base import AzureLLMConfig
from src.haive.core.node.node_factory import NodeFactory


class ReactAgentState(BaseModel):
    """State schema for React Agent."""
    messages: List[Any] = Field(default_factory=list)
    tool_results: List[Dict[str, Any]] = Field(default_factory=list)
    current_tool_calls: List[Dict[str, Any]] = Field(default_factory=list)
    structured_output: Optional[Dict[str, Any]] = None
    error: Optional[str] = None
    done: bool = False


class ReactAgentConfig(BaseAgentConfig):
    """
    Configuration for a React Agent that implements the ReAct pattern.
    
    The agent can:
    1. Reason about the input and decide whether to use tools
    2. Conditionally execute tools when needed
    3. Process tool results and continue reasoning
    4. Optionally provide structured output
    """
    # Override state schema
    state_schema: Type[BaseModel] = ReactAgentState
    
    # Node names
    reasoning_node_name: str = Field(
        default="reasoning",
        description="Name for the reasoning node in the graph."
    )
    
    tool_node_name: str = Field(
        default="tool_execution",
        description="Name for the tool execution node in the graph."
    )
    
    output_node_name: str = Field(
        default="structured_output",
        description="Name for the structured output node in the graph (if used)."
    )
    
    # Tools configuration
    tools: Optional[List[Any]] = Field(
        default=None,
        description="List of tools available to the agent."
    )
    
    tool_choice: str = Field(
        default="auto",
        description="Tool choice strategy: 'auto', 'any', or 'required'."
    )
    
    # Structured output configuration
    structured_output_model: Optional[Type[BaseModel]] = Field(
        default=None,
        description="Optional model for structured output."
    )
    
    max_iterations: int = Field(
        default=10,
        description="Maximum number of iterations before forcing completion."
    )
    
    # System prompts
    reasoning_system_prompt: str = Field(
        default="""You are an intelligent agent that can use tools to help solve problems.
When you need information or want to perform an action, use the appropriate tool.
Think step-by-step and decide if you need to use any tools to fulfill the user's request.
If you need to use a tool, specify the tool name and parameters.
If you don't need a tool, simply respond directly to the user.""",
        description="System prompt for the reasoning node."
    )
    
    # Helper class method for creating from scratch
    @classmethod
    def from_scratch(cls, 
                   system_prompt: Optional[str] = None,
                   model: str = "gpt-4o", 
                   temperature: float = 0.7,
                   tools: Optional[List[Any]] = None,
                   structured_output_model: Optional[Type[BaseModel]] = None,
                   name: Optional[str] = None,
                   **kwargs) -> 'ReactAgentConfig':
        """
        Create a ReactAgentConfig from scratch with basic parameters.
        
        Args:
            system_prompt: Optional system prompt (overrides default reasoning prompt)
            model: Model name
            temperature: Temperature
            tools: Optional list of tools
            structured_output_model: Optional structured output model
            name: Optional agent name
            **kwargs: Additional configuration options
            
        Returns:
            ReactAgentConfig instance
        """
        # Use provided system prompt or default reasoning prompt
        reasoning_system_prompt = system_prompt or cls.__fields__["reasoning_system_prompt"].default
        
        # Create prompt template
        messages = [
            SystemMessage(content=reasoning_system_prompt),
            MessagesPlaceholder(variable_name="messages")
        ]
        prompt = ChatPromptTemplate.from_messages(messages)
        
        # Create LLM config
        llm_config = AzureLLMConfig(
            model=model,
            parameters={"temperature": temperature}
        )
        
        # Create AugLLM config for reasoning
        aug_llm = AugLLMConfig(
            name=name or "react_llm",
            llm_config=llm_config,
            prompt_template=prompt,
            tools=tools
        )
        
        # Create and return config
        return cls(
            name=name or "react_agent",
            engine=aug_llm,
            tools=tools,
            structured_output_model=structured_output_model,
            reasoning_system_prompt=reasoning_system_prompt,
            model=model,
            temperature=temperature,
            **kwargs
        )


@register_agent(ReactAgentConfig)
class ReactAgent(BaseAgent[ReactAgentConfig]):
    """
    React agent that implements the ReAct pattern for reasoning and tool use.
    
    This agent has a graph structure that:
    1. Uses a reasoning node to analyze input and decide on next steps
    2. Conditionally routes to tool execution when needed
    3. Optionally routes to structured output processing
    4. Terminates when the reasoning is complete
    """
    
    def setup_workflow(self) -> None:
        """
        Set up the ReAct workflow with conditional routing based on tool use.
        """
        # Extract configuration
        engine = self.config.engine
        tools = self.config.tools
        structured_output_model = self.config.structured_output_model
        max_iterations = self.config.max_iterations
        
        # Set up tracking for iterations
        self.graph.add_node("track_iterations", self._track_iterations)
        
        # Set up reasoning node
        self.graph.add_node(
            self.config.reasoning_node_name, 
            self._create_reasoning_node()
        )
        
        # Set up tool execution node if tools are provided
        if tools:
            self.graph.add_node(
                self.config.tool_node_name,
                ToolNode(tools)
            )
        
        # Set up structured output node if a model is provided
        if structured_output_model:
            self.graph.add_node(
                self.config.output_node_name,
                self._create_structured_output_node()
            )
        
        # Set the entry point to iteration tracking
        self.graph.set_entry_point("track_iterations")
        
        # Add edge from iteration tracking to reasoning
        self.graph.add_edge("track_iterations", self.config.reasoning_node_name)
        
        # Add conditional edges from reasoning node
        self.graph.add_conditional_edges(
            self.config.reasoning_node_name,
            self._route_from_reasoning,
            {
                "tool_execution": self.config.tool_node_name if tools else END,
                "structured_output": self.config.output_node_name if structured_output_model else END,
                "complete": END
            }
        )
        
        # Add edges from tool execution back to reasoning
        if tools:
            self.graph.add_edge(self.config.tool_node_name, self.config.reasoning_node_name)
        
        # Add edge from structured output to END
        if structured_output_model:
            self.graph.add_edge(self.config.output_node_name, END)
    
    def _track_iterations(self, state: ReactAgentState) -> ReactAgentState:
        """Track iterations to prevent infinite loops."""
        # Initialize iteration counter if not present
        if "iteration_count" not in state.__dict__:
            state.__dict__["iteration_count"] = 0
        else:
            state.__dict__["iteration_count"] += 1
            
        # Check if we've exceeded max iterations
        if state.__dict__["iteration_count"] >= self.config.max_iterations:
            # Add a message indicating we're forcing completion
            state.messages.append(
                AIMessage(content=f"I've reached the maximum number of iterations ({self.config.max_iterations}). "
                          "I'll provide my best response based on what I've learned so far.")
            )
            state.done = True
        
        return state
    
    def _create_reasoning_node(self) -> Callable:
        """Create the reasoning node function."""
        # Extract configuration
        engine = self.config.engine
        
        def reasoning_node(state: ReactAgentState) -> ReactAgentState:
            """Reason about the current state and decide next steps."""
            try:
                # Check if there are new tool results to incorporate
                if state.tool_results:
                    # Get the latest tool results
                    latest_results = state.tool_results[-1]
                    
                    # Add tool result as a message
                    tool_message = f"Tool result: {json.dumps(latest_results, indent=2)}"
                    state.messages.append(AIMessage(content=tool_message))
                
                # Invoke the reasoning engine
                response = engine.invoke({"messages": state.messages})
                
                # Add AI response to messages
                if isinstance(response, AIMessage):
                    state.messages.append(response)
                else:
                    state.messages.append(AIMessage(content=str(response)))
                
                # Extract tool calls from response if present
                latest_message = state.messages[-1]
                if hasattr(latest_message, "tool_calls") and latest_message.tool_calls:
                    state.current_tool_calls = latest_message.tool_calls
                elif hasattr(latest_message, "additional_kwargs") and "tool_calls" in latest_message.additional_kwargs:
                    state.current_tool_calls = latest_message.additional_kwargs["tool_calls"]
                else:
                    # Try to parse tool calls from text content
                    state.current_tool_calls = self._extract_tool_calls_from_text(latest_message.content)
                
                return state
                
            except Exception as e:
                state.error = f"Error in reasoning: {str(e)}"
                state.messages.append(AIMessage(content=f"I encountered an error: {str(e)}"))
                state.done = True
                return state
        
        return reasoning_node
    
    def _extract_tool_calls_from_text(self, text: str) -> List[Dict[str, Any]]:
        """Extract tool calls from text using regex patterns."""
        tool_calls = []
        
        # Look for patterns like: Tool: tool_name({"param": "value"})
        pattern = r'Tool:\s+(\w+)\(({.*?})\)'
        matches = re.findall(pattern, text, re.DOTALL)
        
        for tool_name, args_str in matches:
            try:
                args = json.loads(args_str)
                tool_calls.append({
                    "name": tool_name,
                    "args": args
                })
            except json.JSONDecodeError:
                # If JSON parsing fails, try to extract key-value pairs
                args = {}
                kv_pattern = r'(\w+):\s*"([^"]*)"'
                kv_matches = re.findall(kv_pattern, args_str)
                for key, value in kv_matches:
                    args[key] = value
                
                tool_calls.append({
                    "name": tool_name,
                    "args": args
                })
        
        return tool_calls
    
    def _create_structured_output_node(self) -> Callable:
        """Create the structured output node function."""
        # Extract configuration
        structured_output_model = self.config.structured_output_model
        
        def structured_output_node(state: ReactAgentState) -> ReactAgentState:
            """Process the final AI message into structured output."""
            try:
                # Get the final AI message
                final_messages = [msg for msg in state.messages if isinstance(msg, AIMessage)]
                if not final_messages:
                    state.error = "No AI messages found for structured output."
                    return state
                
                final_message = final_messages[-1]
                
                # Create a parser for the structured output model
                parser = PydanticToolsParser(tools=[structured_output_model])
                
                # Parse the message content
                try:
                    parsed_output = parser.parse(final_message.content)
                    # Take the first result if it's a list
                    if isinstance(parsed_output, list) and parsed_output:
                        state.structured_output = parsed_output[0].dict()
                    else:
                        state.structured_output = parsed_output.dict()
                except Exception as parse_error:
                    # If parsing fails, try to use the model directly
                    try:
                        # Try to extract JSON content
                        json_pattern = r'```json\n(.*?)\n```'
                        json_matches = re.findall(json_pattern, final_message.content, re.DOTALL)
                        
                        if json_matches:
                            json_content = json_matches[0]
                            parsed_data = json.loads(json_content)
                            state.structured_output = structured_output_model(**parsed_data).dict()
                        else:
                            # Try direct instantiation of the model
                            state.structured_output = structured_output_model(
                                **{"content": final_message.content}
                            ).dict()
                    except Exception as e:
                        state.error = f"Failed to create structured output: {str(e)}"
                
                return state
                
            except Exception as e:
                state.error = f"Error in structured output: {str(e)}"
                return state
        
        return structured_output_node
    
    def _route_from_reasoning(self, state: ReactAgentState) -> str:
        """
        Determine where to route next based on reasoning output.
        
        Returns:
            "tool_execution": If there are tool calls to execute
            "structured_output": If we're done and have a structured output model
            "complete": If we're done with no need for structured output
        """
        # Check if we're done (forced completion or no tool calls)
        if state.done:
            if self.config.structured_output_model:
                return "structured_output"
            return "complete"
        
        # Check if there are tool calls to execute
        if state.current_tool_calls:
            return "tool_execution"
        
        # No tool calls, we're done with this request
        if self.config.structured_output_model:
            return "structured_output"
        return "complete"


# Helper functions for easy React agent creation
def create_react_agent(
    system_prompt: Optional[str] = None,
    model: str = "gpt-4o",
    temperature: float = 0.7,
    tools: Optional[List[Any]] = None,
    structured_output_model: Optional[Type[BaseModel]] = None,
    name: Optional[str] = None,
    max_iterations: int = 10,
    visualize: bool = False
) -> ReactAgent:
    """
    Create a React agent from scratch.
    
    Args:
        system_prompt: Optional system prompt
        model: Model name
        temperature: Temperature
        tools: Optional list of tools
        structured_output_model: Optional structured output model
        name: Optional agent name
        max_iterations: Maximum iterations
        visualize: Whether to generate visualization
        
    Returns:
        ReactAgent instance
    """
    config = ReactAgentConfig.from_scratch(
        system_prompt=system_prompt,
        model=model,
        temperature=temperature,
        tools=tools,
        structured_output_model=structured_output_model,
        name=name or "react_agent",
        max_iterations=max_iterations,
        visualize=visualize
    )
    
    return config.build_agent()
````

````python
# examples/react_agent_example.py

from typing import List, Dict, Any, Optional, Type
from pydantic import BaseModel, Field
from langchain_core.tools import tool




# Define a few example tools
@tool
def search_web(query: str) -> str:
    """Search the web for information on a given query."""
    return f"I searched for '{query}' and found relevant information."


@tool
def calculate(expression: str) -> str:
    """Calculate the result of a mathematical expression."""
    try:
        result = eval(expression)
        return f"The result of {expression} is {result}"
    except Exception as e:
        return f"Error calculating {expression}: {str(e)}"


@tool
def get_current_weather(location: str, unit: str = "celsius") -> str:
    """Get the current weather for a location."""
    # This would normally call a weather API
    return f"The weather in {location} is currently sunny and 22 degrees {unit}."


# Define a structured output model (optional)
class ResearchReport(BaseModel):
    """A structured research report."""
    topic: str = Field(..., description="The topic being researched")
    key_findings: List[str] = Field(..., description="Key findings about the topic")
    sources: List[str] = Field(..., description="Sources consulted")
    conclusion: str = Field(..., description="Conclusion of the research")


def main():
    # Create a list of tools
    tools = [search_web, calculate, get_current_weather]
    
    # Create the React agent with custom system prompt
    agent = create_react_agent_from_tools(
        tools=tools,
        model="gpt-4o",
        system_prompt="""You are a helpful research assistant that can search for information, 
        perform calculations, and check the weather to help answer questions.
        When you need information, use the appropriate tool.
        Always explain your reasoning before using a tool.
        After receiving tool results, summarize what you've learned.""",
        structured_output_model=ResearchReport,  # Optional
        name="research_assistant",
        max_iterations=5
    )
    
    # Example 1: Simple question that requires a tool
    question_1 = "What's the square root of 529?"
    print(f"\n\nQuestion: {question_1}")
    agent.run(question_1)
    
    # Example 2: Question that requires multiple tool calls
    question_2 = "I'm planning a trip to Paris. What's the weather like there, and what's the population?"
    print(f"\n\nQuestion: {question_2}")
    agent.run(question_2)
    
    # Example 3: Request for structured output
    question_3 = "Research the impact of climate change on coral reefs and provide a structured report."
    print(f"\n\nQuestion: {question_3}")
    agent.run(question_3)


if __name__ == "__main__":
    main()
````
