```python
from typing import Any, Callable, Dict, List, Optional, Set, Tuple, Union, Type
import json
import inspect
import importlib
import dill as pickle
import base64
from copy import deepcopy
from pydantic import BaseModel, Field, create_model, ValidationError

from langgraph.graph import StateGraph, END
from langgraph.types import Command
from langgraph.checkpoint.memory import MemorySaver
from langgraph.prebuilt import ToolNode


class GraphManager:
    """
    Manager for building, modifying, and compiling a StateGraph.

    This class allows you to add/remove nodes and edges, insert nodes
    in an existing flow, add conditional branches, and compile the graph.
    
    Unlike previous versions, node functions are stored as-is (with a default
    'ends' attribute added if missing) so that you can later retrieve a modifiable
    copy of the underlying StateGraph.
    """

    def __init__(self, state_schema_or_graph: Any):
        """
        Initialize a GraphManager either with a state schema (to create a new StateGraph)
        or an existing StateGraph instance.

        Args:
            state_schema_or_graph: A schema (e.g. Pydantic model or dict) or a StateGraph.
        """
        if isinstance(state_schema_or_graph, StateGraph):
            self.graph = state_schema_or_graph
        else:
            self.graph = StateGraph(state_schema_or_graph)

        # Internal tracking structures.
        self._nodes: Dict[str, Callable] = {}   # node name -> function
        self._tools: Dict[str, List[Any]] = {}    # For tool nodes.
        self._edges: Set[Tuple[str, str]] = set() # (from, to) pairs; "END" is stored as "__end__"
        self._conditional_edges: Dict[str, Tuple[Callable, Dict[str, str]]] = {}
        self._entry_point: Optional[str] = None
        self._compiled_app = None
        self._memory = MemorySaver()
        self._metadata: Dict[str, Any] = {}

        self._load_from_existing_graph()

    def _load_from_existing_graph(self) -> None:
        """
        If the underlying StateGraph already has nodes/edges, load them.
        """
        if hasattr(self.graph, "nodes"):
            for name, node in self.graph.nodes.items():
                self._nodes[name] = node

        if hasattr(self.graph, "edges"):
            self._edges = set()
            for (src, dst) in self.graph.edges:
                self._edges.add((src, "__end__" if dst == END else dst))

        if hasattr(self.graph, "entry_point"):
            self._entry_point = self.graph.entry_point

    def _mark_dirty(self) -> None:
        """
        Mark the compiled app as invalid so that a later compile will rebuild
        the graph.
        """
        self._compiled_app = None

    def _ensure_ends(self, func: Callable) -> Callable:
        """
        Ensure that the node function has an 'ends' attribute. If not, set it to an empty tuple.
        """
        if not hasattr(func, "ends"):
            setattr(func, "ends", ())
        return func

    def add_node(self, name: str, func: Callable) -> 'GraphManager':
        """
        Add a node to the graph.

        Args:
            name: The unique node name.
            func: A callable representing the node's action. If it does not have an 'ends'
                  attribute, one is added automatically.
        """
        wrapped_func = self._ensure_ends(func)
        self.graph.add_node(name, wrapped_func)
        self._nodes[name] = wrapped_func
        self._mark_dirty()
        return self

    def add_tool_node(self, name: str, tools: List[Any]) -> 'GraphManager':
        """
        Add a tool node to the graph. (Tool nodes are created by ToolNode, which you may leave as-is.)

        Args:
            name: Node name.
            tools: List of tool objects.
        """
        tool_node = ToolNode(tools)
        self.graph.add_node(name, tool_node)
        self._nodes[name] = tool_node
        self._tools[name] = tools
        self._mark_dirty()
        return self

    def remove_node(self, name: str) -> 'GraphManager':
        """
        Remove a node (and its connected edges) from the graph.
        """
        if not hasattr(self.graph, "nodes"):
            raise RuntimeError("Cannot modify a compiled graph.")
        self.graph.nodes.pop(name, None)
        self._nodes.pop(name, None)
        self._tools.pop(name, None)
        self._edges = {(s, d) for (s, d) in self._edges if s != name and d != name}
        if hasattr(self.graph, "edges"):
            self.graph.edges = self._edges
        if name in self._conditional_edges:
            self._conditional_edges.pop(name)
        if self._entry_point == name:
            self._entry_point = None
        self._mark_dirty()
        return self

    def add_edge(self, from_node: str, to_node: str) -> 'GraphManager':
        """
        Add a direct edge from one node to another.
        """
        if to_node == "END":
            self.graph.add_edge(from_node, END)
            self._edges.add((from_node, "__end__"))
        else:
            self.graph.add_edge(from_node, to_node)
            self._edges.add((from_node, to_node))
        self._mark_dirty()
        return self

    def remove_edge(self, from_node: str, to_node: str) -> 'GraphManager':
        """
        Remove an edge from the graph.
        """
        if not hasattr(self.graph, "edges"):
            raise RuntimeError("Cannot modify a compiled graph.")
        if to_node in ("END", "__end__"):
            self._edges.discard((from_node, "__end__"))
        else:
            self._edges.discard((from_node, to_node))
        self.graph.edges = self._edges
        self._mark_dirty()
        return self

    def add_conditional_edge(
        self, from_node: str, condition_fn: Callable, routes: Dict[str, str]
    ) -> 'GraphManager':
        """
        Add a conditional edge (branch) from a node.

        Args:
            from_node: The node from which the branch originates.
            condition_fn: Function that returns a key given the current state.
            routes: Mapping from condition keys to target node names.
        """
        self.graph.add_conditional_edges(from_node, condition_fn, {k: (END if v == "END" else v) for k, v in routes.items()})
        self._conditional_edges[from_node] = (condition_fn, routes)
        self._mark_dirty()
        return self

    def set_entry_point(self, node_name: str) -> 'GraphManager':
        """
        Set the entry point of the graph.

        Args:
            node_name: The name of the node where execution starts.
        """
        self.graph.set_entry_point(node_name)
        self._entry_point = node_name
        self._mark_dirty()
        return self

    def insert_node_after(self, target_node: str, new_node_name: str, new_node_func: Callable) -> 'GraphManager':
        """
        Insert a new node immediately after an existing node.
        
        This method removes the existing outgoing edges from the target node,
        adds the new node, and then reconnects the removed edges from the new node.
        
        Args:
            target_node: Name of the existing node.
            new_node_name: Name for the new node.
            new_node_func: Callable for the new node.
        """
        if not hasattr(self.graph, "edges"):
            raise RuntimeError("Cannot modify a compiled graph.")
        outgoing = [(s, d) for (s, d) in self._edges if s == target_node]
        has_cond = target_node in self._conditional_edges
        if has_cond:
            cond_fn, cond_routes = self._conditional_edges.pop(target_node)
            flag = True
        else:
            flag = False
            for (_, dst) in outgoing:
                self.remove_edge(target_node, "END" if dst == "__end__" else dst)
        self.add_node(new_node_name, new_node_func)
        self.add_edge(target_node, new_node_name)
        if flag:
            self.add_conditional_edge(new_node_name, cond_fn, cond_routes)
        else:
            for (_, dst) in outgoing:
                self.add_edge(new_node_name, "END" if dst == "__end__" else dst)
        self._mark_dirty()
        return self

    def replace_node(self, node_name: str, new_func: Callable) -> 'GraphManager':
        """
        Replace the function of an existing node.
        
        Args:
            node_name: Name of the node to replace.
            new_func: New callable; if it lacks 'ends', one is added.
        """
        if not hasattr(self.graph, "nodes"):
            raise RuntimeError("Cannot modify a compiled graph.")
        self.graph.remove_node(node_name)
        self.add_node(node_name, new_func)
        self._mark_dirty()
        return self

    def insert_branch(
        self,
        from_node: str,
        condition_fn: Callable,
        branch_node_name: str,
        branch_node_func: Callable,
        rejoin_node: Optional[str] = None
    ) -> 'GraphManager':
        """
        Insert a conditional branch from a node.
        
        Args:
            from_node: The source node.
            condition_fn: Function that determines the branch (returns a key).
            branch_node_name: Name for the new branch node.
            branch_node_func: Callable for the branch node.
            rejoin_node: Optional node where the branch should rejoin the main flow.
        """
        normal_target = None
        for (s, d) in self._edges:
            if s == from_node:
                normal_target = d
                break
        if not normal_target and from_node not in self._conditional_edges:
            normal_target = "__end__"
        self.add_node(branch_node_name, branch_node_func)
        branch_target = rejoin_node if rejoin_node else ("END" if normal_target == "__end__" else normal_target)
        self.add_edge(branch_node_name, branch_target)
        routes = {"branch": branch_node_name, "continue": (normal_target if normal_target != "__end__" else "END")}
        if normal_target:
            self.remove_edge(from_node, "END" if normal_target == "__end__" else normal_target)
        self.add_conditional_edge(from_node, condition_fn, routes)
        self._mark_dirty()
        return self

    def get_graph_copy(self) -> StateGraph:
        """
        Return a deep copy of the underlying StateGraph.
        This copy can be modified post-compile.
        """
        return deepcopy(self.graph)

    def compile(self, checkpointer: Optional[MemorySaver] = None) -> Any:
        """
        Compile the graph.

        The graph's entry point must be set either via set_entry_point() or already
        defined in the underlying StateGraph.
        """
        if not self._entry_point:
            if hasattr(self.graph, "entry_point") and self.graph.entry_point:
                self._entry_point = self.graph.entry_point
            else:
                raise ValueError("Entry point must be set before compiling the graph.")
        self._compiled_app = self.graph.compile(checkpointer=checkpointer or self._memory)
        return self._compiled_app

    def run(self, inputs: Dict[str, Any], config: Optional[Dict[str, Any]] = None) -> Any:
        """
        Run the compiled graph with the given inputs.
        """
        if not self._compiled_app:
            self.compile()
        return self._compiled_app.invoke(inputs, config=config)

    def stream(self, inputs: Dict[str, Any], config: Optional[Dict[str, Any]] = None) -> Any:
        """
        Stream results from the compiled graph.
        """
        if not self._compiled_app:
            self.compile()
        return self._compiled_app.stream(inputs, stream_mode="values", config=config)

    async def arun(self, inputs: Dict[str, Any], config: Optional[Dict[str, Any]] = None) -> Any:
        """
        Asynchronously run the compiled graph.
        """
        if not self._compiled_app:
            self.compile()
        return await self._compiled_app.ainvoke(inputs, config=config)

    async def astream(self, inputs: Dict[str, Any], config: Optional[Dict[str, Any]] = None) -> Any:
        """
        Asynchronously stream results from the graph.
        """
        if not self._compiled_app:
            self.compile()
        async for output in self._compiled_app.astream(inputs, stream_mode="values", config=config):
            yield output

    def visualize(self, output_file: str = "graph.png") -> None:
        """
        Visualize the compiled graph (requires Graphviz).

        If not already compiled, this triggers a compile.
        """
        if not self._compiled_app:
            self.compile()
        from src.haive.utils.visualize_graph_utils import render_and_display_graph
        render_and_display_graph(self._compiled_app, output_name=output_file)
        print(f"Graph visualization saved to: {output_file}")

    def to_dict(self) -> Dict[str, Any]:
        """
        Convert the graph structure to a serializable dictionary.
        """
        nodes = {}
        for name, node in self._nodes.items():
            if name in self._tools:
                nodes[name] = {
                    "type": "tool_node",
                    "tools": [tool.__class__.__name__ for tool in self._tools[name]]
                }
            else:
                # Serialize the function as source or pickle
                nodes[name] = {
                    "type": "function",
                    "function": self._serialize_function(node)
                }
        edges = [{"from": src, "to": ("END" if dst == "__end__" else dst)} for (src, dst) in self._edges]
        conditional_edges = {}
        for node, (cond_fn, routes) in self._conditional_edges.items():
            conditional_edges[node] = {
                "function": self._serialize_function(cond_fn),
                "routes": routes
            }
        return {
            "nodes": nodes,
            "edges": edges,
            "conditional_edges": conditional_edges,
            "entry_point": self._entry_point,
            "metadata": self._metadata,
        }

    def to_json(self) -> str:
        """
        Serialize the graph to a JSON string.
        """
        return json.dumps(self.to_dict())

    @classmethod
    def from_dict(cls, data: Dict[str, Any], state_schema: Any) -> 'GraphManager':
        """
        Rebuild a GraphManager from a dictionary.
        """
        mgr = cls(state_schema)
        mgr._metadata = data.get("metadata", {})
        for name, node_data in data["nodes"].items():
            node_type = node_data.get("type", "function")
            if node_type == "tool_node":
                mgr.create_command_node(name, update_dict={"tool_node_placeholder": True})
            else:
                func_info = node_data["function"]
                func = mgr._deserialize_function(func_info)
                mgr.add_node(name, func)
        for edge in data["edges"]:
            mgr.add_edge(edge["from"], edge["to"])
        for node, cdata in data.get("conditional_edges", {}).items():
            fn_info = cdata["function"]
            cond_fn = mgr._deserialize_function(fn_info)
            routes = cdata["routes"]
            mgr.add_conditional_edge(node, cond_fn, routes)
        ep = data.get("entry_point")
        if ep:
            mgr.set_entry_point(ep)
        return mgr

    @classmethod
    def from_json(cls, json_str: str, state_schema: Any) -> 'GraphManager':
        """
        Rebuild a GraphManager from a JSON string.
        """
        data = json.loads(json_str)
        return cls.from_dict(data, state_schema)

    def _serialize_function(self, func: Callable) -> Dict[str, Any]:
        """
        Attempt to serialize a function via:
          1) Module import reference,
          2) Source code,
          3) Pickle (base64-encoded).
        """
        if hasattr(func, "__module__") and hasattr(func, "__name__"):
            mod = func.__module__
            name = func.__name__
            if not name.startswith("<") and mod != "__main__":
                try:
                    imported = importlib.import_module(mod)
                    if hasattr(imported, name):
                        return {"type": "import", "module": mod, "name": name}
                except (ImportError, AttributeError):
                    pass
        try:
            source = inspect.getsource(func)
            return {"type": "source", "source": source}
        except (OSError, TypeError):
            pass
        try:
            pickled = base64.b64encode(pickle.dumps(func)).decode("utf-8")
            return {"type": "pickle", "data": pickled}
        except (pickle.PickleError, TypeError) as e:
            raise ValueError(f"Could not serialize function {func}: {e}")

    def _deserialize_function(self, func_info: Dict[str, Any]) -> Callable:
        """
        Reverse the function serialization.
        """
        ftype = func_info.get("type")
        if ftype == "import":
            mod_name = func_info["module"]
            func_name = func_info["name"]
            mod = importlib.import_module(mod_name)
            return getattr(mod, func_name)
        elif ftype == "source":
            source = func_info["source"]
            ns = {}
            try:
                exec(source, ns)
                for obj in ns.values():
                    if callable(obj) and not isinstance(obj, type):
                        return obj
                raise ValueError("No callable found in source.")
            except Exception as e:
                raise ValueError(f"Error deserializing from source: {e}")
        elif ftype == "pickle":
            pickled = func_info["data"]
            return pickle.loads(base64.b64decode(pickled))
        else:
            raise ValueError(f"Unknown function serialization type: {ftype}")

    def get_node_names(self) -> List[str]:
        """Return a list of all node names."""
        return list(self._nodes.keys())

    def get_edge_list(self) -> List[Tuple[str, str]]:
        """Return a list of edges as (from, to) pairs (with 'END' substituted for '__end__')."""
        return [(src, "END" if dst == "__end__" else dst) for (src, dst) in self._edges]

    def get_conditional_edges(self) -> Dict[str, Tuple[Callable, Dict[str, str]]]:
        """Return the mapping of conditional edges."""
        return dict(self._conditional_edges)

    def add_metadata(self, key: str, value: Any) -> 'GraphManager':
        """Add metadata to the graph."""
        self._metadata[key] = value
        return self

    def get_metadata(self, key: str, default: Any = None) -> Any:
        """Retrieve metadata by key."""
        return self._metadata.get(key, default)

    def create_generalized_tool_node(
        self,
        name: str,
        tools: List[Any],
        post_processing_fn: Optional[Callable] = None
    ) -> 'GraphManager':
        """
        Create a node that runs a set of tools and optionally post-processes the result.
        """
        def tool_node_with_processing(state):
            tn = ToolNode(tools)
            result = tn.invoke(state)
            return post_processing_fn(result) if post_processing_fn else result

        self.add_node(name, tool_node_with_processing)
        self._tools[name] = tools
        self._mark_dirty()
        return self

    def add_generalized_tool_routing(
        self,
        node_name: str,
        tool_routing_map: Optional[Dict[str, str]] = None,
        default_tool_node: str = "tools",
        default_end_node: str = "END",
        custom_routing_fn: Optional[Callable] = None,
        messages_key: str = "messages"
    ) -> 'GraphManager':
        """
        Add conditional routing that examines the final AI message for tool calls.
        """
        def generalized_tools_condition(state):
            if isinstance(state, list):
                messages = state
            elif isinstance(state, dict):
                messages = state.get(messages_key, [])
            elif hasattr(state, messages_key):
                messages = getattr(state, messages_key, [])
            else:
                raise ValueError(f"No messages found in input state: {state}")
            if not messages:
                return default_end_node
            ai_message = messages[-1]
            if hasattr(ai_message, "tool_calls") and ai_message.tool_calls:
                for tool_call in ai_message.tool_calls:
                    tool_name = tool_call.get("name") if isinstance(tool_call, dict) else getattr(tool_call, "name", None)
                    if tool_routing_map and tool_name in tool_routing_map:
                        return tool_routing_map[tool_name]
                return default_tool_node
            if custom_routing_fn:
                return custom_routing_fn(ai_message)
            return default_end_node

        routes = {"END": "END"}
        if tool_routing_map:
            routes.update(tool_routing_map)
        routes[default_tool_node] = default_tool_node

        self.add_conditional_edge(node_name, generalized_tools_condition, routes)
        self._mark_dirty()
        return self

    def create_supervisor_branch(
        self,
        condition_fn: Callable,
        supervisor_node_name: str,
        supervisor_func: Callable,
        normal_flow_node: str,
        rejoin_node: Optional[str] = None
    ) -> Tuple['GraphManager', str]:
        """
        Insert a supervisor branch if a condition is met.

        Returns:
            Tuple of (self, branch_node_name)
        """
        self.add_node(supervisor_node_name, supervisor_func)
        target = None
        for src, dst in self._edges:
            if src == normal_flow_node:
                target = dst
                break
        if not target:
            target = "__end__"
        if target == "__end__":
            self.add_edge(supervisor_node_name, "END")
        else:
            self.add_edge(supervisor_node_name, target)

        def branch_router(state):
            return "supervisor" if condition_fn(state) else "normal"

        branch_node_name = f"branch_to_{supervisor_node_name}"

        def branch_node(state):
            return state

        self.add_node(branch_node_name, branch_node)
        self.add_conditional_edge(branch_node_name, branch_router, {"supervisor": supervisor_node_name, "normal": normal_flow_node})
        self._mark_dirty()
        return self, branch_node_name

    def visualize(self, output_file: str = "graph.png") -> None:
        """
        Visualize the compiled graph (requires Graphviz).

        If not compiled, this triggers a compile.
        """
        if not self._compiled_app:
            self.compile()
        from src.haive.utils.visualize_graph_utils import render_and_display_graph
        render_and_display_graph(self._compiled_app, output_name=output_file)
        print(f"Graph visualization saved to: {output_file}")
```

```python
import networkx as nx
import matplotlib.pyplot as plt
from typing import Any, Dict, List, Tuple


class StateGraphManager:
    """
    A manager for extracting metadata and visualizing a StateGraph.
    """

    def __init__(self, graph: Any):
        """
        Initialize the StateGraphManager.

        Args:
            graph (StateGraph): The StateGraph object to manage.
        """
        self.graph = graph
        self.metadata = self.extract_metadata()

    def extract_metadata(self) -> Dict[str, Any]:
        """
        Extract metadata from the StateGraph, including conditional branches.

        Returns:
            Dict[str, Any]: A dictionary containing metadata.
        """
        metadata = {
            "entry_point": getattr(self.graph, "entry_point", None),
            "finish_point": getattr(self.graph, "finish_point", None),
            "nodes": list(getattr(self.graph, "nodes", {}).keys()),
            "edges": list(getattr(self.graph, "edges", set())),
            "conditional_edges": {},  # Extracted below
            "schemas": getattr(self.graph, "schemas", {}),
            "input_schema": getattr(self.graph, "input", None),
            "output_schema": getattr(self.graph, "output", None),
            "compiled": getattr(self.graph, "compiled", False),
            "support_multiple_edges": getattr(self.graph, "support_multiple_edges", True),
        }

        # Extract Conditional Edges (Branches)
        branches = getattr(self.graph, "branches", {})
        for node, conditions in branches.items():
            for condition, target in conditions[1].items():
                if node not in metadata["conditional_edges"]:
                    metadata["conditional_edges"][node] = []
                metadata["conditional_edges"][node].append((condition, target))

        # Ensure entry point exists
        if metadata["entry_point"] is None and metadata["nodes"]:
            metadata["entry_point"] = metadata["nodes"][0]  # Default to first node if not explicitly set

        # Handle the END node
        if metadata["finish_point"] is None:
            for _, target in metadata["edges"]:
                if target == "__end__":
                    metadata["finish_point"] = "__end__"

        return metadata

    def visualize(self, output_file: str = "state_graph.png"):
        """
        Visualize the StateGraph using NetworkX, including conditional branching.

        Args:
            output_file (str): The filename to save the visualization.
        """
        G = nx.DiGraph()
        edge_labels = {}

        # Add nodes
        for node in self.metadata["nodes"]:
            G.add_node(node)

        # Add standard edges
        for src, dst in self.metadata["edges"]:
            if dst == "__end__":
                dst = "END"
            G.add_edge(src, dst)

        # Add conditional branching edges
        for node, conditions in self.metadata["conditional_edges"].items():
            for condition, target in conditions:
                if target == "__end__":
                    target = "END"
                G.add_edge(node, target, style="dashed", color="red")
                edge_labels[(node, target)] = f"{condition}"

        # Draw the graph
        plt.figure(figsize=(12, 7))
        pos = nx.spring_layout(G, seed=42)  # Position nodes dynamically

        # Standard edges
        nx.draw(G, pos, with_labels=True, node_color="lightblue", edge_color="gray", node_size=2200, font_size=10)

        # Conditional edges (dashed red)
        for (src, dst), label in edge_labels.items():
            nx.draw_networkx_edges(G, pos, edgelist=[(src, dst)], edge_color="red", style="dashed")
            nx.draw_networkx_edge_labels(G, pos, edge_labels={(src, dst): label}, font_color="red", font_size=9)

        # Highlight entry and finish points
        entry = self.metadata["entry_point"]
        finish = self.metadata["finish_point"]

        if entry:
            nx.draw_networkx_nodes(G, pos, nodelist=[entry], node_color="green", node_size=2500)  # Entry point
        if finish and finish in self.metadata["nodes"]:
            nx.draw_networkx_nodes(G, pos, nodelist=[finish], node_color="red", node_size=2500)  # Finish point

        # Save and display
        plt.savefig(output_file)
        plt.show()
        print(f"Graph saved as {output_file}")

    def get_metadata(self):
        """Return the extracted metadata."""
        return self.metadata
```
