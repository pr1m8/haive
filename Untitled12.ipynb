```python
from typing import List, Dict, Any, Union, Type, get_args,Literal
from enum import Enum
from pydantic import BaseModel, Field, validator
from langchain_core.messages import HumanMessage, SystemMessage, BaseMessage
from langchain_core.runnables import Runnable
from langgraph.types import Command
from langgraph.graph import END
from src.haive.core.models.llm.base import LLMConfig, AzureLLMConfig
from src.haive.core.aug_llm.base import AugLLMConfig
from langchain.prompts import PromptTemplate
# ==========================
# 🔹 Dynamic Route Choice
# ==========================
class WorkerRole(str, Enum):
    vectorstore = "vectorstore"
    web_search = "web_search"
    finish = "END"

    @classmethod
    def choices(cls) -> List[str]:
        """Return all valid routing choices."""
        return [member.value for member in cls]


# ==========================
# 🔹 RouteChoice Model (For Dynamic Routing)
# ==========================
class RouteChoice(BaseModel):
    name: str
    description: str


# ==========================
# 🔹 SupervisorRouterConfig (Manages Routing Choices & Literals)
# ==========================
# ==========================
# 🔹 SupervisorRouterConfig (Manages Routing Choices & Literals)
# ==========================
class SupervisorRouterConfig(BaseModel):
    """Configuration for SupervisorRouter using structured output."""
    choices: List[RouteChoice] = Field(default=[], description="Available routing choices.")

    def create_literal_type(self) -> Type:
        """Dynamically create a `Literal` type for valid choices."""
        valid_choices = [choice.name for choice in self.choices] + ["END"]
        return Literal[*valid_choices]

    def create_structured_output_model(self) -> Type[BaseModel]:
        """Create a structured output model for routing."""
        literal_type = self.create_literal_type()

        class RouteQuery(BaseModel):
            next: literal_type = Field(..., description="The next node to route to.")

        return RouteQuery

    def create_system_prompt(self) -> ChatPromptTemplate:
        """Dynamically generates a `ChatPromptTemplate` based on available choices."""
        
        choices_str = ", ".join(f'"{c.name}" ({c.description})' for c in self.choices)
        print(ChatPromptTemplate.from_messages([
            ("system", """You are a router responsible for selecting the next step in a workflow.
            You MUST return one of the following options: {choices}.
            
            Your response MUST be a single JSON object with ONLY the "next" key and a valid value.
            
            Example output:
            {{
                "next": "vectorstore"
            }}
            
            If the task is completed, respond with 'END'.""")
        ]).partial(choices=choices_str))
        return ChatPromptTemplate.from_messages([
            ("system", """You are a router responsible for selecting the next step in a workflow.
            You MUST return one of the following options: {choices}.
            
            Your response MUST be a single JSON object with ONLY the "next" key and a valid value.
            
            Example output:
            {{
                "next": "vectorstore"
            }}
            
            If the task is completed, respond with 'END'.""")
        ]).partial(choices=choices_str)
    

        
from pydantic import BaseModel, Field
from langchain_core.messages import HumanMessage
from typing import List, Dict, Any, Union

# ================================
# 🔹 Define Graph State (Includes `options`)
# ================================
class GraphState(BaseModel):
    messages: List[HumanMessage]
    next: str = Field(default="vectorstore", description="Next execution step")
    options: List[str] = Field(default=["vectorstore", "web_search", "api_call"], description="Available execution paths")

# ==========================
# 🔹 SupervisorRouter (Extends AugLLMConfig)
# ==========================
class SupervisorRouter(AugLLMConfig):
    """A dynamic supervisor that routes execution based on LLM decisions."""

    router_config: SupervisorRouterConfig

    def __init__(self, **kwargs):
        super().__init__(**kwargs)
        self.router_config = kwargs.get("router_config")

    @property
    def literal_type(self) -> Type:
        """Compute `Literal` type dynamically when accessed."""
        return self.router_config.create_literal_type()

    def create_agent(self) -> Runnable:
        """Overrides `create_agent()` to use the dynamic routing model."""
        structured_output_model = self.router_config.create_structured_output_model()
        self.structured_output_model = structured_output_model

        return AugLLMConfig(prompt_template=self.router_config.create_system_prompt(),
                            stuctrured_output_model=self.structured_output_model)

    def get_choices(self) -> List[str]:
        """Retrieve the list of available routing options."""
        return [choice.name for choice in self.router_config.choices]

    def format_input(self, input_state: Dict[str, Any]) -> List[HumanMessage]:
        """Converts `input_state` into a valid input format for the LLM."""

        messages = input_state.get("messages", [])
        if not messages:
            return [SystemMessage(content="No messages provided.")]

        formatted_messages = []
        for message in messages:
            if isinstance(message, str):
                formatted_messages.append(HumanMessage(content=message))
            elif isinstance(message, dict) and "role" in message and "content" in message:
                if message["role"] == "system":
                    formatted_messages.append(SystemMessage(content=message["content"]))
                else:
                    formatted_messages.append(HumanMessage(content=message["content"]))
            else:
                raise ValueError(f"Invalid message format: {message}")

        return formatted_messages

    def route(self, input_state: Dict[str, Any]) -> Command:
        """Routes execution based on LLM response."""
        formatted_input = self.format_input(input_state)

        # Invoke LLM agent
        response = self.create_agent().invoke(formatted_input)
        next_node = response.next  # Extract route

        # Validate next_node
        valid_routes = self.get_choices() + ["END"]
        if next_node not in valid_routes:
            print(f"⚠️ Invalid route '{next_node}', defaulting to END.")
            next_node = END
        elif next_node == "END":
            next_node = END

        return Command[Literal[*get_args(self.literal_type)]](update=input_state, goto=next_node)


# ==========================
# 🔹 Helper: Create Supervisor Node
# ==========================
def create_supervisor_node(supervisor) -> callable:
    """
    **Creates a dynamic supervisor node for routing.**
    - ✅ Uses `ChatPromptTemplate` for structured input
    - ✅ Ensures implicit edges & valid routing
    """
    choices = supervisor.get_choices()
    valid_routes = choices + ["END"]

    def supervisor_node(state: Dict[str, Any]) -> Command:
        """LLM-based router with implicit edges."""
        if isinstance(state,dict):
            formatted_input={'messages':state['messages']}
        else:
            formatted_input = {"messages": state.messages}
       

        # Ensure correct input format
        #if not isinstance(state["messages"], list) or not all(isinstance(m, BaseMessage) for m in state["messages"]):
            #raise ValueError("State['messages'] must be a list of `BaseMessage` instances.")
        runnable = supervisor.create_agent().create_runnable()
        print(supervisor.create_agent().prompt_template)
        print(state.messages)
        print(state)
        response = runnable.invoke(formatted_input)
        print(response)
        goto = response.next

        # Ensure valid routing
        if goto not in valid_routes:
            print(f"⚠️ Invalid route '{goto}', defaulting to END.")
            goto = END
        elif goto == "END":
            goto = END

        return Command(update={"next": goto}, goto=goto)

    return supervisor_node
```

```python
from typing import List, Dict, Any, Type
from pydantic import BaseModel, Field
from langgraph.types import Command
from langgraph.graph import END
from langchain_core.messages import HumanMessage


# ================================
# 🔹 Define GraphState (Execution State)
# ================================
from typing import List
from pydantic import BaseModel, Field

class GraphState(BaseModel):
    messages: List[HumanMessage] = Field(default=[], description="Conversation messages")
    next: str = Field(default="vectorstore", description="Next execution step")
    options: List[str] = Field(default_factory=lambda: ["vectorstore", "web_search", "api_call"], description="Available execution paths")

    def add_option(self, name: str):
        """Dynamically add a new execution path."""
        if name not in self.options:
            self.options.append(name)


# ================================
# 🔹 Define MetaState (State Manager)
# ================================
class MetaState(BaseModel):
    """
    Manages:
    - `graph_state`: Current execution state.
    - `supervisor`: SupervisorRouter (keeps the same instance).
    - `llm_configs`: LLM configurations for execution steps.
    """
    graph_state: GraphState
    supervisor: SupervisorRouter
    llm_configs: Dict[str, AugLLMConfig]

    def add_execution_option(self, name: str, description: str, llm_config: AugLLMConfig):
        """
        Adds a new execution option dynamically.
        - Registers a new route in the supervisor config.
        - Updates available execution paths.
        - SupervisorRouter does NOT get rebuilt (just updated).
        """
        if name in self.llm_configs:
            print(f"⚠️ Execution option '{name}' already exists. Skipping.")
            return
        
        # ✅ Add the execution option
        self.llm_configs[name] = llm_config
        
        # ✅ Update SupervisorRouterConfig **without rebuilding**
        self.supervisor.router_config.choices.append(RouteChoice(name=name, description=description))

        print(f"✅ New execution option added: {name} → {description}")

    def get_available_options(self) -> List[str]:
        """Dynamically retrieve all execution options from `llm_configs`."""
        return list(self.llm_configs.keys())  # Always up-to-date
```

```python
# ================================
# 🔹 Define the Supervisor Node (Always Up-to-Date)
# ================================
def supervisor_node(state: GraphState, meta_state: MetaState) -> Command:
    """
    LLM-based router that dynamically selects the next step.
    - Reads `messages` and selects from `meta_state.get_available_options()`.
    - Uses `meta_state.supervisor` (which updates dynamically).
    """
    
    formatted_input = {"messages": state.messages}

    # Ensure correct input format
    if not isinstance(state.messages, list) or not all(isinstance(m, HumanMessage) for m in state.messages):
        raise ValueError("State['messages'] must be a list of `HumanMessage` instances.")

    # Invoke the updated SupervisorRouter
    response = meta_state.supervisor.create_agent().invoke(state.messages)
    print(response)
    next_step = response.next  # Extract the route

    # ✅ Validate next_step dynamically
    valid_routes = meta_state.get_available_options() + ["END"]
    if next_step not in valid_routes:
        print(f"⚠️ Invalid route '{next_step}', defaulting to 'END'.")
        next_step = "END"

    return Command(update={"next": next_step}, goto=next_step)
```

```python
from src.haive.core.aug_llm.base import AugLLMConfig, AugLLMFactory
from src.haive.core.models.llm.base import AzureLLMConfig

# ================================
# 🔹 Define Default Execution Options
# ================================
from langchain_core.prompts import ChatPromptTemplate
from src.haive.core.aug_llm.base import AugLLMConfig

# ================================
# 🔹 Define ChatPromptTemplates for Execution Options
# ================================
default_execution_options = {
    "vectorstore": AugLLMConfig(
        model="gpt-4o",
        prompt_template=ChatPromptTemplate.from_messages([
            ("system", "You are retrieving documents from a vector database."),
            ("human", "{query}")  # Dynamic input placeholder
        ])
    ),
    
    "web_search": AugLLMConfig(
        model="gpt-4o",
        prompt_template=ChatPromptTemplate.from_messages([
            ("system", "You are performing a real-time web search."),
            ("human", "{query}")
        ])
    ),
    
    "api_call": AugLLMConfig(
        model="gpt-4o",
        prompt_template=ChatPromptTemplate.from_messages([
            ("system", "You are fetching real-time data from an external API."),
            ("human", "{query}")
        ])
    )
}

# ================================
# 🔹 Convert Prompts to AugLLMConfig
# ================================
aug_llm_configs=default_execution_options

print("✅ ChatPromptTemplates and AugLLMConfigs Created!")

print(default_execution_options)
# ================================
# 🔹 Setup SupervisorRouterConfig Dynamically
# ================================
router_config = SupervisorRouterConfig(
    choices=[
        RouteChoice(name=name, description=f"Executes {name.replace('_', ' ')} operation.")
        for name in default_execution_options.keys()
    ]
)

# ================================
# 🔹 Initialize SupervisorRouter (Persistent)
# ================================
supervisor = SupervisorRouter(
    name="supervisor_router",
    llm_config=AzureLLMConfig(),
    router_config=router_config
)

# ================================
# 🔹 Initialize MetaState
# ================================
meta_state = MetaState(
    graph_state=GraphState(),
    supervisor=supervisor,
    llm_configs=default_execution_options
)

# ✅ Output Initial Execution Paths
print(f"✅ Initial Execution Options: {meta_state.get_available_options()}")
```

```python
from langgraph.types import Command

# ================================
# 🔹 Define the Dynamic Agent Node (Uses `MetaState`)
# ================================
def agent_node(state: GraphState, meta_state: MetaState) -> Dict[str, Any]:
    """
    Dynamically selects an LLM runnable based on `state.next`, 
    invokes it, and updates messages.
    """
    
    # Ensure `next` is a valid option
    if state.next not in state.options:
        print(f"⚠️ Invalid option '{state.next}', defaulting to 'vectorstore'.")
        state.next = "vectorstore"
    
    # Select the correct LLM pipeline
    llm_config = meta_state.llm_configs.get(state.next, meta_state.llm_configs["vectorstore"])
    
    # Create a runnable from the selected AugLLMConfig
    llm_runnable = AugLLMFactory(llm_config).runnable

    # Invoke the LLM with state messages
    response = llm_runnable.invoke({"messages": state.messages})
    print(response)
    # Return updated state
    return {
        "messages": state.messages + [response],
        "next": state.next,  # Keep next the same for routing
        "options": state.options  # Keep available options in state
    }
```

```python
from langgraph.graph import StateGraph

# ================================
# 🔹 Build LangGraph Dynamically
# ================================
graph = StateGraph(GraphState)

# Add Supervisor Node as the Initial Routing Decision Maker
graph.add_node("supervisor", create_supervisor_node(meta_state.supervisor))

# Add Dynamic Agent Node (Executes the Selected Path)
graph.add_node("agent_node", lambda state: agent_node(state, meta_state))

# Set Supervisor as the Entry Point
graph.set_entry_point("supervisor")

# Define Supervisor Routing Logic (Routes to Agent Node)
graph.add_conditional_edges("supervisor", lambda state: state.next, {opt: "agent_node" for opt in meta_state.graph_state.options})

# Ensure Agent Node Loops Back to Supervisor After Execution
graph.add_edge("agent_node", "supervisor")

# Compile the Graph
compiled_graph = graph.compile()

print("✅ Graph Compiled with Dynamic Supervisor & Agent Nodes!")
```
