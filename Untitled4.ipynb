```python
class StateSchemaManager:
    """
    A dynamic schema manager that:
    - Builds schemas from dicts, lists, or Pydantic models.
    - Merges schemas while preserving uniqueness (no duplicate `messages` field).
    - Tracks validators, properties, computed attributes, class/static methods.
    - Supports recursive field merging for nested schemas.
    - Uses a config dictionary for settings and overrides.
    - Prints schema nicely in Python class format.
    """

    def __init__(self, 
                 data: Optional[Union[Dict[str, Any], Type[BaseModel]]] = None,
                 name: Optional[str] = None,
                 config: Optional[Dict[str, Any]] = None):
        """
        Args:
            data: Dictionary or Pydantic BaseModel.
            name: Custom schema name (defaults to class name if BaseModel).
            config: Optional dictionary for schema customization.
        """
        self.fields = {}
        self.validators = {}
        self.properties = {}
        self.class_methods = {}
        self.static_methods = {}
        self.config = config or {}  # Store user-defined configurations
        self.locked = False  

        if data is None:
            self.name = name or self.config.get("default_schema_name", "UnnamedSchema")
            return  

        if name is None and isinstance(data, type) and issubclass(data, BaseModel):
            self.name = data.__name__  
        else:
            self.name = name or "CustomState"

        if isinstance(data, dict):
            self._load_from_dict(data)
        elif isinstance(data, type) and issubclass(data, BaseModel):
            self._load_from_model(data)
        else:
            raise TypeError(f"Unsupported data type: {type(data)}")

    def _load_from_dict(self, data: Dict[str, Any]) -> None:
        for key, value in data.items():
            inferred_type = self._infer_type(value)
            self.fields[key] = (inferred_type, Field(default=value))

    def _load_from_model(self, model_cls: Type[BaseModel]) -> None:
        for field_name, field_info in model_cls.model_fields.items():
            self.fields[field_name] = (field_info.annotation, field_info)

        for name, method in inspect.getmembers(model_cls, predicate=inspect.isfunction):
            if hasattr(method, "__validator_config__"):
                self.validators[name] = method
            elif isinstance(getattr(model_cls, name, None), property):
                self.properties[name] = method
            elif isinstance(getattr(model_cls, name, None), classmethod):
                self.class_methods[name] = method
            elif isinstance(getattr(model_cls, name, None), staticmethod):
                self.static_methods[name] = method

    def _infer_type(self, value: Any) -> Type:
        """Infer the type of a value, with special handling for collections."""
        if isinstance(value, str):
            return str
        elif isinstance(value, int):
            return int
        elif isinstance(value, float):
            return float
        elif isinstance(value, bool):
            return bool
        elif isinstance(value, list):
            return List[Any]  # We can't guarantee homogeneity so use Any
        elif isinstance(value, dict):
            return Dict[str, Any]
        return Any

    def add_field(self, name: str, type_hint: Type, default: Any = None, required: bool = False) -> 'StateSchemaManager':
        if self.locked:
            raise ValueError("Schema is locked and cannot be modified.")
        field_info = Field(...) if required else Field(default=default)
        self.fields[name] = (type_hint, field_info)
        return self

    def remove_field(self, name: str) -> 'StateSchemaManager':
        if self.locked:
            raise ValueError("Schema is locked and cannot be modified.")
        if name in self.fields:
            del self.fields[name]
        return self

    def modify_field(self, name: str, new_type: Type, new_default: Any = None) -> 'StateSchemaManager':
        if self.locked:
            raise ValueError("Schema is locked and cannot be modified.")
        if name in self.fields:
            self.fields[name] = (new_type, Field(default=new_default))
        return self

    def merge(self, other: Union['StateSchemaManager', Type[BaseModel]]) -> 'StateSchemaManager':
        if self.locked:
            raise ValueError("Schema is locked and cannot be modified.")

        merged = StateSchemaManager(name=f"{self.name}_Merged", config=self.config)
        merged.fields = self.fields.copy()
        merged.validators = self.validators.copy()
        merged.properties = self.properties.copy()
        merged.class_methods = self.class_methods.copy()
        merged.static_methods = self.static_methods.copy()

        if isinstance(other, StateSchemaManager):
            for field, (field_type, field_info) in other.fields.items():
                if field == "messages":
                    continue  
                if field not in merged.fields:
                    merged.fields[field] = (field_type, field_info)

            merged.validators.update(other.validators)
            merged.properties.update(other.properties)
            merged.class_methods.update(other.class_methods)
            merged.static_methods.update(other.static_methods)

        elif isinstance(other, type) and issubclass(other, BaseModel):
            return merged.merge(StateSchemaManager(other, config=self.config))

        return merged

    def get_model(self, lock: bool = False) -> Type[BaseModel]:
        if lock:
            self.locked = True
        model = create_model(self.name, **self.fields)

        for name, validator in self.validators.items():
            setattr(model, name, validator)

        for name, prop in self.properties.items():
            setattr(model, name, prop)

        for name, method in self.class_methods.items():
            setattr(model, name, method)

        for name, method in self.static_methods.items():
            setattr(model, name, method)

        return model

    def pretty_print(self) -> None:
        """
        Print the schema as if it was written as a Python class.
        """
        print(f"class {self.name}(BaseModel):")
        if not any([self.fields, self.properties, self.class_methods, self.static_methods]):
            print("    pass  # No fields defined\n")
            return

        for field_name, (field_type, field_info) in self.fields.items():
            type_str = str(field_type).replace("typing.", "").replace("NoneType", "Optional")
            default_val = field_info.default if field_info.default != ... else "..."
            default_str = f" = {default_val}" if default_val != "..." else ""
            print(f"    {field_name}: {type_str}{default_str}")

        for prop_name in self.properties.keys():
            print(f"\n    @property")
            print(f"    def {prop_name}(self): ...  # Computed property")

        for method_name in self.class_methods.keys():
            print(f"\n    @classmethod")
            print(f"    def {method_name}(cls): ...  # Class method")

        for method_name in self.static_methods.keys():
            print(f"\n    @staticmethod")
            print(f"    def {method_name}(): ...  # Static method")

        print()
     # --- New method to integrate node functionality ---
    def create_node_from_config(
        self,
        config: AugLLMConfig,
        command_goto: str = "execute_step",
        async_mode: bool = True
    ) -> Callable[[Dict[str, Any]], Any]:
        """
        Create a node function from an AugLLMConfig, wrapping it with state validation
        using this schema's base model.
        Returns a node function that accepts a dict, validates it, and then calls the underlying node.
        """
        base_node = create_node_function(config, command_goto, async_mode)
        Model = self.get_model()
        if async_mode:
            async def wrapped_node(state: Dict[str, Any]):
                try:
                    valid_state = Model.model_validate(state).model_dump()
                except ValidationError as e:
                    raise ValueError(f"State validation failed: {e}")
                return await base_node(valid_state)
            return wrapped_node
        else:
            def wrapped_node(state: Dict[str, Any]):
                try:
                    valid_state = Model.model_validate(state).model_dump()
                except ValidationError as e:
                    raise ValueError(f"State validation failed: {e}")
                return base_node(valid_state)
            return wrapped_node

    # --- New helper methods using langgraph.types Command and Send ---
    def create_default_command(
        self,
        update: Optional[Dict[str, Any]] = None,
        goto: Optional[Union[str, Send, List[Union[str, Send]]]] = None,
        resume: Optional[Union[Any, Dict[str, Any]]] = None
    ) -> Command:
        update = update or {}
        goto = goto or ""
        resume = resume or {}
        return Command(update=update, goto=goto, resume=resume)

    def create_send(self, node: str, arg: Any) -> Send:
        return Send(node, arg)

```

```python
from typing import Dict, Any, List, Union, Optional, Type, Callable, Tuple, Literal, Set
from pydantic import BaseModel, Field
import uuid
from functools import partial

from langgraph.graph import StateGraph, END
from langgraph.prebuilt import ToolNode
from langgraph.types import Command, Send
from langchain_core.tools import BaseTool, StructuredTool, Tool
from langchain_core.messages import SystemMessage, filter_messages

from src.haive.core.aug_llm.base import AugLLMConfig, compose_runnable
from src.haive.core.models.llm.base import AzureLLMConfig

# Import our node utilities
#from node_factory_refined import create_node_function, NodeConfig
#from schema_composition import SchemaComposer
from typing import Any, Dict, List, Type, Union, Optional, Callable, Tuple, Literal
from pydantic import BaseModel, Field, create_model
import uuid
import inspect
from langgraph.graph import StateGraph, END
from langgraph.types import Command, Send
from langchain_core.prompts import ChatPromptTemplate
from src.haive.core.aug_llm.base import AugLLMConfig, compose_runnable
from src.haive.core.models.llm.base import AzureLLMConfig


def create_node_function(
    config: AugLLMConfig,
    input_mapping: Optional[Dict[str, str]] = None,
    output_mapping: Optional[Dict[str, str]] = None,
    next_node: Optional[Union[str, Literal["END"]]] = None,
    async_mode: bool = False
):
    """
    Creates a node function from an AugLLMConfig with explicit return type annotations.
    
    Args:
        config: The AugLLM configuration to use for this node
        input_mapping: Maps state fields to AugLLM input fields
        output_mapping: Maps AugLLM output fields to state update fields
        next_node: The next node to route to (if None, will default to END)
        async_mode: Whether to create an async node function
        
    Returns:
        A callable node function for use in a LangGraph with proper type annotations
    """
    # Build the runnable from config
    runnable = compose_runnable(config)
    
    # Default mappings if none provided
    if input_mapping is None:
        # Determine input variables from prompt template if available
        if config.prompt_template:
            input_vars = config.prompt_template.input_variables
            input_mapping = {var: var for var in input_vars}
        else:
            # Default to messages if no prompt template
            input_mapping = {"messages": "messages"}
    
    # Output mapping defaults to empty (direct state updates)
    output_mapping = output_mapping or {}
    
    # Determine target for routing
    goto = next_node if next_node is not None else END
    
    # Create async or sync node function with explicit return annotations
    if async_mode:
        async def node_function(state):
            # Convert Pydantic model to dict if needed
            state_dict = state.model_dump() if hasattr(state, "model_dump") else state
            
            # Prepare input for the runnable based on mapping
            runnable_input = {}
            for state_key, runnable_key in input_mapping.items():
                if state_key in state_dict:
                    runnable_input[runnable_key] = state_dict[state_key]
                else:
                    print(f"Warning: State key '{state_key}' not found in state.")
            
            # Handle empty input case (add debugging)
            if not runnable_input:
                print(f"Warning: Empty input to node. State keys available: {list(state_dict.keys())}")
                print(f"Input mapping: {input_mapping}")
                # Create minimal valid input
                if config.prompt_template and hasattr(config.prompt_template, "input_variables"):
                    for var in config.prompt_template.input_variables:
                        runnable_input[var] = ""
            
            # Invoke the runnable
            result = await runnable.ainvoke(runnable_input)
            
            # Prepare state update
            update = {}
            
            # Special handling for message results (common case)
            if "messages" in state_dict and hasattr(result, "content"):
                update["messages"] = state_dict["messages"] + [result]
            elif "messages" in state_dict and isinstance(result, tuple) and len(result) == 2 and result[0] in ("ai", "assistant"):
                update["messages"] = state_dict["messages"] + [result]
            elif isinstance(result, dict):
                # Handle structured output mapping
                for output_key, state_key in output_mapping.items():
                    if output_key in result:
                        update[state_key] = result[output_key]
                
                # Default behavior: map any matching keys
                for key, value in result.items():
                    if key not in output_mapping and key in state_dict:
                        update[key] = value
            else:
                # Direct mapping for simple outputs
                for key in state_dict:
                    if key in output_mapping:
                        update[output_mapping[key]] = result
            
            # Always return a Command with explicit goto
            return Command(update=update, goto=goto)
        
        return node_function
    else:
        def node_function(state):
            # Convert Pydantic model to dict if needed
            state_dict = state.model_dump() if hasattr(state, "model_dump") else state
            
            # Prepare input for the runnable based on mapping
            runnable_input = {}
            for state_key, runnable_key in input_mapping.items():
                if state_key in state_dict:
                    runnable_input[runnable_key] = state_dict[state_key]
                else:
                    print(f"Warning: State key '{state_key}' not found in state.")
            
            # Handle empty input case
            if not runnable_input:
                print(f"Warning: Empty input to node. State keys available: {list(state_dict.keys())}")
                print(f"Input mapping: {input_mapping}")
                # Create minimal valid input
                if config.prompt_template and hasattr(config.prompt_template, "input_variables"):
                    for var in config.prompt_template.input_variables:
                        runnable_input[var] = ""
            
            # Invoke the runnable
            result = runnable.invoke(runnable_input)
            
            # Prepare state update
            update = {}
            
            # Special handling for message results (common case)
            if "messages" in state_dict and hasattr(result, "content"):
                update["messages"] = state_dict["messages"] + [result]
            elif "messages" in state_dict and isinstance(result, tuple) and len(result) == 2 and result[0] in ("ai", "assistant"):
                update["messages"] = state_dict["messages"] + [result]
            elif isinstance(result, dict):
                # Handle structured output mapping
                for output_key, state_key in output_mapping.items():
                    if output_key in result:
                        update[state_key] = result[output_key]
                
                # Default behavior: map any matching keys
                for key, value in result.items():
                    if key not in output_mapping and key in state_dict:
                        update[key] = value
            else:
                # Direct mapping for simple outputs
                for key in state_dict:
                    if key in output_mapping:
                        update[output_mapping[key]] = result
            
            # Always return a Command with explicit goto
            return Command(update=update, goto=goto)
        
        return node_function


def augment_schema_for_llm(
    schema_manager,  # StateSchemaManager instance
    aug_llm_config: AugLLMConfig
) -> Any:  # Returns StateSchemaManager
    """
    Augments a state schema to accommodate the inputs and outputs of an AugLLM.
    
    Args:
        schema_manager: The state schema manager to augment
        aug_llm_config: The AugLLM configuration
        
    Returns:
        Updated schema manager
    """
    # Ensure we have required fields
    if aug_llm_config.prompt_template:
        input_vars = aug_llm_config.prompt_template.input_variables
        for var in input_vars:
            if var == "messages" and "messages" not in schema_manager.fields:
                # Add messages field with appropriate annotation
                from typing import List, Tuple, Union
                from langchain_core.messages import BaseMessage
                
                schema_manager.add_field(
                    "messages", 
                    List[Union[Tuple[str, str], BaseMessage]], 
                    default=[]
                )
    
    # Add field for structured output if specified
    if aug_llm_config.structured_output_model:
        model_class = aug_llm_config.structured_output_model
        field_name = model_class.__name__.lower()
        
        # Check if the field already exists
        if field_name not in schema_manager.fields:
            # Add field with None default to ensure it's treated as Optional
            schema_manager.add_field(field_name, Optional[model_class], default=None)
    
    return schema_manager


class NodeConfig(BaseModel):
    """Configuration for a node in a LangGraph."""
    name: str = Field(default="default_node")
    engine: AugLLMConfig
    input_mapping: Optional[Dict[str, str]] = None
    output_mapping: Optional[Dict[str, str]] = None
    routing_logic: Optional[Callable] = None
    next_node: Optional[Union[str, Literal["END"]]] = None
    
    def create_node(self) -> Callable:
        """Creates a callable node function based on this configuration."""
        return create_node_function(
            config=self.engine,
            input_mapping=self.input_mapping,
            output_mapping=self.output_mapping,
            next_node=self.next_node
        )


class GraphBuilder:
    """Builder for constructing LangGraph graphs with explicit routing."""
    
    def __init__(self, state_schema):
        """
        Initialize with a state schema.
        
        Args:
            state_schema: Can be a StateSchemaManager, BaseModel, or dict
        """
        # Handle different schema input types
        if hasattr(state_schema, "get_model"):
            # It's a StateSchemaManager
            self.schema_manager = state_schema
            self.state_schema = self.schema_manager.get_model()
        elif isinstance(state_schema, dict):
            # It's a dict, convert to model
            #from state_schema_manager import StateSchemaManager
            self.schema_manager = StateSchemaManager(state_schema)
            self.state_schema = self.schema_manager.get_model()
        elif isinstance(state_schema, type) and issubclass(state_schema, BaseModel):
            # It's already a Pydantic model class
            self.schema_manager = None
            self.state_schema = state_schema
        else:
            # Assume it's a model instance or class
            self.schema_manager = None
            self.state_schema = state_schema
        
        # Initialize the graph
        self.graph = StateGraph(self.state_schema)
        self.nodes = {}
        self.entry_point = None
        
    def add_node(self, 
                 node_config: Union[NodeConfig, AugLLMConfig], 
                 name: Optional[str] = None,
                 next_node: Optional[Union[str, Literal["END"]]] = None):
        """
        Add a node to the graph with explicit next node.
        
        Args:
            node_config: Either a NodeConfig or AugLLMConfig
            name: Optional name for the node (required if using AugLLMConfig)
            next_node: Where this node should route to next (END by default)
            
        Returns:
            self for method chaining
        """
        # Convert AugLLMConfig to NodeConfig if needed
        if isinstance(node_config, AugLLMConfig):
            node_name = name or node_config.name or f"node_{len(self.nodes)}"
            node_config = NodeConfig(
                name=node_name,
                engine=node_config,
                next_node=next_node
            )
        else:
            # If next_node was explicitly provided, override the config
            if next_node is not None:
                node_config.next_node = next_node
                
        # Create and add the node function
        node_fn = node_config.create_node()
        self.nodes[node_config.name] = node_fn
        self.graph.add_node(node_config.name, node_fn)
        
        # Set as entry point if it's the first node
        if self.entry_point is None:
            self.entry_point = node_config.name
            self.graph.set_entry_point(node_config.name)
        
        return self
    
    def add_edge(self, from_node: str, to_node: Union[str, Literal["END"]]):
        """
        Add an explicit edge between nodes, handling END specially.
        
        Args:
            from_node: The source node name
            to_node: The target node name or "END"
            
        Returns:
            self for method chaining
        """
        # Don't add explicit edges to END as they're handled via the node function
        if to_node != END:
            self.graph.add_edge(from_node, to_node)
        return self
    
    def add_conditional_edge(self, 
                           from_node: str, 
                           condition_fn: Callable, 
                           edges_map: Dict[str, Union[str, Literal["END"]]]):
        """
        Add conditional edges based on a routing function.
        
        Args:
            from_node: The source node name  
            condition_fn: Function that takes state and returns a routing key
            edges_map: Mapping from routing keys to target nodes
            
        Returns:
            self for method chaining
        """
        self.graph.add_conditional_edges(from_node, condition_fn, edges_map)
        return self
    
    def build(self, checkpointer=None):
        """
        Compile the graph into an executable app.
        
        Args:
            checkpointer: Optional checkpoint saver
            
        Returns:
            Compiled graph application
        """
        if self.entry_point is None:
            raise ValueError("No entry point defined - add at least one node")
        
        return self.graph.compile(checkpointer=checkpointer)


def build_graph_from_configs(
    configs: Dict[str, AugLLMConfig],
    edges: List[Tuple[str, Union[str, Literal["END"]]]],
    conditional_edges: Optional[List[Tuple[str, Callable, Dict[str, Union[str, Literal["END"]]]]]] = None,
    state_schema=None,  # StateSchemaManager or BaseModel or dict
    entry_point: Optional[str] = None
):
    """
    Builds a LangGraph from AugLLM configurations with explicit routing.
    
    Args:
        configs: Dictionary mapping node names to AugLLM configs
        edges: List of (from_node, to_node) tuples defining graph edges
        conditional_edges: List of (from_node, condition_fn, edge_map) for conditional routing
        state_schema: Base state schema to use
        entry_point: Name of the node to use as entry point (defaults to first node)
        
    Returns:
        Compiled LangGraph application
    """
    # Import only where needed
    #from state_schema_manager import StateSchemaManager
    
    # Initialize state schema
    if state_schema is None:
        schema_manager = StateSchemaManager({"messages": []})
    elif isinstance(state_schema, dict):
        schema_manager = StateSchemaManager(state_schema)
    elif hasattr(state_schema, "get_model"):
        schema_manager = state_schema
    else:
        schema_manager = StateSchemaManager(state_schema)
    
    # Augment schema for all configs
    for node_name, config in configs.items():
        schema_manager = augment_schema_for_llm(schema_manager, config)
    
    # Get the finalized state model
    state_model = schema_manager.get_model()
    
    # Build the graph using the model, not the manager
    builder = GraphBuilder(state_model)
    
    # Add nodes with their next destinations based on edges
    for node_name, config in configs.items():
        # Find the explicit next node from edges
        next_nodes = [to_node for from_node, to_node in edges if from_node == node_name]
        next_node = next_nodes[0] if len(next_nodes) == 1 else END
        
        # Add the node with explicit next_node
        builder.add_node(config, name=node_name, next_node=next_node)
    
    # Set entry point
    if entry_point:
        builder.graph.set_entry_point(entry_point)
    
    # Add conditional edges
    if conditional_edges:
        for from_node, condition_fn, edge_map in conditional_edges:
            builder.add_conditional_edge(from_node, condition_fn, edge_map)
    
    # Compile graph
    return builder.build()


def create_agent_from_aug_llm(
    aug_llm_config: AugLLMConfig,
    state_schema=None  # Can be StateSchemaManager, BaseModel, or dict
):
    """
    Create a simple agent from an AugLLM config with explicit END routing.
    
    Args:
        aug_llm_config: The AugLLM configuration
        state_schema: Optional state schema
        
    Returns:
        Compiled graph application
    """
    # Import only where needed
    #from state_schema_manager import StateSchemaManager
    
    # Initialize schema
    if state_schema is None:
        schema_manager = StateSchemaManager({"messages": []})
    elif isinstance(state_schema, dict):
        schema_manager = StateSchemaManager(state_schema)
    elif hasattr(state_schema, "get_model"):
        schema_manager = state_schema
    else:
        schema_manager = StateSchemaManager(state_schema)
    
    # Augment schema for this config
    schema_manager = augment_schema_for_llm(schema_manager, aug_llm_config)
    
    # Create a node configuration with explicit END routing
    node_config = NodeConfig(
        name="agent_node",
        engine=aug_llm_config,
        next_node=END  # Explicit END routing
    )
    
    # Build the graph
    builder = GraphBuilder(schema_manager)
    builder.add_node(node_config)
    
    # Build and return the executable
    return builder.build()

class GraphManager:
    """
    A flexible manager for building and configuring LangGraph workflows.
    
    Features:
    - Create nodes from AugLLM configs with automatic schema derivation
    - Add generalized tool nodes with post-processing
    - Create conditional branching with custom routing functions
    - Build sequential workflows with automatic END routing
    """
    
    def __init__(self, state_schema=None):
        """
        Initialize the graph manager.
        
        Args:
            state_schema: Optional initial state schema (BaseModel, StateSchemaManager, or dict)
        """
        # Import here to avoid circular imports
        #from state_schema_manager import StateSchemaManager

        # Initialize state schema
        if state_schema is None:
            self.schema_manager = StateSchemaManager({"messages": []})
        elif isinstance(state_schema, dict):
            self.schema_manager = StateSchemaManager(state_schema)
        elif hasattr(state_schema, "get_model"):
            self.schema_manager = state_schema
        else:
            self.schema_manager = StateSchemaManager(state_schema)
            
        # Build the state model
        self.state_model = self.schema_manager.get_model()
        
        # Initialize graph
        self.graph = StateGraph(self.state_model)
        
        # Track nodes, tools, and edges
        self.nodes = {}
        self._tools = {}
        self._edges = []
        self._conditional_edges = []
        self.entry_point = None
        
    def add_node_from_aug_llm(
        self, 
        name: str, 
        config: AugLLMConfig,
        next_node: Optional[Union[str, Literal["END"]]] = None
    ) -> 'GraphManager':
        """
        Add a node from an AugLLM config, with automatic schema derivation.
        
        Args:
            name: Name for the node
            config: AugLLM configuration
            next_node: Optional next node to route to
            
        Returns:
            Self for chaining
        """
        # Derive schema from config and update our schema
        self.schema_manager = SchemaComposer.derive_schema_from_aug_llm(
            config, 
            self.schema_manager
        )
        
        # Rebuild state model with updated schema
        self.state_model = self.schema_manager.get_model()
        self.graph = StateGraph(self.state_model)
        
        # Rebuild existing nodes with new schema
        self._rebuild_existing_nodes()
        
        # Create node function with next_node
        node_fn = create_node_function(
            config=config,
            next_node=next_node
        )
        
        # Add to graph
        self.graph.add_node(name, node_fn)
        self.nodes[name] = node_fn
        
        # Set as entry point if it's the first node
        if self.entry_point is None:
            self.entry_point = name
            self.graph.set_entry_point(name)
            
        # Add edge if next_node is specified
        if next_node is not None and next_node != END:
            self._edges.append((name, next_node))
            self.graph.add_edge(name, next_node)
            
        return self
        
    def add_node(
        self, 
        name: str, 
        node_fn: Callable,
        next_node: Optional[Union[str, Literal["END"]]] = None
    ) -> 'GraphManager':
        """
        Add a custom node function.
        
        Args:
            name: Name for the node
            node_fn: Node function
            next_node: Optional next node to route to
            
        Returns:
            Self for chaining
        """
        # Add to graph
        self.graph.add_node(name, node_fn)
        self.nodes[name] = node_fn
        
        # Set as entry point if it's the first node
        if self.entry_point is None:
            self.entry_point = name
            self.graph.set_entry_point(name)
            
        # Add edge if next_node is specified
        if next_node is not None and next_node != END:
            self._edges.append((name, next_node))
            self.graph.add_edge(name, next_node)
            
        return self
        
    def create_generalized_tool_node(
        self,
        name: str,
        tools: List[Any],
        post_processing_fn: Optional[Callable] = None,
        next_node: Optional[Union[str, Literal["END"]]] = None
    ) -> 'GraphManager':
        """
        Create a tool node with optional post-processing.
        
        Args:
            name: Name for the node
            tools: List of tools for this node
            post_processing_fn: Optional function to process tool outputs
            next_node: Optional next node to route to
            
        Returns:
            Self for chaining
        """
        # Define tool node function
        def tool_node_with_processing(state):
            # Convert Pydantic model to dict if needed
            state_dict = state.model_dump() if hasattr(state, "model_dump") else state
            
            # Create standard tool node
            tool_node = ToolNode(tools)
            
            # Process the state
            result = tool_node.invoke(state_dict)
            
            # Apply post-processing if provided
            if post_processing_fn:
                processed_result = post_processing_fn(result)
                if isinstance(processed_result, Command):
                    return processed_result
                return Command(update=processed_result, goto=next_node or "")
            
            # If no processing or processing didn't return a Command
            if next_node:
                return Command(update=result, goto=next_node)
            return Command(update=result)
            
        # Add to graph
        self.add_node(name, tool_node_with_processing, next_node)
        
        # Track tools
        self._tools[name] = tools
        
        return self
        
    def add_edge(
        self, 
        from_node: str, 
        to_node: Union[str, Literal["END"]]
    ) -> 'GraphManager':
        """
        Add an edge between nodes.
        
        Args:
            from_node: Source node name
            to_node: Target node name or END
            
        Returns:
            Self for chaining
        """
        # Skip if to_node is END (handled by node function)
        if to_node != END:
            self._edges.append((from_node, to_node))
            self.graph.add_edge(from_node, to_node)
            
        return self
        
    def add_conditional_edge(
        self,
        from_node: str,
        condition_fn: Callable,
        edge_map: Dict[str, Union[str, Literal["END"]]]
    ) -> 'GraphManager':
        """
        Add conditional edges based on a routing function.
        
        Args:
            from_node: Source node name
            condition_fn: Function that determines routing
            edge_map: Mapping from condition results to target nodes
            
        Returns:
            Self for chaining
        """
        self._conditional_edges.append((from_node, condition_fn, edge_map))
        self.graph.add_conditional_edges(from_node, condition_fn, edge_map)
        return self
        
    def create_conditional_tool_routing(
        self,
        from_node: str,
        condition_fn: Callable,
        tool_routes: Dict[str, Tuple[List[Any], Optional[Callable], Optional[str]]]
    ) -> 'GraphManager':
        """
        Create conditional routing to different tool nodes based on a condition.
        
        Args:
            from_node: Source node name
            condition_fn: Function that returns a routing key
            tool_routes: Dict mapping routing keys to (tools, post_processor, next_node)
            
        Returns:
            Self for chaining
        """
        # Create tool nodes for each route
        edge_map = {}
        
        for route_key, (tools, processor, next_node) in tool_routes.items():
            # Create unique tool node name
            tool_node_name = f"{from_node}_{route_key}_tools"
            
            # Create the tool node
            self.create_generalized_tool_node(
                name=tool_node_name,
                tools=tools,
                post_processing_fn=processor,
                next_node=next_node
            )
            
            # Add to edge map
            edge_map[route_key] = tool_node_name
            
        # Add conditional edge
        self.add_conditional_edge(from_node, condition_fn, edge_map)
        
        return self
        
    def create_sequential_chain(
        self,
        configs: List[Union[AugLLMConfig, Tuple[str, Callable]]],
        names: Optional[List[str]] = None
    ) -> 'GraphManager':
        """
        Create a sequential chain of nodes with automatic routing.
        
        Args:
            configs: List of AugLLM configs or (name, function) tuples
            names: Optional list of names for AugLLM configs
            
        Returns:
            Self for chaining
        """
        # Track nodes to add edges between them
        chain_nodes = []
        
        # Process each configuration or function
        for i, config_or_fn in enumerate(configs):
            if isinstance(config_or_fn, AugLLMConfig):
                # It's an AugLLM config
                config = config_or_fn
                
                # Determine node name
                if names and i < len(names):
                    node_name = names[i]
                else:
                    node_name = config.name or f"node_{i}"
                    
                # Determine next node
                if i < len(configs) - 1:
                    # Not the last node - will connect to next
                    next_node = None  # We'll add edges later
                else:
                    # Last node - route to END
                    next_node = END
                    
                # Add the node
                self.add_node_from_aug_llm(node_name, config, next_node)
                chain_nodes.append(node_name)
                
            elif isinstance(config_or_fn, tuple) and len(config_or_fn) == 2:
                # It's a (name, function) tuple
                node_name, node_fn = config_or_fn
                
                # Determine next node
                if i < len(configs) - 1:
                    # Not the last node - will connect to next
                    next_node = None  # We'll add edges later
                else:
                    # Last node - route to END
                    next_node = END
                    
                # Add the node
                self.add_node(node_name, node_fn, next_node)
                chain_nodes.append(node_name)
                
            else:
                raise ValueError(f"Unsupported config type: {type(config_or_fn)}")
                
        # Add edges between nodes
        for i in range(len(chain_nodes) - 1):
            self.add_edge(chain_nodes[i], chain_nodes[i + 1])
            
        return self
    
    def set_entry_point(self, node_name: str) -> 'GraphManager':
        """
        Set the entry point for the graph.
        
        Args:
            node_name: Name of the entry point node
            
        Returns:
            Self for chaining
        """
        self.entry_point = node_name
        self.graph.set_entry_point(node_name)
        return self
    
    def build(self, checkpointer=None):
        """
        Build and compile the graph.
        
        Args:
            checkpointer: Optional checkpoint saver
            
        Returns:
            Compiled graph application
        """
        if self.entry_point is None:
            raise ValueError("No entry point defined - add at least one node")
            
        return self.graph.compile(checkpointer=checkpointer)
    
    def _rebuild_existing_nodes(self):
        """Rebuild existing nodes after schema changes."""
        for name, node_fn in self.nodes.items():
            self.graph.add_node(name, node_fn)
            
        for from_node, to_node in self._edges:
            if to_node != END:
                self.graph.add_edge(from_node, to_node)
                
        for from_node, condition_fn, edge_map in self._conditional_edges:
            self.graph.add_conditional_edges(from_node, condition_fn, edge_map)
            
        if self.entry_point:
            self.graph.set_entry_point(self.entry_point)


# Common routing patterns

def message_contains_routing(state, phrase_to_route_map: Dict[str, str], default_route: str = "default"):
    """
    Route based on phrases in the last message.
    
    Args:
        state: The state object
        phrase_to_route_map: Map of phrases to route keys
        default_route: Default route if no phrase matches
        
    Returns:
        Route key
    """
    # Get last message content
    messages = filter_messages(state.messages, exclude_types=[SystemMessage])
    if not messages:
        return default_route
        
    last_message = messages[-1]
    if not hasattr(last_message, "content"):
        return default_route
        
    content = last_message.content.lower()
    
    # Check for phrases
    for phrase, route in phrase_to_route_map.items():
        if phrase.lower() in content:
            return route
            
    return default_route


def tool_name_routing(state, tool_name_to_route_map: Dict[str, str], default_route: str = "end"):
    """
    Route based on the tool name in tool calls.
    
    Args:
        state: The state object
        tool_name_to_route_map: Map of tool names to route keys
        default_route: Default route if no tool calls or no match
        
    Returns:
        Route key
    """
    # Get last message
    messages = filter_messages(state.messages, exclude_types=[SystemMessage])
    if not messages:
        return default_route
        
    last_message = messages[-1]
    
    # Check for tool calls
    if hasattr(last_message, "tool_calls") and last_message.tool_calls:
        for tool_call in last_message.tool_calls:
            tool_name = tool_call.get("name")
            if tool_name in tool_name_to_route_map:
                return tool_name_to_route_map[tool_name]
                
    return default_route


# Example usage:

def example_retrieval_workflow():
    """Example of creating a workflow with conditional retrieval."""
    from langchain_core.prompts import ChatPromptTemplate
    
    # Create AugLLM configs
    planner_config = AugLLMConfig(
        name="planner",
        prompt_template=ChatPromptTemplate.from_messages([
            ("system", "You are a research assistant. If you need to search for information, say 'I need to search for [topic]'."),
            ("user", "{messages}")
        ]),
        llm_config=AzureLLMConfig(model="gpt-4o")
    )
    
    synthesizer_config = AugLLMConfig(
        name="synthesizer",
        prompt_template=ChatPromptTemplate.from_messages([
            ("system", "You are a synthesis assistant. Synthesize the information provided into a coherent response."),
            ("user", "Original query: {original_query}\nSearch results: {search_results}\nCreate a comprehensive response.")
        ]),
        llm_config=AzureLLMConfig(model="gpt-4o")
    )
    
    # Define search tools
    search_tools = [
        Tool(
            name="search",
            description="Search for information on a topic",
            func=lambda x: f"Search results for {x}: [simulated search results]"
        )
    ]
    
    # Define routing function
    def needs_search(state):
        messages = filter_messages(state["messages"], exclude_types=[SystemMessage])
        if not messages:
            return "no_search"
            
        last_message = messages[-1]
        if hasattr(last_message, "content") and "I need to search for" in last_message.content:
            return "use_search"
        return "no_search"
    
    # Create post-processing function
    def process_search_results(state):
        # Extract original query
        original_query = "No query found"
        for msg in state["messages"]:
            if hasattr(msg, "role") and msg.role == "user":
                original_query = msg.content
                break
        
        # Extract search results - assuming last tool call output
        search_results = "No results found"
        for msg in state["messages"]:
            if hasattr(msg, "tool_call_id"):
                search_results = msg.content
        
        # Update state with these fields
        return {
            "messages": state["messages"],
            "original_query": original_query,
            "search_results": search_results
        }
    
    # Create the graph manager
    manager = GraphManager({
        "messages": [],
        "original_query": "",
        "search_results": ""
    })
    
    # Add planner node with conditional routing
    manager.add_node_from_aug_llm("planner", planner_config)
    
    # Create search tool node with post-processing
    manager.create_generalized_tool_node(
        name="search_node",
        tools=search_tools,
        post_processing_fn=process_search_results,
        next_node="synthesizer"
    )
    
    # Add synthesizer node
    manager.add_node_from_aug_llm("synthesizer", synthesizer_config, END)
    
    # Add conditional routing
    manager.add_conditional_edge(
        "planner",
        needs_search,
        {
            "use_search": "search_node",
            "no_search": END
        }
    )
    
    # Set entry point
    manager.set_entry_point("planner")
    
    # Build the graph
    app = manager.build()
    
    return app


def example_multi_tool_routing():
    """Example of routing to different tools based on content."""
    from langchain_core.prompts import ChatPromptTemplate
    
    # Create AugLLM config
    agent_config = AugLLMConfig(
        name="agent",
        prompt_template=ChatPromptTemplate.from_messages([
            ("system", "You are a helpful assistant. For calculation questions, say 'I need to calculate [expression]'. For search, say 'I need to search for [topic]'."),
            ("user", "{messages}")
        ]),
        llm_config=AzureLLMConfig(model="gpt-4o")
    )
    
    # Define different tools
    search_tools = [
        Tool(
            name="search",
            description="Search for information",
            func=lambda x: f"Search results for {x}: [simulated results]"
        )
    ]
    
    calculator_tools = [
        Tool(
            name="calculator",
            description="Perform calculations",
            func=lambda x: f"Result: {eval(x)}"
        )
    ]
    
    # Create routing conditions
    def route_by_need(state):
        messages = filter_messages(state["messages"], exclude_types=[SystemMessage])
        if not messages:
            return "default"
            
        last_message = messages[-1]
        content = last_message.content.lower() if hasattr(last_message, "content") else ""
        
        if "need to calculate" in content:
            return "calculate"
        elif "need to search" in content:
            return "search"
        return "default"
    
    # Create the graph manager
    manager = GraphManager()
    
    # Add agent node
    manager.add_node_from_aug_llm("agent", agent_config)
    
    # Create conditional tool routing
    manager.create_conditional_tool_routing(
        from_node="agent",
        condition_fn=route_by_need,
        tool_routes={
            "search": (search_tools, None, "agent"),
            "calculate": (calculator_tools, None, "agent"),
            "default": ([], None, END)
        }
    )
    
    # Set entry point
    manager.set_entry_point("agent")
    
    # Build the graph
    app = manager.build()
    
    return app
```

```python
from typing import Any, Dict, List, Type, Union, Optional, Callable, Tuple, Literal
from pydantic import BaseModel, Field, create_model
import uuid
import inspect
from langgraph.graph import StateGraph, END
from langgraph.types import Command, Send
from langchain_core.prompts import ChatPromptTemplate
from src.haive.core.aug_llm.base import AugLLMConfig, compose_runnable
from src.haive.core.models.llm.base import AzureLLMConfig


def create_node_function(
    config: AugLLMConfig,
    input_mapping: Optional[Dict[str, str]] = None,
    output_mapping: Optional[Dict[str, str]] = None,
    next_node: Optional[Union[str, Literal["END"]]] = None,
    async_mode: bool = False
):
    """
    Creates a node function from an AugLLMConfig with explicit return type annotations.
    
    Args:
        config: The AugLLM configuration to use for this node
        input_mapping: Maps state fields to AugLLM input fields
        output_mapping: Maps AugLLM output fields to state update fields
        next_node: The next node to route to (if None, will default to END)
        async_mode: Whether to create an async node function
        
    Returns:
        A callable node function for use in a LangGraph with proper type annotations
    """
    # Build the runnable from config
    runnable = compose_runnable(config)
    
    # Default mappings if none provided
    if input_mapping is None:
        # Determine input variables from prompt template if available
        if config.prompt_template:
            input_vars = config.prompt_template.input_variables
            input_mapping = {var: var for var in input_vars}
        else:
            # Default to messages if no prompt template
            input_mapping = {"messages": "messages"}
    
    # Output mapping defaults to empty (direct state updates)
    output_mapping = output_mapping or {}
    
    # Determine target for routing
    goto = next_node if next_node is not None else END
    
    # Create async or sync node function with explicit return annotations
    if async_mode:
        async def node_function(state):
            # Convert Pydantic model to dict if needed
            state_dict = state.model_dump() if hasattr(state, "model_dump") else state
            
            # Prepare input for the runnable based on mapping
            runnable_input = {}
            for state_key, runnable_key in input_mapping.items():
                if state_key in state_dict:
                    runnable_input[runnable_key] = state_dict[state_key]
                else:
                    print(f"Warning: State key '{state_key}' not found in state.")
            
            # Handle empty input case (add debugging)
            if not runnable_input:
                print(f"Warning: Empty input to node. State keys available: {list(state_dict.keys())}")
                print(f"Input mapping: {input_mapping}")
                # Create minimal valid input
                if config.prompt_template and hasattr(config.prompt_template, "input_variables"):
                    for var in config.prompt_template.input_variables:
                        runnable_input[var] = ""
            
            # Invoke the runnable
            result = await runnable.ainvoke(runnable_input)
            
            # Prepare state update
            update = {}
            
            # Special handling for message results (common case)
            if "messages" in state_dict and hasattr(result, "content"):
                update["messages"] = state_dict["messages"] + [result]
            elif "messages" in state_dict and isinstance(result, tuple) and len(result) == 2 and result[0] in ("ai", "assistant"):
                update["messages"] = state_dict["messages"] + [result]
            elif isinstance(result, dict):
                # Handle structured output mapping
                for output_key, state_key in output_mapping.items():
                    if output_key in result:
                        update[state_key] = result[output_key]
                
                # Default behavior: map any matching keys
                for key, value in result.items():
                    if key not in output_mapping and key in state_dict:
                        update[key] = value
            else:
                # Direct mapping for simple outputs
                for key in state_dict:
                    if key in output_mapping:
                        update[output_mapping[key]] = result
            
            # Always return a Command with explicit goto
            return Command(update=update, goto=goto)
        
        return node_function
    else:
        def node_function(state):
            # Convert Pydantic model to dict if needed
            state_dict = state.model_dump() if hasattr(state, "model_dump") else state
            
            # Prepare input for the runnable based on mapping
            runnable_input = {}
            for state_key, runnable_key in input_mapping.items():
                if state_key in state_dict:
                    runnable_input[runnable_key] = state_dict[state_key]
                else:
                    print(f"Warning: State key '{state_key}' not found in state.")
            
            # Handle empty input case
            if not runnable_input:
                print(f"Warning: Empty input to node. State keys available: {list(state_dict.keys())}")
                print(f"Input mapping: {input_mapping}")
                # Create minimal valid input
                if config.prompt_template and hasattr(config.prompt_template, "input_variables"):
                    for var in config.prompt_template.input_variables:
                        runnable_input[var] = ""
            
            # Invoke the runnable
            result = runnable.invoke(runnable_input)
            
            # Prepare state update
            update = {}
            
            # Special handling for message results (common case)
            if "messages" in state_dict and hasattr(result, "content"):
                update["messages"] = state_dict["messages"] + [result]
            elif "messages" in state_dict and isinstance(result, tuple) and len(result) == 2 and result[0] in ("ai", "assistant"):
                update["messages"] = state_dict["messages"] + [result]
            elif isinstance(result, dict):
                # Handle structured output mapping
                for output_key, state_key in output_mapping.items():
                    if output_key in result:
                        update[state_key] = result[output_key]
                
                # Default behavior: map any matching keys
                for key, value in result.items():
                    if key not in output_mapping and key in state_dict:
                        update[key] = value
            else:
                # Direct mapping for simple outputs
                for key in state_dict:
                    if key in output_mapping:
                        update[output_mapping[key]] = result
            
            # Always return a Command with explicit goto
            return Command(update=update, goto=goto)
        
        return node_function


def augment_schema_for_llm(
    schema_manager,  # StateSchemaManager instance
    aug_llm_config: AugLLMConfig
) -> Any:  # Returns StateSchemaManager
    """
    Augments a state schema to accommodate the inputs and outputs of an AugLLM.
    
    Args:
        schema_manager: The state schema manager to augment
        aug_llm_config: The AugLLM configuration
        
    Returns:
        Updated schema manager
    """
    # Ensure we have required fields
    if aug_llm_config.prompt_template:
        input_vars = aug_llm_config.prompt_template.input_variables
        for var in input_vars:
            if var == "messages" and "messages" not in schema_manager.fields:
                # Add messages field with appropriate annotation
                from typing import List, Tuple, Union
                from langchain_core.messages import BaseMessage
                
                schema_manager.add_field(
                    "messages", 
                    List[Union[Tuple[str, str], BaseMessage]], 
                    default=[]
                )
    
    # Add field for structured output if specified
    if aug_llm_config.structured_output_model:
        model_class = aug_llm_config.structured_output_model
        field_name = model_class.__name__.lower()
        
        # Check if the field already exists
        if field_name not in schema_manager.fields:
            # Add field with None default to ensure it's treated as Optional
            schema_manager.add_field(field_name, Optional[model_class], default=None)
    
    return schema_manager


class NodeConfig(BaseModel):
    """Configuration for a node in a LangGraph."""
    name: str = Field(default="default_node")
    engine: AugLLMConfig
    input_mapping: Optional[Dict[str, str]] = None
    output_mapping: Optional[Dict[str, str]] = None
    routing_logic: Optional[Callable] = None
    next_node: Optional[Union[str, Literal["END"]]] = None
    
    def create_node(self) -> Callable:
        """Creates a callable node function based on this configuration."""
        return create_node_function(
            config=self.engine,
            input_mapping=self.input_mapping,
            output_mapping=self.output_mapping,
            next_node=self.next_node
        )


class GraphBuilder:
    """Builder for constructing LangGraph graphs with explicit routing."""
    
    def __init__(self, state_schema):
        """
        Initialize with a state schema.
        
        Args:
            state_schema: Can be a StateSchemaManager, BaseModel, or dict
        """
        # Handle different schema input types
        if hasattr(state_schema, "get_model"):
            # It's a StateSchemaManager
            self.schema_manager = state_schema
            self.state_schema = self.schema_manager.get_model()
        elif isinstance(state_schema, dict):
            # It's a dict, convert to model
            #from state_schema_manager import StateSchemaManager
            self.schema_manager = StateSchemaManager(state_schema)
            self.state_schema = self.schema_manager.get_model()
        elif isinstance(state_schema, type) and issubclass(state_schema, BaseModel):
            # It's already a Pydantic model class
            self.schema_manager = None
            self.state_schema = state_schema
        else:
            # Assume it's a model instance or class
            self.schema_manager = None
            self.state_schema = state_schema
        
        # Initialize the graph
        self.graph = StateGraph(self.state_schema)
        self.nodes = {}
        self.entry_point = None
        
    def add_node(self, 
                 node_config: Union[NodeConfig, AugLLMConfig], 
                 name: Optional[str] = None,
                 next_node: Optional[Union[str, Literal["END"]]] = None):
        """
        Add a node to the graph with explicit next node.
        
        Args:
            node_config: Either a NodeConfig or AugLLMConfig
            name: Optional name for the node (required if using AugLLMConfig)
            next_node: Where this node should route to next (END by default)
            
        Returns:
            self for method chaining
        """
        # Convert AugLLMConfig to NodeConfig if needed
        if isinstance(node_config, AugLLMConfig):
            node_name = name or node_config.name or f"node_{len(self.nodes)}"
            node_config = NodeConfig(
                name=node_name,
                engine=node_config,
                next_node=next_node
            )
        else:
            # If next_node was explicitly provided, override the config
            if next_node is not None:
                node_config.next_node = next_node
                
        # Create and add the node function
        node_fn = node_config.create_node()
        self.nodes[node_config.name] = node_fn
        self.graph.add_node(node_config.name, node_fn)
        
        # Set as entry point if it's the first node
        if self.entry_point is None:
            self.entry_point = node_config.name
            self.graph.set_entry_point(node_config.name)
        
        return self
    
    def add_edge(self, from_node: str, to_node: Union[str, Literal["END"]]):
        """
        Add an explicit edge between nodes, handling END specially.
        
        Args:
            from_node: The source node name
            to_node: The target node name or "END"
            
        Returns:
            self for method chaining
        """
        # Don't add explicit edges to END as they're handled via the node function
        if to_node != END:
            self.graph.add_edge(from_node, to_node)
        return self
    
    def add_conditional_edge(self, 
                           from_node: str, 
                           condition_fn: Callable, 
                           edges_map: Dict[str, Union[str, Literal["END"]]]):
        """
        Add conditional edges based on a routing function.
        
        Args:
            from_node: The source node name  
            condition_fn: Function that takes state and returns a routing key
            edges_map: Mapping from routing keys to target nodes
            
        Returns:
            self for method chaining
        """
        self.graph.add_conditional_edges(from_node, condition_fn, edges_map)
        return self
    
    def build(self, checkpointer=None):
        """
        Compile the graph into an executable app.
        
        Args:
            checkpointer: Optional checkpoint saver
            
        Returns:
            Compiled graph application
        """
        if self.entry_point is None:
            raise ValueError("No entry point defined - add at least one node")
        
        return self.graph.compile(checkpointer=checkpointer)


def build_graph_from_configs(
    configs: Dict[str, AugLLMConfig],
    edges: List[Tuple[str, Union[str, Literal["END"]]]],
    conditional_edges: Optional[List[Tuple[str, Callable, Dict[str, Union[str, Literal["END"]]]]]] = None,
    state_schema=None,  # StateSchemaManager or BaseModel or dict
    entry_point: Optional[str] = None
):
    """
    Builds a LangGraph from AugLLM configurations with explicit routing.
    
    Args:
        configs: Dictionary mapping node names to AugLLM configs
        edges: List of (from_node, to_node) tuples defining graph edges
        conditional_edges: List of (from_node, condition_fn, edge_map) for conditional routing
        state_schema: Base state schema to use
        entry_point: Name of the node to use as entry point (defaults to first node)
        
    Returns:
        Compiled LangGraph application
    """
    # Import only where needed
    #from state_schema_manager import StateSchemaManager
    
    # Initialize state schema
    if state_schema is None:
        schema_manager = StateSchemaManager({"messages": []})
    elif isinstance(state_schema, dict):
        schema_manager = StateSchemaManager(state_schema)
    elif hasattr(state_schema, "get_model"):
        schema_manager = state_schema
    else:
        schema_manager = StateSchemaManager(state_schema)
    
    # Augment schema for all configs
    for node_name, config in configs.items():
        schema_manager = augment_schema_for_llm(schema_manager, config)
    
    # Get the finalized state model
    state_model = schema_manager.get_model()
    
    # Build the graph using the model, not the manager
    builder = GraphBuilder(state_model)
    
    # Add nodes with their next destinations based on edges
    for node_name, config in configs.items():
        # Find the explicit next node from edges
        next_nodes = [to_node for from_node, to_node in edges if from_node == node_name]
        next_node = next_nodes[0] if len(next_nodes) == 1 else END
        
        # Add the node with explicit next_node
        builder.add_node(config, name=node_name, next_node=next_node)
    
    # Set entry point
    if entry_point:
        builder.graph.set_entry_point(entry_point)
    
    # Add conditional edges
    if conditional_edges:
        for from_node, condition_fn, edge_map in conditional_edges:
            builder.add_conditional_edge(from_node, condition_fn, edge_map)
    
    # Compile graph
    return builder.build()


def create_agent_from_aug_llm(
    config: AugLLMConfig,
    state_schema=None  # Can be StateSchemaManager, BaseModel, or dict
):
    """
    Create a simple agent from an AugLLM config with explicit END routing.
    
    Args:
        aug_llm_config: The AugLLM configuration
        state_schema: Optional state schema
        
    Returns:
        Compiled graph application
    """
    # Import only where needed
    #from state_schema_manager import StateSchemaManager
    
    # Initialize schema
    if state_schema is None:
        schema_manager = StateSchemaManager({"messages": []})
    elif isinstance(state_schema, dict):
        schema_manager = StateSchemaManager(state_schema)
    elif hasattr(state_schema, "get_model"):
        schema_manager = state_schema
    else:
        schema_manager = StateSchemaManager(state_schema)
    
    # Augment schema for this config
    schema_manager = augment_schema_for_llm(schema_manager, aug_llm_config)
    
    # Create a node configuration with explicit END routing
    node_config = NodeConfig(
        name="agent_node",
        engine=aug_llm_config,
        next_node=END  # Explicit END routing
    )
    
    # Build the graph
    builder = GraphBuilder(schema_manager)
    builder.add_node(node_config)
    
    # Build and return the executable
    return builder.build()
```

```python
def create_node_function(
    config: AugLLMConfig,
    input_mapping: Optional[Dict[str, str]] = None,
    output_mapping: Optional[Dict[str, str]] = None,
    next_node: Optional[Union[str, Literal["END"]]] = None,
    async_mode: bool = False
):
    """
    Creates a node function from an AugLLMConfig with explicit return type annotations.
    
    Args:
        config: The AugLLM configuration to use for this node
        input_mapping: Maps state fields to AugLLM input fields
        output_mapping: Maps AugLLM output fields to state update fields
        next_node: The next node to route to (if None, will default to END)
        async_mode: Whether to create an async node function
        
    Returns:
        A callable node function for use in a LangGraph with proper type annotations
    """
    # Build the runnable from config
    runnable = compose_runnable(config)
    
    # Default mappings if none provided
    if input_mapping is None:
        # Determine input variables from prompt template if available
        if config.prompt_template:
            input_vars = config.prompt_template.input_variables
            input_mapping = {var: var for var in input_vars}
        else:
            # Default to messages if no prompt template
            input_mapping = {"messages": "messages"}
    
    # Output mapping defaults to empty (direct state updates)
    output_mapping = output_mapping or {}
    
    # Determine target for routing
    goto = next_node if next_node is not None else END
    
    # Create async or sync node function with explicit return annotations
    if async_mode:
        async def node_function(state):
            # Convert Pydantic model to dict if needed
            state_dict = state.model_dump() if hasattr(state, "model_dump") else state
            
            # Prepare input for the runnable based on mapping
            runnable_input = {}
            for state_key, runnable_key in input_mapping.items():
                if state_key in state_dict:
                    runnable_input[runnable_key] = state_dict[state_key]
                else:
                    print(f"DEBUG: State key '{state_key}' not found in state. Available keys: {list(state_dict.keys())}")
            
            # Handle empty input case
            if not runnable_input:
                print(f"DEBUG: Empty input to node. State keys available: {list(state_dict.keys())}")
                print(f"DEBUG: Input mapping: {input_mapping}")
                # Create minimal valid input
                if config.prompt_template and hasattr(config.prompt_template, "input_variables"):
                    for var in config.prompt_template.input_variables:
                        runnable_input[var] = ""
            
            # Debug info before invoking
            print(f"DEBUG: Node input: {runnable_input}")
            
            # Invoke the runnable
            result = await runnable.ainvoke(runnable_input)
            
            # Debug result type
            print(f"DEBUG: Result type: {type(result)}")
            if hasattr(result, "content"):
                print(f"DEBUG: Result has content: {result.content[:100]}...")
            elif isinstance(result, dict):
                print(f"DEBUG: Result keys: {list(result.keys())}")
            else:
                print(f"DEBUG: Result: {str(result)[:100]}...")
            
            # Prepare state update
            update = {}
            
            # IMPORTANT: Handle output mapping based on result type
            # Case 1: Result is a message with content
            if hasattr(result, "content"):
                if output_mapping:
                    # If output mapping exists, map message.content to specified fields
                    for output_key, state_key in output_mapping.items():
                        if output_key == "content":
                            update[state_key] = result.content
                        elif output_key == "*":  # Special case: map entire message
                            update[state_key] = result
                
                # Default behavior: append to messages if exists (no mapping or incomplete mapping)
                if not output_mapping or "content" not in output_mapping:
                    if "messages" in state_dict:
                        update["messages"] = state_dict["messages"] + [result]
            
            # Case 2: Result is a tuple (e.g., ("assistant", "content"))
            elif isinstance(result, tuple) and len(result) == 2:
                if output_mapping:
                    # Map tuple content to specified fields
                    for output_key, state_key in output_mapping.items():
                        if output_key == "content" and result[0] in ("ai", "assistant"):
                            update[state_key] = result[1]
                        elif output_key == "*":  # Special case: map entire tuple
                            update[state_key] = result
                
                # Default behavior: append to messages
                if not output_mapping or "content" not in output_mapping:
                    if "messages" in state_dict:
                        update["messages"] = state_dict["messages"] + [result]
            
            # Case 3: Result is a dictionary (structured output or complex result)
            elif isinstance(result, dict):
                if output_mapping:
                    # Apply output mapping for each key
                    for result_key, result_value in result.items():
                        # If this key is in output mapping, use mapped field name
                        if result_key in output_mapping:
                            update[output_mapping[result_key]] = result_value
                        # Special case: "*" maps the entire result
                        elif "*" in output_mapping:
                            update[output_mapping["*"]] = result
                    
                    # If no keys matched the mapping but we have a "*" mapping
                    if not any(k in result for k in output_mapping) and "*" in output_mapping:
                        update[output_mapping["*"]] = result
                else:
                    # No mapping defined - add all dict fields to state
                    for key, value in result.items():
                        update[key] = value
            
            # Case 4: Direct result (string, number, etc.)
            else:
                if output_mapping:
                    # Apply output mapping for direct result
                    for output_key, state_key in output_mapping.items():
                        if output_key == "*" or output_key == "content":
                            update[state_key] = result
                else:
                    # Default behavior: if messages exists, add as assistant message
                    if "messages" in state_dict:
                        update["messages"] = state_dict["messages"] + [("assistant", str(result))]
            
            # Preserve non-updated state fields (only if no update for that field)
            for key in state_dict:
                if key not in update:
                    update[key] = state_dict[key]
            
            # Debug the update
            print(f"DEBUG: State update keys: {list(update.keys())}")
            
            # Always return a Command with explicit goto
            return Command(update=update, goto=goto)
        
        return node_function
    else:
        def node_function(state):
            # Convert Pydantic model to dict if needed
            state_dict = state.model_dump() if hasattr(state, "model_dump") else state
            
            # Prepare input for the runnable based on mapping
            runnable_input = {}
            for state_key, runnable_key in input_mapping.items():
                if state_key in state_dict:
                    runnable_input[runnable_key] = state_dict[state_key]
                else:
                    print(f"DEBUG: State key '{state_key}' not found in state. Available keys: {list(state_dict.keys())}")
            
            # Handle empty input case
            if not runnable_input:
                print(f"DEBUG: Empty input to node. State keys available: {list(state_dict.keys())}")
                print(f"DEBUG: Input mapping: {input_mapping}")
                # Create minimal valid input
                if config.prompt_template and hasattr(config.prompt_template, "input_variables"):
                    for var in config.prompt_template.input_variables:
                        runnable_input[var] = ""
            
            # Debug info before invoking
            print(f"DEBUG: Node input: {runnable_input}")
            
            # Invoke the runnable
            result = runnable.invoke(runnable_input)
            
            # Debug result type
            print(f"DEBUG: Result type: {type(result)}")
            if hasattr(result, "content"):
                print(f"DEBUG: Result has content: {result.content[:100]}...")
            elif isinstance(result, dict):
                print(f"DEBUG: Result keys: {list(result.keys())}")
            else:
                print(f"DEBUG: Result: {str(result)[:100]}...")
            
            # Prepare state update
            update = {}
            
            # IMPORTANT: Handle output mapping based on result type
            # Case 1: Result is a message with content
            if hasattr(result, "content"):
                if output_mapping:
                    # If output mapping exists, map message.content to specified fields
                    for output_key, state_key in output_mapping.items():
                        if output_key == "content":
                            update[state_key] = result.content
                        elif output_key == "*":  # Special case: map entire message
                            update[state_key] = result
                
                # Default behavior: append to messages if exists (no mapping or incomplete mapping)
                if not output_mapping or "content" not in output_mapping:
                    if "messages" in state_dict:
                        update["messages"] = state_dict["messages"] + [result]
            
            # Case 2: Result is a tuple (e.g., ("assistant", "content"))
            elif isinstance(result, tuple) and len(result) == 2:
                if output_mapping:
                    # Map tuple content to specified fields
                    for output_key, state_key in output_mapping.items():
                        if output_key == "content" and result[0] in ("ai", "assistant"):
                            update[state_key] = result[1]
                        elif output_key == "*":  # Special case: map entire tuple
                            update[state_key] = result
                
                # Default behavior: append to messages
                if not output_mapping or "content" not in output_mapping:
                    if "messages" in state_dict:
                        update["messages"] = state_dict["messages"] + [result]
            
            # Case 3: Result is a dictionary (structured output or complex result)
            elif isinstance(result, dict):
                if output_mapping:
                    # Apply output mapping for each key
                    for result_key, result_value in result.items():
                        # If this key is in output mapping, use mapped field name
                        if result_key in output_mapping:
                            update[output_mapping[result_key]] = result_value
                        # Special case: "*" maps the entire result
                        elif "*" in output_mapping:
                            update[output_mapping["*"]] = result
                    
                    # If no keys matched the mapping but we have a "*" mapping
                    if not any(k in result for k in output_mapping) and "*" in output_mapping:
                        update[output_mapping["*"]] = result
                else:
                    # No mapping defined - add all dict fields to state
                    for key, value in result.items():
                        update[key] = value
            
            # Case 4: Direct result (string, number, etc.)
            else:
                if output_mapping:
                    # Apply output mapping for direct result
                    for output_key, state_key in output_mapping.items():
                        if output_key == "*" or output_key == "content":
                            update[state_key] = result
                else:
                    # Default behavior: if messages exists, add as assistant message
                    if "messages" in state_dict:
                        update["messages"] = state_dict["messages"] + [("assistant", str(result))]
            
            # Preserve non-updated state fields (only if no update for that field)
            for key in state_dict:
                if key not in update:
                    update[key] = state_dict[key]
            
            # Debug the update
            print(f"DEBUG: State update keys: {list(update.keys())}")
            
            # Always return a Command with explicit goto
            return Command(update=update, goto=goto)
        
        return node_function

# Convenience functions for creating different agent types

def create_simple_agent(
    llm_config: AugLLMConfig,
    custom_fields: Optional[Dict[str, Tuple[Type, Any]]] = None
):
    """
    Create a simple single-node agent with automatic schema derivation.
    
    Args:
        llm_config: AugLLMConfig for the agent
        custom_fields: Optional additional fields for the state schema
        
    Returns:
        Compiled graph application
    """
    # Create graph with schema derived from the llm_config
    graph = DynamicGraph([llm_config], custom_fields)
    
    # Add single node with explicit END routing
    graph.add_node("agent", llm_config, END)
    
    # Build the agent
    return graph.build()


def create_sequential_agent(
    configs: Union[List[AugLLMConfig], Dict[str, AugLLMConfig]],
    edge_map: Optional[List[Tuple[str, str]]] = None,
    custom_fields: Optional[Dict[str, Tuple[Type, Any]]] = None
):
    """
    Create a sequential multi-node agent with automatic schema derivation.
    
    Args:
        configs: Either a list of AugLLMConfigs or a dictionary mapping node names to configs
        edge_map: Optional explicit edge mapping (required if configs is a dictionary)
        custom_fields: Optional additional fields for the state schema
        
    Returns:
        Compiled graph application
    """
    # Create graph with schema derived from all configs
    if isinstance(configs, dict):
        config_list = list(configs.values())
    else:
        config_list = configs
        
    graph = DynamicGraph(config_list, custom_fields)
    
    # Handle dictionary of configs with explicit node names
    if isinstance(configs, dict):
        # Edge map is required for dictionary configs
        if not edge_map:
            # Default to sequential if no edge map
            node_names = list(configs.keys())
            edge_map = [(node_names[i], node_names[i+1]) for i in range(len(node_names)-1)]
            # Last node goes to END
            edge_map.append((node_names[-1], END))
        
        # Important: Add all nodes FIRST before adding edges
        for node_name, config in configs.items():
            # Add the node without explicit routing yet
            graph.add_node(node_name, config)
        
        # Now add all the edges explicitly
        for from_node, to_node in edge_map:
            graph.add_edge(from_node, to_node)
            
    # Handle list of configs with auto-generated names
    else:
        # Generate node names
        node_names = [f"node_{i}" for i in range(len(configs))]
        
        # Add nodes first
        for i, config in enumerate(configs):
            graph.add_node(node_names[i], config)
        
        # Then add sequential edges
        for i in range(len(configs) - 1):
            graph.add_edge(node_names[i], node_names[i+1])
        
        # Last node goes to END
        graph.add_edge(node_names[-1], END)
    
    # Print debug info
    print("Nodes added:", list(graph.nodes.keys()))
    print("Entry point:", graph.entry_point)
    print("Edges:", graph.edges)
    
    # Build the agent
    return graph.build()


def create_react_agent(
    llm_config: AugLLMConfig,
    tools: List[BaseTool],
    routing_fn: Optional[Callable] = None,
    custom_fields: Optional[Dict[str, Tuple[Type, Any]]] = None
):
    """
    Create a ReAct agent with automatic schema derivation.
    
    Args:
        llm_config: AugLLMConfig for the agent
        tools: List of tools for the agent to use
        routing_fn: Optional custom routing function (defaults to checking for tool calls)
        custom_fields: Optional additional fields for the state schema
        
    Returns:
        Compiled graph application
    """
    # Create graph with schema derived from llm_config
    graph = DynamicGraph([llm_config], custom_fields)
    
    # Add tool node
    graph.add_tool_node("tool_node", tools, command_goto="agent")
    
    # Add agent node
    graph.add_node("agent", llm_config)
    
    # Define routing function if not provided
    if routing_fn is None:
        def should_use_tools(state):
            # Get the last message
            messages = filter_messages(state["messages"], exclude_types=[SystemMessage])
            if not messages:
                return "end"
            
            last_message = messages[-1]
            
            # Check for tool calls
            if hasattr(last_message, "tool_calls") and last_message.tool_calls:
                return "use_tools"
            return "end"
        
        routing_fn = should_use_tools
    
    # Add conditional routing
    graph.add_conditional_edge(
        "agent",
        routing_fn,
        {
            "use_tools": "tool_node",
            "end": END
        }
    )
    
    # Set entry point
    graph.set_entry_point("agent")
    
    # Build the agent
    return graph.build()
```

```python

class DynamicGraph:
    """
    A dynamic graph builder that automatically derives state schema from components.
    """
    
    def __init__(self, components=None, custom_fields=None,state_schema=None,build_type=None):
        """
        Initialize the graph builder with component-derived schema.
        
        Args:
            components: Optional list of AugLLMConfig objects to derive schema from
            custom_fields: Optional custom fields to add to the schema
        """
        #from state_schema_manager import StateSchemaManager
        from langgraph.graph import START
        
        # Initialize with empty schema if no components
        components = components or []
        
        # Create schema manager from components
        self.schema_manager = SchemaComposer.create_schema_for_components(components)
        
        # Add custom fields if provided
        if custom_fields:
            for name, (type_hint, default) in custom_fields.items():
                self.schema_manager.add_field(name, type_hint, default)
        
        # Get schema model
        self.state_model = self.schema_manager.get_model() if state_schema else None
        
        # Create graph with derived schema
        self.graph = StateGraph(self.state_model)
        
        # Track nodes and edges
        self.nodes = {}
        self.edges = []
        self.entry_point = None
    #if build_type == 'simple':
        
    def add_node(
        self, 
        name: str, 
        config: Union[AugLLMConfig, Callable], 
        command_goto: Optional[Union[str, Literal["END"]]] = None,
        input_mapping: Optional[Dict[str, str]] = None,
        output_mapping: Optional[Dict[str, str]] = None
    ):
        """
        Add a node to the graph.
        
        Args:
            name: Name of the node
            config: AugLLMConfig or callable node function
            command_goto: Where to route after this node
            input_mapping: Maps state fields to node inputs
            output_mapping: Maps node outputs to state fields
            
        Returns:
            Self for chaining
        """
        # If it's an AugLLMConfig, create a node function
        if isinstance(config, AugLLMConfig):
            # Update schema if needed
            self.schema_manager = SchemaComposer.derive_schema_from_aug_llm(config, self.schema_manager)
            self.state_model = self.schema_manager.get_model()
            self.graph = StateGraph(self.state_model)
            
            # Create node function
            node_fn = create_node_function(
                config=config,
                input_mapping=input_mapping,
                output_mapping=output_mapping,
                command_goto=command_goto
            )
        else:
            # It's a callable, use directly
            node_fn = config
        
        # Add to graph
        self.graph.add_node(name, node_fn)
        self.nodes[name] = node_fn
        
        # Set as entry point if first node
        if self.entry_point is None:
            self.entry_point = name
            self.graph.set_entry_point(name)
        
        # IMPORTANT: Remember to rebuild the START edge when we build
        
        # Add edge if command_goto is specified and not END
        if command_goto is not None and command_goto != END:
            self.add_edge(name, command_goto)
        else:
            # Explicitly add edge to END for proper routing
            self.add_edge(name, END)
        
        return self
    
    def add_tool_node(
        self,
        name: str,
        tools: List[BaseTool],
        post_processor: Optional[Callable] = None,
        command_goto: Optional[Union[str, Literal["END"]]] = None
    ):
        """
        Add a tool node to the graph.
        
        Args:
            name: Name of the node
            tools: List of tools to use
            post_processor: Optional function to process tool results
            command_goto: Where to route after this node
            
        Returns:
            Self for chaining
        """
        # Update schema to include tool_results if not present
        if 'tool_results' not in self.schema_manager.fields:
            from typing import Dict, Any, List
            self.schema_manager.add_field('tool_results', List[Dict[str, Any]], default_factory=list)
            self.state_model = self.schema_manager.get_model()
            self.graph = StateGraph(self.state_model)
        
        # Create tool node
        node_fn = NodeFactory.create_tool_node(
            tools=tools,
            post_processor=post_processor,
            command_goto=command_goto
        )
        
        # Add to graph
        self.graph.add_node(name, node_fn)
        self.nodes[name] = node_fn
        
        # Set as entry point if first node
        if self.entry_point is None:
            self.entry_point = name
            self.graph.set_entry_point(name)
        
        # Add edge if command_goto is specified and not END
        if command_goto is not None and command_goto != END:
            self.add_edge(name, command_goto)
        else:
            # Explicitly add edge to END for proper routing
            self.add_edge(name, END)
        
        return self
    
    def add_edge(self, from_node: str, to_node: Union[str, Literal["END"]]):
        """
        Add an edge between nodes.
        
        Args:
            from_node: Source node name
            to_node: Target node name or END
            
        Returns:
            Self for chaining
        """
        # Record the edge
        self.edges.append((from_node, to_node))
        
        # Only add to graph if not END
        if to_node != END:
            self.graph.add_edge(from_node, to_node)
        else:
            # Explicitly add edge to END
            self.graph.add_edge(from_node, END)
        
        return self
    
    def add_conditional_edge(
        self,
        from_node: str,
        condition_fn: Callable,
        routes: Dict[str, Union[str, Literal["END"]]]
    ):
        """
        Add conditional edges based on a routing function.
        
        Args:
            from_node: Source node name
            condition_fn: Function that determines routing
            routes: Mapping from condition results to target nodes
            
        Returns:
            Self for chaining
        """
        self.graph.add_conditional_edges(from_node, condition_fn, routes)
        return self
    
    def set_entry_point(self, node_name: str):
        """
        Set the entry point for the graph.
        
        Args:
            node_name: Name of the entry point node
            
        Returns:
            Self for chaining
        """
        self.entry_point = node_name
        self.graph.set_entry_point(node_name)
        return self
    
    def build(self, checkpointer=None):
        """
        Build and compile the graph.
        
        Args:
            checkpointer: Optional checkpoint saver
            
        Returns:
            Compiled graph application
        """
        from langgraph.graph import START
        
        if self.entry_point is None:
            raise ValueError("No entry point defined - add at least one node")
        
        # CRITICAL: Ensure START edge is present
        self.graph.set_entry_point(self.entry_point)
        
        # Rebuild all edges to ensure they're properly connected
        for from_node, to_node in self.edges:
            if to_node != END:
                self.graph.add_edge(from_node, to_node)
            else:
                # Explicitly add END edge
                self.graph.add_edge(from_node, END)
        
        return self.graph.compile(checkpointer=checkpointer)
    
    def get_schema(self):
        """
        Get the current schema model.
        
        Returns:
            The Pydantic model for the state schema
        """
        return self.state_model
```

```python


class SchemaComposer:
    """
    A utility for dynamically composing state schemas from AugLLM and Agent configurations.
    Uses the merge capabilities of StateSchemaManager to combine schemas.
    """
    
    @staticmethod
    def derive_schema_from_aug_llm(aug_llm_config: AugLLMConfig, schema_manager=None):
        """
        Creates or extends a schema manager with fields derived from an AugLLM configuration.
        """
        if schema_manager is None:
            schema_manager = StateSchemaManager()
    
        # Add required fields based on prompt template
        if aug_llm_config.prompt_template:
            input_vars = aug_llm_config.prompt_template.input_variables
    
            # Ensure "messages" field is added
            if 'messages' in input_vars and 'messages' not in schema_manager.fields:
                from typing import List, Tuple, Union
                from langchain_core.messages import BaseMessage
                
                schema_manager.add_field(
                    "messages", 
                    List[Union[Tuple[str, str], BaseMessage]], 
                    default=[]
                )
    
            # Add other input variables
            for var in input_vars:
                if var != 'messages' and var not in schema_manager.fields:
                    from typing import Any
                    schema_manager.add_field(var, Any, default=None)
        print(f"{aug_llm_config.name} has model: {aug_llm_config.structured_output_model}")
        # Ensure structured output model is added
        if aug_llm_config.structured_output_model:
            print(f"{aug_llm_config.name} has model: {aug_llm_config.structured_output_model}")
            model_class = aug_llm_config.structured_output_model
            field_name = model_class.__name__.lower()
    
            # DEBUGGING: Log field names to check if it’s being added
            print(f"Adding structured output model: {field_name} -> {model_class}")
    
            # Explicitly add structured output model if it's missing
            if field_name not in schema_manager.fields:
                from typing import Optional
                schema_manager.add_field(field_name, Optional[model_class], default=None)
    
        # Ensure tool_results field exists if tools are present
        if aug_llm_config.tools and 'tool_results' not in schema_manager.fields:
            from typing import Dict, Any, List
            schema_manager.add_field('tool_results', List[Dict[str, Any]], default_factory=list)
    
        return schema_manager

    
    @staticmethod
    def create_schema_for_components(components: List[Union[AugLLMConfig, BaseModel]]):
        """
        Creates a schema manager based on multiple components.
        
        Args:
            components: List of configs to derive schema from
            
        Returns:
            A schema manager with fields for all components
        """
        #from state_schema_manager import StateSchemaManager
        
        # Create base schema manager
        schema_manager = StateSchemaManager()
        
        # Process each component
        for component in components:
            print(f"writing for component {component}")
            if isinstance(component, AugLLMConfig):
                #print('check')
                schema_manager = SchemaComposer.derive_schema_from_aug_llm(component, schema_manager)
            else: 
                print(type(component))
        # Ensure messages field exists as a default
        if 'messages' not in schema_manager.fields:
            from typing import List, Tuple, Union
            from langchain_core.messages import BaseMessage
            
            schema_manager.add_field(
                "messages", 
                List[Union[Tuple[str, str], BaseMessage]], 
                default=[]
            )
                
        return schema_manager


# Utility functions for creating agents with composed schemas

def compose_state_schema(components: List[Union[AugLLMConfig, BaseModel]]):
    """
    Compose a state schema from multiple components.
    
    Args:
        components: List of AugLLM configs, Agent configs, etc.
        
    Returns:
        A Pydantic model class for the state
    """
    schema_manager = SchemaComposer.compose_schema_from_components(components)
    return schema_manager.get_model()


def create_base_agent_with_composed_schema(
    aug_llm_config: AugLLMConfig, 
    additional_components: List = None,
    custom_fields: Dict[str, Any] = None
):
    """
    Create a BaseAgent with a state schema composed from the AugLLM config and additional components.
    
    Args:
        aug_llm_config: The primary AugLLM configuration
        additional_components: Optional list of additional components to include in schema
        custom_fields: Optional additional fields to add to the schema
        
    Returns:
        BaseAgent instance
    """
    #from state_schema_manager import StateSchemaManager
    from base_agent_integration import BaseAgentConfig, BaseAgent
    
    # Compose schema from all components
    components = [aug_llm_config]
    if additional_components:
        components.extend(additional_components)
    
    schema_manager = SchemaComposer.compose_schema_for_components(components)
    
    # Add custom fields if provided
    if custom_fields:
        for name, (type_hint, default) in custom_fields.items():
            schema_manager.add_field(name, type_hint, default)
    
    # Get the model
    state_schema = schema_manager.get_model()
    
    # Create agent config
    config = BaseAgentConfig(
        engine=aug_llm_config,
        state_schema=state_schema
    )
    
    return BaseAgent(config)


def create_multi_node_agent_with_composed_schema(
    nodes: Dict[str, Union[AugLLMConfig, BaseModel]],
    edges: List[tuple],
    conditional_edges: Optional[List[tuple]] = None,
    additional_components: List = None,
    custom_fields: Dict[str, Any] = None,
    entry_point: Optional[str] = None
):
    """
    Create a MultiNodeAgent with a state schema composed from all nodes and additional components.
    
    Args:
        nodes: Dictionary mapping node names to node configs
        edges: List of (from_node, to_node) tuples defining edges
        conditional_edges: Optional list of conditional edge definitions
        additional_components: Optional list of additional components to include in schema
        custom_fields: Optional additional fields to add to the schema
        entry_point: Optional entry point node name
        
    Returns:
        MultiNodeAgent instance
    """
    #from state_schema_manager import StateSchemaManager
    #from base_agent_integration import MultiNodeAgentConfig, MultiNodeAgent
    
    # Compose schema from all nodes and additional components
    components = list(nodes.values())
    if additional_components:
        components.extend(additional_components)
    
    schema_manager = SchemaComposer.compose_schema_from_components(components)
    
    # Add custom fields if provided
    if custom_fields:
        for name, (type_hint, default) in custom_fields.items():
            schema_manager.add_field(name, type_hint, default)
    
    # Get the model
    state_schema = schema_manager.get_model()
    
    # Create agent config
    config = MultiNodeAgentConfig(
        nodes=nodes,
        edges=edges,
        conditional_edges=conditional_edges or [],
        entry_point=entry_point,
        state_schema=state_schema
    )
    
    return MultiNodeAgent(config)
```

```python
from typing import List, Dict, Any, Union, Optional, Type
from pydantic import BaseModel, Field
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.tools import Tool
from langchain_core.messages import SystemMessage

from src.haive.core.aug_llm.base import AugLLMConfig
from src.haive.core.models.llm.base import AzureLLMConfig
from src.haive.core.tools.search_tools import tavily_search_tool

# Import our improved agent framework
#from improved_agent_framework import (
#    SchemaComposer,
#    NodeFactory,
#    DynamicGraph,
#    create_simple_agent,
#    create_sequential_agent,
#    create_react_agent
#)


# -------------------------------------------------------------------------------
# Example 1: Simple Agent with Dynamic Schema
# -------------------------------------------------------------------------------

def example_simple_agent():
    """Example of creating a simple agent with dynamically inferred schema."""
    print("\n=== Simple Agent with Dynamic Schema ===")
    
    # Create an AugLLM configuration
    llm_config = AugLLMConfig(
        name="simple_assistant",
        prompt_template=ChatPromptTemplate.from_messages([
            ("system", "You are a helpful AI assistant. Keep your response brief."),
            ("user", "{messages}")
        ]),
        llm_config=AzureLLMConfig(model="gpt-4o")
    )
    
    # Create a simple agent - schema is automatically derived
    agent = create_simple_agent(llm_config)
    from IPython.display import Image, display

    try:
        display(Image(agent.get_graph().draw_mermaid_png()))
    except Exception:
        # This requires some extra dependencies and is optional
        pass
    # Run the agent
    print("Running simple agent with query: 'What is machine learning?'")
    display(Image(agent.app.get_graph().draw_mermaid_png()))
    result = agent.invoke({"messages": [("user", "What is machine learning?")]})
    
    # Print the result
    print("\nResult:")
    if "messages" in result:
        last_message = result["messages"][-1]
        content = last_message[1] if isinstance(last_message, tuple) else last_message.content
        print(f"Response: {content[:150]}...")
    
    return agent


# -------------------------------------------------------------------------------
# Example 2: Sequential Agent with Auto-Derived Schema
# -------------------------------------------------------------------------------

def example_sequential_agent():
    """Example of creating a sequential agent with auto-derived schema."""
    print("\n=== Sequential Agent with Auto-Derived Schema ===")
    
    # Create AugLLM configurations for multiple nodes
    planner_config = AugLLMConfig(
        name="planner",
        prompt_template=ChatPromptTemplate.from_messages([
            ("system", "You are a planning assistant. Create a high-level outline with 3-5 steps."),
            ("user", "{messages}")
        ]),
        llm_config=AzureLLMConfig(model="gpt-4o")
    )
    
    detail_config = AugLLMConfig(
        name="detail",
        prompt_template=ChatPromptTemplate.from_messages([
            ("system", "You are a detail-oriented assistant. Expand each step with specific actions."),
            ("user", "{messages}")
        ]),
        llm_config=AzureLLMConfig(model="gpt-4o")
    )
    
    review_config = AugLLMConfig(
        name="review",
        prompt_template=ChatPromptTemplate.from_messages([
            ("system", "You are a critical reviewer. Evaluate the detailed plan and make it better."),
            ("user", "{messages}")
        ]),
        llm_config=AzureLLMConfig(model="gpt-4o")
    )
    
    # Create a sequential agent - schema is automatically derived
    agent = create_sequential_agent(
        configs=[planner_config, detail_config, review_config],
        #names=["planner", "detailer", "reviewer"]
    )
    
    # Run the agent
    print("Running sequential agent with query: 'Create a plan for a weekend getaway.'")
    display(Image(agent.app.get_graph().draw_mermaid_png()))
    result = agent.invoke({"messages": [("user", "Create a plan for a weekend getaway.")]})
    
    # Print the result
    print("\nResult:")
    if "messages" in result:
        last_message = result["messages"][-1]
        content = last_message[1] if isinstance(last_message, tuple) else last_message.content
        print(f"Response: {content[:150]}...")
    
    return agent

# Example usage with dictionary of configs
def example_sequential_with_dict():
    """Example of creating a sequential agent with dictionary configs."""
    print("\n=== Sequential Agent with Dictionary Configs ===")
    
    # Create AugLLM configurations for multiple nodes
    configs = {
        "planner": AugLLMConfig(
            name="planner",
            prompt_template=ChatPromptTemplate.from_messages([
                ("system", "You are a planning assistant. Create a high-level outline."),
                ("user", "{messages}")
            ]),
            llm_config=AzureLLMConfig(model="gpt-4o")
        ),
        
        "detail": AugLLMConfig(
            name="detail",
            prompt_template=ChatPromptTemplate.from_messages([
                ("system", "You are a detail-oriented assistant. Expand the plan."),
                ("user", "{messages}")
            ]),
            llm_config=AzureLLMConfig(model="gpt-4o")
        ),
        
        "review": AugLLMConfig(
            name="review",
            prompt_template=ChatPromptTemplate.from_messages([
                ("system", "You are a critical reviewer. Improve the detailed plan."),
                ("user", "{messages}")
            ]),
            llm_config=AzureLLMConfig(model="gpt-4o")
        )
    }
    
    # Create a DynamicGraph directly (most reliable approach)
    graph = DynamicGraph(list(configs.values()))
    
    # Add nodes one by one
    graph.add_node("planner", configs["planner"])
    graph.add_node("detail", configs["detail"])
    graph.add_node("review", configs["review"])
    
    # Add edges explicitly
    graph.add_edge("planner", "detail") 
    graph.add_edge("detail", "review")
    graph.add_edge("review", END)
    
    # Set entry point explicitly
    graph.set_entry_point("planner")
    
    # Build the agent
    agent = graph.build()
    
    # Run the agent
    print("Running agent with query: 'Create a plan for a weekend getaway.'")
    result = agent.invoke({"messages": [("user", "Create a plan for a weekend getaway.")]})
    
    # Print the result
    print("\nResult:")
    if "messages" in result:
        last_message = result["messages"][-1]
        content = last_message[1] if isinstance(last_message, tuple) else last_message.content
        print(f"Response: {content[:150]}...")
    
    return agent

# -------------------------------------------------------------------------------
# Example 3: Structured Output with Dynamic Schema
# -------------------------------------------------------------------------------

def example_structured_output_agent():
    """Example of an agent with structured output and dynamically derived schema."""
    print("\n=== Structured Output with Dynamic Schema ===")
    
    # Define a structured output model
    class Analysis(BaseModel):
        """Analysis with insights and recommendations."""
        summary: str = Field(description="Summary of the analysis")
        insights: List[str] = Field(description="Key insights discovered")
        recommendations: List[str] = Field(description="Actionable recommendations")
    
    # Create an AugLLM configuration with structured output
    analyzer_config = AugLLMConfig(
        name="analyzer",
        prompt_template=ChatPromptTemplate.from_messages([
            ("system", "You are an analytical assistant that creates structured analysis reports."),
            ("user", "{messages}")
        ]),
        llm_config=AzureLLMConfig(model="gpt-4o"),
        structured_output_model=Analysis  # Structured output
    )
    
    # Create a DynamicGraph - schema automatically includes Analysis field
    graph = DynamicGraph([analyzer_config])
    
    # Add analyzer node
    graph.add_node("analyzer", analyzer_config, END)
    
    # Build the agent
    agent = graph.build()
    
    # Run the agent
    print("Running structured output agent with query: 'Analyze the impact of AI on healthcare.'")
    result = agent.invoke({"messages": [("user", "Analyze the impact of AI on healthcare.")]})
    
    # Print the result
    print("\nResult:")
    if "messages" in result:
        last_message = result["messages"][-1]
        content = last_message[1] if isinstance(last_message, tuple) else last_message.content
        print(f"Last message: {content[:100]}...")
    
    if "analysis" in result:
        analysis = result["analysis"]
        print("\nStructured Analysis:")
        print(f"Summary: {analysis.summary[:100]}...")
        print(f"Insights: {len(analysis.insights)} insights")
        print(f"Recommendations: {len(analysis.recommendations)} recommendations")
    
    return agent


# -------------------------------------------------------------------------------
# Example 4: ReAct Agent with Dynamic Schema
# -------------------------------------------------------------------------------

def example_react_agent():
    """Example of a ReAct agent with dynamic schema."""
    print("\n=== ReAct Agent with Dynamic Schema ===")
    
    # Define tools
    tools = [
        tavily_search_tool,
        Tool(
            name="calculator",
            description="Useful for performing calculations",
            func=lambda x: f"Result: {eval(x)}"
        )
    ]
    
    # Create an AugLLM configuration for a tool-using agent
    llm_config = AugLLMConfig(
        name="tool_user",
        prompt_template=ChatPromptTemplate.from_messages([
            ("system", """You are a helpful assistant with access to tools. 
When you need to use a tool, use the following format:
<tool_calls>
<tool_name>name of the tool</tool_name>
<tool_input>input to the tool</tool_input>
</tool_calls>

Otherwise, respond directly to the human.
"""),
            ("user", "{messages}")
        ]),
        llm_config=AzureLLMConfig(model="gpt-4o"),
        tools=tools  # Include tools in the config
    )
    
    # Define custom routing based on message content
    def custom_routing(state):
        """Route based on presence of tool calls in the last message."""
        messages = state.get("messages", [])
        if not messages:
            return "end"
            
        last_message = messages[-1]
        
        # Check for tool calls
        if hasattr(last_message, "tool_calls") and last_message.tool_calls:
            return "use_tools"
        # Check for text-based tool call format
        elif hasattr(last_message, "content") and "<tool_calls>" in last_message.content:
            return "use_tools"
        
        return "end"
    
    # Create a ReAct agent with the custom routing
    agent = create_react_agent(llm_config, tools, custom_routing)
    
    # Run the agent
    print("Running ReAct agent with query: 'What is the population of France divided by the population of Canada?'")
    result = agent.invoke({"messages": [("user", "What is the population of France divided by the population of Canada?")]})
    
    # Print the result
    print("\nResult:")
    if "messages" in result:
        print(f"Total messages: {len(result['messages'])}")
        last_message = result["messages"][-1]
        content = last_message[1] if isinstance(last_message, tuple) else last_message.content
        print(f"Final response: {content[:150]}...")
    
    return agent


# -------------------------------------------------------------------------------
# Example 5: Custom DynamicGraph with Mixed Components
# -------------------------------------------------------------------------------

def example_custom_dynamic_graph():
    """Example of a custom graph built with the DynamicGraph builder."""
    print("\n=== Custom DynamicGraph with Mixed Components ===")
    
    # Define custom fields for the state schema
    custom_fields = {
        "context": (str, ""),
        "current_step": (int, 0),
        "completed_steps": (List[str], [])
    }
    
    # Create an information gatherer
    gatherer_config = AugLLMConfig(
        name="gatherer",
        prompt_template=ChatPromptTemplate.from_messages([
            ("system", "You gather information from the user. Ask questions to understand their needs."),
            ("user", "{messages}")
        ]),
        llm_config=AzureLLMConfig(model="gpt-4o")
    )
    
    # Create a solution provider
    provider_config = AugLLMConfig(
        name="provider",
        prompt_template=ChatPromptTemplate.from_messages([
            ("system", "You provide solutions based on gathered information."),
            ("user", "Context: {context}\n\nConversation: {messages}")
        ]),
        llm_config=AzureLLMConfig(model="gpt-4o")
    )
    
    # Create a DynamicGraph with custom fields
    graph = DynamicGraph(
        components=[gatherer_config, provider_config],
        custom_fields=custom_fields
    )
    
    # Define custom context extractor
    def extract_context(state):
        """Extract context from conversation and update state."""
        messages = state.get("messages", [])
        # Simple implementation - concatenate user messages
        context = "\n".join([msg[1] if isinstance(msg, tuple) else msg.content 
                           for msg in messages 
                           if isinstance(msg, tuple) and msg[0] == "user" or 
                              hasattr(msg, "role") and msg.role == "user"])
        
        # Update state
        return {
            "messages": state["messages"],
            "context": context,
            "current_step": state["current_step"] + 1,
            "completed_steps": state["completed_steps"] + [f"Step {state['current_step']}: Context extracted"]
        }
    
    # Define routing logic
    def route_based_on_steps(state):
        """Route based on number of completed steps."""
        if state["current_step"] < 2:
            return "gather"
        return "provide"
    
    # Add nodes to the graph
    graph.add_node("gatherer", gatherer_config)
    
    # Add custom context processing node
    graph.add_node("processor", extract_context, "gatherer")  # Loop back to gatherer
    
    # Add provider node
    graph.add_node("provider", provider_config, END)
    
    # Add conditional routing from gatherer
    graph.add_conditional_edge(
        "gatherer",
        route_based_on_steps,
        {
            "gather": "processor",
            "provide": "provider"
        }
    )
    
    # Build the agent
    agent = graph.build()
    
    # Run the agent
    print("Running custom dynamic graph with query: 'I need help planning my vacation to Italy.'")
    result = agent.invoke({
        "messages": [("user", "I need help planning my vacation to Italy.")],
        "context": "",
        "current_step": 0,
        "completed_steps": []
    })
    
    # Print the result
    print("\nResult:")
    if "messages" in result:
        print(f"Total messages: {len(result['messages'])}")
        last_message = result["messages"][-1]
        content = last_message[1] if isinstance(last_message, tuple) else last_message.content
        print(f"Final response: {content[:150]}...")
    
    print(f"Completed steps: {result['completed_steps']}")
    
    return agent

```

```python
from typing import Dict, Any, List, Union, Optional, Type, Callable, Literal
from pydantic import BaseModel, Field
import uuid
import os
import json
import inspect

from langgraph.graph import StateGraph, END
from langgraph.checkpoint.memory import MemorySaver
from langgraph.types import Command, Send

from src.haive.agents.base import AgentArchitectureConfig, AgentArchitecture, register_agent
from src.haive.core.aug_llm.base import AugLLMConfig, compose_runnable
from src.haive.core.models.llm.base import AzureLLMConfig
from src.haive.utils.visualize_graph_utils import render_and_display_graph
from config.settings import RESOURCES_DIR



class BaseAgentConfig(AgentArchitectureConfig):
    """
    Configuration for the simplest possible agent architecture.
    Wraps an AugLLM in a single-node graph.
    """
    engine: Union[AugLLMConfig, AgentArchitectureConfig] = Field(
        default_factory=lambda: AugLLMConfig(llm_config=AzureLLMConfig(model="gpt-4o")),
        description="LLM configuration for the BaseAgent."
    )
    input_mapping: Optional[Dict[str, str]] = Field(
        default=None,
        description="Maps state fields to LLM input fields."
    )
    output_mapping: Optional[Dict[str, str]] = Field(
        default=None,
        description="Maps LLM output fields to state fields."
    )
    node_name: str = Field(
        default="agent_node",
        description="Name for the single node in the graph."
    )
    should_setup_workflow: bool = Field(
        default=True,
        description="Whether to set up the workflow."
    )
    should_compile: bool = Field(
        default=True,
        description="Whether to compile the graph."
    )
    should_visualize_graph: bool = Field(
        default=False,
        description="Whether to visualize the graph."
    )
    visualize_graph_output_name: Optional[str] = Field(
        default=None,
        description="Output file for graph visualization."
    )
    
    def build_agent(self) -> "BaseAgent":
        """Build the BaseAgent from this configuration."""
        return BaseAgent(config=self)


# The registration will happen at the end of the file
class BaseAgent(AgentArchitecture):
    """
    The simplest possible agent architecture. Wraps an AugLLM in a single-node graph.
    """
    def __init__(self, config: BaseAgentConfig):
        self.input_mapping = config.input_mapping
        self.output_mapping = config.output_mapping
        self.node_name = config.node_name
        
        # Initialize the base agent
        super().__init__(config)
    
    def setup_workflow(self):
        """
        Set up a simple single-node workflow.
        
        The workflow consists of a single node that processes input through the AugLLM
        and routes directly to END.
        """
        # Create the node function with explicit END routing
        node_fn = create_node_function(
            config=self.config.engine if isinstance(self.config.engine, AugLLMConfig) else AugLLMConfig(),
            input_mapping=self.input_mapping,
            output_mapping=self.output_mapping,
            next_node=END  # Explicit END routing
        )
        
        # Add the node to the graph
        self.graph.add_node(self.node_name, node_fn)
        
        # Set the entry point
        self.graph.set_entry_point(self.node_name)
        #if next_node==END:
        self.graph.add_edge(self.node_name, END)

class MultiNodeAgentConfig(AgentArchitectureConfig):
    """
    Configuration for a more complex agent with multiple nodes.
    """
    nodes: Dict[str, Union[AugLLMConfig, NodeConfig]] = Field(
        default_factory=dict,
        description="Dictionary of nodes to include in the graph."
    )
    edges: List[tuple] = Field(
        default_factory=list,
        description="List of (from_node, to_node) tuples defining edges."
    )
    conditional_edges: List[tuple] = Field(
        default_factory=list,
        description="List of (from_node, condition_fn, edge_map) tuples for conditional routing."
    )
    entry_point: Optional[str] = Field(
        default=None,
        description="Name of the entry point node (defaults to first node)."
    )
    
    def build_agent(self) -> "MultiNodeAgent":
        """Build the MultiNodeAgent from this configuration."""
        return MultiNodeAgent(config=self)


# The registration will happen at the end of the file
class MultiNodeAgent(AgentArchitecture):
    """
    A more complex agent with multiple nodes and custom routing.
    """
    def __init__(self, config: MultiNodeAgentConfig):
        self.nodes = config.nodes
        self.edges = config.edges
        self.conditional_edges = config.conditional_edges
        self.entry_point = config.entry_point
        
        # Initialize the base agent
        super().__init__(config)
    
    def setup_workflow(self):
        """
        Set up a multi-node workflow based on the configuration.
        """
        # Convert any AugLLMConfigs to NodeConfigs
        node_configs = {}
        for name, config in self.nodes.items():
            if isinstance(config, AugLLMConfig):
                # Find explicit next node from edges
                next_nodes = [to_node for from_node, to_node in self.edges if from_node == name]
                next_node = next_nodes[0] if len(next_nodes) == 1 else END
                
                node_configs[name] = NodeConfig(
                    name=name,
                    engine=config,
                    next_node=next_node
                )
            else:
                node_configs[name] = config
        
        # Add all nodes to the graph
        for name, config in node_configs.items():
            node_fn = config.create_node()
            self.graph.add_node(name, node_fn)
        
        # Set the entry point
        entry_node = self.entry_point or list(node_configs.keys())[0]
        self.graph.set_entry_point(entry_node)
        
        # Add edges (only for non-END destinations or when multiple edges exist)
        for from_node, to_node in self.edges:
            if to_node != END:
                # Check if there are multiple outgoing edges
                outgoing_edges = [to for f, to in self.edges if f == from_node]
                if len(outgoing_edges) > 1:
                    self.graph.add_edge(from_node, to_node)
        
        # Add conditional edges
        for from_node, condition_fn, edge_map in self.conditional_edges:
            self.graph.add_conditional_edges(from_node, condition_fn, edge_map)


# Register agents
register_agent(BaseAgentConfig, BaseAgent)
register_agent(MultiNodeAgentConfig, MultiNodeAgent)


# Factory functions for easier agent creation
def create_base_agent(aug_llm_config: AugLLMConfig, state_schema=None):
    """
    Create a simple BaseAgent from an AugLLM config.
    
    Args:
        aug_llm_config: The AugLLM configuration
        state_schema: Optional custom state schema
        
    Returns:
        Initialized BaseAgent instance
    """
    config = BaseAgentConfig(
        engine=aug_llm_config,
        state_schema=state_schema or {"messages": []}
    )
    return BaseAgent(config)


def create_multi_node_agent(
    nodes: Dict[str, AugLLMConfig],
    edges: List[tuple],
    conditional_edges: Optional[List[tuple]] = None,
    state_schema=None,
    entry_point: Optional[str] = None
):
    """
    Create a MultiNodeAgent with the specified configuration.
    
    Args:
        nodes: Dictionary mapping node names to AugLLM configs
        edges: List of (from_node, to_node) tuples defining edges
        conditional_edges: Optional list of conditional edge definitions
        state_schema: Optional custom state schema
        entry_point: Optional entry point node name
        
    Returns:
        Initialized MultiNodeAgent instance
    """
    config = MultiNodeAgentConfig(
        nodes=nodes,
        edges=edges,
        conditional_edges=conditional_edges or [],
        entry_point=entry_point,
        state_schema=state_schema or {"messages": []}
    )
    return MultiNodeAgent(config)
```

```python
def augment_schema_for_llm(
    schema_manager: StateSchemaManager,
    aug_llm_config: AugLLMConfig
) -> StateSchemaManager:
    """
    Augments a state schema to accommodate the inputs and outputs of an AugLLM.
    
    Args:
        schema_manager: The state schema manager to augment
        aug_llm_config: The AugLLM configuration
        
    Returns:
        Updated schema manager
    """
    # Ensure we have required fields
    if aug_llm_config.prompt_template:
        input_vars = aug_llm_config.prompt_template.input_variables
        for var in input_vars:
            if var == "messages" and "messages" not in schema_manager.fields:
                # Add messages field with appropriate annotation
                from typing import List, Tuple, Union
                from langchain_core.messages import BaseMessage
                
                schema_manager.add_field(
                    "messages", 
                    List[Union[Tuple[str, str], BaseMessage]], 
                    default=[]
                )
    
    # Add field for structured output if specified
    if aug_llm_config.structured_output_model:
        model_class = aug_llm_config.structured_output_model
        field_name = model_class.__name__.lower()
        
        # Check if the field already exists
        if field_name not in schema_manager.fields:
            schema_manager.add_field(field_name, model_class, default=None)
    
    return schema_manager
```

```python
from typing import Dict, Any, List, Union, Optional, Type, Callable, Tuple, Literal, Set
from pydantic import BaseModel, Field
import uuid
from functools import partial

from langgraph.graph import StateGraph, END
from langgraph.types import Command, Send
from langgraph.checkpoint.memory import MemorySaver

from src.haive.agents.base import AgentArchitectureConfig, AgentArchitecture, register_agent
from src.haive.core.aug_llm.base import AugLLMConfig, compose_runnable
from src.haive.core.models.llm.base import AzureLLMConfig
from src.haive.utils.visualize_graph_utils import render_and_display_graph




class SequentialAgentConfig(BaseAgentConfig):
    """
    Configuration for a sequential multi-node agent that extends the BaseAgent.
    
    This agent chains multiple AugLLM configs together in sequence, with optional
    input/output mappings between them.
    """
    nodes: Union[List[Union[AugLLMConfig, Tuple[AugLLMConfig, Dict[str, str], Dict[str, str]]]], 
                Dict[str, Union[AugLLMConfig, Tuple[AugLLMConfig, Dict[str, str], Dict[str, str]]]]] = Field(
        default_factory=list,
        description="List or Dict of AugLLM configs or (config, input_mapping, output_mapping) tuples"
    )
    node_names: Optional[List[str]] = Field(
        default=None,
        description="Optional node names to use (only if nodes is a list)"
    )
    edge_map: Optional[List[Tuple[str, Union[str, Literal["END"]]]]] = Field(
        default=None,
        description="Optional edge map to define routing (required if nodes is a dict)"
    )
    custom_state: Optional[Dict[str, Any]] = Field(
        default=None,
        description="Optional custom state fields to include in the schema"
    )
    
    def build_agent(self) -> "SequentialAgent":
        """Build a SequentialAgent from this configuration."""
        return SequentialAgent(config=self)


#@register_agent(SequentialAgentConfig, None)  # Will be registered at end of file
class SequentialAgent(BaseAgent):
    """
    An agent that chains multiple components in sequence with custom input/output mappings.
    Extends BaseAgent to inherit core agent functionality.
    """
    def __init__(self, config: SequentialAgentConfig):
        # Store configuration specific to SequentialAgent
        self.node_configs = config.nodes
        self.node_names = config.node_names
        self.edge_map = config.edge_map
        self.custom_state = config.custom_state or {"messages": []}
        
        # Initialize base agent
        super().__init__(config)
        
    def setup_workflow(self):
        """
        Set up a workflow based on the chain configuration.
        Override's BaseAgent's simple workflow.
        """
        #from state_schema_manager import StateSchemaManager
        #from node_factory_refined import augment_schema_for_llm, create_node_function
        
        # Start by creating a schema manager from the custom state
        schema_manager = StateSchemaManager(self.custom_state)
        
        # Extract all AugLLM configs
        if isinstance(self.node_configs, dict):
            # Unpack configs from dictionary with tuples
            configs_list = []
            for name, config_item in self.node_configs.items():
                if isinstance(config_item, tuple) and len(config_item) >= 1:
                    configs_list.append(config_item[0])
                else:
                    configs_list.append(config_item)
        else:
            # Unpack configs from list with tuples
            configs_list = []
            for config_item in self.node_configs:
                if isinstance(config_item, tuple) and len(config_item) >= 1:
                    configs_list.append(config_item[0])
                else:
                    configs_list.append(config_item)
        
        # Derive schema from all configs
        for config in configs_list:
            # Augment schema for each config
            schema_manager = augment_schema_for_llm(schema_manager, config)
        
        # Get the schema model
        self.state_schema = schema_manager.get_model()
        self.graph = StateGraph(self.state_schema)
        
        # Handle dictionary of configs with explicit node names
        if isinstance(self.node_configs, dict):
            # Edge map is required for dictionary configs
            if not self.edge_map:
                # Default to sequential if no edge map
                node_names = list(self.node_configs.keys())
                self.edge_map = [(node_names[i], node_names[i+1]) for i in range(len(node_names)-1)]
                # Last node goes to END
                self.edge_map.append((node_names[-1], END))
            
            # First pass: Add all nodes without explicit next_node
            for node_name, config_item in self.node_configs.items():
                # Extract config and mappings
                if isinstance(config_item, tuple) and len(config_item) >= 1:
                    config = config_item[0]
                    input_mapping = config_item[1] if len(config_item) > 1 else None
                    output_mapping = config_item[2] if len(config_item) > 2 else None
                else:
                    config = config_item
                    input_mapping = None
                    output_mapping = None
                
                # Create node function with END as default next node
                node_fn = create_node_function(
                    config=config,
                    input_mapping=input_mapping,
                    output_mapping=output_mapping,
                    next_node=END  # Default to END, will be overridden by edges
                )
                
                # Add node to graph
                self.graph.add_node(node_name, node_fn)
            
            # Second pass: Add all edges to override the default END routing
            for from_node, to_node in self.edge_map:
                if to_node != END:
                    self.graph.add_edge(from_node, to_node)
                else:
                    # Explicitly add edge to END
                    self.graph.add_edge(from_node, END)
                
        # Handle list of configs with auto-generated or explicit names
        else:
            # Generate node names if not provided
            if self.node_names is None:
                self.node_names = [f"node_{i}" for i in range(len(self.node_configs))]
            elif len(self.node_names) < len(self.node_configs):
                # Extend with default names if needed
                self.node_names.extend([f"node_{i+len(self.node_names)}" for i in range(len(self.node_configs) - len(self.node_names))])
            
            # Add nodes with sequential routing
            for i, config_item in enumerate(self.node_configs):
                # Extract config and mappings
                if isinstance(config_item, tuple) and len(config_item) >= 1:
                    config = config_item[0]
                    input_mapping = config_item[1] if len(config_item) > 1 else None
                    output_mapping = config_item[2] if len(config_item) > 2 else None
                else:
                    config = config_item
                    input_mapping = None
                    output_mapping = None
                
                # Determine node name
                node_name = self.node_names[i]
                
                # Determine next node
                if i < len(self.node_configs) - 1:
                    next_node = self.node_names[i+1]  # Connect to next node in sequence
                else:
                    next_node = END  # Last node goes to END
                
                # Create node function
                node_fn = create_node_function(
                    config=config,
                    input_mapping=input_mapping,
                    output_mapping=output_mapping,
                    next_node=next_node
                )
                
                # Add node to graph
                self.graph.add_node(node_name, node_fn)
                
                # Add edge to next node
                if i < len(self.node_configs) - 1:
                    self.graph.add_edge(node_name, self.node_names[i+1])
                else:
                    # Explicitly add edge to END
                    self.graph.add_edge(node_name, END)
            
        # Set entry point
        if isinstance(self.node_configs, dict):
            # Use first node from edge_map or first node in configs
            entry_candidates = [from_node for from_node, _ in self.edge_map]
            if entry_candidates:
                self.graph.set_entry_point(entry_candidates[0])
            else:
                self.graph.set_entry_point(next(iter(self.node_configs.keys())))
        else:
            # Use first node in the list
            self.graph.set_entry_point(self.node_names[0])
    
    def run(self, input_text: str):
        """
        Execute the agent with the provided input.
        Extends the BaseAgent run method to handle custom state initialization.
        """
        if not self.graph:
            raise RuntimeError("Workflow graph is not set up.")
        if not self.app:
            self.compile_workflow(checkpointer=self.memory)
        
        # Create initial state with input
        input_state = {"messages": [("user", input_text)]}
        
        # Add any custom state fields with defaults
        if isinstance(self.custom_state, dict):
            for key, value in self.custom_state.items():
                if key != "messages" and key not in input_state:
                    input_state[key] = value
        
        # Run the agent with the initialized state
        for output in self.app.stream(
            input_state,
            stream_mode="values",
            config=self.runnable_config,
            debug=True
        ):
            # Print messages as they come in
            if "messages" in output and output["messages"]:
                message = output["messages"][-1]
                if isinstance(message, tuple):
                    print(f"{message[0]}: {message[1]}")
                else:
                    message.pretty_print()
        
        # Auto-save state history after execution
        self.save_state_history()
        
        # Auto-save graph after execution
        self.visualize_graph()
        
        return output


# Register the agent
register_agent(SequentialAgentConfig, SequentialAgent)


# Factory functions for easier creation

def create_sequential_agent(
    configs: Union[List[Union[AugLLMConfig, Tuple[AugLLMConfig, Dict[str, str], Dict[str, str]]]], 
                Dict[str, Union[AugLLMConfig, Tuple[AugLLMConfig, Dict[str, str], Dict[str, str]]]]],
    node_names: Optional[List[str]] = None,
    edge_map: Optional[List[Tuple[str, Union[str, Literal["END"]]]]] = None,
    custom_state: Optional[Dict[str, Any]] = None
):
    """
    Create a sequential agent with the specified configuration.
    
    Args:
        configs: List or Dict of AugLLM configs or (config, input_mapping, output_mapping) tuples
        node_names: Optional node names to use (only if configs is a list)
        edge_map: Optional edge map for routing (required if configs is a dict)
        custom_state: Optional custom state fields
        
    Returns:
        SequentialAgent instance
    """
    # Process custom state
    #from state_schema_manager import StateSchemaManager
    if custom_state:
        schema_manager = StateSchemaManager(custom_state)
        state_schema = schema_manager.get_model()
    else:
        schema_manager = StateSchemaManager({"messages": []})
        state_schema = schema_manager.get_model()
    
    # Create configuration
    config = SequentialAgentConfig(
        engine=AugLLMConfig(llm_config=AzureLLMConfig(model="gpt-4o")),  # Default engine (not used)
        nodes=configs,
        node_names=node_names,
        edge_map=edge_map,
        custom_state=custom_state,
        state_schema=state_schema
    )
    
    return SequentialAgent(config)
```

```python
def build_graph_from_configs(
    configs: Dict[str, AugLLMConfig],
    edges: List[Tuple[str, str]],
    conditional_edges: Optional[List[Tuple[str, Callable, Dict[str, str]]]] = None,
    state_schema: Optional[Union[Type[BaseModel], Dict[str, Any], StateSchemaManager, Type[TypedDict]]] = None,
    input_schema: Optional[Union[Type[BaseModel], Type[TypedDict]]] = None,
    output_schema: Optional[Union[Type[BaseModel], Type[TypedDict]]] = None,
    entry_point: Optional[str] = None
):
    """
    Builds a LangGraph from AugLLM configurations.
    
    Args:
        configs: Dictionary mapping node names to AugLLM configs
        edges: List of (from_node, to_node) tuples defining graph edges
        conditional_edges: List of (from_node, condition_fn, edge_map) for conditional routing
        state_schema: Base state schema to use (will be augmented as needed)
        input_schema: Schema for graph inputs (subset of state)
        output_schema: Schema for graph outputs (subset of state)
        entry_point: Name of the node to use as entry point (defaults to first node)
        
    Returns:
        Compiled LangGraph application
    """
    # Initialize state schema
    if state_schema is None:
        schema_manager = StateSchemaManager({"messages": []})
    elif isinstance(state_schema, dict):
        schema_manager = StateSchemaManager(state_schema)
    elif isinstance(state_schema, StateSchemaManager):
        schema_manager = state_schema
    else:
        schema_manager = StateSchemaManager(state_schema)
    
    # Augment schema for all configs
    for node_name, config in configs.items():
        schema_manager = augment_schema_for_llm(schema_manager, config)
    
    # Create state model
    StateModel = schema_manager.get_model()
    
    # Build graph with appropriate schemas
    if input_schema and output_schema:
        graph = StateGraph(StateModel, input=input_schema, output=output_schema)
    elif input_schema:
        graph = StateGraph(StateModel, input=input_schema)
    elif output_schema:
        graph = StateGraph(StateModel, output=output_schema)
    else:
        graph = StateGraph(StateModel)
    
    # Add nodes to graph
    for node_name, config in configs.items():
        # Find next nodes for this node
        next_nodes = [to_node for from_node, to_node in edges if from_node == node_name]
        
        # If there's exactly one next node, use direct routing
        command_goto = next_nodes[0] if len(next_nodes) == 1 else None
        
        # Create node function
        node_fn = create_node_function(config, command_goto=command_goto)
        graph.add_node(node_name, node_fn)
    
    # Set entry point
    entry_node = entry_point or list(configs.keys())[0]
    graph.set_entry_point(entry_node)
    
    # Add edges (skip if using command_goto)
    for from_node, to_node in edges:
        # Skip if to_node is END (handled by default) or if using direct routing
        if to_node != END:
            next_nodes = [to for from_node_, to in edges if from_node_ == from_node]
            # Only add edge if multiple destinations (otherwise using command_goto)
            if len(next_nodes) > 1:
                graph.add_edge(from_node, to_node)
    
    # Add conditional edges
    if conditional_edges:
        for from_node, condition_fn, edge_map in conditional_edges:
            graph.add_conditional_edges(from_node, condition_fn, edge_map)
    
    # Compile graph
    return graph.compile()


```

```python

class NodeConfig(BaseModel):
    """Configuration for a node in a LangGraph."""
    name: str = Field(default="default_node")
    engine: AugLLMConfig
    input_mapping: Optional[Dict[str, str]] = None
    output_mapping: Optional[Dict[str, str]] = None
    routing_logic: Optional[Callable] = None
    routing_map: Optional[Dict[str, Union[str, List[str]]]] = None
    
    def create_node(self) -> Callable:
        """Creates a callable node function based on this configuration."""
        return create_node_function(
            config=self.engine,
            input_mapping=self.input_mapping,
            output_mapping=self.output_mapping
        )
```

```python
class GraphBuilder:
    """Builder for constructing LangGraph graphs."""
    
    def __init__(
        self, 
        state_schema: Union[Type[BaseModel], StateSchemaManager, Dict[str, Any], Type[TypedDict]],
        input_schema: Optional[Union[Type[BaseModel], Type[TypedDict]]] = None,
        output_schema: Optional[Union[Type[BaseModel], Type[TypedDict]]] = None
    ):
        """
        Initialize with state, input, and output schemas.
        
        Args:
            state_schema: The overall state schema
            input_schema: Schema for graph inputs (subset of state)
            output_schema: Schema for graph outputs (subset of state)
        """
        self.input_schema = input_schema
        self.output_schema = output_schema
        
        # Handle different schema types
        if isinstance(state_schema, StateSchemaManager):
            self.schema_manager = state_schema
            self.state_schema = state_schema.get_model()
        elif isinstance(state_schema, dict):
            self.schema_manager = StateSchemaManager(state_schema)
            self.state_schema = self.schema_manager.get_model()
        elif isinstance(state_schema, type) and (issubclass(state_schema, BaseModel) or 
                                                issubclass(state_schema, TypedDict)):
            self.schema_manager = StateSchemaManager(state_schema)
            self.state_schema = state_schema
        else:
            self.schema_manager = StateSchemaManager(state_schema)
            self.state_schema = state_schema
        
        # Create graph with appropriate schemas
        if input_schema and output_schema:
            self.graph = StateGraph(self.state_schema, input=input_schema, output=output_schema)
        elif input_schema:
            self.graph = StateGraph(self.state_schema, input=input_schema)
        elif output_schema:
            self.graph = StateGraph(self.state_schema, output=output_schema)
        else:
            self.graph = StateGraph(self.state_schema)
            
        self.nodes = {}
        self.entry_point = None
        
    def add_node(self, node_config: Union[NodeConfig, AugLLMConfig], name: Optional[str] = None):
        """Add a node to the graph."""
        if isinstance(node_config, AugLLMConfig):
            node_name = name or node_config.name or f"node_{len(self.nodes)}"
            node_fn = create_node_function(node_config)
        else:
            node_name = node_config.name
            node_fn = node_config.create_node()
        
        self.nodes[node_name] = node_fn
        self.graph.add_node(node_name, node_fn)
        
        # Set as entry point if it's the first node
        if self.entry_point is None:
            self.entry_point = node_name
            self.graph.set_entry_point(node_name)
        
        return self
    
    def add_edge(self, from_node: str, to_node: Union[str, List[str]]):
        """Add an edge between nodes."""
        if isinstance(to_node, list):
            for node in to_node:
                self.graph.add_edge(from_node, node)
        else:
            self.graph.add_edge(from_node, to_node)
        return self
    
    def add_conditional_edge(self, from_node: str, condition_fn: Callable, edges_map: Dict[str, str]):
        """Add conditional edges based on a routing function."""
        self.graph.add_conditional_edges(from_node, condition_fn, edges_map)
        return self
    
    def build(self, checkpointer=None):
        """Compile the graph into an executable app."""
        if self.entry_point is None:
            raise ValueError("No entry point defined - add at least one node")
        
        return self.graph.compile(checkpointer=checkpointer)
```

```python
from typing import Dict, Any, List, Union, Optional, Type, Callable, Set
from pydantic import BaseModel, Field, create_model
import inspect
from src.haive.core.aug_llm.base import AugLLMConfig
from langchain_core.prompts import BasePromptTemplate
from langchain_core.tools import BaseTool

class SchemaComposer:
    """
    A utility for dynamically composing state schemas from AugLLM and Agent configurations.
    Analyzes inputs and outputs of components to create appropriate state fields.
    """
    
    @staticmethod
    def derive_schema_from_aug_llm(
        aug_llm_config: AugLLMConfig,
        schema_manager=None
    ):
        """
        Derives state schema fields from an AugLLM configuration.
        
        Args:
            aug_llm_config: The AugLLM configuration
            schema_manager: Optional existing schema manager to extend
            
        Returns:
            Updated schema manager
        """
        from state_schema_manager import StateSchemaManager
        
        # Initialize schema manager if not provided
        if schema_manager is None:
            schema_manager = StateSchemaManager()
            
        # Create a new schema manager for this config
        config_schema = StateSchemaManager()
        
        # Add required fields based on prompt template
        if aug_llm_config.prompt_template:
            SchemaComposer._add_prompt_input_fields(aug_llm_config.prompt_template, config_schema)
            
        # Add fields for structured output
        if aug_llm_config.structured_output_model:
            SchemaComposer._add_structured_output_field(aug_llm_config.structured_output_model, config_schema)
            
        # Add fields for tool outputs if tools are present
        if aug_llm_config.tools:
            SchemaComposer._add_tool_output_fields(aug_llm_config.tools, config_schema)
            
        # Merge with the existing schema using the built-in merge function
        if schema_manager.fields:
            return schema_manager.merge(config_schema)
        return config_schema
    
    @staticmethod
    def derive_schema_from_agent_config(agent_config, schema_manager=None):
        """
        Derives state schema fields from an agent configuration.
        
        Args:
            agent_config: An AgentArchitectureConfig
            schema_manager: Optional existing schema manager to extend
            
        Returns:
            Updated schema manager
        """
        from state_schema_manager import StateSchemaManager
        
        # Initialize schema manager if not provided
        if schema_manager is None:
            schema_manager = StateSchemaManager()
            
        # Create a new schema manager for this config
        config_schema = StateSchemaManager()
            
        # Process engine if it's an AugLLM config
        if hasattr(agent_config, 'engine'):
            if isinstance(agent_config.engine, AugLLMConfig):
                config_schema = SchemaComposer.derive_schema_from_aug_llm(
                    agent_config.engine, 
                    config_schema
                )
                
        # Process tools if present (e.g., for ReactAgent)
        if hasattr(agent_config, 'tools') and agent_config.tools:
            SchemaComposer._add_tool_output_fields(agent_config.tools, config_schema)
            
        # Process nodes if present (e.g., for MultiNodeAgent)
        if hasattr(agent_config, 'nodes') and agent_config.nodes:
            for node_name, node_config in agent_config.nodes.items():
                if isinstance(node_config, AugLLMConfig):
                    config_schema = SchemaComposer.derive_schema_from_aug_llm(
                        node_config,
                        config_schema
                    )
                
        # Always ensure messages field is present
        if "messages" not in config_schema.fields:
            from typing import List, Tuple, Union
            from langchain_core.messages import BaseMessage
            
            config_schema.add_field(
                "messages", 
                List[Union[Tuple[str, str], BaseMessage]], 
                default=[]
            )
            
        # Merge with the existing schema using the built-in merge function
        if schema_manager.fields:
            return schema_manager.merge(config_schema)
        return config_schema
    
    @staticmethod
    def derive_schema_from_components(components: List[Union[AugLLMConfig, BaseModel]], schema_manager=None):
        """
        Derives a schema from multiple components.
        
        Args:
            components: List of configs or models to derive schema from
            schema_manager: Optional existing schema manager to extend
            
        Returns:
            Updated schema manager
        """
        from state_schema_manager import StateSchemaManager
        
        # Initialize schema manager if not provided
        if schema_manager is None:
            schema_manager = StateSchemaManager()
        
        # Process each component by merging schemas
        for component in components:
            if isinstance(component, AugLLMConfig):
                schema_manager = SchemaComposer.derive_schema_from_aug_llm(
                    component, 
                    schema_manager
                )
            elif hasattr(component, 'engine') or hasattr(component, 'nodes'):
                # It's likely an agent config
                schema_manager = SchemaComposer.derive_schema_from_agent_config(
                    component,
                    schema_manager
                )
                
        # Always ensure messages field is present
        if "messages" not in schema_manager.fields:
            from typing import List, Tuple, Union
            from langchain_core.messages import BaseMessage
            
            schema_manager.add_field(
                "messages", 
                List[Union[Tuple[str, str], BaseMessage]], 
                default=[]
            )
            
        return schema_manager
    
    @staticmethod
    def _add_prompt_input_fields(prompt_template: BasePromptTemplate, schema_manager):
        """Add fields required by a prompt template to the schema."""
        input_vars = prompt_template.input_variables
        
        # Handle special case for 'messages'
        if 'messages' in input_vars and 'messages' not in schema_manager.fields:
            from typing import List, Tuple, Union
            from langchain_core.messages import BaseMessage
            
            schema_manager.add_field(
                "messages", 
                List[Union[Tuple[str, str], BaseMessage]], 
                default=[]
            )
            
        # Add other input variables with default type Any
        for var in input_vars:
            if var != 'messages' and var not in schema_manager.fields:
                from typing import Any
                schema_manager.add_field(var, Any, default=None)
        
        return schema_manager
    
    @staticmethod
    def _add_structured_output_field(model_class: Type[BaseModel], schema_manager):
        """Add field for structured output model."""
        field_name = model_class.__name__.lower()
        
        # Check if the field already exists
        if field_name not in schema_manager.fields:
            # Add field with None default to ensure it's treated as Optional
            schema_manager.add_field(field_name, Optional[model_class], default=None)
            
        return schema_manager
    
    @staticmethod
    def _add_tool_output_fields(tools: List[BaseTool], schema_manager):
        """Add fields for tool outputs."""
        # Add a tool_results field if tools are present and field doesn't exist
        if tools and 'tool_results' not in schema_manager.fields:
            from typing import Dict, Any, List
            schema_manager.add_field('tool_results', List[Dict[str, Any]], default_factory=list)
            
        # You could also add specific fields for each tool if desired
        # for tool in tools:
        #     tool_field = f"{tool.name}_result"
        #     if tool_field not in schema_manager.fields:
        #         schema_manager.add_field(tool_field, Optional[Any], default=None)
            
        return schema_manager


# Integration with existing agent architecture
def compose_state_schema(components: List[Union[AugLLMConfig, BaseModel]]):
    """
    Compose a state schema from multiple components.
    
    Args:
        components: List of AugLLM configs, Agent configs, etc.
        
    Returns:
        A Pydantic model class for the state
    """
    schema_manager = SchemaComposer.derive_schema_from_components(components)
    return schema_manager.get_model()


# Enhanced version of create_base_agent that derives schema from config
def create_base_agent_with_derived_schema(aug_llm_config: AugLLMConfig, custom_fields: Dict[str, Any] = None):
    """
    Create a BaseAgent with a state schema derived from the AugLLM config.
    
    Args:
        aug_llm_config: The AugLLM configuration
        custom_fields: Optional additional fields to add to the schema
        
    Returns:
        BaseAgent instance
    """
    from state_schema_manager import StateSchemaManager
    from base_agent_integration import BaseAgentConfig, BaseAgent
    
    # Derive schema from config
    schema_manager = SchemaComposer.derive_schema_from_aug_llm(aug_llm_config)
    
    # Add custom fields if provided
    if custom_fields:
        custom_schema = StateSchemaManager()
        for name, (type_hint, default) in custom_fields.items():
            custom_schema.add_field(name, type_hint, default)
        
        # Merge with the derived schema
        schema_manager = schema_manager.merge(custom_schema)
    
    # Get the model
    state_schema = schema_manager.get_model()
    
    # Create agent config
    config = BaseAgentConfig(
        engine=aug_llm_config,
        state_schema=state_schema
    )
    
    return BaseAgent(config)


# Enhanced version of create_multi_node_agent that derives schema from nodes
def create_multi_node_agent_with_derived_schema(
    nodes: Dict[str, AugLLMConfig],
    edges: List[tuple],
    conditional_edges: Optional[List[tuple]] = None,
    custom_fields: Dict[str, Any] = None,
    entry_point: Optional[str] = None
):
    """
    Create a MultiNodeAgent with a state schema derived from all nodes.
    
    Args:
        nodes: Dictionary mapping node names to AugLLM configs
        edges: List of (from_node, to_node) tuples defining edges
        conditional_edges: Optional list of conditional edge definitions
        custom_fields: Optional additional fields to add to the schema
        entry_point: Optional entry point node name
        
    Returns:
        MultiNodeAgent instance
    """
    from state_schema_manager import StateSchemaManager
    from base_agent_integration import MultiNodeAgentConfig, MultiNodeAgent
    
    # Derive schema from all nodes
    schema_manager = SchemaComposer.derive_schema_from_components(list(nodes.values()))
    
    # Add custom fields if provided
    if custom_fields:
        custom_schema = StateSchemaManager()
        for name, (type_hint, default) in custom_fields.items():
            custom_schema.add_field(name, type_hint, default)
        
        # Merge with the derived schema
        schema_manager = schema_manager.merge(custom_schema)
    
    # Get the model
    state_schema = schema_manager.get_model()
    
    # Create agent config
    config = MultiNodeAgentConfig(
        nodes=nodes,
        edges=edges,
        conditional_edges=conditional_edges or [],
        entry_point=entry_point,
        state_schema=state_schema
    )
    
    return MultiNodeAgent(config)
```

```python
from typing import Dict, Any, List, Union, Optional, Type, Callable, Tuple, Literal, Set
from pydantic import BaseModel, Field
import uuid
from functools import partial

from langgraph.graph import StateGraph, END
from langgraph.prebuilt import ToolNode
from langgraph.types import Command, Send
from langchain_core.tools import BaseTool, StructuredTool, Tool
from langchain_core.messages import SystemMessage, filter_messages

from src.haive.core.aug_llm.base import AugLLMConfig, compose_runnable
from src.haive.core.models.llm.base import AzureLLMConfig

# Import our node utilities
#from node_factory_refined import create_node_function, NodeConfig
#from schema_composition import SchemaComposer


class GraphManager:
    """
    A flexible manager for building and configuring LangGraph workflows.
    
    Features:
    - Create nodes from AugLLM configs with automatic schema derivation
    - Add generalized tool nodes with post-processing
    - Create conditional branching with custom routing functions
    - Build sequential workflows with automatic END routing
    """
    
    def __init__(self, state_schema=None):
        """
        Initialize the graph manager.
        
        Args:
            state_schema: Optional initial state schema (BaseModel, StateSchemaManager, or dict)
        """
        # Import here to avoid circular imports
        from state_schema_manager import StateSchemaManager

        # Initialize state schema
        if state_schema is None:
            self.schema_manager = StateSchemaManager({"messages": []})
        elif isinstance(state_schema, dict):
            self.schema_manager = StateSchemaManager(state_schema)
        elif hasattr(state_schema, "get_model"):
            self.schema_manager = state_schema
        else:
            self.schema_manager = StateSchemaManager(state_schema)
            
        # Build the state model
        self.state_model = self.schema_manager.get_model()
        
        # Initialize graph
        self.graph = StateGraph(self.state_model)
        
        # Track nodes, tools, and edges
        self.nodes = {}
        self._tools = {}
        self._edges = []
        self._conditional_edges = []
        self.entry_point = None
        
    def add_node_from_aug_llm(
        self, 
        name: str, 
        config: AugLLMConfig,
        next_node: Optional[Union[str, Literal["END"]]] = None
    ) -> 'GraphManager':
        """
        Add a node from an AugLLM config, with automatic schema derivation.
        
        Args:
            name: Name for the node
            config: AugLLM configuration
            next_node: Optional next node to route to
            
        Returns:
            Self for chaining
        """
        # Derive schema from config and update our schema
        self.schema_manager = SchemaComposer.derive_schema_from_aug_llm(
            config, 
            self.schema_manager
        )
        
        # Rebuild state model with updated schema
        self.state_model = self.schema_manager.get_model()
        self.graph = StateGraph(self.state_model)
        
        # Rebuild existing nodes with new schema
        self._rebuild_existing_nodes()
        
        # Create node function with next_node
        node_fn = create_node_function(
            config=config,
            next_node=next_node
        )
        
        # Add to graph
        self.graph.add_node(name, node_fn)
        self.nodes[name] = node_fn
        
        # Set as entry point if it's the first node
        if self.entry_point is None:
            self.entry_point = name
            self.graph.set_entry_point(name)
            
        # Add edge if next_node is specified
        if next_node is not None and next_node != END:
            self._edges.append((name, next_node))
            self.graph.add_edge(name, next_node)
            
        return self
        
    def add_node(
        self, 
        name: str, 
        node_fn: Callable,
        next_node: Optional[Union[str, Literal["END"]]] = None
    ) -> 'GraphManager':
        """
        Add a custom node function.
        
        Args:
            name: Name for the node
            node_fn: Node function
            next_node: Optional next node to route to
            
        Returns:
            Self for chaining
        """
        # Add to graph
        self.graph.add_node(name, node_fn)
        self.nodes[name] = node_fn
        
        # Set as entry point if it's the first node
        if self.entry_point is None:
            self.entry_point = name
            self.graph.set_entry_point(name)
            
        # Add edge if next_node is specified
        if next_node is not None and next_node != END:
            self._edges.append((name, next_node))
            self.graph.add_edge(name, next_node)
            
        return self
        
    def create_generalized_tool_node(
        self,
        name: str,
        tools: List[Any],
        post_processing_fn: Optional[Callable] = None,
        next_node: Optional[Union[str, Literal["END"]]] = None
    ) -> 'GraphManager':
        """
        Create a tool node with optional post-processing.
        
        Args:
            name: Name for the node
            tools: List of tools for this node
            post_processing_fn: Optional function to process tool outputs
            next_node: Optional next node to route to
            
        Returns:
            Self for chaining
        """
        # Define tool node function
        def tool_node_with_processing(state):
            # Convert Pydantic model to dict if needed
            state_dict = state.model_dump() if hasattr(state, "model_dump") else state
            
            # Create standard tool node
            tool_node = ToolNode(tools)
            
            # Process the state
            result = tool_node.invoke(state_dict)
            
            # Apply post-processing if provided
            if post_processing_fn:
                processed_result = post_processing_fn(result)
                if isinstance(processed_result, Command):
                    return processed_result
                return Command(update=processed_result, goto=next_node or "")
            
            # If no processing or processing didn't return a Command
            if next_node:
                return Command(update=result, goto=next_node)
            return Command(update=result)
            
        # Add to graph
        self.add_node(name, tool_node_with_processing, next_node)
        
        # Track tools
        self._tools[name] = tools
        
        return self
        
    def add_edge(
        self, 
        from_node: str, 
        to_node: Union[str, Literal["END"]]
    ) -> 'GraphManager':
        """
        Add an edge between nodes.
        
        Args:
            from_node: Source node name
            to_node: Target node name or END
            
        Returns:
            Self for chaining
        """
        # Skip if to_node is END (handled by node function)
        if to_node != END:
            self._edges.append((from_node, to_node))
            self.graph.add_edge(from_node, to_node)
            
        return self
        
    def add_conditional_edge(
        self,
        from_node: str,
        condition_fn: Callable,
        edge_map: Dict[str, Union[str, Literal["END"]]]
    ) -> 'GraphManager':
        """
        Add conditional edges based on a routing function.
        
        Args:
            from_node: Source node name
            condition_fn: Function that determines routing
            edge_map: Mapping from condition results to target nodes
            
        Returns:
            Self for chaining
        """
        self._conditional_edges.append((from_node, condition_fn, edge_map))
        self.graph.add_conditional_edges(from_node, condition_fn, edge_map)
        return self
        
    def create_conditional_tool_routing(
        self,
        from_node: str,
        condition_fn: Callable,
        tool_routes: Dict[str, Tuple[List[Any], Optional[Callable], Optional[str]]]
    ) -> 'GraphManager':
        """
        Create conditional routing to different tool nodes based on a condition.
        
        Args:
            from_node: Source node name
            condition_fn: Function that returns a routing key
            tool_routes: Dict mapping routing keys to (tools, post_processor, next_node)
            
        Returns:
            Self for chaining
        """
        # Create tool nodes for each route
        edge_map = {}
        
        for route_key, (tools, processor, next_node) in tool_routes.items():
            # Create unique tool node name
            tool_node_name = f"{from_node}_{route_key}_tools"
            
            # Create the tool node
            self.create_generalized_tool_node(
                name=tool_node_name,
                tools=tools,
                post_processing_fn=processor,
                next_node=next_node
            )
            
            # Add to edge map
            edge_map[route_key] = tool_node_name
            
        # Add conditional edge
        self.add_conditional_edge(from_node, condition_fn, edge_map)
        
        return self
        
    def create_sequential_chain(
        self,
        configs: List[Union[AugLLMConfig, Tuple[str, Callable]]],
        names: Optional[List[str]] = None
    ) -> 'GraphManager':
        """
        Create a sequential chain of nodes with automatic routing.
        
        Args:
            configs: List of AugLLM configs or (name, function) tuples
            names: Optional list of names for AugLLM configs
            
        Returns:
            Self for chaining
        """
        # Track nodes to add edges between them
        chain_nodes = []
        
        # Process each configuration or function
        for i, config_or_fn in enumerate(configs):
            if isinstance(config_or_fn, AugLLMConfig):
                # It's an AugLLM config
                config = config_or_fn
                
                # Determine node name
                if names and i < len(names):
                    node_name = names[i]
                else:
                    node_name = config.name or f"node_{i}"
                    
                # Determine next node
                if i < len(configs) - 1:
                    # Not the last node - will connect to next
                    next_node = None  # We'll add edges later
                else:
                    # Last node - route to END
                    next_node = END
                    
                # Add the node
                self.add_node_from_aug_llm(node_name, config, next_node)
                chain_nodes.append(node_name)
                
            elif isinstance(config_or_fn, tuple) and len(config_or_fn) == 2:
                # It's a (name, function) tuple
                node_name, node_fn = config_or_fn
                
                # Determine next node
                if i < len(configs) - 1:
                    # Not the last node - will connect to next
                    next_node = None  # We'll add edges later
                else:
                    # Last node - route to END
                    next_node = END
                    
                # Add the node
                self.add_node(node_name, node_fn, next_node)
                chain_nodes.append(node_name)
                
            else:
                raise ValueError(f"Unsupported config type: {type(config_or_fn)}")
                
        # Add edges between nodes
        for i in range(len(chain_nodes) - 1):
            self.add_edge(chain_nodes[i], chain_nodes[i + 1])
            
        return self
    
    def set_entry_point(self, node_name: str) -> 'GraphManager':
        """
        Set the entry point for the graph.
        
        Args:
            node_name: Name of the entry point node
            
        Returns:
            Self for chaining
        """
        self.entry_point = node_name
        self.graph.set_entry_point(node_name)
        return self
    
    def build(self, checkpointer=None):
        """
        Build and compile the graph.
        
        Args:
            checkpointer: Optional checkpoint saver
            
        Returns:
            Compiled graph application
        """
        if self.entry_point is None:
            raise ValueError("No entry point defined - add at least one node")
            
        return self.graph.compile(checkpointer=checkpointer)
    
    def _rebuild_existing_nodes(self):
        """Rebuild existing nodes after schema changes."""
        for name, node_fn in self.nodes.items():
            self.graph.add_node(name, node_fn)
            
        for from_node, to_node in self._edges:
            if to_node != END:
                self.graph.add_edge(from_node, to_node)
                
        for from_node, condition_fn, edge_map in self._conditional_edges:
            self.graph.add_conditional_edges(from_node, condition_fn, edge_map)
            
        if self.entry_point:
            self.graph.set_entry_point(self.entry_point)


# Common routing patterns

def message_contains_routing(state, phrase_to_route_map: Dict[str, str], default_route: str = "default"):
    """
    Route based on phrases in the last message.
    
    Args:
        state: The state object
        phrase_to_route_map: Map of phrases to route keys
        default_route: Default route if no phrase matches
        
    Returns:
        Route key
    """
    # Get last message content
    messages = filter_messages(state.messages, exclude_types=[SystemMessage])
    if not messages:
        return default_route
        
    last_message = messages[-1]
    if not hasattr(last_message, "content"):
        return default_route
        
    content = last_message.content.lower()
    
    # Check for phrases
    for phrase, route in phrase_to_route_map.items():
        if phrase.lower() in content:
            return route
            
    return default_route


def tool_name_routing(state, tool_name_to_route_map: Dict[str, str], default_route: str = "end"):
    """
    Route based on the tool name in tool calls.
    
    Args:
        state: The state object
        tool_name_to_route_map: Map of tool names to route keys
        default_route: Default route if no tool calls or no match
        
    Returns:
        Route key
    """
    # Get last message
    messages = filter_messages(state.messages, exclude_types=[SystemMessage])
    if not messages:
        return default_route
        
    last_message = messages[-1]
    
    # Check for tool calls
    if hasattr(last_message, "tool_calls") and last_message.tool_calls:
        for tool_call in last_message.tool_calls:
            tool_name = tool_call.get("name")
            if tool_name in tool_name_to_route_map:
                return tool_name_to_route_map[tool_name]
                
    return default_route


# Example usage:

def example_retrieval_workflow():
    """Example of creating a workflow with conditional retrieval."""
    from langchain_core.prompts import ChatPromptTemplate
    
    # Create AugLLM configs
    planner_config = AugLLMConfig(
        name="planner",
        prompt_template=ChatPromptTemplate.from_messages([
            ("system", "You are a research assistant. If you need to search for information, say 'I need to search for [topic]'."),
            ("user", "{messages}")
        ]),
        llm_config=AzureLLMConfig(model="gpt-4o")
    )
    
    synthesizer_config = AugLLMConfig(
        name="synthesizer",
        prompt_template=ChatPromptTemplate.from_messages([
            ("system", "You are a synthesis assistant. Synthesize the information provided into a coherent response."),
            ("user", "Original query: {original_query}\nSearch results: {search_results}\nCreate a comprehensive response.")
        ]),
        llm_config=AzureLLMConfig(model="gpt-4o")
    )
    
    # Define search tools
    search_tools = [
        Tool(
            name="search",
            description="Search for information on a topic",
            func=lambda x: f"Search results for {x}: [simulated search results]"
        )
    ]
    
    # Define routing function
    def needs_search(state):
        messages = filter_messages(state["messages"], exclude_types=[SystemMessage])
        if not messages:
            return "no_search"
            
        last_message = messages[-1]
        if hasattr(last_message, "content") and "I need to search for" in last_message.content:
            return "use_search"
        return "no_search"
    
    # Create post-processing function
    def process_search_results(state):
        # Extract original query
        original_query = "No query found"
        for msg in state["messages"]:
            if hasattr(msg, "role") and msg.role == "user":
                original_query = msg.content
                break
        
        # Extract search results - assuming last tool call output
        search_results = "No results found"
        for msg in state["messages"]:
            if hasattr(msg, "tool_call_id"):
                search_results = msg.content
        
        # Update state with these fields
        return {
            "messages": state["messages"],
            "original_query": original_query,
            "search_results": search_results
        }
    
    # Create the graph manager
    manager = GraphManager({
        "messages": [],
        "original_query": "",
        "search_results": ""
    })
    
    # Add planner node with conditional routing
    manager.add_node_from_aug_llm("planner", planner_config)
    
    # Create search tool node with post-processing
    manager.create_generalized_tool_node(
        name="search_node",
        tools=search_tools,
        post_processing_fn=process_search_results,
        next_node="synthesizer"
    )
    
    # Add synthesizer node
    manager.add_node_from_aug_llm("synthesizer", synthesizer_config, END)
    
    # Add conditional routing
    manager.add_conditional_edge(
        "planner",
        needs_search,
        {
            "use_search": "search_node",
            "no_search": END
        }
    )
    
    # Set entry point
    manager.set_entry_point("planner")
    
    # Build the graph
    app = manager.build()
    
    return app


def example_multi_tool_routing():
    """Example of routing to different tools based on content."""
    from langchain_core.prompts import ChatPromptTemplate
    
    # Create AugLLM config
    agent_config = AugLLMConfig(
        name="agent",
        prompt_template=ChatPromptTemplate.from_messages([
            ("system", "You are a helpful assistant. For calculation questions, say 'I need to calculate [expression]'. For search, say 'I need to search for [topic]'."),
            ("user", "{messages}")
        ]),
        llm_config=AzureLLMConfig(model="gpt-4o")
    )
    
    # Define different tools
    search_tools = [
        Tool(
            name="search",
            description="Search for information",
            func=lambda x: f"Search results for {x}: [simulated results]"
        )
    ]
    
    calculator_tools = [
        Tool(
            name="calculator",
            description="Perform calculations",
            func=lambda x: f"Result: {eval(x)}"
        )
    ]
    
    # Create routing conditions
    def route_by_need(state):
        messages = filter_messages(state["messages"], exclude_types=[SystemMessage])
        if not messages:
            return "default"
            
        last_message = messages[-1]
        content = last_message.content.lower() if hasattr(last_message, "content") else ""
        
        if "need to calculate" in content:
            return "calculate"
        elif "need to search" in content:
            return "search"
        return "default"
    
    # Create the graph manager
    manager = GraphManager()
    
    # Add agent node
    manager.add_node_from_aug_llm("agent", agent_config)
    
    # Create conditional tool routing
    manager.create_conditional_tool_routing(
        from_node="agent",
        condition_fn=route_by_need,
        tool_routes={
            "search": (search_tools, None, "agent"),
            "calculate": (calculator_tools, None, "agent"),
            "default": ([], None, END)
        }
    )
    
    # Set entry point
    manager.set_entry_point("agent")
    
    # Build the graph
    app = manager.build()
    
    return app
```

```python
from typing import Any, Dict, List, Type, Union, Optional, Callable, Tuple, Literal
from pydantic import BaseModel, Field, create_model
import uuid
import inspect
from langgraph.graph import StateGraph, END
from langgraph.types import Command, Send
from langchain_core.prompts import ChatPromptTemplate
from src.haive.core.aug_llm.base import AugLLMConfig, compose_runnable
from src.haive.core.models.llm.base import AzureLLMConfig


def create_node_function(
    config: AugLLMConfig,
    input_mapping: Optional[Dict[str, str]] = None,
    output_mapping: Optional[Dict[str, str]] = None,
    next_node: Optional[Union[str, Literal["END"]]] = None,
    async_mode: bool = False
):
    """
    Creates a node function from an AugLLMConfig with explicit return type annotations.
    
    Args:
        config: The AugLLM configuration to use for this node
        input_mapping: Optional mapping from state fields to AugLLM inputs
        output_mapping: Optional mapping from AugLLM outputs to state fields
        next_node: The next node to route to (if None, will default to END)
        async_mode: Whether to create an async node function
        
    Returns:
        A callable node function for use in a LangGraph with proper type annotations
    """
    # Build the runnable from config
    runnable = compose_runnable(config)
    
    # Default mappings if none provided
    if input_mapping is None:
        # Auto-detect inputs from prompt template if available
        if config.prompt_template:
            input_vars = config.prompt_template.input_variables
            input_mapping = {var: var for var in input_vars}
        else:
            # Default to messages if no prompt template
            input_mapping = {"messages": "messages"}
    
    # Output mapping defaults to empty (direct state updates)
    output_mapping = output_mapping or {}
    
    # Determine target for routing
    goto = next_node if next_node is not None else END
    
    # Create async or sync node function with explicit return annotations
    if async_mode:
        async def node_function(state):
            # Convert Pydantic model to dict if needed
            state_dict = state.model_dump() if hasattr(state, "model_dump") else state
            
            # Prepare input for the runnable based on mapping
            runnable_input = {}
            for state_key, runnable_key in input_mapping.items():
                if state_key in state_dict:
                    runnable_input[runnable_key] = state_dict[state_key]
                else:
                    print(f"Warning: State key '{state_key}' not found in state.")
            
            # Handle empty input case (add debugging)
            if not runnable_input:
                print(f"Warning: Empty input to node. State keys available: {list(state_dict.keys())}")
                print(f"Input mapping: {input_mapping}")
                # Create minimal valid input
                if config.prompt_template and hasattr(config.prompt_template, "input_variables"):
                    for var in config.prompt_template.input_variables:
                        runnable_input[var] = ""
            
            # Invoke the runnable
            result = await runnable.ainvoke(runnable_input)
            
            # Prepare state update
            update = {}
            
            # Special handling for message results (common case)
            if "messages" in state_dict and hasattr(result, "content"):
                update["messages"] = state_dict["messages"] + [result]
            elif "messages" in state_dict and isinstance(result, tuple) and len(result) == 2 and result[0] in ("ai", "assistant"):
                update["messages"] = state_dict["messages"] + [result]
            elif isinstance(result, dict):
                # Handle structured output mapping
                for output_key, state_key in output_mapping.items():
                    if output_key in result:
                        update[state_key] = result[output_key]
                
                # Default behavior: map any matching keys
                for key, value in result.items():
                    if key not in output_mapping and key in state_dict:
                        update[key] = value
            else:
                # Direct mapping for simple outputs
                for key in state_dict:
                    if key in output_mapping:
                        update[output_mapping[key]] = result
            
            # Always return a Command with explicit goto
            return Command(update=update, goto=goto)
        
        return node_function
    else:
        def node_function(state):
            # Convert Pydantic model to dict if needed
            state_dict = state.model_dump() if hasattr(state, "model_dump") else state
            
            # Prepare input for the runnable based on mapping
            runnable_input = {}
            for state_key, runnable_key in input_mapping.items():
                if state_key in state_dict:
                    runnable_input[runnable_key] = state_dict[state_key]
                else:
                    print(f"Warning: State key '{state_key}' not found in state.")
            
            # Handle empty input case
            if not runnable_input:
                print(f"Warning: Empty input to node. State keys available: {list(state_dict.keys())}")
                print(f"Input mapping: {input_mapping}")
                # Create minimal valid input
                if config.prompt_template and hasattr(config.prompt_template, "input_variables"):
                    for var in config.prompt_template.input_variables:
                        runnable_input[var] = ""
            
            # Invoke the runnable
            result = runnable.invoke(runnable_input)
            
            # Prepare state update
            update = {}
            
            # Special handling for message results (common case)
            if "messages" in state_dict and hasattr(result, "content"):
                update["messages"] = state_dict["messages"] + [result]
            elif "messages" in state_dict and isinstance(result, tuple) and len(result) == 2 and result[0] in ("ai", "assistant"):
                update["messages"] = state_dict["messages"] + [result]
            elif isinstance(result, dict):
                # Handle structured output mapping
                for output_key, state_key in output_mapping.items():
                    if output_key in result:
                        update[state_key] = result[output_key]
                
                # Default behavior: map any matching keys
                for key, value in result.items():
                    if key not in output_mapping and key in state_dict:
                        update[key] = value
            else:
                # Direct mapping for simple outputs
                for key in state_dict:
                    if key in output_mapping:
                        update[output_mapping[key]] = result
            
            # Always return a Command with explicit goto
            return Command(update=update, goto=goto)
        
        # Mark this node as one to register as a finish point if it routes to END
        if goto == END:
            # Add a metadata attribute to identify END nodes
            node_function.end_node = True
        
        return node_function


def augment_schema_for_llm(
    schema_manager,  # StateSchemaManager instance
    aug_llm_config: AugLLMConfig
) -> Any:  # Returns StateSchemaManager
    """
    Augments a state schema to accommodate the inputs and outputs of an AugLLM.
    
    Args:
        schema_manager: The state schema manager to augment
        aug_llm_config: The AugLLM configuration
        
    Returns:
        Updated schema manager
    """
    # Ensure we have required fields
    if aug_llm_config.prompt_template:
        input_vars = aug_llm_config.prompt_template.input_variables
        for var in input_vars:
            if var == "messages" and "messages" not in schema_manager.fields:
                # Add messages field with appropriate annotation
                from typing import List, Tuple, Union
                from langchain_core.messages import BaseMessage
                
                schema_manager.add_field(
                    "messages", 
                    List[Union[Tuple[str, str], BaseMessage]], 
                    default=[]
                )
    
    # Add field for structured output if specified
    if aug_llm_config.structured_output_model:
        model_class = aug_llm_config.structured_output_model
        field_name = model_class.__name__.lower()
        
        # Check if the field already exists
        if field_name not in schema_manager.fields:
            # Add field with None default to ensure it's treated as Optional
            schema_manager.add_field(field_name, Optional[model_class], default=None)
    
    return schema_manager


class NodeConfig(BaseModel):
    """Configuration for a node in a LangGraph."""
    name: str = Field(default="default_node")
    engine: AugLLMConfig
    input_mapping: Optional[Dict[str, str]] = None
    output_mapping: Optional[Dict[str, str]] = None
    routing_logic: Optional[Callable] = None
    next_node: Optional[Union[str, Literal["END"]]] = None
    
    def create_node(self) -> Callable:
        """Creates a callable node function based on this configuration."""
        return create_node_function(
            config=self.engine,
            input_mapping=self.input_mapping,
            output_mapping=self.output_mapping,
            next_node=self.next_node
        )


class GraphBuilder:
    """Builder for constructing LangGraph graphs with explicit routing."""
    
    def __init__(self, state_schema):
        """
        Initialize with a state schema.
        
        Args:
            state_schema: Can be a StateSchemaManager, BaseModel, or dict
        """
        # Handle different schema input types
        if hasattr(state_schema, "get_model"):
            # It's a StateSchemaManager
            self.schema_manager = state_schema
            self.state_schema = self.schema_manager.get_model()
        elif isinstance(state_schema, dict):
            # It's a dict, convert to model
            #from state_schema_manager import StateSchemaManager
            self.schema_manager = StateSchemaManager(state_schema)
            self.state_schema = self.schema_manager.get_model()
        elif isinstance(state_schema, type) and issubclass(state_schema, BaseModel):
            # It's already a Pydantic model class
            self.schema_manager = None
            self.state_schema = state_schema
        else:
            # Assume it's a model instance or class
            self.schema_manager = None
            self.state_schema = state_schema
        
        # Initialize the graph
        self.graph = StateGraph(self.state_schema)
        self.nodes = {}
        self.entry_point = None
        
    def add_node(self, 
                 node_config: Union[NodeConfig, AugLLMConfig], 
                 name: Optional[str] = None,
                 next_node: Optional[Union[str, Literal["END"]]] = None):
        """
        Add a node to the graph with explicit next node.
        
        Args:
            node_config: Either a NodeConfig or AugLLMConfig
            name: Optional name for the node (required if using AugLLMConfig)
            next_node: Where this node should route to next (END by default)
            
        Returns:
            self for method chaining
        """
        # Convert AugLLMConfig to NodeConfig if needed
        if isinstance(node_config, AugLLMConfig):
            node_name = name or node_config.name or f"node_{len(self.nodes)}"
            node_config = NodeConfig(
                name=node_name,
                engine=node_config,
                next_node=next_node
            )
        else:
            # If next_node was explicitly provided, override the config
            if next_node is not None:
                node_config.next_node = next_node
                
        # Create and add the node function
        node_fn = node_config.create_node()
        self.nodes[node_config.name] = node_fn
        self.graph.add_node(node_config.name, node_fn)
        
        # Set as entry point if it's the first node
        if self.entry_point is None:
            self.entry_point = node_config.name
            self.graph.set_entry_point(node_config.name)
        
        return self
    
    def add_edge(self, from_node: str, to_node: Union[str, Literal["END"]]):
        """
        Add an explicit edge between nodes, handling END specially.
        
        Args:
            from_node: The source node name
            to_node: The target node name or "END"
            
        Returns:
            self for method chaining
        """
        # Don't add explicit edges to END as they're handled via the node function
        if to_node != END:
            self.graph.add_edge(from_node, to_node)
        return self
    
    def add_conditional_edge(self, 
                           from_node: str, 
                           condition_fn: Callable, 
                           edges_map: Dict[str, Union[str, Literal["END"]]]):
        """
        Add conditional edges based on a routing function.
        
        Args:
            from_node: The source node name  
            condition_fn: Function that takes state and returns a routing key
            edges_map: Mapping from routing keys to target nodes
            
        Returns:
            self for method chaining
        """
        self.graph.add_conditional_edges(from_node, condition_fn, edges_map)
        return self
    
    def build(self, checkpointer=None):
        """
        Compile the graph into an executable app.
        
        Args:
            checkpointer: Optional checkpoint saver
            
        Returns:
            Compiled graph application
        """
        if self.entry_point is None:
            raise ValueError("No entry point defined - add at least one node")
        
        return self.graph.compile(checkpointer=checkpointer)


def build_graph_from_configs(
    configs: Dict[str, AugLLMConfig],
    edges: List[Tuple[str, Union[str, Literal["END"]]]],
    conditional_edges: Optional[List[Tuple[str, Callable, Dict[str, Union[str, Literal["END"]]]]]] = None,
    state_schema=None,  # StateSchemaManager or BaseModel or dict
    entry_point: Optional[str] = None
):
    """
    Builds a LangGraph from AugLLM configurations with explicit routing.
    
    Args:
        configs: Dictionary mapping node names to AugLLM configs
        edges: List of (from_node, to_node) tuples defining graph edges
        conditional_edges: List of (from_node, condition_fn, edge_map) for conditional routing
        state_schema: Base state schema to use
        entry_point: Name of the node to use as entry point (defaults to first node)
        
    Returns:
        Compiled LangGraph application
    """
    # Import only where needed
    #from state_schema_manager import StateSchemaManager
    
    # Initialize state schema
    if state_schema is None:
        schema_manager = StateSchemaManager({"messages": []})
    elif isinstance(state_schema, dict):
        schema_manager = StateSchemaManager(state_schema)
    elif hasattr(state_schema, "get_model"):
        schema_manager = state_schema
    else:
        schema_manager = StateSchemaManager(state_schema)
    
    # Augment schema for all configs
    for node_name, config in configs.items():
        schema_manager = augment_schema_for_llm(schema_manager, config)
    
    # Get the finalized state model
    state_model = schema_manager.get_model()
    
    # Build the graph using the model, not the manager
    builder = GraphBuilder(state_model)
    
    # Add nodes with their next destinations based on edges
    for node_name, config in configs.items():
        # Find the explicit next node from edges
        next_nodes = [to_node for from_node, to_node in edges if from_node == node_name]
        next_node = next_nodes[0] if len(next_nodes) == 1 else END
        
        # Add the node with explicit next_node
        builder.add_node(config, name=node_name, next_node=next_node)
    
    # Set entry point
    if entry_point:
        builder.graph.set_entry_point(entry_point)
    
    # Add conditional edges
    if conditional_edges:
        for from_node, condition_fn, edge_map in conditional_edges:
            builder.add_conditional_edge(from_node, condition_fn, edge_map)
    
    # Compile graph
    return builder.build()


def create_agent_from_aug_llm(
    aug_llm_config: AugLLMConfig,
    state_schema=None  # Can be StateSchemaManager, BaseModel, or dict
):
    """
    Create a simple agent from an AugLLM config with explicit END routing.
    
    Args:
        aug_llm_config: The AugLLM configuration
        state_schema: Optional state schema
        
    Returns:
        Compiled graph application
    """
    # Import only where needed
    #from state_schema_manager import StateSchemaManager
    
    # Initialize schema
    if state_schema is None:
        schema_manager = StateSchemaManager({"messages": []})
    elif isinstance(state_schema, dict):
        schema_manager = StateSchemaManager(state_schema)
    elif hasattr(state_schema, "get_model"):
        schema_manager = state_schema
    else:
        schema_manager = StateSchemaManager(state_schema)
    
    # Augment schema for this config
    schema_manager = augment_schema_for_llm(schema_manager, aug_llm_config)
    
    # Create a node configuration with explicit END routing
    node_config = NodeConfig(
        name="agent_node",
        engine=aug_llm_config,
        next_node=END  # Explicit END routing
    )
    
    # Build the graph
    builder = GraphBuilder(schema_manager)
    builder.add_node(node_config)
    
    # Build and return the executable
    return builder.build()
```

```python
from typing import Dict, Any, List, Union, Optional, Type, Callable, Literal
from pydantic import BaseModel, Field
import uuid
import os
import json
import inspect

from langgraph.graph import StateGraph, END
from langgraph.checkpoint.memory import MemorySaver
from langgraph.types import Command, Send

from src.haive.agents.base import AgentArchitectureConfig, AgentArchitecture, register_agent
from src.haive.core.aug_llm.base import AugLLMConfig, compose_runnable
from src.haive.core.models.llm.base import AzureLLMConfig
from src.haive.utils.visualize_graph_utils import render_and_display_graph
from config.settings import RESOURCES_DIR

# Import our graph/node utilities
#from node_factory_refined import (
#    create_node_function,
#    augment_schema_for_llm,
#    NodeConfig,
#    GraphBuilder
#)


class BaseAgentConfig(AgentArchitectureConfig):
    """
    Configuration for the simplest possible agent architecture.
    Wraps an AugLLM in a single-node graph.
    """
    engine: Union[AugLLMConfig, AgentArchitectureConfig] = Field(
        default_factory=lambda: AugLLMConfig(llm_config=AzureLLMConfig(model="gpt-4o")),
        description="LLM configuration for the BaseAgent."
    )
    input_mapping: Optional[Dict[str, str]] = Field(
        default=None,
        description="Maps state fields to LLM input fields."
    )
    output_mapping: Optional[Dict[str, str]] = Field(
        default=None,
        description="Maps LLM output fields to state fields."
    )
    node_name: str = Field(
        default="agent_node",
        description="Name for the single node in the graph."
    )
    should_setup_workflow: bool = Field(
        default=True,
        description="Whether to set up the workflow."
    )
    should_compile: bool = Field(
        default=True,
        description="Whether to compile the graph."
    )
    should_visualize_graph: bool = Field(
        default=False,
        description="Whether to visualize the graph."
    )
    visualize_graph_output_name: Optional[str] = Field(
        default=None,
        description="Output file for graph visualization."
    )
    
    def build_agent(self) -> "BaseAgent":
        """Build the BaseAgent from this configuration."""
        # Ensure state_schema is a Pydantic model
        if isinstance(self.state_schema, dict):
            #from state_schema_manager import StateSchemaManager
            schema_manager = StateSchemaManager(self.state_schema)
            self.state_schema = schema_manager.get_model()
        return BaseAgent(config=self)


# The registration will happen at the end of the file
class BaseAgent(AgentArchitecture):
    """
    The simplest possible agent architecture. Wraps an AugLLM in a single-node graph.
    """
    def __init__(self, config: BaseAgentConfig):
        self.input_mapping = config.input_mapping
        self.output_mapping = config.output_mapping
        self.node_name = config.node_name
        
        # Initialize the base agent
        super().__init__(config)
    
    def setup_workflow(self):
        """
        Set up a simple single-node workflow.
        
        The workflow consists of a single node that processes input through the AugLLM
        and routes directly to END.
        """
        # Create the node function with explicit END routing
        node_fn = create_node_function(
            config=self.config.engine if isinstance(self.config.engine, AugLLMConfig) else AugLLMConfig(),
            input_mapping=self.input_mapping,
            output_mapping=self.output_mapping,
            next_node=END  # Explicit END routing
        )
        
        # Add the node to the graph
        self.graph.add_node(self.node_name, node_fn)
        
        # Set the entry point
        self.graph.set_entry_point(self.node_name)
        
        # Also set as finish point since it routes to END
        self.graph.set_finish_point(self.node_name)


class MultiNodeAgentConfig(AgentArchitectureConfig):
    """
    Configuration for a more complex agent with multiple nodes.
    """
    nodes: Dict[str, Union[AugLLMConfig, NodeConfig]] = Field(
        default_factory=dict,
        description="Dictionary of nodes to include in the graph."
    )
    edges: List[tuple] = Field(
        default_factory=list,
        description="List of (from_node, to_node) tuples defining edges."
    )
    conditional_edges: List[tuple] = Field(
        default_factory=list,
        description="List of (from_node, condition_fn, edge_map) tuples for conditional routing."
    )
    entry_point: Optional[str] = Field(
        default=None,
        description="Name of the entry point node (defaults to first node)."
    )
    
    def build_agent(self) -> "MultiNodeAgent":
        """Build the MultiNodeAgent from this configuration."""
        return MultiNodeAgent(config=self)


# The registration will happen at the end of the file
class MultiNodeAgent(AgentArchitecture):
    """
    A more complex agent with multiple nodes and custom routing.
    """
    def __init__(self, config: MultiNodeAgentConfig):
        self.nodes = config.nodes
        self.edges = config.edges
        self.conditional_edges = config.conditional_edges
        self.entry_point = config.entry_point
        
        # Initialize the base agent
        super().__init__(config)
    
    def setup_workflow(self):
        """
        Set up a multi-node workflow based on the configuration.
        """
        # Convert any AugLLMConfigs to NodeConfigs
        node_configs = {}
        for name, config in self.nodes.items():
            if isinstance(config, AugLLMConfig):
                # Find explicit next node from edges
                next_nodes = [to_node for from_node, to_node in self.edges if from_node == name]
                next_node = next_nodes[0] if len(next_nodes) == 1 else END
                
                node_configs[name] = NodeConfig(
                    name=name,
                    engine=config,
                    next_node=next_node
                )
            else:
                node_configs[name] = config
        
        # Add all nodes to the graph
        for name, config in node_configs.items():
            node_fn = config.create_node()
            self.graph.add_node(name, node_fn)
            
            # Check if this node routes to END, set as finish point
            if hasattr(node_fn, 'end_node') and node_fn.end_node:
                self.graph.set_finish_point(name)
            # Also check the NodeConfig routing
            elif hasattr(config, 'next_node') and config.next_node == END:
                self.graph.set_finish_point(name)
        
        # Set the entry point
        entry_node = self.entry_point or list(node_configs.keys())[0]
        self.graph.set_entry_point(entry_node)
        
        # Add all edges explicitly without conditionals
        # This is simpler than selective edge addition and avoids branch issues
        for from_node, to_node in self.edges:
            if to_node != END:
                self.graph.add_edge(from_node, to_node)
            # For END nodes, the routing is handled by the node function
        
        # Add conditional edges
        for from_node, condition_fn, edge_map in self.conditional_edges:
            self.graph.add_conditional_edges(from_node, condition_fn, edge_map)
            # Check if any routes in edge_map go to END
            for key, target in edge_map.items():
                if target == END:
                    self.graph.set_finish_point(from_node)


# Register agents
register_agent(BaseAgentConfig, BaseAgent)
register_agent(MultiNodeAgentConfig, MultiNodeAgent)


# Factory functions for easier agent creation
def create_base_agent(aug_llm_config: AugLLMConfig, state_schema=None):
    """
    Create a simple BaseAgent from an AugLLM config.
    
    Args:
        aug_llm_config: The AugLLM configuration
        state_schema: Optional custom state schema
        
    Returns:
        Initialized BaseAgent instance
    """
    # Convert state_schema to a proper model if it's a dict
    if state_schema is None:
        #from state_schema_manager import StateSchemaManager
        schema_manager = StateSchemaManager({"messages": []})
        state_schema = schema_manager.get_model()
    elif isinstance(state_schema, dict):
        #from state_schema_manager import StateSchemaManager
        schema_manager = StateSchemaManager(state_schema)
        state_schema = schema_manager.get_model()
        
    config = BaseAgentConfig(
        engine=aug_llm_config,
        state_schema=state_schema
    )
    return BaseAgent(config)


def create_multi_node_agent(
    nodes: Dict[str, AugLLMConfig],
    edges: List[tuple],
    conditional_edges: Optional[List[tuple]] = None,
    state_schema=None,
    entry_point: Optional[str] = None
):
    """
    Create a MultiNodeAgent with the specified configuration.
    
    Args:
        nodes: Dictionary mapping node names to AugLLM configs
        edges: List of (from_node, to_node) tuples defining edges
        conditional_edges: Optional list of conditional edge definitions
        state_schema: Optional custom state schema
        entry_point: Optional entry point node name
        
    Returns:
        Initialized MultiNodeAgent instance
    """
    # Convert state_schema to a proper model if it's a dict
    if state_schema is None:
        #from state_schema_manager import StateSchemaManager
        schema_manager = StateSchemaManager({"messages": []})
        state_schema = schema_manager.get_model()
    elif isinstance(state_schema, dict):
        #from state_schema_manager import StateSchemaManager
        schema_manager = StateSchemaManager(state_schema)
        state_schema = schema_manager.get_model()
        
    config = MultiNodeAgentConfig(
        nodes=nodes,
        edges=edges,
        conditional_edges=conditional_edges or [],
        entry_point=entry_point,
        state_schema=state_schema
    )
    return MultiNodeAgent(config)
```

```python
from typing import Dict, Any, List, Union, Optional, Callable, Tuple, Literal, Type
from pydantic import BaseModel, Field
import uuid
import inspect

from langgraph.graph import StateGraph, END
from langgraph.types import Command, Send
from langchain_core.prompts import ChatPromptTemplate
from src.haive.core.aug_llm.base import AugLLMConfig, compose_runnable
from src.haive.core.models.llm.base import AzureLLMConfig

class NodeDefinition:
    """
    Defines a node in a graph with its function and edges.
    """
    def __init__(self, 
                 name: str, 
                 function: Callable,
                 outgoing_edges: Optional[Dict[str, str]] = None):
        """
        Initialize a node definition.
        
        Args:
            name: Name of the node
            function: The node function
            outgoing_edges: Optional mapping of condition to target node
        """
        self.name = name
        self.function = function
        self.outgoing_edges = outgoing_edges or {}
        
    def add_edge(self, target: str, condition: Optional[str] = None):
        """
        Add an edge from this node to a target node.
        
        Args:
            target: Target node name
            condition: Optional condition for this edge
            
        Returns:
            Self for chaining
        """
        if condition:
            self.outgoing_edges[condition] = target
        else:
            # Default edge
            self.outgoing_edges["default"] = target
        return self
        
    def add_conditional_edges(self, condition_to_target: Dict[str, str]):
        """
        Add multiple conditional edges.
        
        Args:
            condition_to_target: Mapping from condition to target
            
        Returns:
            Self for chaining
        """
        self.outgoing_edges.update(condition_to_target)
        return self


def create_node_function(
    config: AugLLMConfig,
    input_mapping: Optional[Dict[str, str]] = None,
    output_mapping: Optional[Dict[str, str]] = None,
    result_field: Optional[str] = None
):
    """
    Creates a node function from an AugLLM configuration.
    Returns state updates without explicit routing.
    
    Args:
        config: The AugLLM configuration to use for this node
        input_mapping: Optional mapping from state fields to AugLLM inputs
        output_mapping: Optional mapping from AugLLM outputs to state fields
        result_field: Optional field name to store processing result
        
    Returns:
        A node function that returns a dict of state updates
    """
    # Build the runnable from config
    runnable = compose_runnable(config)
    
    # Default mappings if none provided
    if input_mapping is None:
        # Auto-detect inputs from prompt template if available
        if config.prompt_template:
            input_vars = config.prompt_template.input_variables
            input_mapping = {var: var for var in input_vars}
        else:
            # Default to messages if no prompt template
            input_mapping = {"messages": "messages"}
    
    # Output mapping defaults to empty (direct state updates)
    output_mapping = output_mapping or {}
    
    def node_function(state):
        # Convert Pydantic model to dict if needed
        state_dict = state.model_dump() if hasattr(state, "model_dump") else state
        
        # Prepare input for the runnable based on mapping
        runnable_input = {}
        for state_key, runnable_key in input_mapping.items():
            if state_key in state_dict:
                runnable_input[runnable_key] = state_dict[state_key]
            else:
                print(f"Warning: State key '{state_key}' not found in state.")
        
        # Handle empty input case
        if not runnable_input:
            print(f"Warning: Empty input to node. State keys available: {list(state_dict.keys())}")
            print(f"Input mapping: {input_mapping}")
            # Create minimal valid input
            if config.prompt_template and hasattr(config.prompt_template, "input_variables"):
                for var in config.prompt_template.input_variables:
                    runnable_input[var] = ""
        
        # Invoke the runnable
        result = runnable.invoke(runnable_input)
        
        # Prepare state update
        update = {}
        
        # Special handling for message results (common case)
        if "messages" in state_dict and hasattr(result, "content"):
            update["messages"] = state_dict["messages"] + [result]
        elif "messages" in state_dict and isinstance(result, tuple) and len(result) == 2 and result[0] in ("ai", "assistant"):
            update["messages"] = state_dict["messages"] + [result]
        elif isinstance(result, dict):
            # Handle structured output mapping
            for output_key, state_key in output_mapping.items():
                if output_key in result:
                    update[state_key] = result[output_key]
            
            # Default behavior: map any matching keys
            for key, value in result.items():
                if key not in output_mapping and key in state_dict:
                    update[key] = value
        else:
            # Direct mapping for simple outputs
            for key in state_dict:
                if key in output_mapping:
                    update[output_mapping[key]] = result
        
        # Store full result if result_field is specified
        if result_field:
            update[result_field] = result
            
        return update
    
    return node_function


def create_conditional_router(
    router_function: Callable,
    condition_map: Dict[str, str]
):
    """
    Creates a node that routes based on a condition function.
    
    Args:
        router_function: Function that takes state and returns a routing key
        condition_map: Mapping from routing keys to target nodes
        
    Returns:
        A node function that determines routing
    """
    def router_node(state):
        # Get routing decision
        routing_key = router_function(state)
        
        # Determine target node
        if routing_key in condition_map:
            goto = condition_map[routing_key]
        elif "default" in condition_map:
            goto = condition_map["default"]
        else:
            goto = END
            
        # Return command with no state updates
        return Command(goto=goto)
    
    return router_node


def create_tool_node(
    tools: List[Any],
    post_processing_fn: Optional[Callable] = None
):
    """
    Creates a tool execution node with optional post-processing.
    
    Args:
        tools: List of tools
        post_processing_fn: Optional function to process the results
        
    Returns:
        A node function for tool execution
    """
    from langgraph.prebuilt import ToolNode
    
    def tool_node_function(state):
        # Create the tool node
        tool_node = ToolNode(tools)
        
        # Process state with tool node
        result = tool_node.invoke(state)
        
        # Apply post-processing if provided
        if post_processing_fn and result:
            if isinstance(result, dict):
                return post_processing_fn(result)
            elif hasattr(result, "update"):
                # It's a Command with update
                update_dict = result.update
                processed = post_processing_fn(update_dict)
                if isinstance(processed, dict):
                    # Return the processed dict
                    return processed
                elif hasattr(processed, "update") and hasattr(processed, "goto"):
                    # It's a Command
                    return processed
                else:
                    # Default behavior
                    return update_dict
        
        return result
    
    return tool_node_function


class GraphBuilder:
    """
    Builder for constructing graphs with edge-based routing.
    """
    
    def __init__(self, state_schema: Union[Type[BaseModel], Dict[str, Any], Any] = None):
        """
        Initialize with optional state schema.
        
        Args:
            state_schema: Optional state schema (type, dict, or StateSchemaManager)
        """
        # Process state schema
        if state_schema is None:
            # Default minimal schema
            from state_schema_manager import StateSchemaManager
            state_schema = StateSchemaManager({"messages": []}).get_model()
        elif isinstance(state_schema, dict):
            from state_schema_manager import StateSchemaManager
            state_schema = StateSchemaManager(state_schema).get_model()
        elif hasattr(state_schema, "get_model"):
            state_schema = state_schema.get_model()
            
        # Initialize graph
        self.state_schema = state_schema
        self.graph = StateGraph(state_schema)
        
        # Track nodes and edges
        self.nodes = {}
        self.edges = []
        self.conditional_edges = []
        self.entry_point = None
        
    def add_node(self, name: str, node_function: Callable) -> NodeDefinition:
        """
        Add a node to the graph.
        
        Args:
            name: Node name
            node_function: Node function
            
        Returns:
            NodeDefinition for adding edges
        """
        # Add to graph
        self.graph.add_node(name, node_function)
        
        # Create node definition
        node_def = NodeDefinition(name, node_function)
        self.nodes[name] = node_def
        
        # Set as entry point if first node
        if self.entry_point is None:
            self.entry_point = name
            self.graph.set_entry_point(name)
            
        return node_def
    
    def add_edge(self, from_node: str, to_node: Union[str, Literal["END"]]):
        """
        Add an edge between nodes.
        
        Args:
            from_node: Source node
            to_node: Target node or END
            
        Returns:
            Self for chaining
        """
        if to_node != END:
            self.graph.add_edge(from_node, to_node)
            self.edges.append((from_node, to_node))
            
            # Update node definition
            if from_node in self.nodes:
                self.nodes[from_node].add_edge(to_node)
                
        return self
    
    def add_conditional_edges(self, 
                             from_node: str, 
                             condition_fn: Callable, 
                             condition_map: Dict[str, Union[str, Literal["END"]]]):
        """
        Add conditional edges from a node.
        
        Args:
            from_node: Source node
            condition_fn: Function that determines routing
            condition_map: Mapping of conditions to target nodes
            
        Returns:
            Self for chaining
        """
        # Filter out END nodes from the edge map for the graph
        graph_condition_map = {
            condition: target 
            for condition, target in condition_map.items() 
            if target != END
        }
        
        # Add to graph
        if graph_condition_map:
            self.graph.add_conditional_edges(from_node, condition_fn, graph_condition_map)
            
        # Track conditional edge
        self.conditional_edges.append((from_node, condition_fn, condition_map))
        
        # Update node definition
        if from_node in self.nodes:
            self.nodes[from_node].add_conditional_edges(condition_map)
            
        return self
    
    def set_entry_point(self, node_name: str):
        """
        Set the entry point for the graph.
        
        Args:
            node_name: Name of the entry node
            
        Returns:
            Self for chaining
        """
        self.entry_point = node_name
        self.graph.set_entry_point(node_name)
        return self
    
    def add_aug_llm_node(self, 
                        name: str, 
                        config: AugLLMConfig,
                        input_mapping: Optional[Dict[str, str]] = None,
                        output_mapping: Optional[Dict[str, str]] = None,
                        next_node: Optional[Union[str, Literal["END"]]] = None):
        """
        Add a node based on an AugLLM configuration.
        
        Args:
            name: Node name
            config: AugLLM configuration
            input_mapping: Optional mapping from state to LLM inputs
            output_mapping: Optional mapping from LLM outputs to state
            next_node: Optional next node
            
        Returns:
            NodeDefinition for adding edges
        """
        # Create node function
        node_fn = create_node_function(config, input_mapping, output_mapping)
        
        # Add to graph
        node_def = self.add_node(name, node_fn)
        
        # Add edge if next_node is specified
        if next_node is not None:
            self.add_edge(name, next_node)
            
        return node_def
    
    def add_tool_node(self,
                     name: str,
                     tools: List[Any],
                     post_processing_fn: Optional[Callable] = None,
                     next_node: Optional[Union[str, Literal["END"]]] = None):
        """
        Add a tool node to the graph.
        
        Args:
            name: Node name
            tools: List of tools
            post_processing_fn: Optional function for post-processing
            next_node: Optional next node
            
        Returns:
            NodeDefinition for adding edges
        """
        # Create node function
        node_fn = create_tool_node(tools, post_processing_fn)
        
        # Add to graph
        node_def = self.add_node(name, node_fn)
        
        # Add edge if next_node is specified
        if next_node is not None:
            self.add_edge(name, next_node)
            
        return node_def
    
    def add_router_node(self,
                       name: str,
                       router_function: Callable,
                       routes: Dict[str, Union[str, Literal["END"]]],
                       default_route: Optional[Union[str, Literal["END"]]] = None):
        """
        Add a router node that directs flow based on a function result.
        
        Args:
            name: Node name
            router_function: Function that takes state and returns a routing key
            routes: Mapping from routing keys to target nodes
            default_route: Default route if no match found
            
        Returns:
            NodeDefinition for adding edges
        """
        # Add default route if provided
        if default_route is not None:
            routes["default"] = default_route
            
        # Create router node function
        node_fn = create_conditional_router(router_function, routes)
        
        # Add to graph
        node_def = self.add_node(name, node_fn)
        
        # Add conditional edges
        self.add_conditional_edges(name, router_function, routes)
        
        return node_def
    
    def add_sequence(self,
                    name_prefix: str,
                    nodes: List[Tuple[str, Callable]],
                    final_node: Optional[Union[str, Literal["END"]]] = None):
        """
        Add a sequence of nodes with automatic edge connections.
        
        Args:
            name_prefix: Prefix for generated node names
            nodes: List of (name_suffix, node_function) tuples
            final_node: Optional node to connect the last node to
            
        Returns:
            List of node definitions
        """
        if not nodes:
            return []
            
        node_defs = []
        prev_node = None
        
        # Add each node in sequence
        for i, (name_suffix, node_fn) in enumerate(nodes):
            node_name = f"{name_prefix}_{name_suffix}"
            
            # Add node
            node_def = self.add_node(node_name, node_fn)
            node_defs.append(node_def)
            
            # Connect to previous node
            if prev_node:
                self.add_edge(prev_node, node_name)
                
            prev_node = node_name
            
        # Connect last node to final node if specified
        if final_node is not None and prev_node:
            self.add_edge(prev_node, final_node)
            
        return node_defs
    
    def compile(self, checkpointer=None):
        """
        Compile the graph.
        
        Args:
            checkpointer: Optional checkpointer
            
        Returns:
            Compiled graph
        """
        if not self.entry_point:
            raise ValueError("No entry point defined. Add at least one node first.")
            
        return self.graph.compile(checkpointer=checkpointer)
```

```python
from typing import Dict, Any, List, Union, Optional, Type, Callable, Literal
from pydantic import BaseModel, Field
import uuid
import os
import json
import inspect

from langgraph.graph import StateGraph, END
from langgraph.checkpoint.memory import MemorySaver
from langgraph.types import Command, Send

from src.haive.agents.base import AgentArchitectureConfig, AgentArchitecture, register_agent
from src.haive.core.aug_llm.base import AugLLMConfig, compose_runnable
from src.haive.core.models.llm.base import AzureLLMConfig
from src.haive.utils.visualize_graph_utils import render_and_display_graph
from config.settings import RESOURCES_DIR


class EdgeBasedAgentConfig(AgentArchitectureConfig):
    """
    Configuration for an agent based on edge-defined graphs.
    """
    nodes: Dict[str, Union[AugLLMConfig, Callable]] = Field(
        default_factory=dict,
        description="Dictionary of nodes to include in the graph."
    )
    edges: List[Tuple[str, str]] = Field(
        default_factory=list,
        description="List of (from_node, to_node) tuples defining edges."
    )
    conditional_edges: List[Tuple[str, Callable, Dict[str, str]]] = Field(
        default_factory=list,
        description="List of (from_node, condition_fn, edge_map) tuples for conditional routing."
    )
    entry_point: Optional[str] = Field(
        default=None,
        description="Name of the entry point node (defaults to first node)."
    )
    input_mappings: Dict[str, Dict[str, str]] = Field(
        default_factory=dict,
        description="Mapping of node name to input mappings."
    )
    output_mappings: Dict[str, Dict[str, str]] = Field(
        default_factory=dict,
        description="Mapping of node name to output mappings."
    )
    tools: List[Any] = Field(
        default_factory=list,
        description="Tools available to the agent."
    )
    tool_node_map: Dict[str, List[str]] = Field(
        default_factory=dict,
        description="Mapping of node names to tool names."
    )
    
    def build_agent(self) -> "EdgeBasedAgent":
        """Build the EdgeBasedAgent from this configuration."""
        # Ensure state_schema is a Pydantic model
        if isinstance(self.state_schema, dict):
            from state_schema_manager import StateSchemaManager
            schema_manager = StateSchemaManager(self.state_schema)
            self.state_schema = schema_manager.get_model()
            
        return EdgeBasedAgent(config=self)


class EdgeBasedAgent(AgentArchitecture):
    """
    Agent architecture based on explicit edge definitions and routing.
    """
    def __init__(self, config: EdgeBasedAgentConfig):
        self.node_configs = config.nodes
        self.edges = config.edges
        self.conditional_edges = config.conditional_edges
        self.entry_point = config.entry_point
        self.input_mappings = config.input_mappings
        self.output_mappings = config.output_mappings
        self.tools = config.tools
        self.tool_node_map = config.tool_node_map
        
        # Initialize base agent
        super().__init__(config)
    
    def setup_workflow(self):
        """
        Set up the workflow with edge-based routing.
        """
        # Create a GraphBuilder with our state schema
        builder = GraphBuilder(self.state_schema)
        
        # Add all regular nodes
        for name, config in self.node_configs.items():
            if isinstance(config, AugLLMConfig):
                # Get input/output mappings for this node
                input_mapping = self.input_mappings.get(name)
                output_mapping = self.output_mappings.get(name)
                
                # Create LLM node
                builder.add_aug_llm_node(
                    name, 
                    config, 
                    input_mapping=input_mapping,
                    output_mapping=output_mapping
                )
            elif callable(config):
                # It's a direct function
                builder.add_node(name, config)
        
        # Add tool nodes
        for node_name, tool_names in self.tool_node_map.items():
            # Get tools for this node
            node_tools = [tool for tool in self.tools if getattr(tool, "name", None) in tool_names]
            
            # Create a tool node
            builder.add_tool_node(node_name, node_tools)
        
        # Add all edges
        for from_node, to_node in self.edges:
            builder.add_edge(from_node, to_node)
        
        # Add conditional edges
        for from_node, condition_fn, edge_map in self.conditional_edges:
            builder.add_conditional_edges(from_node, condition_fn, edge_map)
        
        # Set entry point
        if self.entry_point:
            builder.set_entry_point(self.entry_point)
        
        # Store the built graph
        self.graph = builder.graph
    
    def _extract_agent_name(self) -> str:
        """
        Generate a unique agent name based on class and UUID.
        """
        agent_class_name = self.__class__.__name__
        unique_id = str(uuid.uuid4())[:8]
        return f"{agent_class_name}_{unique_id}"


class SimpleAgentConfig(EdgeBasedAgentConfig):
    """
    Configuration for a simple agent with a single LLM node.
    """
    llm_config: AugLLMConfig = Field(
        default_factory=lambda: AugLLMConfig(llm_config=AzureLLMConfig(model="gpt-4o")),
        description="LLM configuration for the agent."
    )
    node_name: str = Field(
        default="agent_node",
        description="Name for the agent node."
    )
    
    def build_agent(self) -> "SimpleAgent":
        """Build SimpleAgent from this configuration."""
        # Ensure state_schema is a Pydantic model
        if isinstance(self.state_schema, dict):
            from state_schema_manager import StateSchemaManager
            schema_manager = StateSchemaManager(self.state_schema)
            self.state_schema = schema_manager.get_model()
            
        return SimpleAgent(config=self)


class SimpleAgent(EdgeBasedAgent):
    """
    Simple agent with a single LLM node.
    """
    def __init__(self, config: SimpleAgentConfig):
        self.llm_config = config.llm_config
        self.node_name = config.node_name
        
        # Prepare EdgeBasedAgentConfig from SimpleAgentConfig
        edge_config = EdgeBasedAgentConfig(
            nodes={config.node_name: config.llm_config},
            entry_point=config.node_name,
            state_schema=config.state_schema,
            input_mappings=config.input_mappings,
            output_mappings=config.output_mappings
        )
        
        # Initialize as EdgeBasedAgent
        super().__init__(config=edge_config)


class SequentialAgentConfig(EdgeBasedAgentConfig):
    """
    Configuration for an agent with a linear sequence of nodes.
    """
    steps: List[Dict[str, Any]] = Field(
        default_factory=list,
        description="List of step configurations."
    )
    step_input_mappings: Dict[str, Dict[str, str]] = Field(
        default_factory=dict,
        description="Input mappings for steps."
    )
    step_output_mappings: Dict[str, Dict[str, str]] = Field(
        default_factory=dict,
        description="Output mappings for steps."
    )
    
    def build_agent(self) -> "SequentialAgent":
        """Build SequentialAgent from this configuration."""
        # Ensure state_schema is a Pydantic model
        if isinstance(self.state_schema, dict):
            from state_schema_manager import StateSchemaManager
            schema_manager = StateSchemaManager(self.state_schema)
            self.state_schema = schema_manager.get_model()
            
        return SequentialAgent(config=self)


class SequentialAgent(EdgeBasedAgent):
    """
    Agent with a linear sequence of steps.
    """
    def __init__(self, config: SequentialAgentConfig):
        self.steps = config.steps
        self.step_input_mappings = config.step_input_mappings
        self.step_output_mappings = config.step_output_mappings
        
        # Process steps into nodes and edges
        nodes = {}
        edges = []
        input_mappings = {}
        output_mappings = {}
        
        prev_node = None
        for i, step in enumerate(self.steps):
            # Get node name
            node_name = step.get("name", f"step_{i}")
            
            # Get config
            if "config" in step:
                node_config = step["config"]
            elif "llm_config" in step:
                node_config = step["llm_config"]
            elif "function" in step:
                node_config = step["function"]
            else:
                raise ValueError(f"Step {i} has no config, llm_config, or function")
                
            # Add to nodes
            nodes[node_name] = node_config
            
            # Add edge from previous node
            if prev_node:
                edges.append((prev_node, node_name))
                
            # Set as previous for next iteration
            prev_node = node_name
            
            # Add input/output mappings
            if node_name in config.step_input_mappings:
                input_mappings[node_name] = config.step_input_mappings[node_name]
            if node_name in config.step_output_mappings:
                output_mappings[node_name] = config.step_output_mappings[node_name]
        
        # Set up edge-based config
        edge_config = EdgeBasedAgentConfig(
            nodes=nodes,
            edges=edges,
            entry_point=self.steps[0].get("name", "step_0") if self.steps else None,
            state_schema=config.state_schema,
            input_mappings=input_mappings,
            output_mappings=output_mappings
        )
        
        # Initialize as EdgeBasedAgent
        super().__init__(config=edge_config)


# Factory functions for easier agent creation
def create_simple_agent(llm_config: AugLLMConfig, state_schema=None):
    """
    Create a simple agent with a single LLM node.
    
    Args:
        llm_config: LLM configuration
        state_schema: Optional state schema
        
    Returns:
        SimpleAgent instance
    """
    # Convert state_schema to a proper model if needed
    if state_schema is None:
        from state_schema_manager import StateSchemaManager
        schema_manager = StateSchemaManager({"messages": []})
        state_schema = schema_manager.get_model()
    elif isinstance(state_schema, dict):
        from state_schema_manager import StateSchemaManager
        schema_manager = StateSchemaManager(state_schema)
        state_schema = schema_manager.get_model()
        
    config = SimpleAgentConfig(
        llm_config=llm_config,
        state_schema=state_schema
    )
    
    return SimpleAgent(config)


def create_sequential_agent(steps: List[Dict[str, Any]], state_schema=None):
    """
    Create an agent with a linear sequence of steps.
    
    Args:
        steps: List of step configurations
        state_schema: Optional state schema
        
    Returns:
        SequentialAgent instance
    """
    # Convert state_schema to a proper model if needed
    if state_schema is None:
        from state_schema_manager import StateSchemaManager
        schema_manager = StateSchemaManager({"messages": []})
        state_schema = schema_manager.get_model()
    elif isinstance(state_schema, dict):
        from state_schema_manager import StateSchemaManager
        schema_manager = StateSchemaManager(state_schema)
        state_schema = schema_manager.get_model()
        
    config = SequentialAgentConfig(
        steps=steps,
        state_schema=state_schema
    )
    
    return SequentialAgent(config)


def create_custom_agent(
    nodes: Dict[str, Union[AugLLMConfig, Callable]],
    edges: List[Tuple[str, str]],
    entry_point: Optional[str] = None,
    state_schema=None
):
    """
    Create a custom agent with arbitrary graph structure.
    
    Args:
        nodes: Dictionary of nodes
        edges: List of edges
        entry_point: Optional entry point (defaults to first node)
        state_schema: Optional state schema
        
    Returns:
        EdgeBasedAgent instance
    """
    # Convert state_schema to a proper model if needed
    if state_schema is None:
        from state_schema_manager import StateSchemaManager
        schema_manager = StateSchemaManager({"messages": []})
        state_schema = schema_manager.get_model()
    elif isinstance(state_schema, dict):
        from state_schema_manager import StateSchemaManager
        schema_manager = StateSchemaManager(state_schema)
        state_schema = schema_manager.get_model()
        
    config = EdgeBasedAgentConfig(
        nodes=nodes,
        edges=edges,
        entry_point=entry_point or next(iter(nodes)) if nodes else None,
        state_schema=state_schema
    )
    
    return EdgeBasedAgent(config)


# Register agent classes
register_agent(EdgeBasedAgentConfig, EdgeBasedAgent)
register_agent(SimpleAgentConfig, SimpleAgent)
register_agent(SequentialAgentConfig, SequentialAgent)
register_agent(SimpleAgentConfig,SimpleAgent)
```

```python
from typing import List, Dict, Any, Union, Optional, Callable, Tuple, Literal
from pydantic import BaseModel, Field
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.tools import Tool, BaseTool
from langchain_core.messages import AIMessage, HumanMessage, BaseMessage

from langgraph.graph import StateGraph, END

from src.haive.core.aug_llm.base import AugLLMConfig
from src.haive.core.models.llm.base import AzureLLMConfig
from src.haive.core.tools.search_tools import tavily_search_tool

# Import our edge-based components


# Example state schema
class AgentState(BaseModel):
    """Basic state schema for examples."""
    messages: List[Union[tuple, BaseMessage]] = Field(default_factory=list)
    research_data: Optional[Dict[str, Any]] = None
    analysis_results: Optional[Dict[str, Any]] = None
    plan: Optional[Dict[str, Any]] = None
    routing: Optional[str] = None


# -------------------------------------------------------------------------------
# Example 1: Simple Single-Node Agent
# -------------------------------------------------------------------------------

def example_simple_agent():
    """Example of a simple single-node agent."""
    
    # Create AugLLM config
    llm_config = AugLLMConfig(
        name="assistant",
        prompt_template=ChatPromptTemplate.from_messages([
            ("system", "You are a helpful AI assistant."),
            ("user", "{messages}")
        ]),
        llm_config=AzureLLMConfig(model="gpt-4o")
    )
    
    # Create simple agent
    agent = create_simple_agent(llm_config, AgentState)
    from IPython.display import Image, display

    try:
        display(Image(agent.app.get_graph().draw_mermaid_png()))
    except Exception:
        # This requires some extra dependencies and is optional
        pass
    # Run the agent
    result = agent.run("Tell me about deep learning.")
    
    return agent


# -------------------------------------------------------------------------------
# Example 2: Sequential Multi-Step Agent
# -------------------------------------------------------------------------------

def example_sequential_agent():
    """Example of an agent with sequential steps."""
    
    # Define step configurations
    steps = [
        {
            "name": "researcher",
            "llm_config": AugLLMConfig(
                name="researcher",
                prompt_template=ChatPromptTemplate.from_messages([
                    ("system", "You are a researcher. Gather key information about the topic."),
                    ("user", "{messages}")
                ]),
                llm_config=AzureLLMConfig(model="gpt-4o")
            )
        },
        {
            "name": "analyzer",
            "llm_config": AugLLMConfig(
                name="analyzer",
                prompt_template=ChatPromptTemplate.from_messages([
                    ("system", "You are an analyst. Analyze the research information."),
                    ("user", "{messages}")
                ]),
                llm_config=AzureLLMConfig(model="gpt-4o")
            )
        },
        {
            "name": "summarizer",
            "llm_config": AugLLMConfig(
                name="summarizer",
                prompt_template=ChatPromptTemplate.from_messages([
                    ("system", "You are a summarizer. Create a concise summary of the analysis."),
                    ("user", "{messages}")
                ]),
                llm_config=AzureLLMConfig(model="gpt-4o")
            )
        }
    ]
    
    # Create sequential agent
    agent = create_sequential_agent(steps, AgentState)
    from IPython.display import Image, display

    try:
        display(Image(agent.app.get_graph().draw_mermaid_png()))
    except Exception:
        # This requires some extra dependencies and is optional
        pass
    # Run the agent
    result = agent.run("Explain the impact of artificial intelligence on healthcare.")
    
    return agent


# -------------------------------------------------------------------------------
# Example 3: Custom Graph with Conditional Routing
# -------------------------------------------------------------------------------

def example_custom_agent_with_routing():
    """Example of a custom agent with conditional routing."""
    
    # Create state schema
    class RoutingState(BaseModel):
        messages: List[Union[tuple, BaseMessage]] = Field(default_factory=list)
        research_needed: bool = Field(default=False)
        topic_area: Optional[str] = None
        routing: Optional[str] = None
    
    # Create AugLLM configs
    initial_config = AugLLMConfig(
        name="initial",
        prompt_template=ChatPromptTemplate.from_messages([
            ("system", """You are an assistant that determines the type of query.
If the query requires research, set research_needed to true and set routing to "research".
Otherwise, set routing to "direct_response".

For research queries, also categorize the topic_area as one of: 
"science", "history", "technology", "business", or "other".
"""),
            ("user", "{messages}")
        ]),
        llm_config=AzureLLMConfig(model="gpt-4o")
    )
    
    research_config = AugLLMConfig(
        name="researcher",
        prompt_template=ChatPromptTemplate.from_messages([
            ("system", "You are a research assistant. Perform thorough research on the topic."),
            ("user", "Research this topic: {messages}")
        ]),
        llm_config=AzureLLMConfig(model="gpt-4o")
    )
    
    direct_config = AugLLMConfig(
        name="direct_responder",
        prompt_template=ChatPromptTemplate.from_messages([
            ("system", "You are a direct response assistant. Answer the query concisely."),
            ("user", "{messages}")
        ]),
        llm_config=AzureLLMConfig(model="gpt-4o")
    )
    
    specialized_config = AugLLMConfig(
        name="specialized",
        prompt_template=ChatPromptTemplate.from_messages([
            ("system", "You are a specialized expert in {topic_area}. Provide expert information."),
            ("user", "{messages}")
        ]),
        llm_config=AzureLLMConfig(model="gpt-4o")
    )
    
    # Create routing function
    def route_by_type(state):
        if hasattr(state, "routing") and state.routing:
            return state.routing
        if hasattr(state, "model_dump"):
            state_dict = state.model_dump()
            return state_dict.get("routing", "direct_response")
        return "direct_response"
    
    # Create specialized router
    def route_by_topic(state):
        if hasattr(state, "topic_area") and state.topic_area:
            return state.topic_area
        if hasattr(state, "model_dump"):
            state_dict = state.model_dump()
            return state_dict.get("topic_area", "other")
        return "other"
    
    # Define nodes and edges
    nodes = {
        "initial": initial_config,
        "research": research_config,
        "direct_response": direct_config,
        "specialized": specialized_config
    }
    
    # Define the basic edges
    edges = [
        # No standard edges - using conditional routing
    ]
    
    # Define conditional edges
    conditional_edges = [
        ("initial", route_by_type, {
            "research": "research",
            "direct_response": "direct_response",
            "default": "direct_response"
        }),
        ("research", route_by_topic, {
            "science": "specialized",
            "history": "specialized",
            "technology": "specialized",
            "business": "specialized",
            "other": "specialized",
            "default": "specialized"
        })
    ]
    
    # Create a custom agent with the graph
    agent = create_custom_agent(
        nodes=nodes,
        edges=edges,
        conditional_edges=conditional_edges,
        entry_point="initial",
        state_schema=RoutingState
    )
    
    # Run the agent
    result = agent.run("What will be the economic impact of quantum computing?")
    
    return agent


# -------------------------------------------------------------------------------
# Example 4: Agent with Tool Integration
# -------------------------------------------------------------------------------

def example_agent_with_tools():
    """Example of an agent that integrates tools."""
    
    # Create state schema
    class ToolState(BaseModel):
        messages: List[Union[tuple, BaseMessage]] = Field(default_factory=list)
        search_results: Optional[List[Dict[str, Any]]] = Field(default=None)
        routing: Optional[str] = None
    
    # Create tools
    tools = [
        tavily_search_tool,
        Tool(
            name="calculator",
            description="Perform mathematical calculations",
            func=lambda expr: str(eval(expr))
        )
    ]
    
    # Create an agent that can detect when to use tools
    detect_config = AugLLMConfig(
        name="detector",
        prompt_template=ChatPromptTemplate.from_messages([
            ("system", """You are an assistant that determines if tools are needed.
If the query requires search, respond with routing="search".
If the query requires calculation, respond with routing="calculator".
Otherwise, respond with routing="direct".
"""),
            ("user", "{messages}")
        ]),
        llm_config=AzureLLMConfig(model="gpt-4o")
    )
    
    # Create a tool handler
    def use_search(state):
        # Convert to dict if needed
        if hasattr(state, "model_dump"):
            state_dict = state.model_dump()
        else:
            state_dict = dict(state)
            
        # Use the search tool
        query = state_dict["messages"][-1][1] if state_dict["messages"] else ""
        search_results = tavily_search_tool.invoke(query)
        
        # Update state
        state_dict["search_results"] = search_results
        state_dict["messages"].append(("system", f"Search results: {search_results[:300]}..."))
        
        return state_dict
    
    def use_calculator(state):
        # Convert to dict if needed
        if hasattr(state, "model_dump"):
            state_dict = state.model_dump()
        else:
            state_dict = dict(state)
            
        # Extract expression from last message
        query = state_dict["messages"][-1][1] if state_dict["messages"] else ""
        
        # Extract mathematical expression (simplified)
        import re
        math_matches = re.findall(r'(\d+[\+\-\*/\d\.\(\)]+\d+)', query)
        
        if math_matches:
            expr = math_matches[0]
            try:
                result = eval(expr)
                state_dict["messages"].append(("system", f"Calculator result: {expr} = {result}"))
            except:
                state_dict["messages"].append(("system", f"Couldn't calculate: {expr}"))
        else:
            state_dict["messages"].append(("system", "No valid calculation found in query."))
            
        return state_dict
    
    # Create AugLLM for responding after tool use
    respond_config = AugLLMConfig(
        name="responder",
        prompt_template=ChatPromptTemplate.from_messages([
            ("system", "You are a helpful assistant. Use the tool results to provide a helpful response."),
            ("user", "{messages}")
        ]),
        llm_config=AzureLLMConfig(model="gpt-4o")
    )
    
    # Create routing function
    def route_to_tool(state):
        if hasattr(state, "routing") and state.routing:
            return state.routing
        if hasattr(state, "model_dump"):
            state_dict = state.model_dump()
            return state_dict.get("routing", "direct")
        return "direct"
    
    # Define nodes and edges
    nodes = {
        "detector": detect_config,
        "search": use_search,
        "calculator": use_calculator,
        "responder": respond_config
    }
    
    # Define edges
    edges = [
        ("search", "responder"),
        ("calculator", "responder")
    ]
    
    # Define conditional edges
    conditional_edges = [
        ("detector", route_to_tool, {
            "search": "search",
            "calculator": "calculator",
            "direct": "responder",
            "default": "responder"
        })
    ]
    
    # Create the agent
    agent = create_custom_agent(
        nodes=nodes,
        edges=edges,
        conditional_edges=conditional_edges,
        entry_point="detector",
        state_schema=ToolState
    )
    
    # Run the agent
    result = agent.run("What is the population of France divided by the population of Canada?")
    
    return agent


# -------------------------------------------------------------------------------
# Example 5: Using Direct GraphBuilder
# -------------------------------------------------------------------------------

def example_direct_graph_builder():
    """Example of using GraphBuilder directly."""
    
    # Create a state schema
    from state_schema_manager import StateSchemaManager
    schema_manager = StateSchemaManager({
        "messages": [],
        "plan": None,
        "analysis": None
    })
    state_schema = schema_manager.get_model()
    
    # Create a GraphBuilder
    builder = GraphBuilder(state_schema)
    
    # Create AugLLM configs
    planner_config = AugLLMConfig(
        name="planner",
        prompt_template=ChatPromptTemplate.from_messages([
            ("system", "You are a planning assistant. Create a high-level plan."),
            ("user", "{messages}")
        ]),
        llm_config=AzureLLMConfig(model="gpt-4o")
    )
    
    analyzer_config = AugLLMConfig(
        name="analyzer",
        prompt_template=ChatPromptTemplate.from_messages([
            ("system", "You are an analyst. Analyze the plan and provide feedback."),
            ("user", "{messages}")
        ]),
        llm_config=AzureLLMConfig(model="gpt-4o")
    )
    
    finalizer_config = AugLLMConfig(
        name="finalizer",
        prompt_template=ChatPromptTemplate.from_messages([
            ("system", "You are a finalizer. Create a final polished plan."),
            ("user", "{messages}")
        ]),
        llm_config=AzureLLMConfig(model="gpt-4o")
    )
    
    # Add nodes
    planner_node = builder.add_aug_llm_node("planner", planner_config)
    analyzer_node = builder.add_aug_llm_node("analyzer", analyzer_config)
    finalizer_node = builder.add_aug_llm_node("finalizer", finalizer_config)
    
    # Add edges
    builder.add_edge("planner", "analyzer")
    builder.add_edge("analyzer", "finalizer")
    builder.add_edge("finalizer", END)
    
    # Compile the graph
    app = builder.compile()
    
    # Run the app
    result = app.invoke({"messages": [("user", "Create a plan for a digital marketing campaign.")]})
    
    return app


# -------------------------------------------------------------------------------
# Run Examples
# -------------------------------------------------------------------------------

if __name__ == "__main__":
    print("Running Simple Agent Example...")
    example_simple_agent()
    
    print("\nRunning Sequential Agent Example...")
    example_sequential_agent()
    
    print("\nRunning Custom Agent with Routing Example...")
    example_custom_agent_with_routing()
    
    print("\nRunning Agent with Tools Example...")
    example_agent_with_tools()
    
    print("\nRunning Direct GraphBuilder Example...")
    example_direct_graph_builder()
```

```python
from typing import Tuple
def build_graph_from_configs(
    configs: Dict[str, AugLLMConfig],
    edges: List[Tuple[str, str]],
    conditional_edges: Optional[List[Tuple[str, Callable, Dict[str, str]]]] = None,
    state_schema: Optional[Union[Type[BaseModel], Dict[str, Any], StateSchemaManager]] = None
):
    """
    Builds a LangGraph from AugLLM configurations.
    
    Args:
        configs: Dictionary mapping node names to AugLLM configs
        edges: List of (from_node, to_node) tuples defining graph edges
        conditional_edges: List of (from_node, condition_fn, edge_map) for conditional routing
        state_schema: Base state schema to use (will be augmented as needed)
        
    Returns:
        Compiled LangGraph application
    """
    # Initialize state schema
    if state_schema is None:
        schema_manager = StateSchemaManager()
    elif isinstance(state_schema, dict):
        schema_manager = StateSchemaManager(state_schema)
    elif isinstance(state_schema, StateSchemaManager):
        schema_manager = state_schema
    else:
        schema_manager = StateSchemaManager(state_schema)
    
    # Augment schema for all configs
    for node_name, config in configs.items():
        augment_schema_for_llm(schema_manager, config)
    
    # Create state model
    StateModel = schema_manager.get_model()
    
    # Build graph
    graph = StateGraph(StateModel)
    
    # Add nodes to graph
    for node_name, config in configs.items():
        # Create node function
        # Find where this node should go (from edges list)
        next_nodes = [to_node for from_node, to_node in edges if from_node == node_name]
        goto = next_nodes[0] if next_nodes else END
        
        node_fn = create_node_function(config, goto)
        graph.add_node(node_name, node_fn)
    
    # Add entry point (first node)
    entry_node = list(configs.keys())[0]
    graph.set_entry_point(entry_node)
    
    # Add edges
    for from_node, to_node in edges:
        if to_node != END:  # We don't add explicit edges to END
            graph.add_edge(from_node, to_node)
    
    # Add conditional edges
    if conditional_edges:
        for from_node, condition_fn, edge_map in conditional_edges:
            graph.add_conditional_edges(from_node, condition_fn, edge_map)
    
    # Compile graph
    return graph.compile()
```

```python
import asyncio
import inspect
import textwrap
from typing import Any, Dict, List, Type, Union, Optional, Callable, get_type_hints
import libcst as cst
from pydantic import BaseModel, Field, create_model, ValidationError

# --- Import your actual models and AugLLM components ---
from src.haive.agents.plan_and_execute.models import Plan, Response, Act
from src.haive.core.aug_llm.base import AugLLMConfig, compose_runnable
from langchain.prompts import ChatPromptTemplate, MessagesPlaceholder

# Import Command and Send from langgraph.types
from langgraph.types import Command, Send

# Global flag for verbose model info
VERBOSE_MODEL_INFO = True

def verbose_model_info(model: Type[BaseModel]) -> str:
    """Return a verbose string with detailed info about a Pydantic model."""
    info_lines = []
    info_lines.append(f"Model Name: {model.__name__}")
    info_lines.append("Fields:")
    fields = model.model_fields if hasattr(model, "model_fields") else model.__fields__
    for field_name, field in fields.items():
        field_type = getattr(field, "annotation", "Unknown")
        required = getattr(field, "is_required", getattr(field, "required", "Unknown"))
        default = field.default if hasattr(field, "default") and field.default is not None else "None"
        try:
            description = field.field_info.description
        except AttributeError:
            description = getattr(field, "description", "")
        info_lines.append(f"  {field_name}: type={field_type}, required={required}, default={default}, description={description}")
    return "\n".join(info_lines)

# --- Define AugLLM configurations for two nodes ---
# Node1 uses the original Plan model.
EXECUTOR_PROMPT = "You are a helpful assistant"
EXECUTOR_PROMPT_TEMPLATE = ChatPromptTemplate.from_messages(
    [("system", EXECUTOR_PROMPT), MessagesPlaceholder(variable_name="messages")]
)
PLANNER_PROMPT = (
    "For the given objective, come up with a simple step by step plan. "
    "This plan should involve individual tasks that yield the correct answer. "
    "Do not add extra steps. The final step is the final answer."
)
PLANNER_PROMPT_TEMPLATE = ChatPromptTemplate.from_messages(
    [("system", PLANNER_PROMPT), ("placeholder", "{messages}")]
)
planner_aug_llm_config = AugLLMConfig(
    name="planner",
    prompt_template=PLANNER_PROMPT_TEMPLATE,
    structured_output_model=Plan,
)

# Node2 uses a modified output model.
class PlanWithExtra(Plan):
    # Note: We remove any default so that the schema meets OpenAI's requirements.
    notes: Optional[str] = Field(..., description="Extra notes for the plan")

planner_aug_llm_config_2 = AugLLMConfig(
    name="planner_extra",
    prompt_template=PLANNER_PROMPT_TEMPLATE,
    structured_output_model=PlanWithExtra,
)

# --- Helper: Extract I/O fields from an AugLLMConfig ---
def extract_io_fields(config: AugLLMConfig) -> Dict[str, Any]:
    """Extracts input and output field information from the given AugLLMConfig."""
    # Input field: usually taken from the prompt template.
    input_field = config.prompt_template.optional_variables[0] if hasattr(config.prompt_template, "optional_variables") and config.prompt_template.optional_variables else "input"
    # For output, if using structured_output_model, list its field names.
    output_fields = []
    if config.structured_output_model:
        model = config.structured_output_model
        fields = model.model_fields if hasattr(model, "model_fields") else model.__fields__
        output_fields = list(fields.keys())
    return {"input_field": input_field, "output_fields": output_fields}

# Print I/O info for both node configurations.
print("Node1 I/O fields:", extract_io_fields(planner_aug_llm_config))
print("Node2 I/O fields:", extract_io_fields(planner_aug_llm_config_2))

# --- Node Factory Functions (same as before) ---
def infer_input_field(config: AugLLMConfig) -> str:
    pt = config.prompt_template
    if hasattr(pt, "optional_variables") and pt.optional_variables:
        for var in pt.optional_variables:
            if var.lower() == "messages":
                return var
        return pt.optional_variables[0]
    elif hasattr(pt, "input_variables") and pt.input_variables:
        return pt.input_variables[0]
    return "input"

def infer_output_mapper(config: AugLLMConfig) -> Callable[[Any], Dict[str, Any]]:
    if config.structured_output_model:
        Model = config.structured_output_model
        if VERBOSE_MODEL_INFO:
            print("Using structured output model:")
            print(verbose_model_info(Model))
        def mapper(result: Any) -> Dict[str, Any]:
            if not isinstance(result, Model):
                try:
                    parsed = Model.model_validate(result)
                except Exception as e:
                    raise ValueError(f"Error parsing result with {Model}: {e}")
            else:
                parsed = result
            plan_val = getattr(parsed, "plan", parsed)
            steps_val = getattr(parsed, "steps", None)
            return {"plan": plan_val, "plan_steps": steps_val}
        return mapper
    elif config.output_parser:
        def mapper(result: Any) -> Dict[str, Any]:
            parsed = config.output_parser.parse(result)
            return {"plan": parsed, "plan_steps": parsed}
        return mapper
    else:
        def mapper(result: Any) -> Dict[str, Any]:
            return {"plan": result, "plan_steps": result}
        return mapper

def create_node_function(
    config: AugLLMConfig,
    command_goto: str = "execute_step",
    async_mode: bool = True,
) -> Callable[[Dict[str, Any]], Any]:
    output_model = config.structured_output_model or config.output_parser
    if output_model:
        print(f"Output model/parser detected: {output_model}")
    else:
        print("No structured output model or output parser provided.")
    node_runnable = compose_runnable(config)
    input_field = infer_input_field(config)
    if hasattr(config.prompt_template, "input_types") and input_field in config.prompt_template.input_types:
        expected_type = config.prompt_template.input_types[input_field]
        print(f"Inferred input field: '{input_field}' with expected type: {expected_type}")
    else:
        print(f"Inferred input field: '{input_field}' (no type info available)")
    def input_mapper(state: Dict[str, Any]) -> Dict[str, Any]:
        value = state.get(input_field)
        if value is None:
            raise ValueError(f"State does not contain required key '{input_field}'")
        if input_field.lower() == "messages":
            if isinstance(value, list):
                if not all(isinstance(x, tuple) and len(x) == 2 for x in value):
                    value = [("user", str(value))]
            else:
                value = [("user", str(value))]
            return {"messages": value}
        return {input_field: value}
    output_mapper = infer_output_mapper(config)
    if async_mode:
        async def node(state: Dict[str, Any]):
            payload = input_mapper(state)
            result = await node_runnable.ainvoke(payload)
            out = output_mapper(result)
            return Command(
                update={"plan": out["plan"]},
                goto=command_goto,
                resume={"plan": out["plan_steps"]}
            )
        return node
    else:
        def node(state: Dict[str, Any]):
            payload = input_mapper(state)
            result = node_runnable.invoke(payload)
            out = output_mapper(result)
            return Command(
                update={"plan": out["plan"]},
                goto=command_goto,
                resume={"plan": out["plan_steps"]}
            )
        return node

# --- Glue Function (async version) ---
async def glue_nodes_async(
    nodes: List[Callable[[Dict[str, Any]], Command]],
    initial_state: Dict[str, Any]
) -> Command:
    state = initial_state.copy()
    last_command = None
    for i, node in enumerate(nodes):
        cmd = await node(state)
        if i < len(nodes) - 1:
            next_node_name = nodes[i+1].__name__
            cmd = Command(update=cmd.update, goto=next_node_name, resume=cmd.resume)
        last_command = cmd
        if last_command.update:
            state.update(last_command.update)
    return last_command

# ------------------------------
# Use your existing StateSchemaManager (assumed defined above)
# ------------------------------
schema_data = {
    "messages": [("user", "default input")],
    "extra": "initial extra"
}
schema_manager = StateSchemaManager(schema_data, name="ExampleState")
print("State Schema:")
schema_manager.pretty_print()

# Get the state model
StateModel = schema_manager.get_model()

# Create node functions from different AugLLM configs.
node1 = schema_manager.create_node_from_config(planner_aug_llm_config, async_mode=True)
node2 = schema_manager.create_node_from_config(planner_aug_llm_config_2, async_mode=True)

# ------------------------------
# Test Execution: Chain nodes to build state from node I/O
# ------------------------------
async def test_state_graph():
    # Start with a minimal state.
    initial_state = {"messages": [("user", "Plan a dinner party for 10")]}
    # Chain node1 and node2 together.
    final_command = await glue_nodes_async([node1, node2], initial_state)
    # Merge final command update into the state.
    final_state = initial_state.copy()
    if final_command.update:
        final_state.update(final_command.update)
    # Validate and pretty-print the final state using the state model.
    state_instance = StateModel.model_validate(final_state)
    print("Final State Model:")
    print(state_instance.model_dump_json(indent=4))

if __name__ == "__main__":
    import nest_asyncio
    nest_asyncio.apply()
    print("=== Testing State Graph Execution ===")
    asyncio.run(test_state_graph())
```

```python
from typing import Dict, Any, List, Union, Optional, Type, Callable, Literal
from pydantic import BaseModel, Field
import uuid

from langgraph.graph import StateGraph, END
from langgraph.prebuilt import ToolNode
from langgraph.types import Command, Send

from langchain_core.tools import BaseTool, StructuredTool, Tool
from langchain_core.messages import SystemMessage, filter_messages
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

from src.haive.agents.base import AgentArchitectureConfig, AgentArchitecture, register_agent
from src.haive.core.aug_llm.base import AugLLMConfig, compose_runnable
from src.haive.core.models.llm.base import AzureLLMConfig
from src.haive.core.tools.search_tools import tavily_search_tool

# Import BaseAgent and BaseAgentConfig
#from base_agent_integration import BaseAgent, BaseAgentConfig, create_node_function

# Default ReAct prompt template
DEFAULT_REACT_PROMPT = ChatPromptTemplate.from_messages([
    ("system", """You are a helpful AI assistant. 
To solve complex problems, you can use tools to help you.

When you need to use a tool, respond in this format:
<tool_calls>
<tool_name>The name of the tool</tool_name>
<tool_input>The input to the tool, matching the tool's JSON schema</tool_input>
</tool_calls>

When you don't need a tool, respond directly to the user's query with helpful information.
    """),
    MessagesPlaceholder(variable_name="messages")
])


# Utility function to determine whether to continue execution
def should_continue(state: Dict[str, Any]) -> str:
    """
    Determine if we should route to tool execution or end.
    Checks if the last message contains tool calls.
    """
    # Get last message
    messages = filter_messages(state.get("messages", []), exclude_types=[SystemMessage])
    if not messages:
        return "end"
        
    last_message = messages[-1]
    
    # Check for tool calls in attribute
    if hasattr(last_message, "tool_calls") and last_message.tool_calls:
        return "continue"
        
    # Check for tool calls in content (text-based format)
    if hasattr(last_message, "content") and "<tool_calls>" in last_message.content:
        return "continue"
        
    return "end"


class ReactAgentConfig(BaseAgentConfig):
    """
    Configuration for a ReAct agent that can use tools.
    Extends BaseAgentConfig with tool-specific configurations.
    """
    tools: List[Union[Tool, BaseTool, StructuredTool]] = Field(
        default_factory=lambda: [tavily_search_tool],
        description="The tools available to the agent."
    )
    tool_node_name: str = Field(
        default="tool_node",
        description="Name for the tool execution node."
    )
    routing_function: Callable = Field(
        default=should_continue,
        description="Function that determines routing based on state."
    )
    routing_map: Dict[str, Union[str, Literal["END"]]] = Field(
        default_factory=lambda: {"continue": "tool_node", "end": END},
        description="Mapping from routing keys to target nodes."
    )
    
    class Config:
        arbitrary_types_allowed = True
    
    def build_agent(self) -> "ReactAgent":
        """Build the ReactAgent from this configuration."""
        return ReactAgent(config=self)


class ReactAgent(BaseAgent):
    """
    An agent that can use tools, implementing the ReAct pattern.
    Extends BaseAgent with tool-specific workflows.
    """
    def __init__(self, config: ReactAgentConfig):
        self.tools = config.tools
        self.tool_node_name = config.tool_node_name
        self.routing_function = config.routing_function
        self.routing_map = config.routing_map
        
        # Make sure tools are attached to the engine if it's an AugLLMConfig
        if isinstance(config.engine, AugLLMConfig):
            # Check if tools are not already set in the engine
            if not config.engine.tools:
                config.engine.tools = self.tools
        
        # Initialize using the BaseAgent's __init__
        super().__init__(config)
    
    def setup_workflow(self):
        """
        Override setup_workflow to create a ReAct pattern with tools.
        
        The workflow consists of two nodes:
        1. Agent node: Reasons and decides when to use tools
        2. Tool node: Executes tools when the agent requests them
        """
        # Create the agent node using BaseAgent's approach, but without END edge
        node_fn = create_node_function(
            config=self.config.engine if isinstance(self.config.engine, AugLLMConfig) else AugLLMConfig(),
            input_mapping=self.input_mapping,
            output_mapping=self.output_mapping
            # No explicit command_goto - we'll use conditional routing
        )
        
        # Create the tool node
        tool_node = ToolNode(self.tools)
        
        # Add nodes to the graph
        self.graph.add_node(self.node_name, node_fn)
        self.graph.add_node(self.tool_node_name, tool_node)
        
        # Set the entry point to the agent node
        self.graph.set_entry_point(self.node_name)
        
        # Add conditional routing from agent to tool/end
        self.graph.add_conditional_edges(
            self.node_name,
            self.routing_function,
            self.routing_map
        )
        
        # Add edge from tool back to agent
        self.graph.add_edge(self.tool_node_name, self.node_name)


# Register the agent
register_agent(ReactAgentConfig, ReactAgent)


# Factory function for easy agent creation
def create_react_agent(
    llm_config: Optional[AugLLMConfig] = None,
    tools: Optional[List[Union[Tool, BaseTool, StructuredTool]]] = None,
    state_schema=None,
    prompt_template=None
):
    """
    Create a ReactAgent with the specified configuration.
    
    Args:
        llm_config: Optional LLM configuration (defaults to GPT-4o)
        tools: Optional list of tools (defaults to tavily_search_tool)
        state_schema: Optional custom state schema
        prompt_template: Optional custom prompt template
        
    Returns:
        Initialized ReactAgent instance
    """
    #from state_schema_manager import StateSchemaManager
    
    # Set up default tools
    default_tools = [tavily_search_tool]
    tools = tools or default_tools
    
    # Set up default llm_config
    if llm_config is None:
        llm_config = AugLLMConfig(
            llm_config=AzureLLMConfig(model="gpt-4o"),
            prompt_template=prompt_template or DEFAULT_REACT_PROMPT,
            tools=tools
        )
    else:
        # Ensure prompt template is set
        if llm_config.prompt_template is None:
            llm_config.prompt_template = prompt_template or DEFAULT_REACT_PROMPT
        # Ensure tools are set
        if llm_config.tools is None:
            # IMPORTANT: Convert any StructuredTool to Tool to avoid validation issues
            if tools:
                safe_tools = []
                for tool in tools:
                    if isinstance(tool, StructuredTool):
                        # Create a Tool that wraps the StructuredTool functionality
                        safe_tools.append(Tool(
                            name=tool.name,
                            description=tool.description,
                            func=tool.func
                        ))
                    else:
                        safe_tools.append(tool)
                tools = safe_tools
            else:
                tools = default_tools
            llm_config.tools = tools
            
    # Set up state schema
    if state_schema is None:
        schema_manager = StateSchemaManager({"messages": []})
        state_schema = schema_manager.get_model()
    elif isinstance(state_schema, dict):
        schema_manager = StateSchemaManager(state_schema)
        state_schema = schema_manager.get_model()
    
    # Create configuration
    config = ReactAgentConfig(
        engine=llm_config,
        tools=tools,
        state_schema=state_schema,
        node_name="agent"  # Use "agent" instead of default "agent_node"
    )
    
    return ReactAgent(config)


# Examples of using the ReactAgent

def example_simple_react_agent():
    """Example of a simple ReactAgent with default tools."""
    print("\n=== Simple ReactAgent with Default Tools ===")
    
    # Create a basic ReactAgent with default settings
    agent = create_react_agent()
    
    # Run the agent
    print("Running agent with query: 'What is the population of Paris, France?'")
    result = agent.run("What is the population of Paris, France?")
    
    return agent


def example_react_agent_with_custom_tools():
    """Example of a ReactAgent with custom tools."""
    print("\n=== ReactAgent with Custom Tools ===")
    
    # Define custom tools
    calculator_tool = Tool(
        name="calculator",
        description="Useful for performing mathematical calculations",
        func=lambda x: f"Result: {eval(x)}"
    )
    
    # Create a ReactAgent with custom tools
    agent = create_react_agent(
        tools=[tavily_search_tool, calculator_tool]
    )
    
    # Run the agent
    print("Running agent with query: 'What is 1234 * 5678?'")
    result = agent.run("What is 1234 * 5678?")
    
    return agent


def example_react_agent_with_custom_state():
    """Example of a ReactAgent with custom state schema."""
    print("\n=== ReactAgent with Custom State Schema ===")
    
    # Define custom state schema
    from state_schema_manager import StateSchemaManager
    
    schema_manager = StateSchemaManager({
        "messages": [],
        "search_history": [],
        "calculation_results": []
    })
    
    # Create a ReactAgent with custom state schema
    agent = create_react_agent(
        state_schema=schema_manager.get_model()
    )
    
    # Run the agent
    print("Running agent with query: 'What is the GDP of Japan divided by its population?'")
    result = agent.run("What is the GDP of Japan divided by its population?")
    
    return agent


# Run examples if script is executed directly
if __name__ == "__main__":
    example_simple_react_agent()
    example_react_agent_with_custom_tools()
    example_react_agent_with_custom_state()
```

```python
from typing import Dict, Any, List, Union, Optional, Type, Callable, Literal
from pydantic import BaseModel, Field
import uuid

from langgraph.graph import StateGraph, END
from langgraph.prebuilt import ToolNode
from langgraph.types import Command

from langchain_core.tools import BaseTool, Tool
from langchain_core.messages import SystemMessage, filter_messages
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

from src.haive.agents.base import AgentArchitectureConfig, AgentArchitecture, register_agent
from src.haive.core.aug_llm.base import AugLLMConfig, compose_runnable
from src.haive.core.models.llm.base import AzureLLMConfig
from src.haive.core.tools.search_tools import tavily_search_tool


# Default ReAct prompt template
DEFAULT_REACT_PROMPT = ChatPromptTemplate.from_messages([
    ("system", """You are a helpful AI assistant. 
To solve complex problems, you can use tools to help you.

When you need to use a tool, respond in this format:
<tool_calls>
<tool_name>The name of the tool</tool_name>
<tool_input>The input to the tool, matching the tool's JSON schema</tool_input>
</tool_calls>

When you don't need a tool, respond directly to the user's query with helpful information.
    """),
    MessagesPlaceholder(variable_name="messages")
])


# Utility function to determine whether to continue execution
def should_continue(state: Dict[str, Any]) -> str:
    """
    Determine if we should route to tool execution or end.
    Checks if the last message contains tool calls.
    """
    # Get last message
    messages = filter_messages(state.get("messages", []), exclude_types=[SystemMessage])
    if not messages:
        return "end"
        
    last_message = messages[-1]
    
    # Check for tool calls in attribute
    if hasattr(last_message, "tool_calls") and last_message.tool_calls:
        return "continue"
        
    # Check for tool calls in content (text-based format)
    if hasattr(last_message, "content") and "<tool_calls>" in last_message.content:
        return "continue"
        
    return "end"


class SimpleReactAgentConfig(AgentArchitectureConfig):
    """
    Configuration for a simplified ReactAgent that avoids Pydantic validation issues.
    """
    engine: AugLLMConfig = Field(
        default_factory=lambda: AugLLMConfig(
            llm_config=AzureLLMConfig(model="gpt-4o"),
            prompt_template=DEFAULT_REACT_PROMPT
        ),
        description="LLM configuration for the ReactAgent."
    )
    # Tools are stored as simple attributes without Pydantic validation
    node_name: str = Field(
        default="agent",
        description="Name for the agent reasoning node."
    )
    tool_node_name: str = Field(
        default="tool_node",
        description="Name for the tool execution node."
    )
    should_setup_workflow: bool = Field(
        default=True, 
        description="Whether to set up the workflow."
    )
    should_compile: bool = Field(
        default=True, 
        description="Whether to compile the graph."
    )
    should_visualize_graph: bool = Field(
        default=False, 
        description="Whether to visualize the graph."
    )
    
    class Config:
        # Allow arbitrary types to prevent validation issues
        arbitrary_types_allowed = True
    
    def build_agent(self) -> "SimpleReactAgent":
        """Build the SimpleReactAgent from this configuration."""
        return SimpleReactAgent(config=self)


class SimpleReactAgent(AgentArchitecture):
    """
    A simplified implementation of a ReAct agent that avoids Pydantic validation issues.
    """
    def __init__(self, config: SimpleReactAgentConfig):
        # Set tools as instance attribute, not part of the config
        self.tools = []  # Will be set by create_simple_react_agent
        
        self.node_name = config.node_name
        self.tool_node_name = config.tool_node_name
        
        # Initialize using the parent class
        super().__init__(config)
    
    def setup_workflow(self):
        """
        Set up a ReAct workflow with agent and tool nodes.
        """
        # Create agent node
        engine = self.config.engine
        runnable = compose_runnable(engine)
        
        def agent_node(state):
            state_dict = state.model_dump() if hasattr(state, "model_dump") else state
            result = runnable.invoke({"messages": state_dict.get("messages", [])})
            return Command(update={"messages": state_dict["messages"] + [result]})
        
        # Create tool node
        tool_node = ToolNode(self.tools)
        
        # Add nodes to graph
        self.graph.add_node(self.node_name, agent_node)
        self.graph.add_node(self.tool_node_name, tool_node)
        
        # Set entry point
        self.graph.set_entry_point(self.node_name)
        
        # Add conditional routing
        self.graph.add_conditional_edges(
            self.node_name,
            should_continue,
            {
                "continue": self.tool_node_name,
                "end": END
            }
        )
        
        # Add edge from tool back to agent
        self.graph.add_edge(self.tool_node_name, self.node_name)


# Factory function for easy agent creation
def create_simple_react_agent(
    llm_config: Optional[AugLLMConfig] = None,
    tools: Optional[List[Union[Tool, BaseTool]]] = None,
    state_schema=None,
    prompt_template=None
):
    """
    Create a SimpleReactAgent with the specified configuration.
    Avoids Pydantic validation issues by handling tools separately.
    
    Args:
        llm_config: Optional LLM configuration (defaults to GPT-4o)
        tools: Optional list of tools (defaults to tavily_search_tool)
        state_schema: Optional custom state schema
        prompt_template: Optional custom prompt template
        
    Returns:
        Initialized SimpleReactAgent instance
    """
    #from state_schema_manager import StateSchemaManager
    
    # Set up default tools
    default_tools = [tavily_search_tool]
    tools = tools or default_tools
    
    # Set up default llm_config
    if llm_config is None:
        llm_config = AugLLMConfig(
            llm_config=AzureLLMConfig(model="gpt-4o"),
            prompt_template=prompt_template or DEFAULT_REACT_PROMPT,
            tools=tools
        )
    else:
        # Ensure prompt template is set
        if llm_config.prompt_template is None:
            llm_config.prompt_template = prompt_template or DEFAULT_REACT_PROMPT
        # Ensure tools are set
        if llm_config.tools is None:
            llm_config.tools = tools
    
    # Set up state schema
    if state_schema is None:
        schema_manager = StateSchemaManager({"messages": []})
        state_schema = schema_manager.get_model()
    elif isinstance(state_schema, dict):
        schema_manager = StateSchemaManager(state_schema)
        state_schema = schema_manager.get_model()
    
    # Create configuration (without tools)
    config = SimpleReactAgentConfig(
        engine=llm_config,
        state_schema=state_schema
    )
    
    # Create agent and set tools directly as instance attribute
    agent = SimpleReactAgent(config)
    agent.tools = tools  # Set tools directly, bypassing Pydantic validation
    
    return agent


# Example usage
def example_simple_react_agent():
    """Example of a simple ReactAgent with default tools."""
    print("\n=== Simple ReactAgent with Default Tools ===")
    
    # Create a basic ReactAgent with default settings
    agent = create_simple_react_agent()
    
    # Run the agent
    print("Running agent with query: 'What is the population of Paris, France?'")
    result = agent.run("What is the population of Paris, France?")
    
    return agent


def example_react_agent_with_custom_tools():
    """Example of a ReactAgent with custom tools."""
    print("\n=== ReactAgent with Custom Tools ===")
    
    # Define custom tools
    calculator_tool = Tool(
        name="calculator",
        description="Useful for performing mathematical calculations",
        func=lambda x: f"Result: {eval(x)}"
    )
    
    # Create a ReactAgent with custom tools
    agent = create_simple_react_agent(
        tools=[tavily_search_tool, calculator_tool]
    )
    
    # Run the agent
    print("Running agent with query: 'What is 1234 * 5678?'")
    result = agent.run("What is 1234 * 5678?")
    
    return agent


def example_react_agent_with_custom_state():
    """Example of a ReactAgent with custom state schema."""
    print("\n=== ReactAgent with Custom State Schema ===")
    
    # Define custom state schema
    from state_schema_manager import StateSchemaManager
    
    schema_manager = StateSchemaManager({
        "messages": [],
        "search_history": [],
        "calculation_results": []
    })
    
    # Create a ReactAgent with custom state schema
    agent = create_simple_react_agent(
        state_schema=schema_manager.get_model()
    )
    
    # Run the agent
    print("Running agent with query: 'What is the GDP of Japan divided by its population?'")
    result = agent.run("What is the GDP of Japan divided by its population?")
    
    return agent


# Run examples if script is executed directly
if __name__ == "__main__":
    example_simple_react_agent()
    example_react_agent_with_custom_tools()
    example_react_agent_with_custom_state()
```
